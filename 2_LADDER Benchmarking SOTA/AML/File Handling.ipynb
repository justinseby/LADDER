```python
import pandas as pd

with_context = pd.read_csv("AMLWithContext_Combined_Annotations.csv")
without_context = pd.read_csv("AMLWithoutContext_Combined_Annotations.csv")

with_context = with_context[
    ["Geneset", "Hu Annotation", "GeneAgent Annotation", "Our Annotation"]
].copy()

without_context = without_context[
    ["Geneset", "Hu Annotation", "GeneAgent Annotation"]
].copy()

with_context = with_context.rename(columns={
    "Hu Annotation": "Hu Annotation (with context)",
    "GeneAgent Annotation": "GeneAgent Annotation (with context)"
})

without_context = without_context.rename(columns={
    "Hu Annotation": "Hu Annotation (without context)",
    "GeneAgent Annotation": "GeneAgent Annotation (without context)"
})

combined = with_context.merge(
    without_context,
    on="Geneset",
    how="outer"
)

combined = combined[
    [
        "Geneset",
        "Hu Annotation (with context)",
        "GeneAgent Annotation (with context)",
        "Hu Annotation (without context)",
        "GeneAgent Annotation (without context)",
        "Our Annotation"
    ]
]

combined.to_csv(
    "AML_Context_vs_NoContext_Annotations.csv",
    index=False
)

print(f"Saved {len(combined)} rows to AML_Context_vs_NoContext_Annotations.csv")
```

```python
import pandas as pd

# Load files
annotations = pd.read_csv("AML_Context_vs_NoContext_Annotations.csv")
msigdb = pd.read_csv("AML_msigdb_descriptions.csv")

msigdb = msigdb[
    [
        "Geneset",
        "MSigDB_Brief_Description",
        "MSigDB_Full_Description"
    ]
]

merged = annotations.merge(
    msigdb,
    on="Geneset",
    how="left"
)

cols = [
    "Geneset",
    "MSigDB_Brief_Description",
    "MSigDB_Full_Description",
    "Hu Annotation (with context)",
    "GeneAgent Annotation (with context)",
    "Hu Annotation (without context)",
    "GeneAgent Annotation (without context)",
    "Our Annotation"
]

cols = [c for c in cols if c in merged.columns]

merged = merged[cols]

# Save
merged.to_csv(
    "AML_All_Annotations_With_MSigDB.csv",
    index=False
)

print(
    f"Saved {len(merged)} rows to AML_All_Annotations_With_MSigDB.csv"
)
```
