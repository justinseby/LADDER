```python
class OptimizedPathwayAnalyzer:
    def __init__(self):
        self.api_key = None
        self.api_url = "https://api.deepseek.com/v1/chat/completions"
        self.checkpoint_file = f"checkpoint_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
        self.results_csv = f"results_{datetime.now().strftime('%Y%m%d_%H%M%S')}.csv"
        self.pathway_gene_mapping = {}
        
    def set_api_key(self, api_key: str = None):
        """
        Set the Deepseek API key
        """
        if api_key:
            self.api_key = api_key
        elif os.environ.get("DEEPSEEK_API_KEY"):
            self.api_key = os.environ.get("DEEPSEEK_API_KEY")
        else:
            raise ValueError("No Deepseek API key provided. Set it via argument or DEEPSEEK_API_KEY environment variable.")
    
    def _create_system_prompt(self) -> str:
        return """You are a bioinformatics expert. Focus on Acute myeloid leukemia (AML). Provide detailed pathway analysis with scientific literature support. Always cite specific papers when discussing pathway relationships."""
    
    def _create_all_in_one_prompt(self, genes: List[str], enrichment_pathways: List[str]) -> str:
        enrichment_section = ""
        if enrichment_pathways and len(enrichment_pathways) > 0:
            enrichment_section = "\nEnrichment Analysis Results:\n" + "\n".join(f"- {pathway}" for pathway in enrichment_pathways)
        
        genes_str = ", ".join(genes)
        
        enrichment_instruction = "Using the enrichment pathways above (if any), provide an analysis of the gene set"
        
        return f"""You are a bioinformatics expert conducting pathway analysis of gene sets. For the gene set: [{genes_str}]{enrichment_section}

Please perform analysis in clearly labeled sections:

===== SECTION 1: ANALYSIS WITH ENRICHMENT =====
{enrichment_instruction}:
1. Propose a concise, descriptive name for the biological process
2. Assign a confidence score (0.00-1.00) for this process
3. Provide explicit reasoning for choosing this pathway/process
4. List the specific genes from the gene set that contribute to the process
5. IMPORTANT: A minimum of TWO genes from the gene set MUST be associated with a common biological process to annotate it as a valid pathway
6. If there isn't sufficient evidence or fewer than two genes are associated with a common pathway, label as "Unknown Pathway" with a low confidence score

Output format for this section:
PROCESS WITH ENRICHMENT: [Process Name] ([Confidence Score])

PATHWAY REASONING WITH ENRICHMENT:
[Your explicit reasoning for choosing this pathway/process]

CONTRIBUTING GENES WITH ENRICHMENT:
[Comma-separated list of contributing genes]

ANALYSIS TEXT WITH ENRICHMENT:
[Detailed analysis text explaining the biological significance and mechanisms of this process, without citations]

===== SECTION 2: ANALYSIS WITHOUT ENRICHMENT =====
Based SOLELY on your knowledge of these genes, without considering the enrichment results:
1. Propose a concise, descriptive name for the biological process
2. Assign a confidence score (0.00-1.00) for this process
3. List the specific genes from the gene set that are involved in the process
4. IMPORTANT: A minimum of TWO genes from the gene set MUST be associated with a common biological process to annotate it as a valid pathway
5. IMPORTANT: If there isn't sufficient evidence or fewer than two genes share a common biological function, label as "Unknown Pathway" with a low confidence score

Output format for this section:
PROCESS WITHOUT ENRICHMENT: [Process Name] ([Confidence Score])

CONTRIBUTING GENES WITHOUT ENRICHMENT:
[Comma-separated list of contributing genes]

PATHWAY REASONING WITHOUT ENRICHMENT:
[Your explicit reasoning for choosing this pathway/process]

ANALYSIS TEXT WITHOUT ENRICHMENT:
[Detailed analysis text explaining the biological significance and mechanisms of this process, without citations]

===== SECTION 3: FINAL PROCESS SELECTION =====
Compare both analyses and provide:
1. Which process (with or without enrichment) has higher confidence and why
2. Final reasoning for the chosen process

Output format for this section:
FINAL PROCESS REASONING:
[Detailed explanation of why you chose the final process, comparing both analyses and explaining which one provides stronger evidence]

Analytical Guidelines:
- Be concise and avoid unnecessary words
- Be factual without editorializing
- Be specific, avoiding overly general statements
- Avoid listing individual protein facts
- Group proteins by similar functions
- Discuss their interplay, synergistic or antagonistic effects
- Focus on functional integration within the system
- Include at least 2-3 specific paper citations when discussing established pathway relationships

Confidence Score Instructions:
- Assign a score from 0.00 to 1.00
- 0.00 indicates lowest confidence
- 1.00 reflects highest confidence
- Base the score on the proportion of genes participating in the identified process"""
    
    def _make_deepseek_request(self, prompt: str) -> str:
        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type": "application/json"
        }
        
        data = {
            "model": "deepseek-chat",
            "messages": [
                {"role": "system", "content": self._create_system_prompt()},
                {"role": "user", "content": prompt}
            ],
            "temperature": 0,
            "max_tokens": 4000
        }
        
        response = requests.post(self.api_url, headers=headers, json=data)
        
        if response.status_code == 200:
            result = response.json()
            return result['choices'][0]['message']['content']
        else:
            raise Exception(f"Deepseek API request failed: {response.status_code} - {response.text}")
    
    def parse_communities(self, input_text: str) -> Dict[str, List[str]]:
        communities = {}
        for line in input_text.strip().split('\n'):
            line = line.strip()
            if 'Refined Community' in line:
                match = re.search(r'Refined Community (\d+): \[(.*?)\]', line)
                if match:
                    community_num = match.group(1)
                    genes_str = match.group(2)
                    genes = [g.strip("'") for g in genes_str.split(', ')]
                    communities[community_num] = genes
        return communities
    
    def perform_enrichment(self, gene_list: List[str]) -> List[str]:

        try:
            databases = ['GO_Biological_Process_2021', 'Reactome_2022', 'KEGG_2021_Human']
            
            all_results = []
            for database in databases:
                enr = gp.enrichr(gene_list=gene_list,
                                gene_sets=[database],
                                organism='Human',
                                outdir=None,
                                no_plot=True,
                                cutoff=0.01)
                
                results_df = enr.results
                if not results_df.empty:
                    results_df = results_df.sort_values('Adjusted P-value')
                    for _, row in results_df.head(5).iterrows():
                        all_results.append(row['Term'])
            
            return list(set(all_results))
        
        except Exception as e:
            print(f"Enrichment analysis failed: {str(e)}")
            return []
    
    def _clean_markdown_formatting(self, text: str) -> str:

        if not text or not isinstance(text, str):
            return text
            
        text = re.sub(r'\*([^*]+)\*', r'\1', text)
        
        text = re.sub(r'\*\*([^*]+)\*\*', r'\1', text)
        
        text = re.sub(r'(?<!\w)\*(?!\w)', '', text)
        
        return text.strip()
    
    def _extract_section(self, text: str, section_name: str) -> str:

        patterns = [
            rf"{re.escape(section_name)}:\s*(.*?)(?=\n\n[\w\s]+:|===|$)",  # Main pattern
            rf"{re.escape(section_name)}:\s*(.*?)(?=\n[A-Z][A-Z\s]+:|$)",
            rf"{re.escape(section_name)}:\s*(.*?)(?=\n\*\*|$)" 
        ]
        
        for pattern in patterns:
            match = re.search(pattern, text, re.DOTALL | re.IGNORECASE)
            if match:
                result = match.group(1).strip()
                if result:
                    return self._clean_markdown_formatting(result)
        
        return f"{section_name} not found"
    
    def _extract_process_info(self, text: str, prefix: str) -> Tuple[str, float]:

        patterns = [
            rf"PROCESS {prefix} ENRICHMENT:\s*([^(]+?)\s*\(([0-9.]+)\)",
            rf"PROCESS {prefix} ENRICHMENT:\s*([^(]*?)\s*\(\s*([0-9.]+)\s*\)",
            rf"PROCESS {prefix} ENRICHMENT:\s*(.*?)\s*\(\s*Confidence Score:\s*([0-9.]+)\s*\)",
            rf"PROCESS {prefix} ENRICHMENT:\s*(.*?)\s*\*\*Confidence Score:\s*([0-9.]+)\*\*",
            rf"PROCESS {prefix} ENRICHMENT:\s*(.*?)\s*\(\s*\*\*Confidence Score:\s*([0-9.]+)\*\*\s*\)"
        ]
        
        for pattern in patterns:
            match = re.search(pattern, text, re.IGNORECASE | re.DOTALL)
            if match:
                process_name = match.group(1).strip()
                try:
                    confidence_score = float(match.group(2))
                    process_name = self._clean_markdown_formatting(process_name)
                    process_name = process_name.strip()
                    if process_name and confidence_score >= 0:
                        return process_name, confidence_score
                except ValueError:
                    continue
        
        simple_pattern = rf"PROCESS {prefix} ENRICHMENT:\s*(.*?)(?=\n|$)"
        match = re.search(simple_pattern, text, re.IGNORECASE)
        if match:
            process_name = match.group(1).strip()
            process_name = self._clean_markdown_formatting(process_name)
            process_name = re.sub(r'\s*\([^)]*$', '', process_name)
            process_name = process_name.strip()
            return process_name, 0.0
        
        return f"Unknown Process {prefix} Enrichment", 0.0
    
    def _extract_final_process(self, text: str) -> str:

        match = re.search(r"FINAL PROCESS REASONING:(.*?)(?=\n\n|$)", text, re.DOTALL | re.IGNORECASE)
        if match:
            result = match.group(1).strip()
            return self._clean_markdown_formatting(result)
        return "Final process reasoning not found"
    
    def analyze_community_optimized(self, comm_id: str, genes: List[str]) -> Dict:

        results = {
            "Community": comm_id,
            "Genes": genes,
            "Genes_String": ", ".join(genes)
        }
        
        try:
            enrichment_pathways = self.perform_enrichment(genes)
            results["Enrichment_Pathways"] = enrichment_pathways
            
            all_in_one_prompt = self._create_all_in_one_prompt(genes, enrichment_pathways)
            
            full_analysis = self._make_deepseek_request(all_in_one_prompt)
            results["Full_Analysis"] = full_analysis

            process_with_enrichment, confidence_with_enrichment = self._extract_process_info(full_analysis, "WITH")
            results["Process_With_Enrichment"] = process_with_enrichment
            results["Confidence_With_Enrichment"] = confidence_with_enrichment

            process_without_enrichment, confidence_without_enrichment = self._extract_process_info(full_analysis, "WITHOUT")
            results["Process_Without_Enrichment"] = process_without_enrichment
            results["Confidence_Without_Enrichment"] = confidence_without_enrichment
            
            try:
                results["Pathway_Reasoning_With_Enrichment"] = self._extract_section(full_analysis, "PATHWAY REASONING WITH ENRICHMENT")
            except Exception as e:
                results["Pathway_Reasoning_With_Enrichment"] = f"Extraction failed: {str(e)}"
                
            try:
                results["Pathway_Reasoning_Without_Enrichment"] = self._extract_section(full_analysis, "PATHWAY REASONING WITHOUT ENRICHMENT")
            except Exception as e:
                results["Pathway_Reasoning_Without_Enrichment"] = f"Extraction failed: {str(e)}"
            
            try:
                results["Contributing_Genes_With_Enrichment"] = self._extract_section(full_analysis, "CONTRIBUTING GENES WITH ENRICHMENT")
            except Exception as e:
                results["Contributing_Genes_With_Enrichment"] = f"Extraction failed: {str(e)}"
                
            try:
                results["Contributing_Genes_Without_Enrichment"] = self._extract_section(full_analysis, "CONTRIBUTING GENES WITHOUT ENRICHMENT")
            except Exception as e:
                results["Contributing_Genes_Without_Enrichment"] = f"Extraction failed: {str(e)}"
            
            try:
                results["Analysis_Text_With_Enrichment"] = self._extract_section(full_analysis, "ANALYSIS TEXT WITH ENRICHMENT")
            except Exception as e:
                results["Analysis_Text_With_Enrichment"] = f"Extraction failed: {str(e)}"
                
            try:
                results["Analysis_Text_Without_Enrichment"] = self._extract_section(full_analysis, "ANALYSIS TEXT WITHOUT ENRICHMENT")
            except Exception as e:
                results["Analysis_Text_Without_Enrichment"] = f"Extraction failed: {str(e)}"
            
            try:
                final_reasoning = self._extract_section(full_analysis, "FINAL PROCESS REASONING")
                results["Final_Process_Reasoning"] = final_reasoning
            except Exception as e:
                results["Final_Process_Reasoning"] = f"Extraction failed: {str(e)}"
            
            if confidence_with_enrichment >= confidence_without_enrichment:
                results["Final_Process"] = process_with_enrichment
                results["Final_Confidence"] = confidence_with_enrichment
                results["Final_Contributing_Genes"] = results.get("Contributing_Genes_With_Enrichment", "Not available")
            else:
                results["Final_Process"] = process_without_enrichment
                results["Final_Confidence"] = confidence_without_enrichment
                results["Final_Contributing_Genes"] = results.get("Contributing_Genes_Without_Enrichment", "Not available")
            
            return results
            
        except Exception as e:
            error_msg = f"Analysis failed: {str(e)}"
            print(f"Error analyzing community {comm_id}: {e}")
            results["Error"] = error_msg
            results.update({
                "Process_With_Enrichment": f"Error: {error_msg}",
                "Confidence_With_Enrichment": 0.0,
                "Process_Without_Enrichment": f"Error: {error_msg}",
                "Confidence_Without_Enrichment": 0.0,
                "Final_Process": f"Error: {error_msg}",
                "Final_Confidence": 0.0,
                "Full_Analysis": f"Error: {error_msg}",
                "Pathway_Reasoning_With_Enrichment": "Error occurred",
                "Pathway_Reasoning_Without_Enrichment": "Error occurred",
                "Contributing_Genes_With_Enrichment": "Error occurred",
                "Contributing_Genes_Without_Enrichment": "Error occurred",
                "Analysis_Text_With_Enrichment": "Error occurred",
                "Analysis_Text_Without_Enrichment": "Error occurred",
                "Final_Process_Reasoning": "Error occurred",
                "Final_Contributing_Genes": "Error occurred"
            })
            return results
    
    def analyze_communities_batch(self, communities: Dict[str, List[str]], batch_size: int = 1) -> List[Dict]:

        all_results = []
        community_items = list(communities.items())
        
        from tqdm import tqdm
        
        progress_bar = tqdm(total=len(community_items), desc="Analyzing communities", unit="community")
        
        for i in range(0, len(community_items), batch_size):
            batch = community_items[i:i+batch_size]
            
            for comm_id, genes in batch:
                result = self.analyze_community_optimized(comm_id, genes)
                all_results.append(result)

                progress_bar.update(1)

                processed = len(all_results)
                total = len(community_items)
                progress_bar.set_description(f"Processed {processed}/{total} communities")
                
                progress_bar.set_postfix({"Current": f"Community {comm_id}", "Genes": len(genes)})
        
        progress_bar.close()
        
        return all_results
    
    def create_detailed_dataframe(self, communities_text: str, batch_size: int = 1) -> pd.DataFrame:

        communities = self.parse_communities(communities_text)
        all_results = self.analyze_communities_batch(communities, batch_size)
        
        full_df = pd.DataFrame(all_results)
        
        requested_columns = [
            "Community",
            "Genes_String",
            "Enrichment_Pathways",
            "Process_With_Enrichment",
            "Confidence_With_Enrichment", 
            "Pathway_Reasoning_With_Enrichment",
            "Contributing_Genes_With_Enrichment",
            "Process_Without_Enrichment",
            "Confidence_Without_Enrichment",
            "Pathway_Reasoning_Without_Enrichment",
            "Contributing_Genes_Without_Enrichment", 
            "Final_Process",
            "Final_Confidence",
            "Final_Contributing_Genes",
            "Final_Process_Reasoning",
            "Full_Analysis"
        ]
        
        available_columns = [col for col in requested_columns if col in full_df.columns]
        return full_df[available_columns]

def run_optimized_analysis(input_text: str, api_key: str = None, detailed_csv: str = None, batch_size: int = 1) -> pd.DataFrame:

    analyzer = OptimizedPathwayAnalyzer()
    analyzer.set_api_key(api_key)
    
    detailed_df = analyzer.create_detailed_dataframe(input_text, batch_size)
    
    if detailed_csv:
        detailed_df.to_csv(detailed_csv, index=False)
    
    return detailed_df

if __name__ == "__main__":
    input_text = """
Refined Community 1: ['PALM', 'PF4', 'CCR2', 'SLC35D2', 'MTARC2', 'RIBC2', 'DHRS3', 'TMEM87A', 'NR4A1', 'TPSAB1', 'PUDP', 'CCRL2', 'SFTPB', 'HSPA5', 'PGM1', 'TOR1A', 'TRHR', 'VEGFA', 'PML', 'EXOG', 'CTRL', 'SRPK2', 'RHOA', 'CDC40', 'GLA', 'MAPKBP1', 'MMP9', 'LRP8', 'UCP3', 'LDHB', 'PSMA6', 'SLC6A11', 'RGS17', 'H2BC21', 'PPP1R16B', 'KCNS1', 'BCL2', 'SHANK3', 'AREG', 'CD52', 'GOSR2', 'MRTFA', 'TFCP2', 'LFNG', 'STK25', 'NF1', 'CD3G', 'TWF1', 'IL4R', 'MAPKAPK2', 'TMEM109', 'SMIM15', 'CLEC10A', 'SASH3', 'MYH11', 'NKX3-2', 'SEZ6L', 'ICAM5', 'NOS3', 'ZNF318', 'PLXNB1', 'PPIL2', 'TACC1', 'RAP2A', 'HMCES', 'RPL37', 'VPS72', 'PXDC1', 'GPA33', 'ALDH4A1', 'IGLJ3', 'CDK11B', 'TSHR', 'SCHIP1', 'LY6G6C', 'CITED2', 'S1PR2', 'SLC10A3', 'SPEN', 'ATF6B', 'ZNF646', 'F7', 'KIFC1', 'MNAT1', 'TADA2B', 'SELPLG', 'ATP9A', 'GTPBP6', 'EXOSC2', 'DIDO1', 'ALX1', 'ACOX3', 'CBARP', 'MIR22HG', 'PGBD5', 'TRAF5', 'AOC1', 'RER1', 'IL1RAP', 'ARHGAP25', 'C3AR1', 'NDUFS3', 'ACKR3', 'PCP4', 'ZFR', 'SH3BP4', 'H2AC16', 'TPT1', 'TRIM16', 'MS4A2', 'ICA1', 'SLX4IP', 'C2CD5', 'ZIC1', 'DTNA', 'NACA', 'CD99P1', 'CMTM3', 'CPXM1', 'BMP7', 'TRPV3', 'RAD51C', 'MRPS18B', 'BLOC1S5', 'TRIO', 'PTGER2', 'TGM5', 'ARL15', 'IGF2R', 'RPS17', 'MT2A', 'ATP2B1', 'CKB', 'IDH3B', 'SFTPD', 'ME3', 'DGKD', 'DHFR', 'PTPRE', 'ZDHHC3', 'TGFBR1', 'BANK1', 'YBX3', 'ZCCHC14', 'CSNK1D', 'MN1', 'RRAD', 'NUP210', 'DGAT1', 'GTF2H4', 'PECAM1', 'TRPC6', 'SPRY2', 'ANGEL2', 'CAP2', 'NINJ1', 'CCKAR', 'CLDND1', 'HAP1', 'CASP2', 'GNG11', 'EEF1A1', 'CHEK1', 'VIM', 'GAD1', 'TRBC2', 'SRRT', 'CDC42BPA', 'TSPO', 'BIN1', 'CXCL10', 'ARID5A', 'DTYMK', 'MFAP5', 'MTMR4', 'HDGFL3', 'CLCN3', 'GTF2F2', 'PRKACG', 'AOC4P', 'ADK', 'SLC25A40', 'S100A6', 'CDS2', 'FASTKD2', 'CDC42EP4', 'S100A10', 'OSMR', 'DOK6', 'EZR', 'RUNX1', 'ETV3', 'PRH1', 'MINPP1', 'DEAF1', 'COBLL1', 'CSNK1E', 'SH3GL2', 'UBC', 'CKMT1B', 'TIAL1', 'FCMR', 'MEST', 'PYGM', 'PGLYRP1', 'TOR1B', 'RBM39', 'FOXO1', 'CDR1', 'HPS5', 'HGF', 'ZNF354A', 'SCAF8', 'TMEM263', 'CDKN3', 'MYH9', 'CENPV', 'PPT2', 'OGFOD1', 'AGT', 'GH1', 'HNRNPK', 'PRKCH', 'PLAGL1', 'CUL9', 'ENDOD1', 'SLC27A6', 'ARHGEF7', 'NFIB', 'PTGIR', 'SPTLC1', 'ALDH1A1', 'RPS20', 'RLIG1', 'RPS13', 'ANPEP', 'XBP1', 'ARHGEF12', 'KRT32', 'PKD2', 'COX8A', 'DEFA3', 'XRCC4', 'ERP29', 'FLRT1', 'SDC4', 'HAVCR2', 'COQ2', 'SLC18A1', 'HSPA1A', 'S100A2', 'TLN2', 'GBP2', 'ACOT13', 'SAP30BP', 'ZNF783', 'PRKY', 'CEACAM8', 'ST18', 'CDKL5', 'MMP1', 'NELFE', 'FASLG', 'FOXC1', 'STXBP1', 'HDAC1', 'TPPP3', 'DHRS12', 'ABCB9', 'ATP8A1', 'FKBP2', 'DUSP14', 'MAP2K5', 'GPR83', 'ELOVL2', 'LAMB2', 'OAS1', 'PDHA2', 'ZNF143', 'TGFBI', 'RAMP1', 'ALB', 'MAML1', 'CD1C', 'STARD8', 'KXD1', 'IDS', 'IRS1', 'PBX2', 'NAA10', 'TAF7', 'DNAJC7', 'ARIH1', 'ALDOB', 'AP1B1', 'PARP2', 'H2AZ2', 'RB1CC1', 'ARHGAP45', 'SRP19', 'CD9', 'NEK9', 'UPK2', 'ABCC2', 'KRT13', 'DNAH3', 'SPTB', 'ARL2', 'PCMT1', 'CASP12', 'MAP4K4', 'SKIDA1', 'RPLP0', 'CD2', 'MYO5C', 'TNFRSF18', 'PPP2R2A', 'RPL13A', 'DHX38', 'SAT1', 'LRIG1', 'GZMB', 'MED6', 'SMC4', 'AMOTL1', 'UBE2C', 'COX7A1', 'PCK1', 'TCP11', 'BCL2L2', 'CLMN', 'HTR7', 'IL7R', 'RPS4X', 'C19orf33', 'MON2', 'HLA-DMA', 'RERE', 'RPL19', 'TIMP2', 'LTB', 'BPI', 'ADNP2', 'NCALD', 'PVR', 'FGL2', 'RABGGTB', 'MTUS2', 'RAB27B', 'CTDSPL', 'INPP5F', 'SCN4A', 'POU5F1', 'RPS16', 'B2M', 'FKBP6', 'TPM2', 'RPLP1', 'KDM5B', 'SNTA1', 'SMPDL3A', 'PLPBP', 'RPL32', 'PPIH', 'RPS6KA5', 'PEBP1', 'PSMB3', 'FOSL2', 'GNA15', 'RAB22A', 'DAB2', 'PI4KB', 'HSPB1', 'RPL37A', 'DUS4L', 'BCL6', 'KCNN4', 'CEP68', 'GNGT2', 'UNC119', 'EML4', 'CXCR4', 'ACOX1', 'NTRK1', 'PFKP', 'LY6E', 'WT1', 'SLC8A3', 'RBM3', 'USP19', 'GTF2H2C', 'RS1', 'ARF6', 'EFNB3', 'BAZ2A', 'GHRHR', 'MDFIC', 'F2', 'POU2F2', 'CLIP3', 'DARS1', 'ETF1', 'ACSL6', 'DGKB', 'RTL8A', 'CCDC9B', 'BIVM', 'BATF2', 'NUP50', 'SGCG', 'SLC4A7']
Refined Community 2: ['SMG1', 'ORAI2', 'GUCY1A1', 'TAF1', 'FOXC1', 'GALR3', 'OXSR1', 'HIF1A', 'TPD52L2', 'TYMS', 'ERF', 'NUP210', 'MTARC2', 'HADHA', 'UCK2', 'MYRF', 'ATXN1', 'PSMB3', 'SACS', 'PLIN2', 'NAA80', 'TNFRSF21', 'IL10RB', 'RRAGD', 'GRHPR', 'TRIO', 'ITIH1', 'FBXO21', 'TNFSF13', 'KDM3B', 'FAM131A', 'EXT2', 'DLGAP2', 'MLLT11', 'SEL1L3', 'CD2AP', 'CXCL10', 'DDX39A', 'NAALAD2', 'GLI2', 'GBA1LP', 'ZKSCAN4', 'LTB', 'CD74', 'CSTB', 'NBAS', 'LIN7A', 'RS1', 'HLA-DPB1', 'TEK', 'CCR2', 'HLA-DRA', 'U2AF1', 'ZNF143', 'CNTROB', 'IL1RAP', 'SLC35D2', 'FBXO38', 'ZAP70', 'TIAL1', 'HNRNPK', 'GPC4', 'ADK', 'CDH2', 'PFKP', 'IL1R2', 'MYO6', 'HLA-DMA', 'QPCT', 'DCAF11', 'MPO', 'GLO1', 'SUPT5H', 'TMF1', 'TSC22D4', 'RRBP1', 'SFN', 'ARID4A', 'MCL1', 'SORL1', 'CAV2', 'KRT1', 'MTSS1', 'TNFAIP2', 'ZNF273', 'HMGA2', 'PRSS57', 'VWF', 'ANKLE2', 'ADCY7', 'CCL20', 'CEP162', 'GIP', 'CLINT1', 'S100A4', 'UBE2C', 'CD68', 'LILRA5', 'BMI1', 'SASH3', 'ENPP4', 'GRM8', 'CCT8', 'MPPE1', 'SELPLG', 'CA11', 'UBE2G2', 'RALBP1', 'ADO', 'STAT5B', 'ITGAM', 'NELFE', 'MAN1A1', 'TMBIM6', 'C3orf38', 'LBR', 'RBPMS2', 'ROBO1', 'STIM1', 'CHD7', 'SH3TC1', 'DUSP6', 'HDAC1', 'PALM', 'RPN2', 'FAM30A', 'CYP4A11', 'NDFIP1', 'NIT1', 'IL6', 'SIRPB1', 'ADAM18', 'PRR5L', 'PELI1', 'MN1', 'STARD8', 'ANKRD28', 'MAP2K5', 'ZNF274', 'NT5E', 'H2BC12L', 'SLC25A1', 'PPP2CA', 'ELK3', 'NR4A2', 'STK10', 'HLA-DRB4', 'SLC18A2', 'HLA-DPA1', 'NKG7', 'ENSG00000235059', 'CAP2', 'IL2RG', 'PARP1', 'MDK', 'ZNF117', 'NLRP10', 'KRR1', 'VDR', 'PLAC8', 'AEBP1', 'U2AF2', 'CACNA2D2', 'GH1', 'ZC3H3', 'CLDND2', 'DDAH2', 'ZNF254', 'ELANE', 'RUNX1', 'C2CD5', 'VOPP1', 'JAK1', 'PRLR', 'GPR6', 'CYBB', 'ID2', 'FCGR2A', 'NT5C2', 'COX8A', 'CAT', 'ACAA2', 'SPHK2', 'EPCAM', 'FKBP8', 'NOD2', 'TAF3', 'B2M', 'ZC3H13', 'SLC36A1', 'SYT13', 'LFNG', 'NAA10', 'RMND5B', 'RPP30', 'UBXN2B', 'DHRS3', 'HLA-DQB1', 'KRT10', 'YBX1P5', 'KDM4B', 'GABRE', 'THOC2', 'PPP1R7', 'DDT', 'SMARCD2', 'ZNF271P', 'COL11A1', 'AREG', 'EPB41L3', 'NCK1', 'N4BP2L1', 'CDKN2D', 'SORD', 'PXDN', 'GGCX', 'MTMR9', 'CD19', 'IFI30', 'HLA-DMB', 'TIMP1', 'DARS1', 'NTRK3', 'PIK3R1', 'SLC10A3', 'ZNNT1', 'FBLN5', 'COX5B', 'NUMA1', 'ADRM1', 'PROZ', 'PCK1', 'ECI1', 'ACSM3', 'RB1', 'FZD5', 'TSPAN7', 'GPM6B', 'UFD1', 'VCL', 'CLU', 'ZNF667', 'FCGRT', 'FOXN3', 'ZNF43', 'ABCC1', 'PAPSS2', 'TSPOAP1', 'CTDSPL', 'GYPA', 'ASF1A', 'PLAGL1', 'CIITA', 'CLDND1', 'TCN1', 'WDR31', 'ZNF281', 'F13A1', 'CAPG', 'CLEC4A', 'FOXM1', 'CD200', 'GPR68', 'EIF4EBP2', 'SCRN1', 'CSNK1G2', 'SMARCD1', 'PKP3', 'ELOVL5', 'AURKA', 'PRKCD', 'ARHGEF6', 'CDKN2C', 'PSMD12', 'SCML1', 'AGXT', 'FOS', 'ACTN2', 'EPB41L2', 'TRAT1', 'IRAG2', 'BTK', 'NFKB1', 'CEACAM4', 'ALDH5A1', 'ETHE1', 'MS4A2', 'ELF4', 'PLTP', 'VLDLR', 'FEV', 'PTPRM', 'HSPB1', 'BCL2L1', 'PI4KA', 'IL5RA', 'IGKV1D-17', 'SLC1A4', 'HLA-DRB1', 'RANBP2', 'C11orf21', 'GATA1', 'CD34', 'PPP2R5D', 'ELF3', 'MMP9', 'EPB41', 'CALCOCO2', 'SLA', 'POU2F2', 'CYP7B1', 'IL11RA', 'NUDCD3', 'CD59', 'ATP5MG', 'RASL12', 'ZBTB16', 'ERP29', 'ADAMTS3', 'FYB1', 'IFITM2', 'ABR', 'MYO16', 'PTPRCAP', 'NUP50', 'RBM39', 'DOK5', 'FURIN', 'PLXND1', 'APOB', 'TLE5', 'DGAT1', 'HBA1', 'DSTN', 'DDR1', 'GIMAP1', 'WDR82', 'TIMELESS', 'XAGE1B', 'GDE1', 'TBC1D15', 'MAGED2', 'SFI1', 'RRM2', 'TMPRSS11D', 'HADH', 'CDK9', 'IFI35', 'SIPA1L1', 'FOXO1', 'CTRL', 'GP1BA', 'IFITM1', 'TGFBR1', 'C2', 'CCND2', 'PKMYT1', 'OSBPL1A', 'CERS6', 'TNFRSF14', 'RASA4', 'APOOL', 'SMARCD3', 'ENOSF1', 'AIFM1', 'DRAM1', 'ZNF582', 'CTNND1', 'GAPDHS', 'CRLF3', 'VAMP5', 'BLVRA', 'RAB38']
Refined Community 3: ['CHMP4C', 'KCNQ1', 'TAFA4', 'CD2AP', 'ST18', 'LGALS1', 'TPPP3', 'CSTA', 'ATPAF2', 'DENND10P1', 'CHST12', 'PPP1R2', 'LCP1', 'UBN1', 'CSNK1D', 'PRR14', 'COX7A1', 'GAN', 'MRTFA', 'PLAUR', 'HYOU1', 'IL6R', 'DHCR7', 'HOMER2', 'NFIX', 'HOXA2', 'SLC16A5', 'NEK4', 'CEP41', 'TMC6', 'PPP2R5D', 'NR2C1', 'PIR', 'CTIF', 'SEL1L3', 'CDH2', 'RS1', 'ERAP1', 'CAMK1', 'SNX17', 'TNFRSF17', 'HEXB', 'TGM5', 'PRKAG1', 'MSRB3', 'TXNDC5', 'SYDE2', 'CD79A', 'CHRNA2', 'ANGEL1', 'PVR', 'PRKCB', 'PTP4A2', 'NDUFA1', 'KCNH2', 'BCL2', 'HMCES', 'TBX1', 'TACSTD2', 'KCNN2', 'GIMAP6', 'SAP30BP', 'CRTC3', 'IMPG1', 'DCHS1', 'RB1', 'GDE1', 'CYP4A11', 'IGF2R', 'NPRL3', 'ATXN1', 'VRK2', 'KRT10', 'GMDS', 'GNA12', 'EIF2S2', 'LANCL1', 'SPAG6', 'SIGLEC7', 'TOP2B', 'THBS1', 'CTDSPL', 'CCN2', 'BCAM', 'LARS2', 'GLT8D1', 'UNC5D', 'SMPDL3B', 'OPCML', 'CYP4F12', 'SRSF8', 'MLF2', 'NBR2', 'CCN3', 'STS', 'CLASP2', 'ITM2B', 'RBM15B', 'FSTL1', 'FZD6', 'RARRES2', 'SST', 'KEL', 'LALBA', 'OSTF1', 'SAMD4A', 'MACO1', 'IGF2BP3', 'OVOL2', 'PPP1CA', 'JAK3', 'FCN1', 'LAMA5', 'CDC20', 'GNG11', 'SLC6A9', 'FLNB', 'CDK9', 'CEBPG', 'TNFRSF18', 'CTNND1', 'PIK3CD', 'ZCCHC12', 'ENOSF1', 'ELP1', 'TNNI1', 'TIMELESS', 'ALOX5AP', 'TBR1', 'RGS10', 'AGTPBP1', 'PIP4P2', 'QKI', 'PRUNE1', 'PBX3', 'BLNK', 'JTB', 'TCIRG1', 'MANBA', 'POU4F1', 'CHURC1', 'DUSP7', 'POLR2G', 'CS', 'ITM2A', 'ERP29', 'CRISP3', 'PUDP', 'MAPKAPK3', 'MAPK9', 'SLC25A36', 'ANAPC10', 'TNPO1', 'ARHGEF18', 'SLF2', 'NPDC1', 'PPP1R7', 'MAP2K5', 'NDUFB3', 'GRIK2', 'DLEU1', 'FCMR', 'DYNLT1', 'NUMB', 'ADCY9', 'DEPDC5', 'FBXO21', 'PPT2', 'FEZ1', 'RGS19', 'EEF1A2', 'HNF4A', 'DMXL2', 'UVRAG', 'DHX16', 'BMP7', 'DGKD', 'CALCOCO2', 'INSR', 'ENPP4', 'GPR68', 'TRIB1', 'PCDHGC3', 'POU4F2', 'TTLL12', 'CARS1', 'CYP7B1', 'PML', 'CYBB', 'ARHGAP45', 'FBXO46', 'ANGEL2', 'PLXNA4', 'SC5D', 'GIMAP5', 'CDC42BPB', 'CHKB', 'SMAD1', 'IFNA2', 'ITPR2', 'HIP1', 'IL17RA', 'PROM1', 'SMC1A', 'CSTB', 'HMGA2', 'HPS5', 'C2CD5', 'KHK', 'BCL2L1', 'OGFOD1', 'ATP1B1', 'SLC26A9', 'POP5', 'GDI2', 'LDAF1', 'NIT1', 'RBBP4', 'TAF1B', 'NDUFS3', 'PPP1R16B', 'HMGCS1', 'AP3B1', 'NRP2', 'KRT32', 'UTP20', 'ITPA', 'ST3GAL6', 'ATP6V1B2', 'GLUL', 'ANXA5', 'HADH', 'IL9R', 'IKZF2', 'LDB1', 'SLC5A1', 'PREP', 'MARCKS', 'SLC7A1', 'PI4KB', 'MYH9', 'BABAM2', 'NUDT3', 'NF1', 'SLC16A1', 'UBE2N', 'JMJD1C', 'ZNF667', 'ITGA7', 'RUFY1', 'LRP10', 'CSRP2', 'PCYT1B', 'NBAS', 'COL21A1', 'EXT2', 'SORL1', 'MAML1', 'RPS29', 'CDS2', 'ADA', 'MBNL1', 'SGO1', 'SMCO4', 'GPM6A', 'ATF3', 'HLA-DRB1', 'DIP2C', 'CBX1', 'LAPTM5', 'PRDX1', 'RANBP2', 'PHKA2', 'FCGR1A', 'TBXA2R', 'GATA2', 'PLCG2', 'NDUFB6', 'PIM1', 'DDAH2', 'WFS1', 'HFE', 'JOSD1', 'TAL1', 'RBPMS', 'HDAC2', 'SYF2', 'ZNF280A', 'CAMTA2', 'GYPB', 'SPATA31E1', 'ACOX3', 'SH3GLB1', 'ELK3', 'PSMD7', 'YWHAH-AS1', 'SPAG11B', 'ADCY3', 'DPP4', 'MMP8', 'PCMT1', 'PARP2', 'LYSET', 'ZNF274', 'ATP5PD', 'RBM3', 'ACSL6', 'SPP1', 'AMOTL1', 'BAP1', 'ZNF343', 'SOCS2', 'CLEC5A', 'CTNNAL1', 'RPS6KC1', 'IL1RAP', 'DOP1B', 'BMI1', 'GGT1', 'PHF2', 'BANK1', 'IGF2BP2', 'MPL', 'SLC36A1', 'MMP17', 'TTC27', 'RABGGTB', 'NUP214', 'STX16', 'ALDH3A2', 'GYPC', 'MAN1A1', 'SLC25A13', 'SORD', 'HILPDA', 'RUFY3', 'ZNF91', 'EGR3', 'PRKACG', 'CSF3R', 'KBTBD11', 'TBXAS1', 'ZNF454']
Refined Community 4: ['VPREB1', 'LAMP5', 'FOLR1', 'CEP15', 'ENSG00000269210', 'CD3D', 'RRAGD', 'NAMPT', 'ADAMDEC1', 'CYTL1', 'PRODH', 'CCR8', 'CEACAM4', 'PREB', 'POU4F1', 'SLC25A36', 'HPCAL1', 'TSPAN13', 'ZBTB39', 'RAPGEF1', 'ITGA4', 'TNFRSF1A', 'SERINC5', 'COL4A5', 'IL10RA', 'NAALAD2', 'LHFPL2', 'GATA3', 'SLC16A3', 'PXDN', 'KRT1', 'SORD', 'ADIPOR1', 'TLR1', 'CYP2E1', 'SPEN', 'SELPLG', 'ARHGAP1', 'F2', 'NEDD4', 'GNAI1', 'ADAMTS3', 'HOXB6', 'MYH11', 'FAM171A1', 'GCA', 'ARL15', 'SPTBN1', 'LGALS2', 'EZR', 'RS1', 'EGR3', 'GMPR', 'IL5RA', 'TKTL1', 'CCPG1', 'MKNK1', 'LPAR6', 'DPEP2', 'CCND2', 'IGFBP7', 'FGF13', 'ABLIM1', 'CHRDL1', 'KRR1', 'HNRNPAB', 'MS4A3', 'TLE1', 'MEST', 'ITGB2', 'MPL', 'CD1C', 'ATN1', 'HLA-DQA1', 'TRBC2', 'ARHGEF12', 'TPP1', 'SPHK2', 'ARHGAP45', 'PRSS16', 'PTGER4', 'KCNN4', 'GNG7', 'PRTN3', 'SERPING1', 'AIM2', 'ANXA8', 'DHX38', 'FCER1A', 'MORF4L1', 'ZNF280A', 'PALM2AKAP2', 'ICAM4', 'ELANE', 'FZD6', 'THRB', 'TMEM204', 'FARP1', 'SYNGR1', 'HLA-DRA', 'GALC', 'HSPB1', 'RGS2', 'UGCG', 'ITM2C', 'LRP6', 'DEPTOR', 'TGFB1I1', 'GPA33', 'FAM30A', 'THSD7A', 'ZHX2', 'SAG', 'TRH', 'HAP1', 'PDE4B', 'EREG', 'DDX39A', 'DPYSL4', 'ATXN10', 'SELENOP', 'VPS72', 'RBPMS', 'VPS13D', 'FOXO1', 'TMC6', 'TCF4', 'CD38', 'IFITM2', 'RAP2A', 'SSX2', 'BLOC1S1', 'ZNF84', 'CORO2A', 'NPDC1', 'CACNA2D2', 'XYLT1', 'HIGD1A', 'HOXA1', 'ENSA', 'MEG3', 'TRIB2', 'SACS', 'ASS1', 'EVL', 'TCF3', 'TSPAN7', 'SERPINF1', 'DPY19L1', 'NR2F2', 'SORL1', 'PRR16', 'RHOBTB3', 'ALOX5', 'SPRYD7', 'THOC5', 'ENDOD1', 'TACSTD2', 'DPYSL2', 'CDX4', 'CLIP3', 'CD48', 'POP5', 'PLEKHG4B', 'ZNF354A', 'STOM', 'CD69', 'ZNF454', 'MN1', 'LCN2', 'CHKA', 'STXBP2', 'OMD', 'HOXB3', 'TGIF2', 'CST7', 'CIAO3', 'SPRY1', 'MECOM', 'TIAM1', 'HPGDS', 'EGFL7', 'PTGIR', 'DDAH2', 'DUSP6', 'OSBPL1A', 'UGP2', 'PRNP', 'PRKAG1', 'TUSC3', 'SPON1', 'PRG3', 'HADHB', 'PDE3B', 'TNFRSF17', 'NT5E', 'NFKBIA', 'SSBP2', 'MFAP4', 'WASF1', 'LRP3', 'BANK1', 'MPZL1', 'MYH10', 'CCRL2', 'BLCAP', 'ECM1', 'CLEC5A', 'WT1-AS', 'MLLT3', 'YPEL1', 'ERG', 'CSGALNACT1', 'DLK1', 'TST', 'QSOX1', 'HLA-DRB4', 'MMP2', 'RPS6KA2', 'FARP2', 'H2BC8', 'ARAP2', 'FZD2', 'OTULINL', 'NPR3', 'MTMR11', 'DSE', 'KMT2A', 'ALAS2', 'RPN1', 'PRAME', 'ZAP70', 'HBG1', 'PTPRM', 'CSF2RB', 'VLDLR', 'LDAH', 'RETN', 'NCF1', 'MREG', 'CAVIN1', 'MEF2C', 'CAPS2', 'ST18', 'GRB10', 'EVI2A', 'LAMC1', 'BMP7', 'PROM1', 'SLC8A3']
Refined Community 5: ['VCL', 'ZC3H13', 'GATA2', 'CSF3R', 'AIF1', 'DEAF1', 'NR4A2', 'MINPP1', 'LAMA5', 'STAC', 'HSPE1', 'SC5D', 'GIPC1', 'TJP2', 'LY86', 'CD24', 'NKX2-8', 'TBPL1', 'TOM1L1', 'MAX', 'AHNAK', 'PREB', 'USP8', 'MVP', 'VPREB3', 'PROS1', 'DOK5', 'SORL1', 'GP1BA', 'TCF4', 'LILRB1', 'SEC31A', 'SQLE', 'PBX1', 'F7', 'CYP1B1', 'NCAM1', 'PTGER2', 'ZNF783', 'BAMBI', 'PRDX1', 'SERPINB2', 'PRSS3', 'PCNA', 'PCDHB10', 'F2', 'AMOTL1', 'SEZ6L', 'PIR', 'BCL2L1', 'VIPR2', 'CISH', 'SLC16A5', 'HOXA9', 'OS9', 'MORF4L2', 'BSG', 'CLIP3', 'TMBIM6', 'PKIA', 'MAMDC2', 'PAFAH1B3', 'GP1BB', 'GAD1', 'PLA2G4A', 'SLC38A1', 'TUSC3', 'CAPN2', 'IL11RA', 'POLR3G', 'ARID5B', 'SPNS2', 'NR4A3', 'FBXL7', 'ALB', 'AKR1C1', 'GYPC', 'SLC6A9', 'HDGFL3', 'ITGAX', 'HERC2P9', 'CD58', 'TFDP2', 'TMEM185B', 'PYGO1', 'S100A10', 'RGS2', 'HBA1', 'TRPC6', 'CPM', 'CD200', 'RYR3', 'AHCYL1', 'KIF2A', 'SIRPAP1', 'RFC4', 'INO80C', 'COL21A1', 'SLC7A8', 'MYO16', 'RBL1', 'TMSB15A', 'ATXN1', 'PYGM', 'HOXB6', 'ITM2C', 'ADGRG1', 'ULK2', 'ELK3', 'MST1', 'CSTA', 'AOAH', 'DDAH2', 'NAP1L3', 'CCDC9B', 'FAM174B', 'MYL4', 'RPL27A', 'EZH2', 'C3orf38', 'PIGZ', 'LGALS1', 'LDLRAD4', 'RUNX1T1', 'KDM5B', 'CAV1', 'CAST', 'TRIB2', 'PCP4', 'DDB1', 'GUCY1A1', 'POU4F1', 'VPS13D', 'UCP3', 'PCCB', 'TAF3', 'PFDN2', 'AMD1', 'MECOM', 'F2R', 'KRT4', 'MED21', 'TLE5', 'TGFBR1', 'TPST2', 'PLOD2', 'SLC66A1LP', 'GOLGA8A', 'FCMR', 'NUP50', 'CASP1', 'ZBTB25', 'SOCS3', 'CDC42BPA', 'GABRE', 'RSL1D1', 'TYMP', 'PDCD10', 'DRAM1', 'ICAM2', 'MOB3B', 'COL11A1', 'CCL4', 'GET3', 'GNE', 'YBX1P5', 'SLC25A13', 'SPTLC1', 'ANKLE2', 'SOD2', 'PIK3R3', 'KEL', 'PDHX', 'FGFR3', 'TCIRG1', 'CD164', 'PSMG1', 'ATP6V1G2', 'FAM20B', 'ARHGEF18', 'CLTA', 'APOB', 'TBX1', 'PIM2', 'NAE1', 'CTSW', 'ADAMTS3', 'FGFR1', 'NLRP3', 'FUT11', 'NET1', 'ERF', 'GOSR2']
Refined Community 6: ['IER3', 'PROM1', 'TTC27', 'SOD2', 'PBX3', 'LILRB2', 'PDGFD', 'ITGB3', 'CCL4', 'C1QA', 'H3C4', 'TMEM176B', 'LTB', 'ABAT', 'KDM5B', 'MMP2', 'ADCY2', 'BASP1', 'FTO', 'RPS18', 'SELENOP', 'HMOX1', 'GDF15', 'DTYMK', 'PHACTR4', 'ANGEL2', 'G0S2', 'RUNX1', 'HOXB2', 'RPS13', 'PRPF40A', 'STAU2', 'ATP6V0E1', 'FKRP', 'ADO', 'CLU', 'SLC2A1', 'H1-2', 'SLC9A8', 'NLRP1', 'HOXB6', 'HOXA1', 'CX3CR1', 'SLC15A3', 'CXCR6', 'ETS2', 'CHD1', 'VWA5A', 'TBXAS1', 'TNFSF10', 'CD200', 'ITGA4', 'GTF2A1', 'TRIM44', 'SDHB', 'HIP1', 'F2', 'ELAVL3', 'CDK6', 'CAV1', 'OAZ3', 'PRKD1', 'H2BC21', 'STK3', 'TRIB1', 'VNN1', 'APIP', 'FBP1', 'LTBP3', 'PPBP', 'SAT1', 'SNCAIP', 'RBM25', 'SORT1', 'FCGR3B', 'DUSP14', 'JAG1', 'GNS', 'CHST15', 'ANXA11', 'NAMPT', 'RUNX2', 'VAT1', 'CAVIN2', 'KCNJ2', 'SMIM24', 'LY6E', 'B4GALT6', 'MED6', 'VLDLR', 'SLC25A42', 'HNRNPR', 'S100A6', 'CD79B', 'COL4A5', 'HOXB5', 'GRM8', 'CFP', 'ATP2B1', 'KCNS1', 'ST18', 'ZNNT1', 'SMG1', 'BEX3', 'SLC25A40', 'LGALS3', 'COL5A2', 'FPR1', 'ELANE', 'RASGRP3', 'SNX10', 'NAGA', 'FGL2', 'ADGRL4', 'MVP', 'LAPTM4B', 'PLA2G4A', 'HOMER3', 'TUBB6', 'PHKA2', 'TREM1', 'VNN2', 'MTMR11', 'FAM20B', 'TLE5', 'ZNF646', 'TNF', 'LILRB1', 'ADGRE1', 'SCCPDH', 'SERPINB2', 'TSHR', 'CD14', 'SECTM1', 'FGF20', 'SCHIP1', 'SMC4', 'CD22', 'MON2', 'NR4A2', 'CYP1B1', 'ATM', 'GTPBP1', 'CCR1', 'BMP2K', 'COL9A2', 'DPEP2', 'MTARC2', 'CXCL2', 'GCA', 'ACSL1', 'CA11', 'CD86', 'SRPK2', 'MST1', 'PELI1', 'MAFB', 'MAGED2', 'SCPEP1', 'PRKAR2B', 'MYO6', 'SLC2A6', 'TLR4', 'PTAFR', 'CASP1', 'H3C10', 'APOC1', 'PRKCD', 'MTHFD1', 'PILRA', 'MEIS1', 'SLC38A10', 'EPB41L3', 'GAN', 'FBXO46', 'DEFB1', 'IL6', 'TMEM176A', 'CYTL1', 'ZNF663P', 'TRGC1', 'SLC16A3', 'KCNK5', 'MPO', 'TNFAIP2', 'FSTL1', 'FLNB']
Refined Community 7: ['LPAR1', 'ACVR1', 'HLA-DRA', 'KDM4B', 'DHFR', 'ALOX5AP', 'RGS10', 'CYFIP2', 'FRMPD1', 'MED6', 'HDAC9', 'GRAMD1B', 'IER3', 'DUSP2', 'HGF', 'ARL2', 'SERPING1', 'GYPC', 'PTPN12', 'GATA3', 'COL3A1', 'H2BC12L', 'S100B', 'DHRS12', 'RUNX3', 'NRP1', 'PLEKHG4B', 'TMEM106C', 'CSGALNACT1', 'MCM2', 'HNMT', 'CDC6', 'TK1', 'RNH1', 'PCK1', 'GAMT', 'PITPNC1', 'GAPVD1', 'GTF3C1', 'MST1P2', 'ASB9', 'ATN1', 'VCAN', 'CDKN3', 'MFSD1', 'RBM3', 'FARP1', 'CDKN2C', 'EXO1', 'ITM2C', 'LILRB2', 'CAVIN2', 'GET3', 'NINJ2', 'CD34', 'POLR3G', 'MORF4L1', 'TMF1', 'GPA33', 'CIITA', 'DGKD', 'MOB3B', 'SPATA31E1', 'HADH', 'SERPINB6', 'EPS8', 'ZNF580', 'BIRC5', 'AGTPBP1', 'HAAO', 'CEBPG', 'ATP6V0E1', 'KIF23', 'BLVRB', 'ILF2', 'HBB', 'MAN1A1', 'IMPA1', 'SPRED2', 'XYLT1', 'POLE2', 'UBXN8', 'SYNGR1', 'PSEN1', 'HSF4', 'WDR62', 'MLF2', 'KIAA0930', 'IFITM1', 'CHAF1A', 'PPBP', 'ITM2A', 'QKI', 'LTBP3', 'XPA', 'KRT32', 'ABCB1', 'SERPINI1', 'TYMS', 'ALDH4A1', 'RBM14', 'TPX2', 'GBX2', 'IGFBP7', 'NXT2', 'POLD1', 'ATP8A1', 'MTSS1', 'CHEK1', 'DEFA1B', 'RSAD2', 'RPS6KA2', 'SMG1P5', 'PRG3', 'AHCYL1', 'PTGER2', 'NUP210', 'PRRC2C', 'CDH16', 'IL1R2', 'GSN', 'BANK1', 'SELE', 'RASAL1', 'DHRS3', 'SRGN', 'ZCCHC2', 'RRM2', 'KCNIP2', 'CRHBP', 'MBNL1', 'GJB6', 'GGT1', 'VIM', 'KCTD7', 'FBXO38', 'KRT81', 'GYPA', 'H4C9', 'CHURC1', 'PCDHB8', 'RAB4A', 'TESC', 'TPSAB1', 'CCNB2', 'TRIM16', 'DRAM1', 'POLE', 'HNRNPD', 'DAAM1', 'DMXL2', 'HBZ', 'CCRL2', 'CENPM', 'FGL2', 'SLC4A7', 'TRIB2', 'TNFRSF14', 'CORO2A', 'INPP4B', 'SERPINF1', 'RHOBTB3', 'PLXND1', 'MICAL1', 'SPAG6', 'ANK1', 'RPS6KC1', 'KRT76', 'REL', 'ICAM5', 'CEP41', 'TFDP1', 'CD96', 'CDC25C', 'DTYMK', 'FARP2', 'ACAA2', 'ZBTB11', 'SERPINH1', 'NPR3', 'REPS2', 'TSPAN13', 'USP13', 'PBX2', 'KIFC1', 'ACAT1']
Refined Community 8: ['MED6', 'SLX4IP', 'PCDHB10', 'STAT5B', 'YTHDF2', 'RPS4X', 'PLPBP', 'RBPMS', 'SMIM15', 'EREG', 'HPGDS', 'PRF1', 'PCDHA11', 'PNRC1', 'OAZ3', 'ABLIM1', 'OXA1L', 'GUCY1A1', 'NARS1', 'MORF4L1', 'TNPO3', 'ELAVL2', 'GUCY1B1', 'LEMD2', 'IFNGR2', 'PCDHB16', 'ZNF560', 'GTF2A1', 'NPDC1', 'KIF4A', 'HOMER2', 'PCP4', 'TNFRSF1A', 'PCDHB17P', 'SERPINH1', 'KDM4B', 'ZNF264', 'CEP41', 'EIF2B4', 'RIPPLY3', 'GSN-AS1', 'PRRG1', 'NRXN2', 'NFIB', 'ANAPC10', 'SLC9A8', 'EPDR1', 'PRR18', 'POU2F2', 'CUL9', 'NAMPT', 'CALCOCO1', 'TUBB6', 'POP5', 'SPART', 'PKP3', 'CD1E', 'CFAP184', 'KRTCAP3', 'ISCA1', 'HNRNPD', 'ABR', 'ARHGAP44', 'TGM5', 'CDC123', 'FAM83B', 'SLC2A1', 'TRAPPC1', 'PCDHGA11', 'CHCHD2', 'PCDHGA8', 'CCDC34', 'STX18', 'ZNF154', 'SAMD4A', 'CDX4', 'MTHFD1', 'KRT10', 'RPL22', 'IDH3B', 'PRRC2C', 'ZFP3', 'ZNF667', 'ZFHX2', 'CNTROB', 'TNFSF8', 'ZIC5', 'AKAP13', 'SLC16A3', 'ZNF646', 'ZBTB45', 'TMEM94', 'CD79A', 'HDAC2', 'BSG', 'SRP9', 'ZNF875', 'TTC27', 'REPS2', 'PACRGL', 'TRIM10', 'GPRASP3', 'LILRA5', 'GLIPR1L2', 'BPGM', 'C3orf38', 'THOC2', 'TCEA2', 'GPAT4', 'TAF3', 'GTPBP6', 'PCDHGA5', 'H2AC18', 'RECQL5', 'SMAD6', 'COL5A2', 'PTP4A2', 'CDR1', 'N4BP1', 'MAP4K4', 'NIM1K', 'IL2', 'TBL2', 'APIP', 'FSTL1', 'ZNF343', 'PCDHGB7', 'MAP7D3', 'PNN', 'PABPN1', 'CEACAM8', 'NRP2', 'GATA1', 'DHRS12', 'POU3F4', 'FADS3', 'SMARCD1', 'ZNF84', 'UHRF1', 'GPR149', 'MT2A', 'FNDC3B', 'PCDHB18P', 'CAPN3', 'TTLL12', 'KMT2A', 'ZNF662', 'VWA5A', 'HILPDA', 'PYGO1', 'PCDHB11', 'ZBTB14', 'SIX3', 'KRT81', 'AGO1', 'RFC3', 'PAK1', 'ANKLE2', 'C2CD2', 'KATNIP', 'PNLIPRP1', 'BLOC1S1', 'TANC2', 'HOXB3', 'MCTS1', 'THBS1', 'ARHGEF18', 'RPS6KA3', 'TUBB2B', 'ARHGEF12']
Refined Community 9: ['LRIG1', 'GYPE', 'SAC3D1', 'HOXA6', 'H2BC6', 'PMAIP1', 'SPATA6', 'ADAM18', 'TREM1', 'VSIG4', 'FBXO28', 'CD70', 'PPP1R26', 'HOXB7', 'SPTBN1', 'HNMT', 'YTHDF2', 'HOXA2', 'WARS1', 'ITGA7', 'ZBTB46', 'TRAF5', 'PIEZO2', 'PYGO1', 'HOXB9', 'HOXA7', 'JOSD1', 'CFH', 'HDAC2', 'P2RY2', 'ATF2', 'RLIG1', 'PSMD7', 'STAC', 'GATA2', 'FZD7', 'SGPL1', 'WASHC2C', 'HBBP1', 'VANGL1', 'RPS6KC1', 'SCHIP1', 'ABCC6', 'STAB1', 'NNAT', 'FBXL7', 'HOXA9', 'ECHDC2', 'CS', 'BAP1', 'PLEKHA4', 'CD19', 'SNX11', 'LMNA', 'MKKS', 'TPST2', 'TRIM32', 'KCTD12', 'BCL10', 'KRTAP5-8', 'TSPOAP1', 'TSR3', 'RAB13', 'AQP3', 'PIK3R1', 'SMPDL3A', 'NFE2L3', 'MEF2A', 'MYL4', 'DLEC1', 'FOXC1', 'DGCR2', 'GRHPR', 'ICA1', 'DACH1', 'CAVIN1', 'RUNX1', 'CYTH1', 'MTMR9', 'AGXT', 'CLDN15', 'F13A1', 'MEIS1', 'KCNN4', 'CENPX', 'DPEP2', 'CAVIN2', 'CORO1B', 'MMP2', 'H2BC4', 'ACVR1B', 'H2AC18', 'SC5D', 'CCL5', 'SRY', 'HOXA1', 'HLA-DRA', 'NKX3-1', 'MACO1', 'ITGA5', 'SELENOP', 'PREP', 'MTARC2', 'LTB', 'TNFSF8', 'MAGEB2', 'HBB', 'CCL20', 'RASGRP3', 'DSC2', 'ADAM17', 'ACOT13', 'CAPN2', 'LAPTM4B', 'BRINP1', 'AKR7A2', 'HOXB3', 'RAMP1', 'CTNND1', 'HOXB6', 'ADGRE1', 'HOXA10', 'GSTM3', 'DPYSL3', 'ZNF667', 'UGGT2', 'LUM', 'PPP2R5D', 'SEMA4A', 'KRT18', 'WDR91', 'GJA4', 'TBL2', 'FAM110B', 'NAB2', 'CFD', 'TRIB1', 'IGHA1', 'STRN3', 'UBA1']
Refined Community 10: ['HNF4A', 'POU5F1', 'DYNC1LI1', 'HSPA1A', 'CNTROB', 'GPAT4', 'MANBA', 'CUL9', 'ACOX3', 'CD9', 'MSRB1', 'GRHPR', 'CISH', 'GTF2A1', 'TP53INP2', 'TMEM185B', 'SEC31A', 'NQO2', 'TGFBR1', 'VWF', 'KCNAB3', 'PPP2CA', 'CDC123', 'STX18', 'HLTF', 'GET3', 'GRIK2', 'SNRNP70', 'FBXL7', 'HLA-DMA', 'CBLN4', 'BLVRB', 'SMARCD3', 'PHF13', 'PCDHGA11', 'LRP3', 'IRAK3', 'STX11', 'CFAP184', 'SYDE2', 'CIAO1', 'ARIH1', 'HNRNPD', 'UNC5D', 'DHRS12', 'GLT8D1', 'ZSCAN1', 'NEUROG1', 'RPL19', 'WDFY3', 'STK3', 'ZNF454', 'PRRT1', 'CENPV', 'FUT9', 'CASP2', 'GJA4', 'BPGM', 'IL11RA', 'BTBD3', 'CRHBP', 'OSBPL1A', 'CRTC3', 'MAGED1', 'CASP12', 'FNBP4', 'FARP2', 'SLC25A51', 'COCH', 'KRT81', 'GH2', 'ZCCHC14', 'RIPPLY3', 'PTAFR', 'MCTP1', 'TSC22D4', 'WBP1L', 'CACNA2D3', 'S1PR1', 'CLEC10A', 'ATM', 'HSPB1', 'STMN1', 'MAFB', 'RECQL5', 'ELL2', 'PSMA6', 'ZNF280A', 'PPIL6', 'INO80C', 'ICA1L', 'BTAF1', 'SGPP2', 'PIGZ', 'RNASE2', 'REXO2', 'TSC22D1', 'KDM5B', 'IL1B', 'ARAF', 'TTC22', 'SLC66A1LP', 'PHC1', 'SPHK2', 'CYP26C1', 'PELI2', 'PIGH', 'GPRASP3', 'MT2A', 'POLR2H', 'PFN2', 'GATA1', 'AK4', 'PDGFD', 'CANT1', 'EXOSC3', 'CDK6', 'PTPN14', 'PXMP2', 'ANKRD28', 'NNAT', 'ZBTB14', 'C2', 'ARHGAP1', 'HMHB1', 'HBBP1', 'SMAD1', 'SLC7A7', 'RPS18', 'PCNA', 'SCG5', 'EXOSC2', 'UGT2B11', 'UGCG', 'PDHA2', 'JOSD1', 'ADRA2C', 'DLC1', 'GUCY1B1', 'AEBP1']
Refined Community 11: ['PRKAR2B', 'DOK5', 'HOXB5', 'OAS1', 'SLC27A3', 'HPS4', 'ATP2B1', 'AUTS2', 'PLCG2', 'WARS1', 'MUSK', 'GRK5', 'THOC5', 'NTRK3', 'NDUFS6', 'UBE2N', 'HNRNPAB', 'POLD1', 'CD300C', 'H4C9', 'CEMIP', 'ZSCAN32', 'SPRY1', 'CD69', 'CYP2E1', 'NLRP3', 'CTRL', 'SERTAD2', 'GNB5', 'ZBTB11', 'LANCL1', 'GGCX', 'PLPBP', 'SLC25A36', 'ABCC2', 'IFI35', 'APOC2', 'CACNB1', 'MTERF4', 'ATXN10', 'ZNF560', 'CEP350', 'TRA2A', 'JMJD1C', 'ADCY3', 'FOXC1', 'MRPL33', 'PHIP', 'GNAI1', 'MEIS1', 'MR1', 'TCL1A', 'RBBP6', 'IGHM', 'IFI44L', 'SGCG', 'ANPEP', 'HYAL2', 'IRX5', 'CSRP2', 'ACTR2', 'IDH3A', 'JTB', 'PML', 'DDX39A', 'TSC2', 'SRGN', 'GJC3', 'TGFBI', 'NELFE', 'AK2', 'KRT81', 'GAPDHS', 'FZD2', 'MLF2', 'NARS1', 'SOCS2', 'PALLD', 'TGM5', 'TRIM44', 'ELP4', 'CYFIP1', 'CDX4', 'CYP7B1', 'FEZ1', 'RANBP6', 'NKX2-8', 'AGXT', 'FLNB', 'PHYH', 'SF1', 'CYLD', 'MX2', 'CCL23', 'ADGRE1', 'PIGH', 'SFI1', 'DACH1', 'PREPL', 'DGCR2', 'EOMES', 'GAN', 'RFPL3S', 'TNPO1', 'GNE', 'TWF1', 'BEX3', 'OXSR1', 'GNG12', 'PBX3', 'ATP6V0B', 'GLO1', 'IMPG1', 'CAV2', 'ADPRHL1', 'PLAC8', 'PHKA2', 'DNAJC7', 'RUVBL1', 'MYCN', 'MSLN', 'EFNB2', 'PLXNB2', 'RPS17', 'SATB1', 'EVI5', 'TREX2', 'TWF2', 'CERS6', 'CKM', 'CD200']
Refined Community 12: ['C3orf38', 'ATG13', 'TIA1', 'GNG12', 'CDC42BPB', 'PCBP3', 'EIF3A', 'ZC3H3', 'RRAD', 'GTF2IRD2', 'PRR5L', 'NAP1L3', 'CTR9', 'PSEN1', 'NAA10', 'PIK3R3', 'PUDP', 'ZNF271P', 'HMGCS1', 'XPO1', 'CD14', 'PXMP2', 'WT1-AS', 'MYCN', 'CSRP2', 'DNM2', 'LGALS2', 'NPC2', 'BST1', 'SPNS2', 'PDCD10', 'PDE4DIP', 'PSTPIP1', 'H4C9', 'GLRA1', 'PLAC8', 'SLC25A4', 'TRAF2', 'ST3GAL5', 'GDI2', 'DNAJC28', 'VPS11', 'CTDSPL', 'BRD8', 'ATP6V0B', 'PRLR', 'LBR', 'WSB2', 'PKIG', 'ECE2', 'CORO2A', 'ANP32E', 'SLC9A7', 'TAFA4', 'UBE2C', 'CCL1', 'PPBP', 'STAU2', 'ATP1B1', 'PTP4A2', 'PRPF4', 'DMXL2', 'CALCR', 'NACC2', 'KIF2A', 'CRYZ', 'AZU1', 'TFAP4', 'CDK6', 'JAG1', 'MAP7', 'SLC66A1LP', 'PRKAR1B', 'TLL1', 'UBXN7', 'PLBD1', 'ATF6B', 'CEACAM6', 'VDAC1', 'GRIK5', 'LUZP1', 'PGD', 'ADAM15', 'LPAR1', 'MAGED1', 'SPCS3', 'HNF4A', 'PREPL', 'ENSA', 'GTF2F2', 'IMMT', 'ENPP1', 'ALDH2', 'CTNNBL1', 'NDFIP1', 'MED21', 'TNFRSF17', 'CLIP2', 'ARAF', 'TAF1', 'TES', 'CEP68', 'RSKR', 'RARB', 'CCRL2', 'XBP1', 'VAT1', 'CTDP1', 'CRLF3', 'TPM2', 'RAB32', 'DRAP1', 'DCHS1', 'DPP4', 'ADAM17', 'XKRY', 'CLINT1', 'EED', 'MMP8', 'SDC4', 'RASAL1', 'U2AF2', 'GPR182']
Refined Community 13: ['ZNF264', 'CYP26C1', 'TTC22', 'MOS', 'PFN2', 'TLR2', 'ASB9', 'LBX2', 'ATXN1', 'PCDHGA5', 'ZNF766', 'TOR1A', 'CXCL1', 'TUSC3', 'ZFP3', 'KRT10', 'RSKR', 'WBP1L', 'JCAD', 'SYDE2', 'MRPL55', 'SMIM15', 'FKBP2', 'ZNF662', 'CERS6', 'SPRY1', 'AIFM1', 'TBC1D16', 'EREG', 'NIM1K', 'MAP7D3', 'TIGD3', 'SCO2', 'DHRS12', 'MCTS1', 'SAP30BP', 'IFNA2', 'PCDHB16', 'RAB38', 'PKMYT1', 'LMO4', 'CUL1', 'SCG5', 'RAP1GAP', 'TNFRSF1A', 'KRT81', 'SLC9A7', 'CISH', 'USP13', 'RAB13', 'UBXN2B', 'SLC38A10', 'UCP3', 'RANBP1', 'AZU1', 'VPS53', 'RRM2', 'CANT1', 'GNG11', 'KATNIP', 'WT1', 'GPR149', 'DEPDC5', 'RIPPLY3', 'ITM2A', 'TXNRD3', 'PCDHB9', 'HOXB6', 'SLC39A14', 'TUBB6', 'VPREB3', 'OFD1', 'RPL23A', 'GPAT4', 'NKX2-8', 'POU5F1', 'MFAP5', 'FRMPD1', 'TIE1', 'SAMHD1', 'TBR1', 'MARCKS', 'RBPMS', 'XYLT1', 'ABLIM1', 'AOC1', 'PTHLH', 'SH2D1A', 'OS9', 'CNOT8', 'FAM83B', 'PXMP2', 'ZBTB14', 'HTR2A', 'CENPV', 'SLC8A3', 'FSTL1', 'CFAP184', 'CD70', 'VWA5A', 'CRHBP', 'ECE2', 'NR2C1', 'ZNF875', 'COL24A1', 'CHCHD2', 'CRIM1', 'DHPS', 'TRAK2', 'NEK4', 'RPL13A', 'ZNF560', 'ISCA1', 'GPRASP3', 'KIAA0930', 'EXO1', 'NUP50', 'PRSS3', 'RECQL5', 'NTN5', 'TEK', 'DMXL2']
Refined Community 14: ['CEP78', 'TMOD1', 'N4BP1', 'KDM5B', 'CEP70', 'HTR4', 'KATNIP', 'CD9', 'NR4A1', 'DPYSL4', 'DSE', 'SYNGR1', 'TGFB1I1', 'MORF4L1', 'FZD7', 'FAM3A', 'U2AF1', 'SIGLEC7', 'TIE1', 'TRPV3', 'NEUROG1', 'TUBB6', 'GNE', 'CUL9', 'TRH', 'SLC8A3', 'GAPDHS', 'RHOT1', 'TLR2', 'BTBD3', 'MCM2', 'ZFHX2', 'CD48', 'ADCY9', 'FCGR2A', 'FSTL1', 'CANT1', 'SPOP', 'YBX3', 'NT5DC3', 'IL11RA', 'ADH5', 'HBB', 'SLC25A51', 'RIMKLB', 'BTN2A1', 'ABLIM1', 'TIA1', 'TRPC1', 'RNASEH2B', 'TIGD3', 'EDEM1', 'CCPG1', 'SORL1', 'GPRASP3', 'DNM3', 'SNRNP200', 'GAPVD1', 'OAZ3', 'HMHB1', 'IL6ST', 'CYP26C1', 'TAFA4', 'SNRNP70', 'LGALS9', 'ADGRG6', 'COCH', 'MKKS', 'ARL15', 'BTAF1', 'NRP2', 'RGS19', 'MYO7A', 'ADAM28', 'CFAP184', 'RSKR', 'INO80C', 'ITGAV', 'VWF', 'ISCA1', 'PAQR5', 'TEKT2', 'HMGCS1', 'AGPAT2', 'VIPR2', 'PIGZ', 'CNTROB', 'CAV1', 'CACNB1', 'ANK1', 'SLC6A9', 'NNAT', 'MRPL9', 'OSBPL1A', 'ZNF281', 'RHAG', 'NAA80', 'ACSL4', 'NDUFS3', 'SRP9', 'PLPBP', 'ELAVL2', 'RGP1', 'SRSF8', 'RECQL5', 'ENSG00000269210', 'WBP1L', 'DHRS12', 'FAM131A', 'NDUFS6', 'TYMP', 'CTSF', 'APPBP2', 'ZNF560', 'SRY', 'IFNAR2', 'THSD7A', 'HMCES']
Refined Community 15: ['IPO8', 'TNPO3', 'ZFHX2', 'LHCGR', 'FSTL1', 'ADAMTS3', 'CSTB', 'MOAP1', 'CHMP4C', 'FAM83B', 'OAZ3', 'MCTS1', 'TNFRSF1B', 'ZNF281', 'GSN-AS1', 'ARHGAP22', 'SDC4', 'HMCES', 'MACROD2', 'TAF4B', 'TRIB2', 'TADA2B', 'NR3C1', 'ATP6V1G2', 'ATP12A', 'ZFP28', 'PABPN1', 'CAMK1', 'TLR2', 'STAT4', 'LANCL1', 'ATP8B4', 'CAV2', 'KCNJ2', 'PLPBP', 'LAMB2', 'ZFP3', 'EXOSC3', 'ADGRG6', 'CKMT1B', 'PPT2', 'LARGE2', 'SERPINA10', 'SPCS2', 'RECQL5', 'TMPRSS11D', 'PKD2', 'KRT81', 'THBS4', 'LDAF1', 'INKA1', 'ILF2', 'IFNAR2', 'CD36', 'LPAR6', 'PPIH', 'MAP7D2', 'DDX5', 'WBP1L', 'SLC8A3', 'SCG5', 'MYO5C', 'NEK9', 'CNTROB', 'CD59', 'ZNF560', 'HNF4A', 'SGCG', 'HK3', 'TNFAIP8L2', 'TIGD3', 'CDK8', 'HILPDA', 'ZNF667', 'CYP26C1', 'COMT', 'TGFBR1', 'FSCN1', 'CCND2', 'VDR', 'MMP1', 'PPP1CA', 'PIP4P2', 'ADNP2', 'CXCL5', 'ELAVL2', 'POU2AF1', 'ZC3H3', 'SKIDA1', 'ZBTB14', 'SNCA', 'FZD5', 'PMAIP1', 'ELANE', 'COCH', 'NEUROG1', 'SLC39A4', 'KCNS1', 'ITGA4', 'HADHA', 'CHCHD2', 'PAQR5', 'PRSS16', 'CEBPG', 'NFKBIL1', 'VAMP5', 'LBX2', 'LAPTM5', 'YBX3', 'GRAMD1B', 'PIGZ', 'ABLIM1']
Refined Community 16: ['TSC1', 'NNAT', 'BCL11A', 'EEF1A1', 'ZCCHC12', 'DNM1', 'MAP7', 'EZR', 'CD3D', 'SELENBP1', 'PTPRM', 'WDR82', 'IFI27', 'GLULP4', 'S1PR1', 'NR2F2', 'NBL1', 'ZNNT1', 'BMP7', 'LDAH', 'SLC38A10', 'TSC2', 'SKAP2', 'HOXA11', 'SNX9', 'ANGPT1', 'CDC42EP4', 'PF4', 'MACROD2', 'TMEM158', 'GGT1', 'MSI2', 'SAMSN1', 'EDEM1', 'SSBP2', 'HOXA4', 'CCN4', 'TLE5', 'HOXB2', 'COL1A1', 'IFNGR2', 'NKX2-8', 'IL6ST', 'TRPS1', 'ISG20', 'ASNS', 'TBXAS1', 'ENSG00000274253', 'U2AF1', 'PMEPA1', 'FARP1', 'NFIB', 'TSPAN32', 'PBX3', 'ANXA2', 'SLC7A7', 'TACSTD2', 'MSRB1', 'LGALS3BP', 'RRM2', 'MAP4', 'APOE', 'ALCAM', 'CHRNA2', 'AGRN', 'MAFB', 'FOXO1', 'ENDOD1', 'SRP19', 'IRAK3', 'TSPAN7', 'HPGDS', 'IL10RB', 'CHURC1', 'NYNRIN', 'FLT3', 'HLA-DQA1', 'PIK3R4', 'UGP2', 'S100P', 'ADGRB2']
Refined Community 17: ['SLC22A4', 'SCAP', 'MEF2C', 'ANXA5', 'DEXI', 'CCN3', 'TLR8', 'ZNF160', 'MYO6', 'TIMP1', 'ACSL4', 'TWSG1', 'SAGE1', 'ITGA7', 'CAPN3', 'NAA10', 'GAS7', 'FCGR1A', 'SIRPA', 'MRPL34', 'SRGN', 'AKR7A2', 'DMWD', 'DACH1', 'AK2', 'ADCY3', 'HOXA9', 'TMEM185B', 'IKZF2', 'LGALS1', 'TFEB', 'TMEM37', 'PLEK', 'ENSG00000274253', 'GOLGA8N', 'TRPM4', 'SH3TC1', 'CAPG', 'GGA2', 'C2', 'GLG1', 'ATN1', 'PPID', 'GYPA', 'VNN1', 'STXBP2', 'SWAP70', 'TFAP4', 'LEMD2', 'SNX10', 'PGAM1', 'FTO', 'APOC2', 'SOCS2', 'SLC16A1', 'MBNL1', 'GTPBP1', 'PREB', 'NDUFA1', 'HK3', 'RNASE2', 'ALDH3B1', 'MICAL1', 'SRY', 'ZNF280A', 'RNASE3', 'ITGAX', 'RARRES2', 'ATG13', 'RNH1', 'FES', 'CD3D', 'CCL5', 'CEMIP', 'HOXB7', 'CTNNA1', 'TNFAIP3', 'CYFIP2', 'FCMR', 'PHKA2']
Refined Community 18: ['ASNS', 'TLE5', 'ANKRD28', 'CPA3', 'CDK11B', 'HBB', 'MEG3', 'CTSW', 'MRC2', 'MAGED2', 'CDKN1C', 'S100B', 'MPO', 'CAV2', 'CLSTN1', 'LILRB1', 'P4HB', 'AGAP4', 'LRFN4', 'CEACAM8', 'JAG1', 'GJA4', 'CMAHP', 'MEGF6', 'HSPA5', 'SOCS2', 'AKR7A2', 'MAP1A', 'ZSCAN1', 'TNS3', 'TLE1', 'COL2A1', 'PGLYRP1', 'MMP2', 'CALR', 'TNFRSF4', 'CLASP2', 'NAMPT', 'YTHDF2', 'KIF23', 'PTGDS', 'NCOA6', 'MAP4', 'MXRA7', 'EXOC1', 'ID3', 'NDST2', 'ARFGEF1', 'RCN1', 'SMARCD2', 'AGRN', 'AUTS2', 'CTRL', 'GALNS', 'ARHGAP25', 'TCF4', 'TRIB2', 'FCGRT', 'LY6E', 'GRB10', 'SLC25A38', 'CHPF2', 'CLEC16A', 'DENND5A', 'PHKA2', 'HYOU1', 'TFEB', 'ALDOB', 'STAB1', 'ATG13', 'GLO1', 'DEFA3', 'TMEM94', 'SLC39A4', 'IGFBP2', 'SLC25A36', 'HGF', 'TTLL12', 'SEC31A', 'CAST']
Refined Community 19: ['MPPE1', 'EGR3', 'FCGR2B', 'STK32B', 'H2BC5', 'CCR1', 'IFNAR2', 'SLC9A7', 'ADD2', 'MDM4', 'OFD1', 'CCND1', 'CYP2A13', 'HPGDS', 'HCAR3', 'H2BC10', 'CACNA2D2', 'AP1B1', 'ARHGEF6', 'CD19', 'SLCO4A1', 'MYB', 'TSC22D1', 'GCLC', 'RNASE2', 'POP5', 'ELMO1', 'SEMA4A', 'CDK20', 'POU4F1', 'KLF11', 'RASAL1', 'NIT2', 'RENBP', 'NCALD', 'KDM4B', 'DLG5', 'ZNF143', 'WSB2', 'YTHDF2', 'EPCAM', 'CIITA', 'RRM2', 'PRAME', 'PCDHB17P', 'TRAF3IP2', 'VOPP1', 'XYLT1', 'OAZ3', 'C15orf39', 'IL7R', 'RASA4', 'DGCR2', 'H2BC7', 'CLEC5A', 'SYDE2', 'RUNX1T1', 'COLEC12', 'FKRP', 'PTGIR', 'RCBTB1', 'ARHGAP1', 'TMX4', 'TWSG1', 'USP10', 'WFDC1', 'CYTH1', 'LILRA3', 'CLINT1', 'RCN1', 'TBXA2R', 'VAMP5', 'ADRM1', 'ENSG00000275616', 'IL5RA', 'TCEA2', 'PRSS16', 'IER2', 'PELI2']
Refined Community 20: ['FKBP8', 'ADCY6', 'HLTF', 'GATA1', 'LAPTM4B', 'CYTL1', 'DNM3', 'ARMC8', 'ATP5MJ', 'FBLN5', 'HIGD1A', 'CDC42EP4', 'GNAI1', 'IFITM2', 'APOC1', 'LDHB', 'RHAG', 'RBBP4', 'MRPS12', 'TAL1', 'NBR2', 'BMP2K', 'MYG1', 'GIPC1', 'BEX3', 'RAB32', 'CUTA', 'TNIK', 'UBE2C', 'CDK9', 'TIMP3', 'PTGS1', 'SOD1', 'DNAJC9', 'PROS1', 'RASA1', 'FUT9', 'RETN', 'PCDH9', 'ZMYND8', 'TFR2', 'THOC2', 'KIAA2013', 'CCNF', 'PNMT', 'DHPS', 'TEK', 'RPS6KA3', 'MNAT1', 'GNGT2', 'RYR3', 'PDCD10', 'NET1', 'TNF', 'CLASP2', 'FOXN3', 'NEO1', 'SLC18A1', 'NDUFB6', 'AIF1', 'DRAP1', 'MYL4', 'GNS', 'POFUT2', 'LUZP1', 'TNS1', 'CCDC85A', 'KCNH2', 'FHL2', 'EIF2B4']
Refined Community 21: ['GFI1', 'GTF2A1', 'OAZ3', 'CCDC34', 'MATR3', 'CDK8', 'DNM3', 'GTF3C1', 'ARHGAP27', 'PFN2', 'HEXB', 'SLC39A14', 'VPREB3', 'ASPH', 'SLC6A8', 'CDH19', 'COL21A1', 'PMAIP1', 'CYP26C1', 'PLA2G4A', 'CCKAR', 'ZNF582', 'MAP4', 'KRT81', 'MOS', 'SLC2A6', 'ZNF667', 'ASH2L', 'SAP30BP', 'GTPBP6', 'MORF4L2', 'TFCP2L1', 'EFHD2', 'CORO2A', 'CDX4', 'ICAM3', 'TIGD3', 'ARHGAP44', 'TM9SF4', 'TSPAN4', 'SELENBP1', 'EXOSC3', 'CD52', 'TRIM10', 'PIGZ', 'PCDHGA11', 'FOXC1', 'EXOSC2', 'SYT13', 'DLEC1', 'GNGT2', 'MAP7D3', 'NARS1', 'TBC1D31']
Refined Community 22: ['STK3', 'CD302', 'LST1', 'SERPINE2', 'FNBP1', 'ATP2B4', 'RHOT1', 'PRG2', 'SGPL1', 'RNASE2', 'C11orf21', 'GH2', 'GCH1', 'HSPA6', 'NKG7', 'OS9', 'PTGER4', 'PCMT1', 'SCRN1', 'LYZ', 'FARP2', 'PSME3IP1', 'S100A8', 'CCND1', 'RNASE3', 'GOLGA8A', 'UBE2D1', 'CMTM3', 'RASGRP2', 'ENTPD1', 'CD24', 'TCF3', 'SLA', 'CDH19', 'STX16', 'PCCB', 'VIM', 'MANBA', 'AZU1', 'CST7', 'SELPLG', 'SPCS3', 'TWF2', 'BTG1', 'GAPVD1', 'HDGFL3', 'F2', 'TRGC2', 'ADO', 'CADM1', 'PSTPIP1', 'RTL8C', 'BAZ2A']
Refined Community 23: ['KCTD12', 'F2', 'HLA-DQA1', 'ATXN1', 'CTSS', 'CCR7', 'PIK3R1', 'GRHPR', 'MDFIC', 'DEFA1', 'ITGA2B', 'ST18', 'PTPRM', 'AK5', 'ELP1', 'IL2RG', 'CBFB', 'BTN2A1', 'HPCAL1', 'VPS11', 'MFAP5', 'NISCH', 'GYPA', 'LPAR1', 'SPARC', 'WDR91', 'CLEC7A', 'MX1', 'SLC38A1', 'DYNLT1', 'IL10RA', 'BCL10', 'S100A8', 'PRKCH', 'FARP2', 'GOLGA8N', 'NCK2', 'TES', 'CHI3L1', 'PGM1', 'LAIR1', 'IGLL1', 'S100A9', 'GIMAP4', 'STX16', 'TNFAIP2', 'MYH11', 'FAM90A1', 'NRP1', 'NACC2', 'GAN', 'PLXNB2']
Refined Community 24: ['TPM1', 'SLC26A9', 'DHX16', 'MPO', 'NR2C1', 'CLDND2', 'TUB', 'MARVELD1', 'TRPC4AP', 'TMEM37', 'GPR17', 'PNLIPRP2', 'FAM90A1', 'SH3RF2', 'OGFOD1', 'FUT11', 'SPMIP5', 'LYRM1', 'SERPINB9', 'ASB10', 'TACSTD2', 'ASXL3', 'KPNA5', 'NKG7', 'ATP6V0E2', 'TIAM1', 'CLINT1', 'SFI1', 'FKRP', 'FCGRT', 'PTPRM', 'ITGB2', 'SELENBP1', 'CD3D', 'CD48', 'SLX4IP', 'FADS1', 'TXNDC5', 'GIMAP5', 'TPST2', 'ULK2', 'RAPGEF1', 'DHCR7', 'WFDC1', 'SNX11', 'PCK1', 'TACR3']
Refined Community 25: ['COL24A1', 'NIT1', 'TRIO', 'RHCE', 'RPLP0', 'SOCS2', 'KAT6A', 'MACROH2A2', 'IL11RA', 'SLC15A3', 'EPDR1', 'NYNRIN', 'AIF1', 'TNFRSF1B', 'GUCY1A1', 'RARRES2', 'EMP1', 'RYR1', 'GPSM1', 'ANGPT1', 'PRSS57', 'ACADM', 'LAPTM4B', 'HCAR3', 'BTAF1', 'CACNB1', 'KCNK17', 'SPINK2', 'ATF6B', 'SPNS2', 'BCL7A', 'UGT2B11', 'HOXA5', 'ZBTB46', 'BIVM', 'FAM30A', 'NBAS', 'SHANK3', 'HOXA2', 'HOXA6', 'SMG1P5', 'STARD8', 'ZNF710-AS1', 'OPTN', 'VWF', 'DNMT3B', 'CDK6']
Refined Community 26: ['SLC25A42', 'NQO2', 'CEBPD', 'DDHD2', 'PCDHGA11', 'PCP4', 'NEMF', 'H2BC11', 'SEZ6L', 'KRT1', 'PRF1', 'ASB10', 'CGB3', 'LHFPL2', 'NUDT5', 'TLR1', 'H2AC11', 'TRIM32', 'NFATC3', 'HDGFL3', 'TACR3', 'RPS20', 'TIAM1', 'SCML2', 'CDS2', 'SLC35E3', 'MFAP4', 'MED25', 'VAMP8', 'PCNA', 'ATP8B4', 'SH2D1A', 'PDCD1', 'CD68', 'MOSPD1', 'DNM1P35', 'PPIF', 'CEP350', 'CD48', 'NLRC4', 'FAM171A1', 'TSC1', 'RPL28', 'AHNAK', 'MBNL1', 'KDM4B']
Refined Community 27: ['ADGRG1', 'LAMC1', 'ETV6', 'BRINP1', 'PBX3', 'NFE2', 'HOXB6', 'PIEZO2', 'CHRNA7', 'KAT6A', 'CXCR3', 'ADCY2', 'SNRPN', 'TSC1', 'COL4A5', 'ENPP2', 'AGXT', 'FAM110B', 'SELENOP', 'EGFL7', 'APP', 'YWHAH-AS1', 'TRPC2', 'JAK3', 'SERPINB6', 'DOC2A', 'LYRM1', 'MAP1A', 'PIK3C2B', 'CHKB', 'GPC4', 'KCNK5', 'MED13L', 'KRR1', 'HOXB2', 'MYCN', 'TRIM16', 'LAPTM4B', 'CYSLTR2', 'MAGED1', 'MN1']
Refined Community 28: ['MARVELD1', 'NECAP2', 'TSFM', 'TOR1B', 'IGLJ3', 'PYCARD', 'RPP25', 'SLC26A9', 'CILK1', 'MFSD1', 'HOXA2', 'EPHA2', 'NOS3', 'ARHGAP10', 'ITPR2', 'ZSCAN1', 'KDM4B', 'H2AC11', 'CRTC3', 'MAML1', 'ZBTB43', 'MYG1', 'TLR1', 'SH3BP4', 'SPRED2', 'HOXA7', 'DEFA3', 'TMBIM6', 'AHNAK2', 'ZNF580', 'RSL1D1', 'STARD8', 'AOC1', 'ASB10', 'PPARD', 'EGFL7', 'TTC22', 'SLC6A11', 'SLC45A1', 'PKDCC']
Refined Community 29: ['GRB10', 'SOCS3', 'ITGAE', 'ENTPD1', 'IL10', 'PARP1', 'C1QA', 'FFAR2', 'CXCR3', 'CREM', 'CASP4', 'CAPS2', 'CD84', 'PRPF18', 'SGO1', 'NFE2L3', 'TNFRSF18', 'GMPR', 'RBL1', 'DEPTOR', 'HOXA7', 'CLIC4', 'MTMR4', 'GPSM1', 'IL17RA', 'CRHBP', 'IRF4', 'ITGA2B', 'CCR7', 'IL17A', 'QPRT', 'SPMIP5', 'CCL5', 'AQP3', 'KCNN4', 'CCR2', 'FOXN3', 'HLTF', 'IZUMO1R', 'PPARD']
Refined Community 30: ['CIITA', 'VEGFA', 'FYN', 'EVL', 'OMD', 'VRK2', 'IGKV1D-17', 'ITGA6', 'UGP2', 'GYPB', 'SASH3', 'REEP5', 'MYG1', 'ZNF117', 'SNRPN', 'CLIC4', 'TRAPPC1', 'NIM1K', 'HTR4', 'ST3GAL5', 'ALDH2', 'KCNH2', 'SELENBP1', 'SDHB', 'IL1B', 'COL3A1', 'SPARC', 'TFAP2B', 'MDFIC', 'GIMAP6', 'SERPINB9', 'CISH', 'DAB2', 'RAD23B', 'DNM1', 'LPAR6', 'EGFL7', 'PRKD1']
Refined Community 31: ['FKRP', 'GRTP1', 'POU4F2', 'UNC5D', 'PSMG1', 'WFS1', 'KCNH2', 'DLEU1', 'CHKB', 'NLRP10', 'TOP2B', 'ZNF662', 'ITGB4', 'DEFA1B', 'AHNAK2', 'BLVRA', 'FBLN5', 'KRTAP5-9', 'RECQL5', 'NUP50', 'TAL1', 'SLC26A9', 'NBL1', 'PLEK', 'DEFA3', 'CFH', 'TCN1', 'PLD3', 'IL2RG', 'AIM2', 'ZNF663P', 'SAP30BP', 'ASXL3', 'RPLP0', 'CSTB', 'STRN4', 'GPR39']
Refined Community 32: ['RAD51C', 'MORF4L1', 'UCP3', 'LDAF1', 'NUDT21', 'HSPB1', 'IGF2BP2', 'SUCLG2', 'PGLYRP1', 'CMAS', 'TNS3', 'TUBB6', 'IRAG2', 'CAMP', 'CTNNA1', 'HOXB2', 'ELP4', 'LILRA3', 'DDAH2', 'HPGDS', 'CXCL2', 'PRPF18', 'TRDC', 'CEACAM8', 'TBL1X', 'LCN2', 'GALC', 'ITM2A', 'CACNA2D3', 'CDR2L', 'HOXA9', 'CAPS2', 'ARHGEF3', 'TACC1', 'CLU', 'LTBP1', 'GRAMD4']
Refined Community 33: ['TBC1D31', 'PLXNB2', 'CLIP2', 'CEBPG', 'VASN', 'DHRS3', 'CLIP3', 'AK5', 'RPS6KA2', 'CPXM1', 'VIM', 'HNF4A', 'BAALC', 'ZFHX2', 'PGM1', 'SHANK3', 'TPPP3', 'COLEC12', 'MGLL', 'HLA-DQA1', 'CD81', 'CD1C', 'RUNX1T1', 'NEO1', 'NT5E', 'NRP1', 'FAM171A1', 'CBFB', 'TPX2', 'BPGM', 'MYH11', 'EMID1', 'PTPRCAP', 'CD59', 'ACOT7']
Refined Community 34: ['ZYX', 'ICAM2', 'GRIK5', 'VDAC1', 'SH3GL2', 'TYMP', 'POLR2H', 'RAB32', 'ANAPC10', 'HAP1', 'TKT', 'CCND3', 'PILRA', 'OGT', 'KRTAP5-9', 'HYAL2', 'PGD', 'ELK3', 'VAT1', 'KCTD7', 'CD14', 'CASP12', 'HOMER3', 'TNF', 'ENSA', 'NPC2', 'BST1', 'APLP2', 'SLC25A38', 'GTF2H2C', 'TEKT2', 'HNRNPD', 'FRMD4B', 'AOAH', 'LRP10']
Refined Community 35: ['DRD4', 'DCHS1', 'AGO1', 'GTPBP1', 'SSX2', 'HNF4A', 'LRIG1', 'DACH1', 'MKNK1', 'STK3', 'TMEM94', 'HAVCR2', 'SLC27A6', 'OPCML', 'CD302', 'SERPINF1', 'TPM1', 'P2RY14', 'CSF2', 'TRIM10', 'CCRL2', 'SMARCD3', 'SDC4', 'VRK2', 'PKMYT1', 'TLN2', 'SH3BP2', 'SLC7A8', 'NOP14', 'TUBG1', 'SPINK2', 'NOP14-AS1', 'ATP2B1', 'HSF4']
Refined Community 36: ['PML', 'GFRA1', 'ADCY2', 'SERPINB8', 'VPREB3', 'RPL28', 'NKX3-2', 'SCN9A', 'WT1', 'SYT13', 'IL17RA', 'CEBPD', 'DOCK1', 'EPHB4', 'NPR3', 'CTDSPL', 'MYH11', 'BLVRA', 'TRIM16', 'LAPTM4B', 'TMX4', 'ENPP4', 'IL6ST', 'CCL1', 'GJA4', 'FCGRT', 'NUDT21', 'MAP7', 'CRYBG1', 'TRHR', 'SELENOP', 'USP8', 'CCNB1', 'PLS1']
Refined Community 37: ['SNCAIP', 'SPCS2', 'CEP70', 'PPP2R5C', 'SMC4', 'HSF2', 'P2RX5', 'FAM110B', 'XPA', 'FOXF2', 'HLA-DRB1', 'DPPA4', 'CREM', 'DSC2', 'PIGZ', 'SRP19', 'DTYMK', 'GOSR2', 'JOSD1', 'DARS1', 'LTBP1', 'HIGD1A', 'LGALS2', 'CYP4F12', 'HLA-DPA1', 'PECAM1', 'CD74', 'RPN1', 'EPDR1', 'PFN2', 'BST2', 'PLXNB1', 'KRT81', 'ADCY2']
Refined Community 38: ['PCP4', 'F2RL1', 'CYP2C8', 'SMAGP', 'CIAO3', 'RNASE2', 'CRIM1', 'MN1', 'CFD', 'IGHM', 'PRKD2', 'AZU1', 'HSD17B4', 'ATXN1', 'QSOX1', 'FTO', 'BCL7A', 'HOXB9', 'MPZ', 'IFNAR2', 'PPIL6', 'MLLT3', 'PCDHB11', 'GNAI1', 'IL13RA1', 'STK10', 'SLC8A3', 'EEF1A2', 'ARHGEF17', 'BAALC', 'TBC1D16', 'NBL1', 'PIK3C2B']
Refined Community 39: ['CACNA2D3', 'EFHD2', 'RPS6', 'RASSF4', 'SLC15A3', 'SRY', 'SIRPAP1', 'CXCL5', 'LILRB2', 'PILRA', 'STX18', 'LILRB1', 'OPCML', 'TMEM109', 'SIGLEC7', 'STS', 'MAFB', 'IFNGR2', 'CSNK1D', 'KIT', 'RGS17', 'EPB41L3', 'MEGF6', 'VDR', 'PSAP', 'NBAS', 'ZNF22', 'TM7SF2', 'CEBPE', 'NOD2', 'ASB9', 'UBE2D1']
Refined Community 40: ['MVP', 'VOPP1', 'GTF2H2C', 'TBX1', 'LARGE2', 'CACNA2D2', 'P2RY14', 'ITGB4', 'TRH', 'CYFIP2', 'RABGGTB', 'ADAM23', 'PNMT', 'GRK5', 'C15orf39', 'TGFB1I1', 'NBL1', 'LCP1', 'PSD3', 'C11orf21', 'SRPK2', 'ZNNT1', 'RPS16', 'SMARCD2', 'PTPRM', 'NCALD', 'INPP5E', 'KDM4B', 'ROBO1', 'ZNF271P', 'VLDLR']
Refined Community 41: ['IFNA2', 'TUBB6', 'RAD17', 'NYNRIN', 'SNCAIP', 'PIGZ', 'HSPB1', 'ATN1', 'FADS1', 'GNA12', 'CYTH1', 'LDAF1', 'EPB41L2', 'SUCLG2', 'CTNNA1', 'SLC15A3', 'PROZ', 'CD52', 'RNH1', 'IGF2BP2', 'MEIS1', 'IGHM', 'HOXB2', 'PRKCB', 'ARHGEF3', 'HOXA9', 'ARHGAP44', 'H2AC18', 'EPS8', 'PPP1CA', 'LYN']
Refined Community 42: ['PCBP3', 'NAALADL1', 'VCL', 'BAHCC1', 'STIM1', 'MFNG', 'TMEM87A', 'S100A2', 'NRXN2', 'ARHGAP4', 'NKX3-2', 'AIF1L', 'CDC123', 'SIX3', 'LGALS9', 'THOC2', 'LAMC1', 'HGF', 'SLC18A1', 'FGF13', 'PSMD12', 'EVI5', 'RAB5B', 'ENDOD1', 'CYP26C1', 'KLF9', 'PTGER1', 'MEG3', 'PALLD', 'ULK2']
Refined Community 43: ['IFNA5', 'GMPR', 'TRDC', 'CEACAM4', 'PSMD2', 'IGLL1', 'P2RX5', 'CD7', 'RAB13', 'PRR5L', 'ECM1', 'IKZF2', 'GP9', 'ITGA4', 'APOBEC3B', 'ABCB1', 'FZD6', 'VOPP1', 'CFD', 'ACTN2', 'CYFIP1', 'DTX2P1-UPK3BP1-PMS2P11', 'MCL1', 'PMS2P3', 'BPI', 'MRTFA', 'CTNNA1', 'SPCS2', 'ETV6', 'DRAM1']
Refined Community 44: ['SPAG6', 'BHLHE41', 'RABGGTB', 'TEKT2', 'AEBP1', 'PCDHB14', 'BMI1', 'SLC38A1', 'SGPP2', 'FUT11', 'SOCS2', 'COBLL1', 'INPP5F', 'NR4A2', 'ZNF343', 'SNX17', 'FKBP1B', 'DNAJC6', 'PCDH9', 'DPP4', 'KIF17', 'KAT2A', 'GAMT', 'BLNK', 'IRF7', 'INSL3', 'TRPS1', 'KCNA5', 'PPID']
Refined Community 45: ['SIRPB1', 'IL2RA', 'BAIAP3', 'EZR', 'ARHGAP22', 'KIAA0930', 'TRIM16', 'PTP4A3', 'SRSF8', 'SLC5A1', 'RPS6KA3', 'HBA1', 'NFATC3', 'DRD4', 'HOXB2', 'GLI2', 'GRB10', 'IMMT', 'GOLGA8A', 'VCAN', 'CHD1', 'TRPC2', 'CCRL2', 'ACTR2', 'CCL1', 'TCF3', 'GUCY1A1', 'NIT2', 'FNDC3B']
Refined Community 46: ['BIRC5', 'LRP8', 'CACNA2D3', 'EPHA2', 'ITPR2', 'CYTL1', 'HSPA5', 'ADD2', 'CD70', 'MBNL1', 'CTDSPL', 'IFITM2', 'ITGA7', 'CTRL', 'BABAM2', 'SERTAD2', 'DACH1', 'AKR7A2', 'APOC2', 'ADAM17', 'LAMA5', 'LAMP5', 'CLIP2', 'CEP15', 'TGM5', 'RPP25', 'ZNF454', 'CADM1']
Refined Community 47: ['IGHM', 'DAB2', 'RAP1GAP', 'TGFBR1', 'CALCOCO2', 'HLA-DMB', 'SLC6A8', 'DEPDC5', 'GAPVD1', 'SLC6A9', 'NAALAD2', 'GDF15', 'RYR3', 'TAL1', 'CLCN4', 'FAXDC2', 'MYL4', 'HBZ', 'IL11RA', 'PKP3', 'SLC2A1', 'DGCR2', 'RHCE', 'FCGRT', 'SPTB', 'OSBP2', 'ACAT1']
Refined Community 48: ['ANK1', 'TMX4', 'SELENBP1', 'CDH1', 'APOBEC3A', 'MOSPD1', 'TMEM158', 'CLSTN1', 'SCML1', 'PRKCB', 'FECH', 'MYO7A', 'SLC6A8', 'CDC42BPA', 'TTLL12', 'NARS1', 'OSBP2', 'GP1BA', 'TMOD1', 'MTMR9', 'PBX1', 'CD164', 'RHCE', 'CD38', 'SNCA']
Refined Community 49: ['FKRP', 'PAWR', 'RNASE3', 'CAMK1', 'MN1', 'ENOSF1', 'FAM30A', 'DBNDD2', 'CD58', 'SEPTIN11', 'SMAGP', 'SPRY1', 'DMXL2', 'PIK3C2B', 'GPR3', 'NPDC1', 'EPB41L3', 'AZU1', 'SLC9A7', 'CFD', 'ALX1', 'RPS6KA2', 'SPTBN1', 'HNMT', 'SH3BP4']
Refined Community 50: ['CD79A', 'ABCC4', 'AKR1C3', 'PLXNB1', 'PCDHGB4', 'PHF13', 'OMD', 'ZNF22', 'MYL6B', 'SMARCA4', 'SRP9', 'FOS', 'ATP2A3', 'IL7R', 'CCND3', 'RABGGTB', 'DYNLL1', 'NUP88', 'TCF3', 'IGLJ3', 'ACADM', 'MYH9', 'ATP6V1B2', 'TRAPPC3']
Refined Community 51: ['IFI30', 'RGS2', 'FASLG', 'GZMB', 'JUN', 'P2RY2', 'PDCD1LG2', 'MECP2', 'VCY', 'DUSP7', 'DRAP1', 'PKMYT1', 'NKG7', 'NT5C2', 'PIEZO2', 'TANK', 'IFNA5', 'DDX5', 'EOMES', 'LRP10', 'P2RY14', 'PFN2', 'PDCD1', 'DLK1']
Refined Community 52: ['LTC4S', 'GLI2', 'SRSF8', 'SRGN', 'ACVR1B', 'NFKBIA', 'SEMA4A', 'MCL1', 'CKMT1B', 'SSX2', 'CFD', 'GFRA1', 'ZYX', 'TTC22', 'CCR1', 'AGT', 'ITGAX', 'CXCL8', 'IFNG', 'CAT', 'CD33', 'EMID1', 'SQSTM1']
Refined Community 53: ['CES1', 'HIF1A', 'KCNA5', 'RET', 'MST1', 'CACNA2D3', 'MAMDC2', 'ZNF254', 'TNFRSF1B', 'FKBP6', 'XAGE1B', 'CD70', 'CLEC5A', 'EXOSC3', 'ADRM1', 'STXBP2', 'LAPTM4B', 'TRPM4', 'TKTL1', 'MSLN', 'PCDHGA11', 'LAMP5', 'ACSL4']
Refined Community 54: ['TMEM87A', 'MDH2', 'RUNX1T1', 'MROH5', 'BAALC', 'TMC6', 'SLC9A8', 'ADAM12', 'EDIL3', 'MEF2C', 'ENTPD6', 'CD34', 'PIP4P2', 'TNFRSF1A', 'PGBD5', 'MTSS1', 'RANBP2', 'BAIAP3', 'DEPTOR']
Refined Community 55: ['ELOVL6', 'GALNT2', 'TM9SF4', 'MORF4L1', 'GNGT2', 'NQO2', 'CLINT1', 'PTPRS', 'ABI3', 'CSGALNACT1', 'EPHA2', 'MKNK2', 'GBX2', 'BRINP1', 'KCNK5']
Refined Community 56: ['PPP2R5D', 'H2AC16', 'ZNF646', 'CXCR4', 'PAFAH1B3', 'CD99', 'GIMAP6', 'ZFHX2', 'CITED2', 'HADHA', 'SNX17', 'ANAPC10', 'SCAP', 'PLCG1']
Refined Community 57: ['NRP2', 'GTF2A1', 'ZNF667', 'HMCES', 'KIFC1', 'GIMAP5', 'NPRL3', 'CANT1', 'IL21', 'VASP', 'PABPN1', 'CIAO1']
Refined Community 58: ['EZR', 'CYP2J2', 'HSPA5', 'KIF23', 'ZNF582', 'RAC2', 'LARGE2', 'GIMAP5', 'CDC14A', 'RPL28', 'ADCY9', 'TNFRSF21']
Refined Community 59: ['HLTF', 'DEFA3', 'NCOR1', 'MPO', 'SIN3A', 'ACVR1', 'SELL', 'CSF2', 'ACAT1', 'SRD5A1', 'NDUFA2']
Refined Community 60: ['MVP', 'CD3G', 'MRPS12', 'IDS', 'AIM2', 'PMP22', 'ARAF', 'LINC00242', 'PPP1R3B']
Refined Community 61: ['CKM', 'BTN2A1', 'MBD1', 'FOXA1', 'HOXB5', 'CTSS', 'HBA2', 'SCG5']
Refined Community 62: ['PECAM1', 'ASXL3', 'ARHGEF17', 'TACR3', 'GDI2']
Refined Community 63: ['MSRB1', 'CD14', 'TRGC2', 'RUNX1', 'DOCK1']
Refined Community 64: ['SKIDA1', 'RUNX1T1', 'ROBO1', 'PRR18', 'LRFN4']
"""    
    
    detailed_df = run_optimized_analysis(
        input_text=input_text,
        api_key=os.environ.get("DEEPSEEK_API_KEY"), 
        detailed_csv="AML Annotation partial.csv"
    )
    
    pd.set_option('display.max_colwidth', None)
    print("\nPathway Analysis Results (Clean Text):")

    main_columns = ['Community', 'Process_With_Enrichment', 'Confidence_With_Enrichment', 
                   'Process_Without_Enrichment', 'Confidence_Without_Enrichment', 
                   'Final_Process', 'Final_Confidence']
    available_main_columns = [col for col in main_columns if col in detailed_df.columns]
    print(detailed_df[available_main_columns])
    
    print(f"\nTotal communities analyzed: {len(detailed_df)}")
    print(f"Available columns: {list(detailed_df.columns)}")
    
    if 'Contributing_Genes_With_Enrichment' in detailed_df.columns:
        print(f"\nSample contributing genes (cleaned):")
        for idx, row in detailed_df.head(3).iterrows():
            print(f"Community {row['Community']}: {row.get('Contributing_Genes_With_Enrichment', 'N/A')}")
```

### Clean Process Name Columns — AML Annotation

Removes trailing embedded confidence-score artifacts from `Process_With_Enrichment` and
`Process_Without_Enrichment`, e.g.:

- `Name ([0.42])`
- `Name (0.68)`
- `Name (confidence score0.65)`
- `Name (confidence score: 0.65)`

Genuine parenthetical content in the middle of a name (e.g. `... (SREBP) Signaling`) is left alone,
since the pattern only matches a trailing parenthetical block that contains a number.

```python
import re
import pandas as pd

COLUMNS_TO_CLEAN = [
    "Process_With_Enrichment",
    "Process_Without_Enrichment",
]

SCORE_SUFFIX_PATTERN = re.compile(
    r"""
    \s*
    \(+
    \s*
    (?:confidence\s*score\s*:?\s*)?
    \[?\s*
    (-?\d*\.\d+|-?\d+)
    \s*\]?
    \s*
    \)+
    \s*$
    """,
    re.IGNORECASE | re.VERBOSE,
)


def clean_process_name(value):
    if not isinstance(value, str):
        return value

    return SCORE_SUFFIX_PATTERN.sub("", value).strip()


def clean_annotation_file(input_path: str, output_path: str):
    df = pd.read_csv(input_path)
    df.columns = [c.strip() for c in df.columns]

    for col in COLUMNS_TO_CLEAN:
        if col not in df.columns:
            print(f"Skipping '{col}' - column not found")
            continue

        df[col] = df[col].apply(clean_process_name)
        print(f"{col}: cleaned")

    df.to_csv(output_path, index=False)
    print(f"Saved cleaned file to {output_path}")

    return df



INPUT_CSV = "AML Annotation partial.csv"
OUTPUT_CSV = "AML Annotation partial.csv"

df = clean_annotation_file(INPUT_CSV, OUTPUT_CSV)
df.head()
```

### Validation Pipeline

```python
import pandas as pd
import json
import os
import re
import time
import requests
from typing import List, Dict, Any, Optional, Tuple
from tqdm import tqdm

UNKNOWN_PROCESS_LABEL = "unknown process"

def _is_unknown(process_name: str) -> bool:
    """Return True if the process name is a variant of 'unknown process'."""
    return process_name.strip().lower() == UNKNOWN_PROCESS_LABEL


class GeneSetValidator:
    def __init__(
        self,
        gene_sets_path: str,
        papers_csv: str = "AML_Paper_DB.csv",
        api_key: Optional[str] = None,
        output_path: str = "validated_gene_sets.csv",
    ):
        self.gene_sets_path = gene_sets_path
        self.papers_csv     = papers_csv
        self.output_path    = output_path
        self.api_key        = api_key or os.environ.get("DEEPSEEK_API_KEY")

        if not self.api_key:
            print("Warning: No DeepSeek API key provided.")

        self.gene_sets_df      = None
        self.papers_df         = None
        self.validated_results = []

        # ── Load Clarivate JIF journal list ───────────────────────────────────
        try:
            jif_df = pd.read_csv("journals_filtered_JIF_ge_4.csv", sep=",")
            jif_df.columns = [c.strip() for c in jif_df.columns]

            self._hq_issn_set  = set()
            self._hq_eissn_set = set()
            self.TOP_JOURNALS  = []

            for _, jr in jif_df.iterrows():
                issn  = str(jr.get("ISSN",  "")).strip().replace("-", "").upper()
                eissn = str(jr.get("eISSN", "")).strip().replace("-", "").upper()
                name  = str(jr.get("Journal name", "")).strip()

                if issn  and issn  != "NAN": self._hq_issn_set.add(issn)
                if eissn and eissn != "NAN": self._hq_eissn_set.add(eissn)
                if name:                     self.TOP_JOURNALS.append(name)

            self._top_journal_set_normalized = {
                self.normalize_journal(j) for j in self.TOP_JOURNALS
            }

            print(f"Loaded {len(jif_df)} high-quality journals from Clarivate JIF CSV "
                  f"({len(self._hq_issn_set)} ISSNs, {len(self._hq_eissn_set)} eISSNs)")

        except FileNotFoundError:
            print("Warning: journals_filtered_JIF_ge_4.csv not found. No quality filter applied.")
            self._hq_issn_set                = set()
            self._hq_eissn_set               = set()
            self.TOP_JOURNALS                = []
            self._top_journal_set_normalized = set()


    def normalize_journal(self, name) -> str:
        if pd.isna(name):
            return ""
        name = str(name).lower()
        name = name.split(" : ")[0].split(" - ")[0]
        name = re.sub(r"[^\w\s]", "", name)
        name = re.sub(r"\s+", " ", name)
        return name.strip()

    def _normalize_id(self, val) -> str:
        return str(val).strip().replace("-", "").upper() if pd.notna(val) else ""

    def _is_high_quality(self, row) -> bool:
        """Match in priority order: ISSN -> eISSN -> exact normalized journal name only."""
        issn = self._normalize_id(row.get("issn"))
        if issn and issn in self._hq_issn_set:
            return True
        eissn = self._normalize_id(row.get("eissn"))
        if eissn and eissn in self._hq_eissn_set:
            return True
        norm = self.normalize_journal(row.get("journal", ""))
        if norm and norm in self._top_journal_set_normalized:
            return True
        return False

    def _load_papers_for_genes(self, genes: List[str]) -> pd.DataFrame:
        if self.papers_df is None or self.papers_df.empty:
            return pd.DataFrame()

        genes_upper = {g.upper() for g in genes if g}
        mask   = self.papers_df["gene"].str.upper().isin(genes_upper)
        subset = self.papers_df[mask].copy()

        if subset.empty:
            return subset

        if "full_text" not in subset.columns:
            subset["full_text"] = "Full text not available via PMC"

        subset["matched_journal"] = subset.apply(
            lambda row: "HQ" if self._is_high_quality(row) else None, axis=1
        )

        hq  = subset["matched_journal"].notna().sum()
        tot = len(subset)
        print(f"  Paper filter: {hq}/{tot} high-quality papers for gene set")
        return subset


    def load_data(self):
        print("Loading gene sets data...")
        self.gene_sets_df = pd.read_csv(self.gene_sets_path)
        self.gene_sets_df.columns = [col.strip() for col in self.gene_sets_df.columns]
        print(f"Loaded {len(self.gene_sets_df)} gene sets from {self.gene_sets_path}")

        print(f"Loading AML paper database from {self.papers_csv}...")
        if not os.path.exists(self.papers_csv):
            print(f"Paper CSV '{self.papers_csv}' not found.")
            self.papers_df = pd.DataFrame()
        else:
            self.papers_df = pd.read_csv(self.papers_csv, low_memory=False)
            self.papers_df.columns = [col.strip() for col in self.papers_df.columns]
            if "gene" not in self.papers_df.columns:
                print("'gene' column not found in paper CSV - gene filtering disabled.")
                self.papers_df = pd.DataFrame()
            else:
                self.papers_df["gene"] = self.papers_df["gene"].astype(str).str.strip()
                print(f"Loaded {len(self.papers_df)} papers ({self.papers_df['gene'].nunique()} unique genes)")


    def extract_genes_from_set(self, gene_set_row: pd.Series) -> List[str]:
        genes = []
        possible_gene_columns = [
            "Genes_String", "Contributing_Genes", "Genes", "Final_Contributing_Genes",
        ]
        for col in possible_gene_columns:
            if col in gene_set_row and pd.notna(gene_set_row[col]):
                gene_data = gene_set_row[col]
                if isinstance(gene_data, list):
                    genes = gene_data
                elif isinstance(gene_data, str):
                    if "," in gene_data:
                        genes = [g.strip() for g in gene_data.split(",")]
                    elif ";" in gene_data:
                        genes = [g.strip() for g in gene_data.split(";")]
                    else:
                        genes = [gene_data.strip()]
                break
        return [g for g in genes if g and g.strip()]


    def extract_gene_related_abstracts(
        self,
        genes: List[str],
        papers_df: pd.DataFrame,
        max_papers: int = 50,
    ) -> Tuple[List[Dict], List[str]]:
        """
        Filter to HQ papers only, then prioritise by gene_count descending.

        gene_count = number of query genes found in the COMBINED
        abstract + full_text corpus (never short-circuited by the gene column).
        The gene column is used only as a safety net to ensure the primary
        gene is never missed if the text search fails.

        """
        if isinstance(genes, str):
            genes = [g.strip() for g in genes.split(",")]

        genes_set = {g.upper() for g in genes if g}

        hq_papers = []

        for _, paper in papers_df.iterrows():

            # ── HQ gate ───────────────────────────────────────────────────────
            if pd.isna(paper.get("matched_journal")):
                continue

            abstract  = str(paper.get("abstract",  ""))
            full_text = str(paper.get("full_text", ""))

            if abstract  in ("nan", "Abstract not available", ""):
                abstract = ""
            if full_text in ("nan", "Full text not available via PMC", ""):
                full_text = ""

            if not abstract and not full_text:
                continue

            # ── Always scan abstract + full_text combined corpus ──────────────
            search_corpus = (abstract + " " + full_text).strip()
            mentioned_genes = [
                g for g in genes
                if g and re.search(rf"\b{re.escape(g)}\b", search_corpus, re.IGNORECASE)
            ]
            gene_from_col = str(paper.get("gene", "")).strip()
            if gene_from_col and gene_from_col.upper() in genes_set:
                if gene_from_col not in mentioned_genes:
                    mentioned_genes.append(gene_from_col)

            if not mentioned_genes:
                continue

            truncated_abstract  = (abstract[:500]  + "...") if len(abstract)  > 500  else abstract
            truncated_full_text = (full_text[:3000] + "...") if len(full_text) > 3000 else full_text

            hq_papers.append({
                "title":           paper.get("title",   "No title"),
                "authors":         paper.get("authors", "No authors"),
                "journal":         paper.get("journal", "No journal"),
                "year":            paper.get("year",    "Unknown"),
                "abstract":        truncated_abstract,
                "full_text":       truncated_full_text,
                "doi":             paper.get("doi",     "No DOI"),
                "pmid":            paper.get("pmid",    "No PMID"),
                "genes_mentioned": mentioned_genes,
                "gene_count":      len(mentioned_genes),
            })

        hq_papers.sort(key=lambda x: x["gene_count"], reverse=True)
        relevant_papers = hq_papers[:max_papers]

        top_journal_papers_used = [
            f"Paper {idx}: '{p['title']}', {p['authors']}"
            for idx, p in enumerate(relevant_papers, 1)
        ]

        return relevant_papers, top_journal_papers_used

    def create_validation_prompt(
        self,
        gene_set_id: str,
        genes: List[str],
        process_with_enrichment: str,
        confidence_with_enrichment: float,
        process_without_enrichment: str,
        confidence_without_enrichment: float,
        analysis_with_enrichment: str,
        analysis_without_enrichment: str,
        relevant_papers: List[Dict],
    ) -> str:

        genes_str      = ", ".join(genes) if isinstance(genes, list) else str(genes)
        papers_section = ""
        for i, paper in enumerate(relevant_papers, 1):
            papers_section += (
                f"PAPER {i}:\n"
                f"Title: {paper['title']}\n"
                f"Authors: {paper['authors']}\n"
                f"Journal: {paper['journal']} ({paper['year']})\n"
                f"Genes Mentioned: {', '.join(paper['genes_mentioned'])}\n"
                f"Abstract: {paper['abstract']}\n"
                f"\n"
            )

        return f"""You are a scientific expert in genomics and bioinformatics tasked with validating gene set analysis results using ONLY the provided literature.

GENE SET ID: {gene_set_id}
GENES: {genes_str}

ORIGINAL ANALYSIS WITH ENRICHMENT:
Process Name: {process_with_enrichment}
Original Confidence Score: {confidence_with_enrichment}
Analysis: {analysis_with_enrichment[:1000]}...

ORIGINAL ANALYSIS WITHOUT ENRICHMENT:
Process Name: {process_without_enrichment}
Original Confidence Score: {confidence_without_enrichment}
Analysis: {analysis_without_enrichment[:1000]}...

PROVIDED LITERATURE (USE ONLY THESE STUDIES):
{papers_section}

VALIDATION TASK:
Based STRICTLY on the provided literature above, evaluate both analyses and provide updated confidence scores.

**ABSOLUTE REQUIREMENT FOR PAPER CITATIONS:**
Whenever you reference a paper anywhere in your response, always use this exact format:
"Paper X (FirstAuthorLastName et al.)"
For example: "Paper 3 (Nakamura et al.) shows..." or "supported by Paper 1 (Chen et al.) and Paper 4 (Okafor et al.)"
Do NOT use bare numbers like "Paper 3" alone, and do NOT spell out full titles or complete author lists in-line — the first-author-et-al form is sufficient everywhere in the response.

CRITICAL REQUIREMENTS:
1. Use ONLY the provided papers - do not add external knowledge
2. For each process, check if the genes are supported by the literature
3. Provide updated confidence scores based on evidence strength
4. Select the better-supported process as the final choice
5. **MANDATORY:  When referencing papers by number anywhere in your analysis text, always use the "Paper X (FirstAuthor et al.)" format defined above — never bare numbers, never full citations..**
6. The final selected process MUST be either:
   (a) exactly one of the two original process names, OR
   (b) "Neither process" ONLY if both updated confidence scores (Updated Confidence With Enrichment, Updated Confidence Without Enrichment) are <= 0.05.

7. The final confidence MUST follow:
   - If a process is selected: Final Confidence = the updated confidence of that selected process.
   - If "Neither process" is selected: Final Confidence = max(Updated Confidence With Enrichment, Updated Confidence Without Enrichment).

FORMAT YOUR RESPONSE EXACTLY AS FOLLOWS:

VALIDATION OF ENRICHMENT ANALYSIS:
Evidence Assessment: [Detailed assessment based strictly on provided papers]
Original Confidence: {confidence_with_enrichment}
Updated Confidence: [Your revised score 0.00-1.00 based on literature evidence]
Supporting Papers: [List paper numbers that support this process]

VALIDATION OF DIRECT ANALYSIS:
Evidence Assessment: [Detailed assessment based strictly on provided papers]
Original Confidence: {confidence_without_enrichment}
Updated Confidence: [Your revised score 0.00-1.00 based on literature evidence]
Supporting Papers: [List paper numbers that support this process]

FINAL PROCESS SELECTION:
Selected Process: [Choose the better-supported process name]
Final Confidence: [The updated confidence score for your selected process]
Selection Reasoning: [Explain why this process has stronger literature support]

CONFLICT ANALYSIS:
Review all papers for contradictory evidence about gene functions, pathway assignments, or experimental results.
CONFLICTING_EVIDENCE_FOUND: [TRUE/FALSE]
CONFLICT_DESCRIPTION: [Brief description of any conflicts found, or "No conflicts detected"]

SUPPORTING CITATIONS:
[List citations in format: "Title, Authors" for papers that support the final selected process]

VALIDATION ANALYSIS TEXT:
[Comprehensive summary of all changes made, reasoning for confidence adjustments, and evidence from the provided studies that led to the final process selection.
Use "Paper X (FirstAuthor et al.)" format for any paper references.]"""

    def call_deepseek_api(self, prompt: str) -> str:
        if not self.api_key:
            raise ValueError("DeepSeek API key is required.")

        api_url = "https://api.deepseek.com/v1/chat/completions"
        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type":  "application/json",
        }
        data = {
            "model": "deepseek-chat",
            "messages": [
                {
                    "role":    "system",
                    "content": (
                        "You are a scientific expert in genomics and bioinformatics "
                        "specializing in gene set analysis validation. "
                        "Base your analysis strictly on the provided literature."
                    ),
                },
                {"role": "user", "content": prompt},
            ],
            "temperature": 0,
            "max_tokens":  8000,
        }

        max_retries = 3
        retry_delay = 5

        for attempt in range(max_retries):
            try:
                response = requests.post(api_url, headers=headers, json=data, timeout=180)

                if response.status_code == 200:
                    return response.json()["choices"][0]["message"]["content"]

                elif response.status_code == 429:
                    wait_time = int(response.headers.get("Retry-After", retry_delay * 2))
                    print(f"Rate limited. Waiting {wait_time}s...")
                    time.sleep(wait_time)

                elif 500 <= response.status_code < 600:
                    print(f"Server error {response.status_code}. Retrying in {retry_delay}s...")
                    time.sleep(retry_delay)
                    retry_delay *= 2

                else:
                    raise Exception(f"API Error {response.status_code}: {response.text}")

            except requests.exceptions.Timeout:
                print(f"Timeout on attempt {attempt + 1}/{max_retries}")
                if attempt < max_retries - 1:
                    time.sleep(retry_delay)
                    retry_delay *= 2
                else:
                    raise Exception(f"API timeout after {max_retries} attempts")

            except requests.exceptions.RequestException as e:
                if attempt < max_retries - 1:
                    time.sleep(retry_delay)
                    retry_delay *= 2
                else:
                    raise Exception(f"Request failed after {max_retries} attempts: {e}")

        raise Exception(f"Failed to get a valid response after {max_retries} attempts")

    def parse_llm_response(self, response: str) -> Dict[str, Any]:
        results = {}

        m = re.search(
            r"VALIDATION OF ENRICHMENT ANALYSIS:.*?Updated Confidence:\s*([\d.]+)",
            response, re.DOTALL,
        )
        if m:
            results["confidence_with_enrichment_after"] = float(m.group(1))

        m = re.search(
            r"VALIDATION OF DIRECT ANALYSIS:.*?Updated Confidence:\s*([\d.]+)",
            response, re.DOTALL,
        )
        if m:
            results["confidence_without_enrichment_after"] = float(m.group(1))

        m = re.search(r"Selected Process:\s*(.+?)(?=\n|$)", response)
        if m:
            results["final_process"] = m.group(1).strip()

        m = re.search(r"Final Confidence:\s*([\d.]+)", response)
        if m:
            results["final_confidence"] = float(m.group(1))

        m = re.search(r"CONFLICTING_EVIDENCE_FOUND:\s*(TRUE|FALSE)", response, re.IGNORECASE)
        results["conflicting_evidence_found"] = (
            m.group(1).upper() == "TRUE" if m else False
        )

        m = re.search(
            r"CONFLICT_DESCRIPTION:\s*(.+?)(?=\n\n|SUPPORTING CITATIONS:|$)",
            response, re.DOTALL,
        )
        results["conflict_description"] = m.group(1).strip() if m else "No conflicts detected"

        m = re.search(
            r"SUPPORTING CITATIONS:\s*(.+?)(?=\n\n|VALIDATION ANALYSIS TEXT:|$)",
            response, re.DOTALL,
        )
        if m:
            results["supporting_citations"] = [
                line.strip()
                for line in m.group(1).strip().split("\n")
                if line.strip() and not line.strip().startswith(("-", "*"))
            ]
        else:
            results["supporting_citations"] = []

        m = re.search(r"VALIDATION ANALYSIS TEXT:\s*(.+?)$", response, re.DOTALL)
        if m:
            results["validation_analysis_text"] = m.group(1).strip()

        return results

    def validate_gene_set(self, gene_set_row: pd.Series) -> Dict[str, Any]:
        set_id = str(
            gene_set_row.get("Community", "")
            or gene_set_row.get("Set_ID", "")
            or gene_set_row.get("community", "")
            or gene_set_row.get("index", "")
        )

        genes = self.extract_genes_from_set(gene_set_row)
        if not genes:
            print(f"  No genes found for set {set_id}")
            return self._error_result(set_id, "No genes found for this set")

        process_with    = str(gene_set_row.get("Process_With_Enrichment", "")
                              or gene_set_row.get("Final_Process", ""))
        conf_with       = float(gene_set_row.get("Confidence_With_Enrichment", 0)
                                or gene_set_row.get("Final_Confidence", 0))
        process_without = str(gene_set_row.get("Process_Without_Enrichment", ""))
        conf_without    = float(gene_set_row.get("Confidence_Without_Enrichment", 0))

        analysis_with    = str(gene_set_row.get("Analysis_Text_With_Enrichment", "")
                               or gene_set_row.get("Pathway_Reasoning_With_Enrichment", "")
                               or gene_set_row.get("Full_Analysis", ""))
        analysis_without = str(gene_set_row.get("Analysis_Text_Without_Enrichment", "")
                               or gene_set_row.get("Pathway_Reasoning_Without_Enrichment", ""))

        # ── Unknown process guard ─────────────────────────────────────────────
        with_is_unknown    = _is_unknown(process_with)
        without_is_unknown = _is_unknown(process_without)

        if with_is_unknown and without_is_unknown:
            print(f"  Both processes are 'unknown process' for set {set_id} - skipping.")
            return self._both_unknown_result(set_id, genes, gene_set_row)

        if with_is_unknown:
            print(f"  'With enrichment' is unknown process for set {set_id} - zeroing out, validating direct only.")
            process_with  = UNKNOWN_PROCESS_LABEL
            conf_with     = 0.0
            analysis_with = "Unknown process - not validated."

        if without_is_unknown:
            print(f"  'Without enrichment' is unknown process for set {set_id} - zeroing out, validating enrichment only.")
            process_without  = UNKNOWN_PROCESS_LABEL
            conf_without     = 0.0
            analysis_without = "Unknown process - not validated."

        papers_df = self._load_papers_for_genes(genes)

        relevant_papers, top_journal_info = self.extract_gene_related_abstracts(
            genes, papers_df, max_papers=50
        )

        if not relevant_papers:
            print(f"  No matching high-quality papers for set {set_id}")
            return self._no_papers_result(set_id, genes, gene_set_row)

        prompt = self.create_validation_prompt(
            gene_set_id=set_id,
            genes=genes,
            process_with_enrichment=process_with,
            confidence_with_enrichment=conf_with,
            process_without_enrichment=process_without,
            confidence_without_enrichment=conf_without,
            analysis_with_enrichment=analysis_with,
            analysis_without_enrichment=analysis_without,
            relevant_papers=relevant_papers,
        )

        os.makedirs("prompts", exist_ok=True)
        with open(f"prompts/prompt_{set_id}.txt", "w", encoding="utf-8") as f:
            f.write(prompt)

        try:
            response = self.call_deepseek_api(prompt)

            os.makedirs("responses", exist_ok=True)
            with open(f"responses/response_{set_id}.txt", "w", encoding="utf-8") as f:
                f.write(response)

            parsed = self.parse_llm_response(response)

            conf_with_after    = parsed.get("confidence_with_enrichment_after",   conf_with)
            conf_without_after = parsed.get("confidence_without_enrichment_after", conf_without)
            if with_is_unknown:
                conf_with_after = 0.0
            if without_is_unknown:
                conf_without_after = 0.0

            return {
                "Set_ID":                                set_id,
                "Genes":                                 genes,
                "Process_With_Enrichment_Original":      process_with,
                "Process_Without_Enrichment_Original":   process_without,
                "Confidence_With_Enrichment_Before":     conf_with,
                "Confidence_Without_Enrichment_Before":  conf_without,
                "Confidence_With_Enrichment_After":      conf_with_after,
                "Confidence_Without_Enrichment_After":   conf_without_after,
                "Final_Process":                         parsed.get("final_process",   process_with),
                "Final_Confidence":                      parsed.get("final_confidence", conf_with),
                "Validation_Analysis_Text":              parsed.get("validation_analysis_text", "No analysis provided"),
                "Supporting_Citations":                  parsed.get("supporting_citations", []),
                "Conflicting_Evidence_Found":            parsed.get("conflicting_evidence_found", False),
                "Conflict_Description":                  parsed.get("conflict_description", "No conflicts detected"),
                "Total_Papers_Found":                    len(relevant_papers),
            }

        except Exception as e:
            print(f"  API error for set {set_id}: {e}")
            return self._error_result(set_id, f"API error: {e}", genes, gene_set_row)

    def validate_all_gene_sets(self):
        if self.gene_sets_df is None:
            self.load_data()

        results    = []
        total_sets = len(self.gene_sets_df)
        print(f"\nStarting validation of {total_sets} gene sets...")

        pbar = tqdm(
            total=total_sets,
            desc="Validating gene sets",
            bar_format="{desc}: {percentage:3.0f}%|{bar}| {n_fmt}/{total_fmt} [{elapsed}<{remaining}]",
            ncols=100,
        )

        for i, (_, row) in enumerate(self.gene_sets_df.iterrows()):
            set_id = str(
                row.get("Community", "")
                or row.get("Set_ID", "")
                or row.get("community", "")
                or i
            )
            try:
                result = self.validate_gene_set(row)
                results.append(result)
            except Exception as e:
                print(f"\nUnexpected error on set {set_id}: {e}")

            pbar.update(1)
            pbar.set_postfix_str(f"Set {set_id}")

            if (i + 1) % 5 == 0:
                self.validated_results = results
                self.save_results(f"{self.output_path}.partial")

        pbar.close()
        self.validated_results = results
        self.save_results(self.output_path)
        print(f"\nValidation complete! Processed {len(results)} gene sets")
        return results

    def save_results(self, output_path: str = None):
        if not self.validated_results:
            print("No results to save")
            return

        path = output_path or self.output_path
        df   = pd.DataFrame(self.validated_results)

        required_columns = [
            "Set_ID", "Genes",
            "Process_With_Enrichment_Original", "Process_Without_Enrichment_Original",
            "Confidence_With_Enrichment_Before", "Confidence_Without_Enrichment_Before",
            "Confidence_With_Enrichment_After",  "Confidence_Without_Enrichment_After",
            "Final_Process", "Final_Confidence",
            "Validation_Analysis_Text", "Supporting_Citations",
            "Conflicting_Evidence_Found", "Conflict_Description",
            "Total_Papers_Found",
        ]
        for col in required_columns:
            if col not in df.columns:
                df[col] = None
        df = df[required_columns]
        df.to_csv(path, index=False)
        print(f"Saved {len(df)} validated results to {path}")


    def generate_summary_report(self) -> str:
        if not self.validated_results:
            return "No validation results available"

        total = len(self.validated_results)

        conflict_count     = sum(1 for r in self.validated_results if r.get("Conflicting_Evidence_Found"))
        total_paper_counts = [r.get("Total_Papers_Found", 0)           for r in self.validated_results]
        final_confidences  = [r.get("Final_Confidence", 0)             for r in self.validated_results]

        conf_delta_enrich = []
        conf_delta_direct = []
        for r in self.validated_results:
            be, ae = r.get("Confidence_With_Enrichment_Before", 0),   r.get("Confidence_With_Enrichment_After", 0)
            bd, ad = r.get("Confidence_Without_Enrichment_Before", 0), r.get("Confidence_Without_Enrichment_After", 0)
            if be > 0 and ae > 0: conf_delta_enrich.append(ae - be)
            if bd > 0 and ad > 0: conf_delta_direct.append(ad - bd)

        lines = [
            "Gene Set Validation Summary Report",
            "=" * 50,
            f"Total gene sets processed: {total}",
            "",
            "CONFLICT ANALYSIS:",
            f"  With conflicts:    {conflict_count} ({conflict_count/total*100:.1f}%)",
            f"  Without conflicts: {total - conflict_count} ({(total-conflict_count)/total*100:.1f}%)",
            "",
            "PAPER DISCOVERY:",
            f"  Avg papers/set:  {sum(total_paper_counts)/len(total_paper_counts):.2f}",
            f"  Sets with papers: {sum(1 for c in total_paper_counts if c > 0)} ({sum(1 for c in total_paper_counts if c > 0)/total*100:.1f}%)",
            "",
            "CONFIDENCE:",
            f"  Avg final confidence: {sum(final_confidences)/len(final_confidences):.2f}",
            f"  Range: {min(final_confidences):.2f} - {max(final_confidences):.2f}",
        ]
        if conf_delta_enrich:
            lines.append(f"  Avg enrichment confidence delta: {sum(conf_delta_enrich)/len(conf_delta_enrich):+.3f}")
        if conf_delta_direct:
            lines.append(f"  Avg direct confidence delta:     {sum(conf_delta_direct)/len(conf_delta_direct):+.3f}")

        return "\n".join(lines)


    def _error_result(
        self,
        set_id: str,
        error_msg: str,
        genes: List[str] = None,
        row: pd.Series = None,
    ) -> Dict[str, Any]:
        genes = genes or []
        return {
            "Set_ID":                                set_id,
            "Genes":                                 genes,
            "Process_With_Enrichment_Original":      row.get("Process_With_Enrichment", "") if row is not None else "",
            "Process_Without_Enrichment_Original":   row.get("Process_Without_Enrichment", "") if row is not None else "",
            "Confidence_With_Enrichment_Before":     row.get("Confidence_With_Enrichment", 0) if row is not None else 0,
            "Confidence_Without_Enrichment_Before":  row.get("Confidence_Without_Enrichment", 0) if row is not None else 0,
            "Confidence_With_Enrichment_After":      0,
            "Confidence_Without_Enrichment_After":   0,
            "Final_Process":                         "ERROR",
            "Final_Confidence":                      0,
            "Validation_Analysis_Text":              error_msg,
            "Supporting_Citations":                  [],
            "Conflicting_Evidence_Found":            False,
            "Conflict_Description":                  "Error occurred during validation",
            "Total_Papers_Found":                    0,
        }

    def _no_papers_result(
        self, set_id: str, genes: List[str], row: pd.Series
    ) -> Dict[str, Any]:
        conf_with = float(row.get("Confidence_With_Enrichment", 0) or row.get("Final_Confidence", 0))
        return {
            "Set_ID":                                set_id,
            "Genes":                                 genes,
            "Process_With_Enrichment_Original":      str(row.get("Process_With_Enrichment", "") or row.get("Final_Process", "")),
            "Process_Without_Enrichment_Original":   str(row.get("Process_Without_Enrichment", "")),
            "Confidence_With_Enrichment_Before":     conf_with,
            "Confidence_Without_Enrichment_Before":  float(row.get("Confidence_Without_Enrichment", 0)),
            "Confidence_With_Enrichment_After":      conf_with,
            "Confidence_Without_Enrichment_After":   float(row.get("Confidence_Without_Enrichment", 0)),
            "Final_Process":                         str(row.get("Process_With_Enrichment", "") or row.get("Final_Process", "")),
            "Final_Confidence":                      conf_with,
            "Validation_Analysis_Text":              "No high-quality papers found for this gene set",
            "Supporting_Citations":                  [],
            "Conflicting_Evidence_Found":            False,
            "Conflict_Description":                  "No high-quality papers available",
            "Total_Papers_Found":                    0,
        }

    def _both_unknown_result(
        self, set_id: str, genes: List[str], row: pd.Series
    ) -> Dict[str, Any]:
        return {
            "Set_ID":                                set_id,
            "Genes":                                 genes,
            "Process_With_Enrichment_Original":      UNKNOWN_PROCESS_LABEL,
            "Process_Without_Enrichment_Original":   UNKNOWN_PROCESS_LABEL,
            "Confidence_With_Enrichment_Before":     0,
            "Confidence_Without_Enrichment_Before":  0,
            "Confidence_With_Enrichment_After":      0,
            "Confidence_Without_Enrichment_After":   0,
            "Final_Process":                         UNKNOWN_PROCESS_LABEL,
            "Final_Confidence":                      0,
            "Validation_Analysis_Text":              "Both processes are unknown - validation skipped.",
            "Supporting_Citations":                  [],
            "Conflicting_Evidence_Found":            False,
            "Conflict_Description":                  "Both processes unknown - no validation performed",
            "Total_Papers_Found":                    0,
        }



def main():
    gene_sets_path = "AML Annotation partial.csv"
    papers_csv     = "AML_Paper_DB.csv"
    output_path    = "AML Validation partial.csv"

    print("AML Gene Set Validation Pipeline")
    print("=" * 50)

    api_key = os.environ.get("DEEPSEEK_API_KEY")
    if not api_key:
        print("No DEEPSEEK_API_KEY env var found.")
        print("Set it with:  export DEEPSEEK_API_KEY=your_key")
        return

    print("API key found")

    validator = GeneSetValidator(
        gene_sets_path=gene_sets_path,
        papers_csv=papers_csv,
        api_key=api_key,
        output_path=output_path,
    )

    validator.load_data()

    test_mode = False

    if test_mode:
        print("\nTest mode - validating first gene set only")
        result = validator.validate_gene_set(validator.gene_sets_df.iloc[0])
        validator.validated_results = [result]
        validator.save_results("test_validation.csv")
        for k, v in result.items():
            print(f"  {k}: {len(v) if isinstance(v, list) else v}")
    else:
        validator.validate_all_gene_sets()

    report = validator.generate_summary_report()
    print("\n" + report)
    with open("validation_summary.txt", "w") as f:
        f.write(report)

    print("\nDone!")


if __name__ == "__main__":
    main()
```
