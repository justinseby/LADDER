### Annotation Pipeline

```python
class OptimizedPathwayAnalyzer:
    def __init__(self):
        self.api_key = None
        self.api_url = "https://api.deepseek.com/v1/chat/completions"
        self.checkpoint_file = f"checkpoint_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
        self.results_csv = f"results_{datetime.now().strftime('%Y%m%d_%H%M%S')}.csv"
        self.pathway_gene_mapping = {}
        
    def set_api_key(self, api_key: str = None):
        """
        Set the Deepseek API key
        """
        if api_key:
            self.api_key = api_key
        elif os.environ.get("DEEPSEEK_API_KEY"):
            self.api_key = os.environ.get("DEEPSEEK_API_KEY")
        else:
            raise ValueError("No Deepseek API key provided. Set it via argument or DEEPSEEK_API_KEY environment variable.")
    
    def _create_system_prompt(self) -> str:
        return """You are a bioinformatics expert. Focus on Lung Cancer. Provide detailed pathway analysis with scientific literature support. Always cite specific papers when discussing pathway relationships."""
    
    def _create_all_in_one_prompt(self, genes: List[str], enrichment_pathways: List[str]) -> str:
        enrichment_section = ""
        if enrichment_pathways and len(enrichment_pathways) > 0:
            enrichment_section = "\nEnrichment Analysis Results:\n" + "\n".join(f"- {pathway}" for pathway in enrichment_pathways)
        
        genes_str = ", ".join(genes)
        
        enrichment_instruction = "Using the enrichment pathways above (if any), provide an analysis of the gene set"
        
        return f"""You are a bioinformatics expert conducting pathway analysis of gene sets. For the gene set: [{genes_str}]{enrichment_section}

Please perform analysis in clearly labeled sections:

===== SECTION 1: ANALYSIS WITH ENRICHMENT =====
{enrichment_instruction}:
1. Propose a concise, descriptive name for the biological process
2. Assign a confidence score (0.00-1.00) for this process
3. Provide explicit reasoning for choosing this pathway/process
4. List the specific genes from the gene set that contribute to the process
5. IMPORTANT: A minimum of TWO genes from the gene set MUST be associated with a common biological process to annotate it as a valid pathway
6. If there isn't sufficient evidence or fewer than two genes are associated with a common pathway, label as "Unknown Pathway" with a low confidence score

Output format for this section:
PROCESS WITH ENRICHMENT: [Process Name] ([Confidence Score])

PATHWAY REASONING WITH ENRICHMENT:
[Your explicit reasoning for choosing this pathway/process]

CONTRIBUTING GENES WITH ENRICHMENT:
[Comma-separated list of contributing genes]

ANALYSIS TEXT WITH ENRICHMENT:
[Detailed analysis text explaining the biological significance and mechanisms of this process, without citations]

===== SECTION 2: ANALYSIS WITHOUT ENRICHMENT =====
Based SOLELY on your knowledge of these genes, without considering the enrichment results:
1. Propose a concise, descriptive name for the biological process
2. Assign a confidence score (0.00-1.00) for this process
3. List the specific genes from the gene set that are involved in the process
4. IMPORTANT: A minimum of TWO genes from the gene set MUST be associated with a common biological process to annotate it as a valid pathway
5. IMPORTANT: If there isn't sufficient evidence or fewer than two genes share a common biological function, label as "Unknown Pathway" with a low confidence score

Output format for this section:
PROCESS WITHOUT ENRICHMENT: [Process Name] ([Confidence Score])

CONTRIBUTING GENES WITHOUT ENRICHMENT:
[Comma-separated list of contributing genes]

PATHWAY REASONING WITHOUT ENRICHMENT:
[Your explicit reasoning for choosing this pathway/process]

ANALYSIS TEXT WITHOUT ENRICHMENT:
[Detailed analysis text explaining the biological significance and mechanisms of this process, without citations]

===== SECTION 3: FINAL PROCESS SELECTION =====
Compare both analyses and provide:
1. Which process (with or without enrichment) has higher confidence and why
2. Final reasoning for the chosen process

Output format for this section:
FINAL PROCESS REASONING:
[Detailed explanation of why you chose the final process, comparing both analyses and explaining which one provides stronger evidence]

Analytical Guidelines:
- Be concise and avoid unnecessary words
- Be factual without editorializing
- Be specific, avoiding overly general statements
- Avoid listing individual protein facts
- Group proteins by similar functions
- Discuss their interplay, synergistic or antagonistic effects
- Focus on functional integration within the system
- Include at least 2-3 specific paper citations when discussing established pathway relationships

Confidence Score Instructions:
- Assign a score from 0.00 to 1.00
- 0.00 indicates lowest confidence
- 1.00 reflects highest confidence
- Base the score on the proportion of genes participating in the identified process"""
    
    def _make_deepseek_request(self, prompt: str) -> str:
        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type": "application/json"
        }
        
        data = {
            "model": "deepseek-chat",
            "messages": [
                {"role": "system", "content": self._create_system_prompt()},
                {"role": "user", "content": prompt}
            ],
            "temperature": 0,
            "max_tokens": 4000
        }
        
        response = requests.post(self.api_url, headers=headers, json=data)
        
        if response.status_code == 200:
            result = response.json()
            return result['choices'][0]['message']['content']
        else:
            raise Exception(f"Deepseek API request failed: {response.status_code} - {response.text}")
    
    def parse_communities(self, input_text: str) -> Dict[str, List[str]]:
        communities = {}
        for line in input_text.strip().split('\n'):
            line = line.strip()
            if 'Refined Community' in line:
                match = re.search(r'Refined Community (\d+): \[(.*?)\]', line)
                if match:
                    community_num = match.group(1)
                    genes_str = match.group(2)
                    genes = [g.strip("'") for g in genes_str.split(', ')]
                    communities[community_num] = genes
        return communities
    
    def perform_enrichment(self, gene_list: List[str]) -> List[str]:

        try:
            databases = ['GO_Biological_Process_2021', 'Reactome_2022', 'KEGG_2021_Human']
            
            all_results = []
            for database in databases:
                enr = gp.enrichr(gene_list=gene_list,
                                gene_sets=[database],
                                organism='Human',
                                outdir=None,
                                no_plot=True,
                                cutoff=0.01)
                
                results_df = enr.results
                if not results_df.empty:
                    results_df = results_df.sort_values('Adjusted P-value')
                    for _, row in results_df.head(5).iterrows():
                        all_results.append(row['Term'])
            
            return list(set(all_results))
        
        except Exception as e:
            print(f"Enrichment analysis failed: {str(e)}")
            return []
    
    def _clean_markdown_formatting(self, text: str) -> str:

        if not text or not isinstance(text, str):
            return text
            
        text = re.sub(r'\*([^*]+)\*', r'\1', text)
        
        text = re.sub(r'\*\*([^*]+)\*\*', r'\1', text)
        
        text = re.sub(r'(?<!\w)\*(?!\w)', '', text)
        
        return text.strip()
    
    def _extract_section(self, text: str, section_name: str) -> str:

        patterns = [
            rf"{re.escape(section_name)}:\s*(.*?)(?=\n\n[\w\s]+:|===|$)",  # Main pattern
            rf"{re.escape(section_name)}:\s*(.*?)(?=\n[A-Z][A-Z\s]+:|$)",
            rf"{re.escape(section_name)}:\s*(.*?)(?=\n\*\*|$)" 
        ]
        
        for pattern in patterns:
            match = re.search(pattern, text, re.DOTALL | re.IGNORECASE)
            if match:
                result = match.group(1).strip()
                if result:
                    return self._clean_markdown_formatting(result)
        
        return f"{section_name} not found"
    
    def _extract_process_info(self, text: str, prefix: str) -> Tuple[str, float]:

        patterns = [
            rf"PROCESS {prefix} ENRICHMENT:\s*([^(]+?)\s*\(([0-9.]+)\)",
            rf"PROCESS {prefix} ENRICHMENT:\s*([^(]*?)\s*\(\s*([0-9.]+)\s*\)",
            rf"PROCESS {prefix} ENRICHMENT:\s*(.*?)\s*\(\s*Confidence Score:\s*([0-9.]+)\s*\)",
            rf"PROCESS {prefix} ENRICHMENT:\s*(.*?)\s*\*\*Confidence Score:\s*([0-9.]+)\*\*",
            rf"PROCESS {prefix} ENRICHMENT:\s*(.*?)\s*\(\s*\*\*Confidence Score:\s*([0-9.]+)\*\*\s*\)"
        ]
        
        for pattern in patterns:
            match = re.search(pattern, text, re.IGNORECASE | re.DOTALL)
            if match:
                process_name = match.group(1).strip()
                try:
                    confidence_score = float(match.group(2))
                    process_name = self._clean_markdown_formatting(process_name)
                    process_name = process_name.strip()
                    if process_name and confidence_score >= 0:
                        return process_name, confidence_score
                except ValueError:
                    continue
        
        simple_pattern = rf"PROCESS {prefix} ENRICHMENT:\s*(.*?)(?=\n|$)"
        match = re.search(simple_pattern, text, re.IGNORECASE)
        if match:
            process_name = match.group(1).strip()
            process_name = self._clean_markdown_formatting(process_name)
            process_name = re.sub(r'\s*\([^)]*$', '', process_name)
            process_name = process_name.strip()
            return process_name, 0.0
        
        return f"Unknown Process {prefix} Enrichment", 0.0
    
    def _extract_final_process(self, text: str) -> str:

        match = re.search(r"FINAL PROCESS REASONING:(.*?)(?=\n\n|$)", text, re.DOTALL | re.IGNORECASE)
        if match:
            result = match.group(1).strip()
            return self._clean_markdown_formatting(result)
        return "Final process reasoning not found"
    
    def analyze_community_optimized(self, comm_id: str, genes: List[str]) -> Dict:

        results = {
            "Community": comm_id,
            "Genes": genes,
            "Genes_String": ", ".join(genes)
        }
        
        try:
            enrichment_pathways = self.perform_enrichment(genes)
            results["Enrichment_Pathways"] = enrichment_pathways
            
            all_in_one_prompt = self._create_all_in_one_prompt(genes, enrichment_pathways)
            
            full_analysis = self._make_deepseek_request(all_in_one_prompt)
            results["Full_Analysis"] = full_analysis

            process_with_enrichment, confidence_with_enrichment = self._extract_process_info(full_analysis, "WITH")
            results["Process_With_Enrichment"] = process_with_enrichment
            results["Confidence_With_Enrichment"] = confidence_with_enrichment

            process_without_enrichment, confidence_without_enrichment = self._extract_process_info(full_analysis, "WITHOUT")
            results["Process_Without_Enrichment"] = process_without_enrichment
            results["Confidence_Without_Enrichment"] = confidence_without_enrichment
            
            try:
                results["Pathway_Reasoning_With_Enrichment"] = self._extract_section(full_analysis, "PATHWAY REASONING WITH ENRICHMENT")
            except Exception as e:
                results["Pathway_Reasoning_With_Enrichment"] = f"Extraction failed: {str(e)}"
                
            try:
                results["Pathway_Reasoning_Without_Enrichment"] = self._extract_section(full_analysis, "PATHWAY REASONING WITHOUT ENRICHMENT")
            except Exception as e:
                results["Pathway_Reasoning_Without_Enrichment"] = f"Extraction failed: {str(e)}"
            
            try:
                results["Contributing_Genes_With_Enrichment"] = self._extract_section(full_analysis, "CONTRIBUTING GENES WITH ENRICHMENT")
            except Exception as e:
                results["Contributing_Genes_With_Enrichment"] = f"Extraction failed: {str(e)}"
                
            try:
                results["Contributing_Genes_Without_Enrichment"] = self._extract_section(full_analysis, "CONTRIBUTING GENES WITHOUT ENRICHMENT")
            except Exception as e:
                results["Contributing_Genes_Without_Enrichment"] = f"Extraction failed: {str(e)}"
            
            try:
                results["Analysis_Text_With_Enrichment"] = self._extract_section(full_analysis, "ANALYSIS TEXT WITH ENRICHMENT")
            except Exception as e:
                results["Analysis_Text_With_Enrichment"] = f"Extraction failed: {str(e)}"
                
            try:
                results["Analysis_Text_Without_Enrichment"] = self._extract_section(full_analysis, "ANALYSIS TEXT WITHOUT ENRICHMENT")
            except Exception as e:
                results["Analysis_Text_Without_Enrichment"] = f"Extraction failed: {str(e)}"
            
            try:
                final_reasoning = self._extract_section(full_analysis, "FINAL PROCESS REASONING")
                results["Final_Process_Reasoning"] = final_reasoning
            except Exception as e:
                results["Final_Process_Reasoning"] = f"Extraction failed: {str(e)}"
            
            if confidence_with_enrichment >= confidence_without_enrichment:
                results["Final_Process"] = process_with_enrichment
                results["Final_Confidence"] = confidence_with_enrichment
                results["Final_Contributing_Genes"] = results.get("Contributing_Genes_With_Enrichment", "Not available")
            else:
                results["Final_Process"] = process_without_enrichment
                results["Final_Confidence"] = confidence_without_enrichment
                results["Final_Contributing_Genes"] = results.get("Contributing_Genes_Without_Enrichment", "Not available")
            
            return results
            
        except Exception as e:
            error_msg = f"Analysis failed: {str(e)}"
            print(f"Error analyzing community {comm_id}: {e}")
            results["Error"] = error_msg
            results.update({
                "Process_With_Enrichment": f"Error: {error_msg}",
                "Confidence_With_Enrichment": 0.0,
                "Process_Without_Enrichment": f"Error: {error_msg}",
                "Confidence_Without_Enrichment": 0.0,
                "Final_Process": f"Error: {error_msg}",
                "Final_Confidence": 0.0,
                "Full_Analysis": f"Error: {error_msg}",
                "Pathway_Reasoning_With_Enrichment": "Error occurred",
                "Pathway_Reasoning_Without_Enrichment": "Error occurred",
                "Contributing_Genes_With_Enrichment": "Error occurred",
                "Contributing_Genes_Without_Enrichment": "Error occurred",
                "Analysis_Text_With_Enrichment": "Error occurred",
                "Analysis_Text_Without_Enrichment": "Error occurred",
                "Final_Process_Reasoning": "Error occurred",
                "Final_Contributing_Genes": "Error occurred"
            })
            return results
    
    def analyze_communities_batch(self, communities: Dict[str, List[str]], batch_size: int = 1) -> List[Dict]:

        all_results = []
        community_items = list(communities.items())
        
        from tqdm import tqdm
        
        progress_bar = tqdm(total=len(community_items), desc="Analyzing communities", unit="community")
        
        for i in range(0, len(community_items), batch_size):
            batch = community_items[i:i+batch_size]
            
            for comm_id, genes in batch:
                result = self.analyze_community_optimized(comm_id, genes)
                all_results.append(result)

                progress_bar.update(1)

                processed = len(all_results)
                total = len(community_items)
                progress_bar.set_description(f"Processed {processed}/{total} communities")
                
                progress_bar.set_postfix({"Current": f"Community {comm_id}", "Genes": len(genes)})
        
        progress_bar.close()
        
        return all_results
    
    def create_detailed_dataframe(self, communities_text: str, batch_size: int = 1) -> pd.DataFrame:

        communities = self.parse_communities(communities_text)
        all_results = self.analyze_communities_batch(communities, batch_size)
        
        full_df = pd.DataFrame(all_results)
        
        requested_columns = [
            "Community",
            "Genes_String",
            "Enrichment_Pathways",
            "Process_With_Enrichment",
            "Confidence_With_Enrichment", 
            "Pathway_Reasoning_With_Enrichment",
            "Contributing_Genes_With_Enrichment",
            "Process_Without_Enrichment",
            "Confidence_Without_Enrichment",
            "Pathway_Reasoning_Without_Enrichment",
            "Contributing_Genes_Without_Enrichment", 
            "Final_Process",
            "Final_Confidence",
            "Final_Contributing_Genes",
            "Final_Process_Reasoning",
            "Full_Analysis"
        ]
        
        available_columns = [col for col in requested_columns if col in full_df.columns]
        return full_df[available_columns]

def run_optimized_analysis(input_text: str, api_key: str = None, detailed_csv: str = None, batch_size: int = 1) -> pd.DataFrame:

    analyzer = OptimizedPathwayAnalyzer()
    analyzer.set_api_key(api_key)
    
    detailed_df = analyzer.create_detailed_dataframe(input_text, batch_size)
    
    if detailed_csv:
        detailed_df.to_csv(detailed_csv, index=False)
    
    return detailed_df

if __name__ == "__main__":
    input_text = """
Refined Community 1: ['POLR2C', 'DPY19L1', 'MB', 'UBLCP1', 'YWHAB', 'FIGNL1', 'KCNMB1', 'EMSY', 'KIF15', 'NME7', 'SFMBT2', 'TMEM61', 'LAMA2', 'CARD19', 'SNAI2', 'IGLV3-25', 'PLXND1', 'EFNA3', 'RGS5', 'LSP1P5', 'CADM1', 'PLSCR3', 'TCN1', 'ENSG00000278932', 'SPEG', 'FGF10', 'LGALS1', 'CXCR5', 'ADAMTS12', 'FLNC', 'IRAG1', 'KCTD1', 'CAPN2', 'SORT1', 'CPNE1', 'NAP1L2', 'NXN', 'MAGI3', 'MRPL36', 'H2AC6', 'CTNND2', 'VSNL1', 'GPC1', 'CENPI', 'UCHL3', 'ITGB5', 'CORO6', 'CALD1', 'CD1C', 'FITM2', 'GARNL3', 'ATAD3B', 'CEP41', 'AOPEP', 'MXRA8', 'UBA52', 'ACTA2', 'SLC50A1', 'LIMS2', 'PMP22', 'TIMP3', 'NEK5', 'MPRIP', 'ANO1', 'PAPOLA', 'HBA1', 'WIF1', 'MAFF', 'BAP1', 'ELOB', 'LAGE3', 'UGT2B15', 'SNRPN', 'PRKG1', 'NDUFB9', 'COL1A2', 'PDCD4', 'DIPK1B', 'CXCL8', 'RTN4', 'SMTN', 'STX1A', 'TMEM192', 'UACA', 'PDCD6', 'UAP1L1', 'LRRN1', 'RABGAP1', 'STARD13', 'CDC25C', 'CALML4', 'AHNAK', 'FLNA', 'TAF2', 'TMEM201', 'CAV1', 'WRAP53', 'CGAS', 'RFLNB', 'TMEM35B', 'CNPY4', 'IFI6', 'CCDC15', 'FAXDC2', 'TSC22D3', 'NPR2', 'MAPK7', 'C9orf152', 'BARX1', 'MYH9', 'MYH10', 'BOC', 'CCDC50', 'G3BP2', 'IPO5', 'TTC28', 'HMGCL', 'CLN8', 'CNN2', 'KCTD21', 'SPP1', 'C6orf132', 'GALNT6', 'DNAAF9', 'SNHG14', 'EDN2', 'TMEM134', 'SLC4A3', 'MYLK', 'NARS2', 'NDUFV1', 'VPS37D', 'BEX5', 'UPP1', 'CAPN5', 'PRICKLE1', 'CBR3-AS1', 'COPS8', 'MYL9', 'EPHA3', 'OBSL1', 'ENPP1', 'AOC3', 'ZBTB43', 'CX3CR1', 'LILRB3', 'ULK4', 'SEMA3F', 'TTC7A', 'USP34', 'NOSTRIN', 'EDNRA', 'FAIM2', 'ERBB4', 'MYO5C', 'SLC25A36', 'DLC1', 'KLHDC10', 'PHF21A', 'RNF182', 'PPHLN1', 'KIDINS220', 'ALDH1A1', 'EXTL2', 'CSNK1D', 'SH3PXD2A', 'ZNF385D', 'AQP3', 'MAFK', 'AGPAT2', 'APBA1', 'H3C4', 'CKS1B', 'GSTA1', 'FBXO32', 'CD93', 'LIN28A', 'MRPL12', 'TLN2', 'ABCA5', 'LAP3', 'LGR6']
Refined Community 2: ['C1RL', 'FAM209B', 'SLK', 'MYCN', 'PPA1', 'GREB1L', 'MT1H', 'RAD9A', 'SYNE4', 'MICB', 'ZFP42', 'PCSK6', 'TAP1', 'CBS', 'DTX3L', 'ICAM1', 'SKAP1', 'WDR24', 'RND1', 'RAPGEF1', 'GNA15', 'TPD52L2', 'GTSE1', 'LEMD2', 'CASS4', 'CSNK1G3', 'MPPED2', 'SLC7A1', 'SNTG2', 'WRAP53', 'GABPB1', 'FABP4', 'TNPO2', 'BMI1', 'LGALSL', 'E2F5', 'INTS13', 'PLGRKT', 'MYNN', 'PLCL2', 'IL32', 'ABCA3', 'PARVB', 'BLOC1S6', 'ZSCAN18', 'GZMA', 'LIMS2', 'SLC30A8', 'EEF1D', 'STX16', 'NCKAP5L', 'TAPBPL', 'TRHR', 'APOBEC3A', 'VCAM1', 'BMAL2', 'CGAS', 'PSMA5', 'MCM10', 'CA4', 'LAGE3', 'FGF18', 'FBXO46', 'TBC1D30', 'CLBA1', 'CASP10', 'PMVK', 'ANXA7', 'PTPN6', 'NUP93', 'ADAM8', 'SPOPL', 'NOP16', 'UHMK1', 'ITGB7', 'GRK3', 'SNHG12', 'MAP2K1', 'FLNC', 'IL20RB', 'MRFAP1L1', 'RARS1', 'MORF4L2', 'CD48', 'BTN3A3', 'LAG3', 'BUB3', 'TMX2', 'FGF3', 'CRISP3', 'SLIT3', 'DRAM1', 'FBXO5', 'RPS21', 'ME2', 'CCDC88C', 'TMEM35B', 'MRTO4', 'PNKD', 'BTN3A1', 'RDH10', 'HSH2D', 'NFKBIZ', 'ZNF750', 'FAM53B', 'DDX11', 'ACSL5', 'ZNF524', 'RCL1', 'SCAMP2', 'PPP1R15B', 'TBC1D9', 'SEC23IP', 'IGLC2', 'FERRY3', 'RELB', 'CENPA', 'STMN2', 'TAC1', 'CEP55', 'ARHGAP32', 'PHF19', 'C9orf72', 'GYG1', 'YBX3', 'MAP1LC3B', 'SIAH2', 'RASGEF1A', 'FGF7', 'ANKRD22', 'BCLAF3', 'SNHG26', 'SLC9A2', 'CYC1', 'CALHM6', 'OVOL2', 'TMC8', 'PICALM', 'SNRPF', 'TREX1', 'YEATS2', 'MOXD1', 'ZCCHC9', 'NKX6-3', 'TYSND1', 'MINPP1', 'CHASERR', 'C2CD2', 'TMEM123', 'PIK3R6', 'PUS1', 'GABBR1', 'ASF1A', 'GRWD1', 'SLC2A6', 'SLC3A2', 'CD14', 'MAGOHB', 'CCL11', 'LAP3', 'SGPL1', 'COMMD5', 'PCGF2', 'ATXN1', 'DNAJB14', 'ZNF512B', 'ADH1B', 'PGAP2', 'CHCHD1', 'PQBP1', 'TNFRSF1B', 'CD38', 'TNFAIP2', 'APOE', 'IL15', 'CD96', 'TLR2', 'GALM', 'CUL2', 'TMPRSS2', 'RELA', 'RASGRP2', 'ADAMTS1', 'CEBPD', 'ETV6', 'SLC6A19', 'FOXM1', 'OBP2A', 'DECR2', 'C1RL-AS1', 'CCL19', 'PSMC5', 'COP1', 'KDM6A', 'CCDC69', 'RGP1', 'MCAM', 'ATG5', 'GID8', 'ATP8A1', 'MCMBP', 'SLC2A1', 'CD274', 'RANBP3', 'MAP3K13', 'IL27RA', 'HSD17B1']
Refined Community 3: ['NRARP', 'IGKV1D-8', 'DIXDC1', 'FLT1', 'ZNF469', 'HEY2', 'KRTAP9-3', 'FEN1', 'TBL3', 'JAG1', 'ETV2', 'TNFRSF19', 'PTDSS1', 'APLP2', 'PRELP', 'KRT5', 'NPR3', 'PLK2', 'PLAT', 'TCL1A']
Refined Community 4: ['NAP1L2', 'PFKP', 'HRH1', 'HNRNPDL', 'RBM24', 'FTSJ3', 'SCG2', 'CHD2', 'TRAF3', 'ZEB1', 'ABCC3', 'RRS1', 'ITPKB', 'ZNF394', 'CIBAR1', 'FOXQ1', 'DPY19L1', 'BBIP1', 'MAP1B', 'MMP16', 'SHOX2', 'POLR2B', 'TIMP4', 'TUBB2A', 'CTDSP1', 'GULP1', 'SNX27', 'CASP8AP2', 'GFPT2', 'MAGI3', 'LBH', 'GNG11', 'ADAMTS12', 'CCL14', 'PRKAG1', 'GRB10', 'ZFP36L2', 'STRADB', 'PLXDC1', 'C1QC', 'ARHGEF40', 'CNKSR3', 'AK5', 'EDIL3', 'FAM171A1', 'CMTM3', 'PNMA2', 'KCND3', 'CXCL12', 'GRK5', 'BCAT1', 'DCLK1']
Refined Community 5: ['LITAF', 'PPP1R14C', 'ELF1', 'SERPINA3', 'GLS', 'VRK3', 'MBNL3', 'LPL', 'OVOL1', 'FAM83A', 'DBNDD1', 'TRIP11', 'SH2D3A', 'ELF3', 'RADIL', 'APOBEC3B', 'ENTPD2', 'PPM1E', 'GULP1', 'VBP1', 'PEX1', 'MRFAP1L2', 'C1orf116', 'EFHD1', 'ARNT2', 'FAM131B', 'ZNF124', 'AGGF1', 'BET1', 'OVOL2', 'S100P', 'MUC1', 'EMC10', 'CENPX', 'LILRB2', 'CBLC', 'DMKN', 'BSPRY', 'EIF5A', 'IRX2-DT', 'HLA-DRB4', 'EWSR1', 'ARNT', 'RNF207', 'CSE1L', 'LAD1', 'FAM83B', 'TMEM183A', 'CCL14', 'CCNL1', 'SPINT2', 'KRT16', 'NHSL3', 'MICALL1', 'DYSF', 'ATP1A1', 'IRF6', 'MACC1', 'KIAA0040', 'RASSF5', 'C6orf132', 'SLC3A2', 'SIRT1', 'STEAP4', 'PKP3', 'SEPHS2', 'KRT5', 'CDH1', 'SLC49A4', 'PPP1R12C', 'HEATR3', 'GINS2', 'MYO5B', 'SLC45A4', 'KLK6', 'PPL', 'KLRG2', 'C1orf115', 'GFRA4', 'CRYBG1', 'EVL', 'PDZK1IP1', 'KCNK1', 'MAPK3', 'ESRP2', 'SMAD9', 'NAB1', 'TMEM87A', 'QKI', 'IQCE', 'SYTL1', 'PLXND1', 'SSH3', 'CD24P2', 'MYRF', 'ELMO3', 'SPINT1', 'PRELID3B', 'LRRN3', 'RHOD', 'WWP1', 'ST14', 'CISH', 'CD5', 'COPZ2', 'LARP1', 'CDH3', 'COX5B', 'MLF1', 'EHF', 'TLCD1', 'ACSL1', 'EPHA1', 'ZFP36', 'RBL2', 'RTKN', 'TXN', 'MCUR1', 'LIMK2', 'TAPBPL', 'KLK5', 'FGFBP1', 'USP3', 'F11R', 'HLA-DPA1', 'BARD1']
Refined Community 6: ['ITGA1', 'CISH', 'AP5Z1', 'AR', 'SLC6A4', 'KRT15', 'WBP2', 'IFI16', 'ACAD10', 'LAMC2', 'CDH2', 'TTC12', 'HSD17B11', 'YBX2', 'STAT1', 'SLC6A15', 'PACSIN3', 'TWIST2', 'PRNP', 'HOTAIRM1', 'VAMP3', 'PLS3', 'ATP13A3', 'SLC25A13', 'RAB17', 'EPHA4', 'LRRC45', 'GTF2B', 'SPRY4', 'ZNF436-AS1', 'TGFBR1', 'LARP4B', 'PPP3CC', 'ISG20L2', 'CGN', 'RALB', 'TCF7L1', 'PCLAF', 'IL1RAP', 'PDGFC', 'ZNF394', 'TUBA4A', 'TGIF2', 'MBP', 'CFI', 'PCK1', 'RPL7', 'GART', 'LPAR2', 'KRT16', 'FXYD5', 'KMO', 'ARRDC1', 'ETFA', 'MOGS', 'SEMG2', 'EVA1A', 'SPTBN1', 'PLXNB2', 'CCDC80', 'TGFA', 'C1S', 'WWC2', 'ZFAND1', 'MTMR2', 'ALDH1A3', 'YTHDF3', 'TLE4', 'CD44', 'F2RL1', 'GAS2', 'TMEM65', 'RUNX3', 'CD24', 'CIBAR1', 'SMAD3', 'MYL6B', 'PXDC1', 'YIPF5', 'JPT2', 'ACAA1', 'SCPEP1', 'FAM222B', 'WARS1', 'ICAM1', 'CAV2', 'ANKRD33B', 'SNAI2', 'KLHL29', 'MPV17L2', 'TG', 'SMCHD1', 'CD59', 'BATF', 'SLC16A1', 'FMNL2', 'ANXA2P2', 'EDN2', 'CAVIN1', 'BUB1B', 'TRMT11', 'ITIH5', 'CYP26A1', 'HMGA2', 'FLOT2', 'HK2', 'SUMO2', 'ORMDL1', 'GBP3', 'SOCS5', 'PEF1', 'NFE2L2', 'PIK3CD', 'ANKH', 'CDK6', 'NSMAF', 'SREBF1', 'CAMLG', 'LOX', 'CLDN1', 'TRIM24', 'CFB', 'SELENOF', 'C1R', 'MYL12A', 'SIRPAP1', 'ATP1B3', 'HTRA1', 'RAPH1', 'TM2D1', 'METTL25B', 'ATP1A1', 'KLF5', 'PUS1', 'MAP3K7CL', 'PPP1R14C', 'RIOK3', 'TRIP10', 'APOBEC3G', 'SBNO2', 'MYO5B', 'IFNGR1', 'RFC4', 'PKN2', 'NUDT15', 'CARD6', 'MAP4K4', 'SQOR', 'ASXL1', 'UBE2E3', 'MME', 'RBMS3', 'SFN', 'NR3C1', 'DSC3', 'IGKV1D-39', 'TP63', 'RBBP4', 'OVOL1', 'MAP7D3', 'PTK7', 'MTAP', 'FGF16', 'MARK2', 'ANXA8', 'SLC16A7', 'SLC49A4', 'RPL7P9', 'PM20D2', 'LASP1', 'BTG3', 'DUOX1', 'LAMA3', 'PSMB8', 'FITM2', 'RBMS1', 'TMEM105', 'TGFBR2', 'NAMPT', 'MCTP2', 'CORO2B', 'ZNF568', 'TCOF1', 'SGCA', 'TIMM17A', 'NELFCD', 'OCIAD1', 'CSNK1E', 'GJB3', 'YES1', 'FGFBP1', 'LGR5', 'FRAT1', 'GAS6', 'AKTIP', 'GLIPR1', 'FBLIM1', 'LUZP1', 'NCKAP5L', 'DOK4', 'MARCHF9', 'CERS2', 'IGFALS', 'ZC3H3', 'ARHGAP5', 'S100A2', 'SERPINE2', 'CD53', 'SEC14L2', 'ADAM12', 'HPX', 'CCDC106', 'RGL1', 'RNFT2', 'SMARCA4', 'TPSD1', 'SLC39A10', 'FLT3LG', 'LAMC1', 'CAVIN3', 'RFLNB', 'CDK5RAP3', 'STIP1', 'KLF10', 'DGKA', 'TCHP', 'CCNB1', 'MAFK', 'NSFL1C', 'IL20', 'EPHA2', 'EVPL', 'TMEM117', 'CELSR2', 'COL18A1-AS1', 'PRSS21', 'CRY2', 'C17orf67', 'LIPG', 'FRMD6', 'VAT1', 'COL8A1', 'SPINK4', 'PRSS16', 'ANXA13', 'AKT1S1', 'KCNQ2', 'C2orf68', 'EVA1C', 'ACTR1A', 'PLSCR1', 'ANXA3', 'KRT6A', 'ALPK2', 'GPSM2', 'SLPI', 'ERAP2', 'SRI', 'WDR1', 'IFI30', 'TMEM104', 'STING1', 'TNFAIP3', 'CRK', 'DSE', 'ETF1', 'MT1F', 'UCK1', 'RBM7', 'NNMT', 'TRIM22', 'CEACAM1', 'CPA3', 'MPG', 'CD58', 'CACNG1', 'FANCG', 'B3GNT5', 'RAB22A', 'DVL3', 'RNF7', 'SGK1', 'TNFAIP8', 'RPL31', 'ATP8B1', 'KCTD1', 'SYK', 'KRT6B', 'TRMT1L', 'GPRC5B', 'GPS1', 'CD14', 'HOXA5', 'PTGS2', 'C3', 'PRIM1', 'GMPS', 'DAG1', 'RHOA', 'TASOR2', 'TMED5', 'C1orf74', 'DPYD', 'SP100', 'CYP3A5', 'PSCA', 'ETS1', 'CYBRD1', 'HP1BP3', 'OGFRL1', 'MBNL2', 'EMC10', 'PTGFR', 'ANAPC15', 'BAIAP2', 'CDH19', 'LOXL1', 'YBX1', 'IGF2BP2', 'ITGB8', 'SNX8', 'BICD2', 'KRT5', 'AKR1C1', 'ARHGAP23', 'PI3', 'CTSC', 'FGF2', 'RALBP1', 'PRKD3', 'PNLIPRP3', 'NCAPG2', 'CLMN', 'MOCOS', 'INIP', 'MAPT', 'GRIPAP1', 'CLIP4', 'NUP50', 'NXN', 'LAMB3', 'IL7R', 'RPL18', 'BIN1', 'MSH6', 'ELK3', 'PTPN14', 'RBP4', 'DSG3', 'DSG2', 'DMWD', 'ZC3H12C', 'HOXA1', 'PI4KA', 'CAPN5', 'CMPK1', 'IL7', 'TLR2', 'CALU', 'TLN2', 'ATXN7', 'SPRY2', 'CCDC6', 'ANXA4', 'CBLC', 'CYLD', 'ACADL', 'CXCL1', 'PSAT1', 'ZNF160', 'STAT3', 'BICC1', 'KIRREL1', 'AKR1B1', 'PKP2', 'TOX2', 'RBM5', 'MAOA', 'PSD3', 'EMP1', 'SELENOO', 'NOVA1', 'DIPK1A', 'TRIM29', 'PDZK1IP1', 'RERG', 'BACH1', 'CASZ1', 'ENTPD2', 'ITPRID2', 'CHST3', 'ANXA1', 'RIPK4', 'MRPL38', 'FKBP1A', 'RBM39', 'MYO1E', 'CCNA1', 'ADGRE5', 'LARP6', 'IL1A', 'RHEB', 'ZBTB16', 'GNAI1', 'MPZL1', 'GAS1', 'GATA2', 'ATG12', 'IGHV4-61', 'LMBRD2', 'LINC01355', 'HBEGF', 'PTPRM', 'RGCC', 'SKAP2', 'DNAJB4', 'DOCK5', 'PCGF6', 'PRG2', 'KRT83', 'SGO2', 'CRYBG1', 'WASHC4', 'PPP1R1A', 'KANK1', 'GNG12', 'H4C8', 'MICALL1', 'FAS', 'GBP1', 'MAP4K3', 'ADRB2', 'LMO2', 'TAC1', 'RAPGEF1', 'CDC42EP3', 'SERPINB2', 'MT1H', 'UBE2V1', 'FZD1', 'MFAP4', 'EXT1', 'SH3BP4', 'SEPTIN10', 'CASP1', 'EHBP1', 'TMEM245', 'CAPN9', 'CORO1C', 'AKR1C3', 'B2M', 'PGA5', 'DACT1', 'PLA2G4A', 'CCDC82', 'CST5']
Refined Community 7: ['ACKR1', 'ARFGEF3', 'SCYL3', 'CACYBP', 'RAB3D', 'FGFR4', 'FGFR1OP2', 'COL18A1', 'LINC01128', 'DDX42', 'SPAG1', 'ONECUT2', 'ULK1', 'LRP3', 'RBAK', 'ZNF318', 'CCNL1', 'PDE2A', 'KRTAP4-2', 'EPRS1', 'TLE1', 'ATP5MF', 'RCC1', 'ARHGEF6', 'TFF3', 'TCERG1', 'PYCR1', 'GATB', 'MFAP4', 'KPNA2', 'INTS15', 'NAV1', 'CMTM7', 'ACADS', 'CAPN9', 'KCTD2', 'BTBD9', 'PPFIA1', 'NAF1', 'MPHOSPH10P1', 'HECTD4', 'CSNK1D', 'KIFC2', 'FEM1B', 'MXRA8', 'CRNKL1', 'ZMIZ1', 'ADAMTS1', 'TMEM86B', 'PACSIN3', 'PRLR', 'CENPI', 'NCOR1', 'ACAA1', 'MRPS33', 'DNAJA4', 'BANF1', 'C4orf19', 'INTS8', 'GADD45B', 'TSPAN13', 'ANXA9', 'POGZ', 'VAMP1', 'SFMBT2', 'SCAND1', 'RMND5B', 'KAT6B', 'RDH13', 'ASH1L', 'TMX2', 'LARGE1', 'TBC1D16', 'IRF8', 'BGN', 'PGR', 'TTC9', 'KIFC1', 'NF2', 'SHANK2', 'ADAM8', 'PSMG3', 'SRRM2', 'EIF3B', 'DEFB1', 'PRSS16', 'CISH', 'MYEF2', 'NR6A1', 'HACD2', 'EVL', 'SCN4A', 'CACNA1D', 'EPS8L1', 'HNRNPA2B1', 'PCP4', 'LZTR1', 'SLC12A9', 'CACNB3', 'POMT1', 'ATXN1', 'MEGF9', 'EPS8', 'RSRC1', 'SPMIP5', 'PATZ1', 'NPW', 'DES', 'TSPOAP1', 'BAZ2A', 'THSD4', 'TMEM79', 'PDCL3', 'RGL2', 'TRPS1', 'BRIP1', 'BAIAP2L1', 'KDELR2', 'ANG', 'FUS', 'C9orf152', 'PCBP2', 'SMAD2', 'BLNK', 'CTXN1', 'EDIL3', 'CD40', 'SPTLC2', 'ZNF703', 'SLC38A1', 'SECISBP2', 'IVD', 'ARF3', 'TLE3', 'SLC16A6', 'ARHGEF26', 'TET3', 'P4HTM', 'TRIL', 'ASAH1', 'KAT8', 'NMT2', 'PRR14', 'ISG20', 'DLEU2', 'PRRT2', 'NEFH', 'DUSP5', 'RETSAT', 'CELSR2', 'MAGEA9', 'NUDT4', 'CSGALNACT1', 'ZNF461', 'NTHL1', 'TTC6', 'WFS1', 'FGFR1', 'BAG4', 'CCDC90B', 'GATA3-AS1', 'WWC2', 'ARRB1', 'MEMO1', 'RPL8', 'FAM53B', 'SELPLG', 'AR', 'ENTR1', 'RRP1B', 'CDKN1C', 'CSNK2A2', 'GATA3', 'NACA', 'HLA-DQB1', 'DMD', 'HID1', 'REPS2', 'SMC2', 'CTNND2', 'RHPN1', 'HAGH', 'TPGS2', 'HMGCS2', 'CERS2', 'CA5A', 'SYDE2', 'SCG2', 'CACNA2D2', 'SFI1', 'ZNF267', 'APPBP2', 'MACO1', 'BCL11A', 'SLC1A4', 'LUC7L3', 'BHLHE40', 'DACH1', 'KLF2', 'SLC24A3', 'ZNF12', 'LTBP3', 'DHRS13', 'SH3BP5-AS1', 'ICA1', 'ARID3A', 'MS4A2', 'DSCAM-AS1', 'RAB40C', 'DLAT', 'NXPH3', 'LRRC24', 'VIPR1', 'TMEM184A', 'TCAF1', 'ZBTB7B', 'TFPI', 'THUMPD1', 'MAST3', 'PDGFC', 'HSPB1', 'SIDT2', 'SH3GLB2', 'CIRBP', 'TC2N', 'TESK1', 'SPATA2L', 'DENND1A', 'RAB11FIP3', 'SEC16A', 'STARD10', 'VPS72', 'KIF12', 'SRPX2', 'LRRN1', 'MAGED2', 'PLEK', 'TAPT1', 'RAI2', 'NPDC1', 'IRAG1', 'ZNF16', 'CHIT1', 'BCAS1', 'DLG5', 'ADAM33', 'NFKBIZ', 'SLC26A11', 'MGAT4A', 'IGLV3-19', 'USP42', 'F7', 'TENT5C', 'MTERF2', 'CEP95', 'CEP57', 'LZTFL1', 'CHTOP', 'FBXL18', 'PIP4P2', 'NME3', 'TMEM80', 'RASL10A', 'GPD1L', 'CA12', 'SLC7A8', 'ATXN7L3B', 'INHBB', 'NECTIN2', 'CACFD1', 'FHL1', 'KDM4B', 'CAPN2', 'EVA1A', 'CLCA2', 'CERS6', 'GTF3C1', 'ADGRG5', 'TRAC', 'SYNJ1', 'ZNF467', 'ENPP2', 'DEGS2', 'ECT2', 'LONRF2', 'SPCS2', 'PPP1R14C', 'KDM7A', 'C3orf18', 'CRACD', 'PBX1', 'EIF5A2', 'FKBP4', 'SOX8', 'PRRT3', 'MAB21L4', 'LFNG', 'DNAJC1', 'B3GNT5', 'SLC44A4', 'TRGC1', 'TGFB3', 'LLGL2', 'PLA2G7', 'SOX13', 'HEXD', 'RSAD1', 'SPDEF', 'RNASE6', 'ASPH', 'TSHZ2', 'ATRNL1', 'MDM4', 'SMARCE1', 'FTX', 'SH3BP5L', 'SNX27', 'KIF19', 'DAAM1', 'KPNA4', 'CSNK2A1', 'MGRN1', 'CD19', 'CSRNP2', 'ZBTB42', 'RUSC1-AS1', 'GP1BB', 'SOX12', 'PI4KA', 'ZNF552', 'FAM110C', 'PRRC2C', 'KRT16', 'VPS37C', 'DOCK10', 'PLEKHH1', 'ATF1', 'VRK3', 'GPR160', 'ARMT1', 'SRGAP2', 'RAP2B', 'LPAR2', 'RECQL4', 'LMCD1', 'PRIM1', 'HPN', 'AHCY', 'TNRC18', 'IRGQ', 'PON2', 'CNTROB', 'MAPK15', 'CMIP', 'BCOR', 'G6PD', 'SMG8', 'KATNIP', 'SPP1', 'PPP1R16A', 'ABAT', 'KRT19', 'C17orf58', 'ARFIP2', 'CYP2D6', 'ZMIZ2', 'GRAMD4', 'UAP1L1', 'RALY', 'TBC1D30', 'ALG12', 'NEK2', 'PNPO', 'CCL18', 'FOLR1', 'CHDH', 'UNCX', 'EIF4EBP1', 'CCND1', 'SNCAIP', 'FZD4', 'PLIN2', 'TAF1D', 'FBN1']
Refined Community 8: ['GSE1', 'RAPGEFL1', 'SPRY2', 'PMP22', 'TSPAN31', 'CAMLG', 'ARHGDIA', 'IGF2BP2', 'DSE', 'DNM1', 'TMEM204', 'PCOLCE2', 'TNFRSF11B', 'ANAPC1', 'RAB31', 'PABPC4', 'MAK', 'GLIPR2', 'MSN', 'CEP164', 'SLC2A10', 'ASAP1', 'PECAM1', 'PLAT', 'ZNF217', 'OSMR', 'FANCD2', 'CCDC50', 'PALM2AKAP2', 'CDK6', 'RUBCN', 'PTGS2', 'BIN1', 'BACH1', 'ADAMTS5', 'ECHDC1', 'LRRC45', 'ANKRD30A', 'GRK5', 'NAT1', 'CDC27', 'TAPBPL', 'FLRT2', 'SV2A', 'FZD6', 'GPR87', 'MYLK', 'NHERF1', 'SGTB', 'RGS2', 'MCM2', 'PRSS22', 'FBN1', 'UPF3B', 'CORO1C', 'NUP50', 'CD44', 'COL6A2', 'CALD1', 'COL6A3', 'TUBB', 'UPP1', 'RASSF2', 'ANXA5', 'NCAPG', 'SMIM14', 'RALBP1', 'ERFE', 'RSAD1', 'SLC16A7', 'ASXL1', 'IGF2BP3', 'EXTL2', 'UBE2S', 'FZD7', 'GALNT7', 'SHOX2', 'HRH1', 'WLS', 'SNX7', 'F2RL2', 'PRR16', 'PRKD3', 'SH3KBP1', 'CUL9', 'PSMD12', 'RPS6KA1', 'GULP1', 'EPN2', 'ZNF587', 'NDC80', 'SEPTIN6', 'ATP10D', 'ANLN', 'SNRK', 'WWTR1', 'SH3GLB1', 'GSDME', 'RGS4', 'POGLUT3', 'NKAIN4', 'HEXD', 'NUP88', 'CNN1', 'ALPK2', 'TTL', 'BASP1', 'PAFAH1B1', 'VPS45', 'GFOD2', 'RUNX2', 'ABCC3', 'KCNQ3', 'TFPT', 'ATP5MF', 'HEG1', 'COL5A1', 'FEZ2', 'PDE7B', 'ULK2', 'MTMR2', 'CENPV', 'NMT2', 'SGK1', 'MYLK3', 'GTF2H5', 'MIR100HG', 'SMAD7', 'PSPH', 'DDX39A', 'RHOA', 'QKI', 'EVA1B', 'CNTNAP4', 'RGS20', 'SPHK1', 'ELOVL2', 'AEBP1', 'CSRNP2', 'FOXC1', 'LUC7L', 'DZIP1', 'GRIFIN', 'NPEPL1', 'THRSP', 'TGFBR2', 'SDCBP', 'AKR1B1', 'ZNF831', 'SLC25A37', 'ATP2A3', 'COL13A1', 'STX2', 'PFAS', 'TLCD3A', 'LAYN', 'WNT5A', 'SEMA3A', 'EIF5A2', 'TMEM65', 'UBLCP1', 'SRPX2', 'ZCCHC24', 'UTP23', 'FEM1C', 'GNG11', 'NEDD4', 'ARHGAP21', 'ARHGEF40', 'TMEM245', 'HIF1A', 'MAP2K1', 'USHBP1', 'NANP', 'MSL2', 'ZBP1', 'TIMP1', 'CD24', 'SELENOF', 'POGZ', 'PAPOLB', 'SPIN1', 'BDNF', 'BTAF1', 'BCHE', 'CTDSPL2', 'MALT1', 'HEY2', 'COL25A1', 'SPTBN1', 'MAP4K4', 'PPP1R18', 'CCNO', 'EDIL3', 'NOC3L', 'GLIS3', 'SLC29A2', 'FAM210B', 'NEIL1', 'ADD3', 'NDEL1', 'GRAMD1C', 'ETS1', 'CDC20', 'POP1', 'TMEM158', 'NR3C1', 'SLC10A5', 'HPCAL1', 'RBL2', 'TNPO2', 'ITSN1', 'KIF18B', 'SOS1', 'FRMD6', 'ANXA2', 'WNT5B', 'SYK', 'RGPD5', 'IFI16', 'GPRC5B', 'CYP26B1', 'FSTL1', 'MATN2', 'ULK3', 'PGA3', 'INCENP', 'IL1R2', 'COG3', 'DVL2', 'TMEM33', 'BCAT1', 'TGFB1I1', 'CYP24A1', 'IGFBP6', 'PLK1', 'VIM', 'CENPB', 'PSMB9', 'ADD1', 'KLHL22', 'ACTN1', 'ORMDL2', 'TOX', 'GSPT1', 'MSANTD3', 'ELAPOR2', 'SPMIP5', 'ANKRD22', 'HEBP1', 'ZIC2', 'HOOK2', 'CDKN2C', 'EEF1D', 'EPHA2', 'PPIP5K1', 'CDKN1C', 'SNTG2', 'ETV7', 'POSTN', 'TYMS', 'NONO', 'ELAVL1', 'GSPT2', 'RANGAP1', 'GSTP1', 'AXIN1', 'LZTS3', 'SSR1', 'BOLA1', 'FKBP1A', 'CLIP2', 'FNDC5', 'IL6', 'CPNE2', 'GLS', 'KIRREL1', 'KRT26', 'FYN', 'GAS8', 'AKR1C3', 'LILRA4', 'DENND5A', 'EIF4E3', 'DLK2', 'WDR90', 'LONRF2', 'ELP5', 'MEIOB', 'BANK1', 'NMNAT2', 'ACTG1', 'WRAP53', 'KATNAL1', 'FST', 'DKK3', 'DUOX1', 'ANTXR2', 'BCOR', 'TEX14', 'CRIM1', 'GDF7', 'TRIM29', 'EXT1', 'BICDL1', 'TMED4', 'ITGB7', 'GDF6', 'ANKRD33B', 'STAMBPL1', 'DCAF8', 'IGFBPL1', 'PRSS12', 'MPRIP', 'ABCC5', 'TRIO', 'ADCY7', 'ZDHHC2', 'STIL', 'LCP1', 'FHOD3', 'ZC3H7B', 'C1RL', 'TEX2', 'IVD', 'STMN2', 'TWSG1', 'CLMP', 'ELOVL5', 'VCL', 'OR12D2', 'RFLNB', 'COL6A6', 'TBX2', 'CDC6', 'EZH2', 'LAMB1', 'TNFRSF10B', 'MRC2', 'LPAR1', 'FXYD5', 'CCDC6', 'WIPF1', 'SERF1A', 'ITPKB', 'AIP', 'LIX1L', 'IFT81', 'MCAM', 'DUOXA2', 'NPW', 'TMEM35B', 'CALU', 'SLC37A1', 'ACOT9', 'HS3ST6', 'SNRNP35', 'PTTG1', 'WNT10A', 'PAPPA', 'AVIL', 'ACAD11', 'DHTKD1', 'HPN', 'PSMB6', 'OFD1', 'SLF2', 'NECTIN3', 'MAP1B', 'ATM', 'OPRL1', 'GAL', 'SLIT2', 'AMACR', 'CTNNAL1', 'UQCC1', 'CCDC88A', 'GATA2', 'FAT4', 'PDE10A', 'FOSL2', 'INPP1', 'NEXN', 'GPD1L', 'TNFAIP3', 'DHX33', 'GNB4', 'PPP1R16A', 'TPM1', 'ANKH', 'FAS', 'HNMT', 'S100P', 'SMAD3', 'ZEB1', 'ODF1', 'RTN4', 'NAP1L1', 'LCP2', 'CFL2', 'CPSF1', 'BIRC5', 'KLHL29', 'P4HTM', 'LYN', 'VRK1', 'LRRK1', 'LUC7L3', 'GNAI1', 'AP3M2', 'PSMD1', 'IQCG', 'DUSP9', 'MSRB3', 'LDHB', 'C16orf74', 'DSN1', 'ANXA2P2', 'LOXL2', 'POPDC3', 'TIMP2', 'LTBP2', 'NEFL', 'CYB5R3', 'P3H2', 'ARSJ', 'TLK1', 'TMEM41B', 'RAC1', 'FNDC4', 'NAV1', 'HOXA9', 'CARD6', 'MMP2', 'ADAMTS12', 'ENG', 'TFPI', 'HYAL1', 'GIMAP5', 'CUL4B', 'EPB41L2', 'SKP2', 'PNMA2', 'NRP1', 'GABRQ', 'LARP6', 'EIF4A1', 'NAP1L5', 'ME3', 'FAHD1', 'PPP2R5A', 'KCNJ3', 'TPSD1', 'GART', 'TNFRSF10D', 'PYGL', 'ALKBH5', 'DNAJB4', 'CEP72', 'MAP2', 'TSPAN13', 'GINS1', 'PRKCZ', 'OVOL1', 'GNB1']
Refined Community 9: ['SSTR2', 'SLC25A29', 'FUT1', 'TRPM4', 'SUOX', 'OPLAH', 'EFNA4', 'KRT23', 'RGP1', 'VWA5A', 'RAB17', 'PPP1R3D', 'PI4KA', 'STARD10', 'DDX39A', 'SLC7A8', 'PNKD', 'ARHGEF5', 'PSAP', 'RNASE4', 'ESRP2', 'ZBTB7B', 'PPP1R1B', 'ST6GALNAC2', 'FZD5', 'HEXIM1', 'GRINA', 'ARL3', 'NRBP2', 'SPINT2', 'SETD6', 'LAD1', 'NMT2', 'BLVRB', 'TMEM184A', 'IGBP1', 'PER2', 'SERF2', 'ARFGAP2', 'GRHL2-DT', 'RHPN1', 'CKMT1B', 'SLC37A1', 'CACNG4', 'INPP5J', 'TMEM229B', 'SPATA2L', 'EEIG1', 'APH1A', 'BEX5', 'TMPRSS2', 'IGHV3-23', 'TMIGD1', 'CHMP4C', 'ATP1B3', 'ZNF165', 'MAFB', 'CBLC', 'ARRDC4', 'CEP43', 'SLC23A2', 'ANKRD22', 'KATNIP', 'SEPTIN8', 'CFAP410', 'FLAD1', 'CDH1', 'CRK', 'PAQR4', 'ATP6V0E2', 'ZXDC', 'ENPP5', 'CACNB3', 'TFF1', 'CPNE1', 'GALNT6', 'SLC66A2', 'CENPV', 'ASB8', 'HIPK1', 'BUB3', 'ZNF611', 'CACNA1D', 'BBS4', 'TRAPPC6A', 'TFF3', 'ACVR1B', 'DENND3', 'FKBP8', 'LNX1', 'UCHL3', 'HLA-DRB4', 'ATP2C2', 'CEBPA', 'CREB3L4', 'CD24', 'GGT1', 'ESRP1', 'HID1', 'RGS1', 'MYL12A', 'LINC00482', 'XKR9', 'FBXW2', 'PRSS54', 'ANTXR1', 'EMG1', 'SLC8A2', 'PLEKHA6', 'OPN3', 'MGAM', 'PRODH', 'TTC7A', 'GNS', 'ULK1', 'PGA3', 'PLEKHF2', 'CHCHD1', 'PTGS2', 'BCOR', 'CCR2', 'PIR', 'IFRD1', 'EPCAM', 'LOXL1', 'C8orf33', 'GGT6', 'ULK3', 'CHAD', 'INTS7', 'DEPTOR', 'MSMO1', 'ELF3', 'FOXN1', 'CLN3', 'CNNM4', 'OCLN', 'ELL3', 'SIRT3', 'SUN1', 'ZFC3H1', 'SERPINI1', 'BICDL2', 'AUTS2', 'HIP1R', 'PSMD1', 'CMC2', 'CCDC97', 'DENND1B', 'LAMC1', 'DAB2', 'ABAT', 'ABHD11', 'SIDT1', 'AHI1', 'PDE9A', 'ENSG00000280119', 'SGMS1', 'SLC30A8', 'CST5', 'ERBB3', 'TMBIM4', 'MDK', 'ERP29', 'ZNF367', 'RPSAP47', 'GTPBP1', 'ARPC1A', 'BICDL1', 'STAT5B', 'PGAP3', 'FGD3', 'CD300E', 'TEX19', 'PRR15', 'ERBB2', 'ITGB2', 'LYSMD4', 'MLPH', 'EPB41L4A', 'ETFB', 'PLEKHA1', 'GSK3B', 'ACOT11', 'CAPN8', 'NECTIN4', 'JAG2', 'ADCY6', 'RAD21', 'ABCC11', 'CIRBP', 'RAB25', 'VPS45', 'UBLCP1', 'CCNE2', 'SYNE4', 'PEX3', 'JAG1', 'EPN3', 'NHERF1', 'CARNS1', 'KIFC2', 'TAPT1', 'PNPLA4', 'EPHB3', 'C2CD2L', 'ZKSCAN1', 'TTC9', 'MRFAP1L2', 'FREM2', 'NUP107', 'HGF', 'RABGAP1', 'TMEM208', 'ABCG1', 'FAM91A1', 'CLPTM1L', 'PTGDS', 'NOL3', 'CTH', 'HNRNPA3', 'DCAF11', 'MAP7', 'CYP4B1', 'RPS6KA5', 'OVOL2', 'ISG20L2', 'TAX1BP1', 'LINC02627', 'CA5BP1', 'LSR', 'DEPDC5', 'NOP2', 'KRTAP4-3', 'ZHX2', 'IGFBP3', 'DEGS2', 'FOXA1', 'TRIM3', 'SECISBP2', 'OPRL1', 'SPCS3', 'SBNO2', 'SSH3', 'MPPED2', 'GRP', 'SPINT1', 'MKNK2', 'MPP7', 'C4orf19', 'SYTL1', 'WDR24', 'VAV3', 'UQCRB', 'RERG', 'PDS5B', 'NUDT8', 'C11orf52', 'EMC10', 'EPB41L5', 'DNAL1', 'ATOSA', 'GATA6', 'MMEL1', 'MYCBPAP', 'SBNO1', 'ARFGEF3', 'GRAMD4', 'ZBTB42', 'UNC119', 'DDHD1', 'MCCC2', 'TINAGL1', 'GTPBP10', 'DDR1', 'CD24P4', 'MTERF1', 'ISOC1', 'ARHGEF7', 'CYB5R2', 'PDE5A', 'EHF', 'UGT2B11', 'SYMPK', 'ASAH1', 'CLDN6', 'SORL1', 'CDS1', 'CACFD1', 'SHANK2', 'FRAT2', 'GIT2', 'GAREM1', 'GNA15', 'IDH2', 'ZNF24', 'CCDC28A', 'CPSF1', 'RRM2', 'ZNF703', 'MAPK7', 'TMPRSS13', 'CASZ1', 'LLGL2', 'BTBD3', 'EMP2', 'CCNG2', 'ARHGEF28', 'ITCH', 'TNNT1', 'JMJD7-PLA2G4B', 'HOXB3', 'CCN2', 'DCAF6', 'FAM174B', 'USP15', 'TRDMT1', 'PLEC', 'RNF144B', 'GPR160', 'ULK2', 'MUC1', 'ST14', 'RALB', 'AAGAB', 'MYBL2', 'PLK3', 'MYO1E', 'ARFIP2', 'TSPAN1', 'SPDEF', 'RCBTB2', 'TRAF4', 'CCNE1', 'SCAMP2', 'HILPDA', 'PRLR', 'VIPR1', 'MRE11', 'IGHV3-33', 'PCK2', 'PHKG2', 'KRT222', 'BGN', 'PYGB', 'TSTD1', 'PRRG4', 'PREX1', 'HGS', 'TMBIM6', 'C1orf43', 'TAX1BP3', 'HERC5', 'ATL3', 'SFI1', 'EPHX3', 'TRPS1', 'IMPA1', 'ERGIC2', 'CCT5', 'MS4A6A', 'TRA2B', 'MARVELD2', 'PKP3', 'TUFT1', 'MIR29B2', 'FGFR1', 'HOOK2', 'ADAM17', 'EIF5A2', 'NEK8', 'TMEM276', 'NME3', 'SPTBN2', 'MRFAP1L1', 'PATZ1', 'SSR4', 'TFCP2L1', 'SLC35E2B', 'ME3', 'SELENOS', 'PPFIA1', 'RHO', 'LPCAT2', 'DAAM1', 'NR2F6', 'BCAS1', 'EPHA1', 'CABLES2', 'PCDH1', 'CHMP2A', 'INHBB', 'IL7', 'SPTLC2', 'EYA3', 'DUSP16', 'SEMA4A', 'MED24', 'DPY19L4', 'GLIS3', 'DDR2', 'FXYD3', 'ATP6AP1', 'UNG', 'EPS8L1', 'SNHG14', 'ANXA9', 'CHSY1', 'DSCAM-AS1', 'TMEM79', 'ANK2', 'TSPAN13', 'PLA2G12A', 'TMC4', 'RBM47', 'SYNGR2', 'CRABP2', 'ANKRD13C-DT', 'TMEM183A', 'EMILIN1', 'PPCS', 'DBP', 'TRPC4AP', 'CD82', 'CD37', 'EPHA10', 'PABPN1L', 'TCF7L2', 'CXCL16', 'RNF103', 'RHBDF1', 'SH3GLB1', 'CSF1', 'SELENBP1', 'ERCC6L2', 'ROGDI', 'SEMA3F', 'SEPHS2', 'TLK2', 'MYO5B', 'TM7SF2', 'CLDN7', 'SULT2B1', 'FCAMR', 'P4HTM', 'FTL', 'ALDH4A1', 'CDH3', 'IGFBP2', 'MRPS17', 'TADA2B', 'EXPH5', 'VRK3', 'BOLA1', 'FBXL16']
Refined Community 10: ['HLA-DPA1', 'GNG13', 'PDIA3', 'LLGL1', 'HRAS', 'SAP30BP', 'ANKRD11', 'TMEM117', 'MLST8', 'PNMA2', 'CDH13', 'FEZ2', 'CTSB']
Refined Community 11: ['ASB13', 'ERBB2', 'FAT2', 'PEBP1', 'HS3ST3A1', 'TOP2A', 'TNFRSF6B', 'CD1E', 'MFGE8', 'PDE4DIP', 'TGFB2', 'TESPA1', 'OLFML3', 'LRRC32', 'NCOA3', 'THRA', 'LAMC2', 'MYC', 'ZNF217', 'ACAA1', 'TNFAIP2', 'AURKA', 'TRAF4', 'CYP24A1', 'EXT1', 'ENPP5', 'WNT10B', 'C11orf54']
Refined Community 12: ['FNDC3A', 'XRCC5', 'SOX7', 'CASP9', 'ZMYND8', 'LZTFL1', 'PDGFRL', 'NRG1']
Refined Community 13: ['CCL7', 'ABCA7', 'MTSS1', 'ERBB2', 'SH3BP2', 'PREX1', 'PAK1', 'BCL6', 'FGFR1', 'NPR3', 'PTPRT', 'EXT1', 'JAG1', 'JUN', 'C1orf122', 'TRMT13', 'BOLA1', 'TRPA1', 'SOX9', 'FGF8', 'DACT3', 'EBAG9', 'ATP2B4']
Refined Community 14: ['LINC03042', 'IGHV3-33', 'DSC2', 'MARCHF3', 'ADAM32', 'CACNA2D3', 'NCOA1', 'CTSV', 'EPHA2', 'EEF1D', 'WDR33', 'ATP6V0A4', 'SPPL2B', 'ATP11A', 'CALU', 'DEK', 'PAPSS1', 'LSM14B', 'INHBB', 'SPX', 'NLRC5', 'PYGL', 'PIP5K1A', 'KRT16', 'C16orf95', 'SGCB', 'LMO4', 'BAG2', 'HSDL2', 'PLOD2', 'PNN', 'CHST3', 'SLC16A1']
Refined Community 15: ['LRRC14', 'SULT1A2', 'ABCA12', 'RGS22', 'ZNF704', 'FAM174B', 'ALDH4A1', 'CFAP97', 'TMC4', 'RND1', 'CD1E', 'CDK12', 'HMGCS2', 'PEX11A', 'ATP11B', 'MDC1', 'NEIL1', 'E2F8', 'COL4A5', 'SCUBE2', 'ZNF318', 'FA2H', 'LRIG1', 'CRAT', 'APOD', 'SOX15', 'MLPH', 'MET', 'CHST7', 'TRIL', 'UACA', 'PLSCR1', 'CERS4', 'TRIM3', 'LRRC31', 'H4C12', 'PIP', 'EMC8', 'SFRP4', 'RPSAP47', 'ALCAM', 'PDCD4', 'PBLD', 'GGT1', 'CRISP3', 'SERPINB5', 'KDM5A', 'DNAI2', 'KLK5', 'KMO', 'TNXA', 'LAMA5', 'AGTR1', 'ABI3', 'ADGRL3', 'AGR2', 'CDKN2C', 'SPDEF', 'A2M', 'CEACAM6', 'TRMT11', 'TOX3', 'NFIA', 'ABCC6', 'CYP4F8', 'POLR3H', 'TFF3', 'G6PD', 'ESM1', 'CD7', 'ECHDC2', 'KRT222']
Refined Community 16: ['PAX6', 'MCRIP2', 'HMGCR', 'PSAT1', 'LUZP1', 'UGT8', 'BNC2', 'PALS2', 'CLCN7', 'SERPINA3', 'ART3', 'PGBD5', 'SFRP1', 'UPP1', 'PARD6B', 'NSUN4', 'KRT6B', 'SOX11', 'PON2', 'PLAAT1', 'RNASE6', 'ECE1', 'BMAL2', 'FYCO1', 'KCTD13', 'ELF5', 'GJB3', 'FTH1P5', 'PDIA2', 'CRYAB', 'ANAPC15', 'SPDL1', 'UNC13D', 'RARRES1', 'CXCL8', 'ROPN1', 'CHST3', 'TPSG1', 'BCL11A', 'MSLN', 'RNPS1', 'DOCK10', 'MMP7', 'DLX3', 'CCDC9B', 'LY6D', 'FOXK1']
Refined Community 17: ['AMDHD2', 'TUBGCP6', 'ACOX1', 'PIGT', 'SEPHS1', 'SIDT1', 'CFD', 'GPRC5A', 'HTRA4', 'PIP', 'RAMP1', 'MAPKAP1', 'CA12', 'GNB5', 'STX1A', 'SCGB1D2', 'CLGN', 'SYT1', 'MYH10', 'STK32B', 'FKBP11', 'CLSTN2', 'GLI4', 'ADORA2A', 'ARMT1', 'AREG', 'NINJ2', 'CPSF7', 'SRP19', 'SLURP1', 'NCF1', 'AGR2', 'PNMT', 'ELAPOR1', 'EEF1A2', 'DSN1', 'CACNA1D', 'YTHDF1', 'KRT18', 'GATA3', 'CRABP2', 'TFF1', 'PBX1', 'GPR162', 'GPC1-AS1', 'TTC39A', 'ESR1', 'MIRLET7D', 'RBM24', 'RELL1', 'CCDC74A', 'MAPK9', 'RGS20', 'TPSAB1', 'SEMA3B', 'CACNG4', 'SLC16A6', 'SYTL2', 'SYBU', 'ARRDC4', 'ELOVL2', 'KRTAP5-9', 'PGR', 'AKR1C1', 'PGS1', 'DACH1', 'C1orf21', 'TRIB3', 'SYNGR3', 'PNPLA4', 'AZGP1', 'IFT74', 'SPDEF', 'MAP3K12', 'SLC39A6', 'HEPH', 'IGKV3-20', 'SLC44A4', 'STC2', 'SCGB2A2', 'CDCA8', 'EVI2B', 'ANXA9', 'TOX3', 'IFT81', 'SFPQ', 'RSPH1', 'TCHP', 'CHURC1', 'SLC27A2', 'GPRC5C', 'CRIP1', 'GFRA1', 'SOCS3', 'LINC00115', 'DNAJC12', 'CYBRD1', 'CASC3', 'ACOX2', 'TJP3', 'ETFRF1', 'PIERCE1', 'ACADSB', 'ERBB3', 'BRD9', 'LAMB2', 'TBC1D9', 'PAAF1', 'CHAD', 'SSTR2', 'CYP2B6']
Refined Community 18: ['TMEM47', 'CDKN2C', 'KCNK5', 'ACOX2', 'CCNL1', 'CHST15', 'EPB41L2', 'SDC2', 'CAPZA2', 'RETSAT', 'CEP55', 'KRTAP9-8', 'GCAT', 'OR2AT4', 'PSIP1', 'IFT20', 'IGF2BP2', 'EEF1AKMT3', 'TNXA', 'GPD1L', 'GSE1', 'CD52', 'PLAGL1', 'GLIPR1', 'SIDT1', 'CCDC81', 'ELOVL5', 'FAM174B', 'CLN3', 'ELAPOR1', 'CADPS2', 'TOP2A', 'MLLT1', 'STARD3', 'MLXIP', 'ETFB', 'PUM3', 'TUBB6', 'HMGN4', 'CSRP2', 'RABEP2', 'RPS6KB1', 'C16orf87', 'KMO', 'HHIPL2', 'GGT1', 'ACACA', 'HOXB2', 'HACD3', 'TAF5', 'ADGRL4', 'MYO1B', 'P2RY14', 'TP53TG1', 'SAMD14', 'WNT4', 'EP300', 'NPLOC4', 'PFDN4', 'FBXO46', 'TASOR2', 'BLVRB', 'FBXL16', 'CCDC74B', 'MLF1', 'CDKN2A', 'GINS1', 'NT5C2', 'LMO4', 'CNPY4', 'KRT7', 'SEPHS2', 'ARPC4', 'DPEP1', 'C4A', 'CEP57', 'SRRM2', 'CIRBP', 'SYNCRIP', 'FAM110A', 'TMEM143', 'EVI5', 'S100A4', 'CDKN1C', 'NAT2', 'SLC38A1', 'TP53BP2', 'GSPT2', 'IL17C', 'RNF149', 'CASD1', 'FMR1', 'TLR7', 'BCL11A', 'E2F3', 'EN1', 'MED13L', 'ZFAND1', 'EPHA10', 'STYX', 'CRIPT', 'WTAP', 'CNKSR1', 'MBOAT2', 'GLS', 'MSX2', 'MRPL9', 'SPTLC2', 'IDH2', 'NCOR1', 'OLFML3', 'PLAAT2', 'KIF12', 'SIK1', 'TLCD1', 'FCN2', 'ENOSF1', 'CYP2J2', 'BRINP3', 'CROT', 'RAD50', 'CCDC117', 'CYP51A1', 'FABP5', 'SPRED2', 'ZKSCAN1', 'CYB561', 'KRT6A', 'GSTZ1', 'DAPK1', 'GAS2', 'CCNO', 'BTG3', 'KRT6B', 'SPINT1', 'C1QBP', 'PNISR', 'ELOVL2', 'SH3BP4', 'MAOA', 'HCK', 'LRP1', 'PIMREG', 'QDPR', 'AVL9', 'CHRNA9', 'HGS', 'MLPH', 'EFHD1', 'DDX27', 'TYROBP', 'TNFSF10', 'H4C12', 'ZMYND19', 'VWA5A', 'CTBS', 'HSPA6', 'LIMA1', 'ADAM18', 'LOXL2', 'DCAF7', 'MRTO4', 'KLF5', 'DNAJC12', 'COL26A1', 'MSR1', 'NR2F1', 'PRKACB', 'NCK1', 'NHERF1', 'USP36', 'CHI3L2', 'STK26', 'IRF6', 'ACSS3', 'TMEM87A', 'C17orf58', 'GPR161', 'KIF14', 'AMACR', 'CCL2', 'ETFA', 'PRICKLE1', 'KCNMA1', 'MMP15', 'MMP24', 'IFRD1', 'CHST1', 'TMEM135', 'CERS4', 'ADAM28', 'SOCS5', 'C16orf74', 'LDHB', 'MPHOSPH6', 'ZBTB42', 'CES1', 'PROM1', 'TRPV6', 'TTLL12', 'GYG1', 'HERC5', 'CLDND1', 'SREBF1', 'GPX7', 'FZD8', 'AXIN1', 'ATP1B3', 'MAP2K6', 'MPHOSPH9', 'TAF1A', 'NRAS', 'SCHIP1', 'RIN3', 'ACSL3', 'CLCA2', 'MCUR1', 'ZNF638', 'ARFIP2', 'SPAG1', 'ABCC6', 'PSMG1', 'TRAK2', 'CRISP3', 'MAPT', 'CFI', 'PRKX', 'TACSTD2', 'TMEM41B', 'RHBDF1', 'EPHA4', 'NUP88', 'ADCY7', 'RSPH3', 'FBL', 'PLCL2', 'BTBD3', 'TPD52', 'TGFB3', 'FUT8', 'MYL9', 'MYPOP', 'LITAF', 'PPY2P', 'APLP2', 'FOXC1', 'PTK6', 'LIMCH1', 'LEFTY2', 'PIK3C2A', 'BCAS3', 'DMD', 'GABRP', 'DHCR7', 'SEC62', 'LRBA', 'LGALS1', 'SUPT16H', 'CLDN8', 'TSPAN13', 'NFIX', 'TMPRSS2', 'SLC2A10', 'RNF141', 'UPF3B', 'TP63', 'DLK2', 'TPGS2', 'DCAF6', 'JRK', 'RTL10', 'CYBRD1', 'BMERB1', 'CTNNA1', 'NYNRIN', 'SMG7', 'GALNT7', 'ZWILCH', 'PDLIM1', 'ADAM33', 'SLC16A1', 'MMP11', 'CYB5A', 'CC2D1B', 'ITGA4', 'AR', 'EGFL6', 'TFF1', 'GDPD5', 'SIGIRR', 'RALGPS1', 'SLC26A3', 'PNMA1', 'RNF220', 'CD247', 'HBQ1', 'KIF11', 'UNG', 'AKR1D1', 'VWA1', 'SNRPA', 'CEP164', 'MCTS1', 'MYC', 'IDI1', 'CYBC1', 'CASP3', 'TSPAN3', 'TTC13', 'NRXN1', 'BRD1', 'SAR1B', 'MTUS1', 'WDR75', 'TXNIP', 'KRTAP9-2', 'RGS10', 'HOXA5', 'SH3YL1', 'OPTN', 'TRIL', 'TFAP2B', 'STOML2', 'PAGR1', 'DSE', 'FCGR3A', 'NANS', 'RNF138', 'PCLO']
Refined Community 19: ['FBF1', 'POLD3', 'ABLIM3', 'PLCH1', 'GINS1', 'RAMP1', 'FAM217B', 'KRTAP5-7', 'CCNB2', 'LRP12', 'PMAIP1', 'IGLV3-25', 'HDGFL3', 'LAMB2', 'RASGEF1A', 'STK17B', 'BMI1', 'ABRACL', 'UGCG', 'TK1', 'LGR5', 'HEMK1', 'TFAP2B', 'TCAP', 'MPHOSPH6', 'GADD45B', 'SWAP70', 'PIP', 'ESR1', 'FLRT3', 'COL9A3', 'ZYX', 'PPARD', 'BNIP3', 'SUPT16H', 'CLIC3', 'GPR160', 'MICB', 'ZNF750', 'TRIM36', 'BDNF', 'PEX3', 'GHR', 'SLPI', 'TAL1', 'BST2', 'AGFG1', 'ALDH1A3', 'EFCAB2', 'LPCAT4', 'EGF', 'AKR7A3', 'POLQ', 'DDAH2', 'PRSS12', 'ANXA2', 'RBM48', 'COMT', 'GREB1', 'NKX1-2', 'MYO10', 'SC5D', 'DAP', 'TNFRSF10D', 'ADIRF', 'CEP55', 'KIT', 'SCG2', 'SLF2', 'LDOC1', 'TFF3', 'APLP2', 'CRNDE', 'GPRC5B', 'PAN2', 'TSPAN8', 'BTG2', 'FDFT1', 'MYLIP', 'IGFBP4', 'UGT2B11', 'TMEFF1', 'GPAA1', 'HMGN4', 'NUDT4', 'SEMA3F', 'ABCA12', 'CFAP45', 'RBBP8', 'CPA3', 'TWF1', 'NCAM1', 'DIP2C', 'TCN2', 'AEBP1', 'WSB2', 'ERO1A', 'EPHX3', 'TMEM87A', 'OPTN', 'SKAP2', 'ADAMDEC1', 'DCBLD2', 'DCAF10', 'RPS6KA1', 'PSME4', 'MBD3', 'TPM2', 'NDRG1', 'SCUBE2', 'REEP5', 'KPNB1', 'CSTB', 'PACRG', 'IKBKB', 'BEX1', 'UGT2B15', 'VEZF1', 'RAB38', 'MDM2', 'TRMT1L', 'CLDN1', 'SERPINA1', 'GALNT3', 'TREX1', 'TACC2', 'SMAD3', 'MTUS1', 'WDR19', 'ZNF394', 'TMEM265', 'SNX5', 'RMND1', 'GMPS', 'RSKR', 'IGHV3-33', 'MT2A', 'DMD', 'LAPTM4B', 'MIR21', 'NPY1R', 'PNMA1', 'ABCC5', 'PSIP1', 'TTYH1', 'TAC4', 'EDAR', 'GADD45A', 'ACE2', 'ACSL3', 'FOSL1', 'SHC2', 'POSTN', 'NBR1', 'FABP7', 'ASS1', 'TLE1', 'FADD', 'PRKAA1', 'AMACR', 'RESF1', 'PWP1', 'PDLIM1', 'OCA2', 'MAP3K5', 'AFF1', 'TBC1D9', 'ENSG00000293341', 'CHST1', 'CRYBG1', 'PLXNA2', 'HGD', 'RLN2', 'MKNK2', 'RRM2', 'POLG2', 'GFPT2', 'PNP', 'GFUS', 'SPOP', 'ETFA', 'CNOT3', 'SMARCE1', 'LRRFIP1', 'PON2', 'SUZ12', 'LBP', 'C1QBP', 'CPQ', 'PKP2', 'PHKB', 'EGR3', 'PABPC1', 'RASSF3', 'ERGIC2', 'FAM20A', 'C16orf95', 'PPIF', 'SNRPN', 'POGLUT1', 'BAG1', 'TGFBR2', 'RPL13', 'LINC01133', 'SLC27A3', 'CAPN9', 'ADAM9', 'IGF1R', 'PI3', 'DDC', 'E2F1', 'CCL18', 'CCL23', 'HSD17B4', 'SEMA3C', 'TSPAN3', 'ST6GALNAC2', 'DLGAP4', 'TUBG2', 'PLCL1', 'ZDHHC4', 'ASB13', 'LIMCH1', 'ACOT7', 'GTF2H2C', 'PAM', 'TANK', 'EIF2AK2', 'DBNDD2', 'MEOX2', 'G6PD', 'PELI2', 'SLC31A1', 'SGO1', 'EPB41', 'GBP1', 'ELN', 'RTN1', 'LAT2', 'THAP1', 'TRPS1', 'NAMPT', 'SAR1B', 'PSME3IP1', 'MT-CO2', 'CROT', 'KAT7', 'KRTAP9-3', 'SERPINC1', 'ABHD13', 'PALS2', 'SIGMAR1', 'CHN2', 'MCF2', 'MAPT', 'FAR2', 'PER3', 'PGRMC1', 'RECK', 'THBS1', 'PFKM', 'FNDC3B', 'RNF166', 'TIMELESS', 'FKBP5', 'MIA3', 'ITGA3', 'CCDC6', 'GATA3', 'QPRT', 'ATP7A', 'MED13', 'ECHDC1', 'MAOA', 'CDYL2', 'TMT1A', 'RNF24', 'NUFIP1', 'MKI67', 'ADH5', 'SREK1', 'MSMO1', 'FAAH', 'TRIM65', 'AARD', 'ZKSCAN5', 'PLIN2', 'DST', 'MEIS3P1', 'TNPO2', 'ABCC6', 'ITCH', 'STEAP3', 'C2CD5', 'CLDN4', 'IL15', 'MZT2A', 'ADAMTS17', 'EIF3B', 'CLK4', 'HEBP2', 'MORC4', 'CHRNB2', 'ELOVL2', 'ECHDC2', 'INPP5F', 'DNALI1', 'IDI1', 'LARGE1', 'CLSTN2', 'ORC3', 'CXCL16', 'SLC27A2', 'CEP131', 'MAP2K6', 'TCEAL9', 'VLDLR', 'NQO1', 'TFEC', 'BRINP3', 'DMAC1', 'NTSR1', 'RPS20P22', 'RIOX2', 'CYP2J2', 'HPRT1', 'TNFRSF25']
Refined Community 20: ['WWP1', 'NOTCH4', 'CAPN9', 'MOAP1', 'FBXL5', 'KRT27', 'CNTROB', 'SLC43A3', 'ADD2', 'PFKP', 'PNLIPRP3', 'RFLNB', 'HSD17B4', 'PTP4A2', 'DMD', 'BCL2A1', 'MMP2', 'SKP2', 'TMT1A', 'DALRD3', 'SF3B1', 'LRBA', 'CD36', 'ANKRD27', 'SLC9A6', 'CTNNB1', 'KIF5C', 'WNK4', 'PCNT', 'INPP5J', 'NOSTRIN', 'TBC1D9', 'PPP1CB', 'IGF1', 'KLHL2', 'ANK1', 'ST8SIA1', 'PPP1CA', 'GPR20', 'SREBF1', 'WNT9B', 'SCCPDH', 'ALG8', 'RAP1BL', 'RHOF', 'LASP1', 'BUB1', 'NHERF1', 'LLGL2', 'GALNT10', 'HS3ST1', 'RALGAPA1', 'UNC5D', 'GDF15', 'IGF2BP2', 'FGB', 'KIF16B', 'MATN3', 'PALM2AKAP2', 'CASP8AP2', 'CD46', 'GIMAP4', 'PUM3', 'KRTAP9-3', 'GBE1', 'SOD2', 'TRPA1', 'CDT1', 'SUPT4H1', 'YBX3', 'REEP1', 'SERPINA5', 'PSME4', 'GLMN', 'WNT3A', 'SYNC', 'POPDC3', 'ACP1', 'APH1A', 'CYP2B7P', 'MAPT', 'CSF3R', 'NOL8', 'ARHGEF40', 'HPGDS', 'CTCFL', 'KLF5', 'DSC2', 'TFPI', 'EMSY', 'GABBR2', 'YAP1', 'CYP7A1', 'TUBB6', 'EYA1', 'PSPH', 'TFF3', 'LRRC17', 'FYCO1', 'GZMH', 'AMACR', 'MCM5', 'C1QTNF3', 'MTMR2', 'EVI5', 'MAK', 'ZNF727', 'CACNA2D2', 'SYNCRIP', 'TOM1L1', 'AKR7A3', 'SLC35E2B', 'IFIT3', 'UGT8', 'VGLL1', 'TAL1', 'NT5C2', 'SEPHS1', 'POP1', 'MAGED2', 'PLXNB1', 'IL18', 'EMP2', 'PRKX', 'CEBPB', 'TFAP2A', 'PLSCR1', 'ADCY1', 'LRP6', 'CSRP2', 'CHEK1', 'MYC', 'TAF1', 'ADAM33', 'RNF43', 'SNTB1', 'MRTFB', 'CILP', 'SERPINA3', 'HEG1', 'SIRPAP1', 'EXT1', 'TRMT11', 'CHST2', 'AR', 'FASN', 'PRDM14', 'CSTB', 'GPM6B', 'CA12', 'CTDSPL2', 'FCN2', 'LSM14B', 'ZFAND6', 'SSR2', 'DHRS7', 'AKAP12', 'LIX1L', 'PTGER3', 'MROH5', 'IGLV2-14', 'CYBA', 'GABRP', 'NR2F6', 'KLHL7', 'ETFA', 'RIF1', 'RARA', 'CDK5RAP3', 'NME3', 'KDM4A', 'CROT', 'ZNF34', 'LDHB', 'TSHZ3', 'CERS6', 'C1orf131', 'DACH1', 'KDM4B', 'ITPA', 'NPY1R', 'MICALL1', 'TAB1', 'TP53INP1', 'SCNN1A', 'YEATS2', 'FIGNL1', 'LPP', 'MT1H', 'CDK2', 'PSMD2', 'OFD1', 'EPHA4', 'LONP2', 'PRNP', 'FUT8', 'DNALI1', 'SPATA20', 'NABP1', 'RSU1', 'EREG', 'WASHC5', 'FAM174B', 'SLC2A5', 'TBC1D12', 'TMEM135', 'RGS5', 'RETSAT', 'COX16', 'PTPRN2', 'CRIP2', 'CD2AP', 'CDK17', 'COL24A1', 'PHACTR4', 'ADGRB1', 'CXCL8', 'PKP1', 'KLHL22', 'EIF1AX', 'CIRBP', 'CENPA', 'ROBO3', 'SPHK1', 'ARMT1', 'MYO10', 'GPC1-AS1', 'TBRG1', 'TRIM2', 'UBA3', 'LBR', 'G6PC3', 'KIF14', 'HSPB1', 'RNF138', 'ADM', 'CD180', 'ADH5', 'TMEM125', 'LGALS8', 'ZNF527', 'CCNC', 'STIL', 'PDSS1', 'OBP2B', 'PI3', 'DNAJC12', 'ITGB5', 'EID1', 'RUNX1', 'MEOX1', 'TCEAL4', 'S100A2', 'OSGIN2', 'PEX11A', 'EFCAB2', 'AZGP1', 'HEBP2', 'HMGCS2', 'CCL2', 'SLC39A3', 'NFKBIZ', 'TFF1', 'MARCO', 'MRFAP1L1', 'TESC', 'GATA3', 'FOXC1', 'CYP3A7', 'IL6ST', 'NECTIN4', 'RBM4', 'SERTAD4', 'CEBPG', 'NFE2L3', 'CFLAR', 'ARMC7', 'UBASH3B', 'PTX3', 'RAD51AP1', 'VPS37C', 'FBXL7', 'IRX6', 'FMO5', 'PPP1R15B', 'KRT16', 'HYCC1', 'RET', 'GAL', 'PDGFD', 'SPTLC2', 'GALNT9', 'ASF1A', 'APBB2', 'TM7SF2', 'NDC80', 'KIF4A', 'BCCIP', 'NDUFAF4', 'UBN1', 'MID1', 'U2SURP', 'TCF7L1', 'TSN', 'C2CD2L', 'ARMC8', 'PLPP5', 'ANP32A', 'ECI2', 'PAAF1', 'GATB', 'DPM1', 'GSTM3', 'HGF', 'FBXO32', 'CERS2', 'PRSS8', 'AGR2', 'GALNT6', 'FGF1', 'BBS1', 'CLCN4', 'MLPH', 'GSTZ1', 'MAGOH', 'SLC35B4', 'MSN', 'ST14', 'TCEAL1']
Refined Community 21: ['DPM3', 'TNFRSF17', 'OAS2', 'ADGRL3', 'RSAD2', 'GIMAP4', 'ZSCAN18', 'ISG15', 'PLAU', 'GPR171', 'CD38', 'DNAAF9', 'OSER1', 'OASL', 'POU2AF1', 'HERC6', 'GZMK', 'NFIX', 'CST7', 'CCR2', 'TMEM147-AS1', 'IFIT1', 'HERC5', 'ERAL1', 'IGLV2-14', 'S100A13', 'PRG2', 'CD2', 'IGLL3P', 'PRRX2', 'TRBC2', 'DSCC1', 'IGKV1D-13', 'ALKBH5', 'MDM4', 'ZFP64', 'PPHLN1', 'IFIT3', 'LCP2', 'IL10RA', 'DHRS4-AS1', 'ACYP1', 'IGHG1', 'ESYT2', 'MIA3', 'NUP50', 'SH2D1A', 'ITGAV']
Refined Community 22: ['KIF4A', 'SLC38A1', 'TRPA1', 'CCNB2', 'CEP55', 'PRKDC', 'ITGB4', 'MELK', 'NEK2', 'MTERF3', 'BUB1B', 'RET', 'B4GAT1', 'NCAPG', 'SNX9', 'SLC25A32', 'LRRFIP1', 'SLC6A14', 'CD36', 'GZMH', 'RAD54B', 'FNDC3B', 'RRS1', 'UCK1', 'FAM234B', 'ESRP1', 'BAZ2A', 'ARMC1', 'NRIP1', 'MRPL15', 'RSKR', 'KIF15', 'MAP7D2']
Refined Community 23: ['STEAP3', 'ERO1A', 'NSMCE2', 'PDZK1IP1', 'CLDN1', 'ADM', 'CTSS', 'AP2A2', 'VPS37C', 'MEF2D', 'NAMPT', 'STAU2', 'MALL', 'AGFG1', 'AQP11', 'C1orf43']
Refined Community 24: ['TENM4', 'THBS2', 'TCN2', 'CTSK', 'CTSW', 'FAP', 'SRPX2', 'PAXBP1', 'COL5A1', 'DACT1', 'PDP1', 'PDCL3', 'BTBD3', 'SERPINF1', 'FTCD', 'GRIFIN', 'ADAM12', 'ACADS', 'COL5A2']
Refined Community 25: ['SUPT4H1', 'CUL9', 'RAD51C', 'PTRH2', 'FOXQ1', 'TENT5C', 'TACO1', 'RAPGEFL1', 'PCK2', 'PPM1D', 'PABPC4L', 'RNF43', 'BRIP1', 'CYB561', 'CA9', 'SRSF11', 'TUBD1', 'MTMR4', 'ENSG00000301761']
Refined Community 26: ['RHOB', 'RCE1', 'LINC01973', 'NAP1L1', 'ESR1', 'PIDD1', 'CERS6', 'CCT2', 'SRPX2', 'TBC1D9', 'SLC39A6', 'GPD1L', 'EMP2', 'AGR2']
Refined Community 27: ['MID1IP1', 'HNRNPU', 'TRIM36', 'AARD', 'TAGLN2', 'SLC35A3', 'SREBF1', 'EPCAM', 'GSE1', 'ZACN', 'IRX5', 'CYP11B2', 'LRRC31', 'MSX2', 'ESRP2', 'SUCO', 'PERP', 'BLVRB', 'CDH1', 'RND1']
Refined Community 28: ['GABBR2', 'AIP', 'GSDMB', 'PGAP3', 'C8orf76', 'KLRC1', 'PNMT']
Refined Community 29: ['RUNX3', 'TM7SF2', 'TRBV5-4', 'TMEM201', 'CXXC4', 'SPP1', 'CD8A', 'TYSND1', 'S100A14', 'TIAL1', 'CXCL14', 'PSMB3', 'TTC39A', 'RPTOR', 'G0S2', 'GIMAP5', 'FAXC', 'PLEK', 'LCP1', 'F2RL2', 'HOXA10', 'SLC40A1', 'FTCD', 'NDUFAF8', 'CD48', 'LAMB1']
Refined Community 30: ['RAB2A', 'FGFR3', 'DNALI1', 'EPHA10', 'OTOP2', 'DCLK1', 'CYP2D6', 'MAP3K12', 'IKBKB', 'PDZD8', 'GOSR1', 'GAK', 'IGF1R', 'RET', 'EREG', 'ERBB3', 'ADGRA2', 'TUBA1C', 'FGF2', 'COL11A1', 'ERBB4']
Refined Community 31: ['DYRK1A', 'CYP26A1', 'NEK4', 'CHUK', 'HK2', 'SCYL2', 'SNX1', 'SNRK', 'SH2D3A', 'STAT3', 'NDUFAF4', 'PIP4K2B', 'SLK', 'CYP2B6', 'CSNK1G3']
Refined Community 32: ['ITK', 'SNHG12', 'CRTC3', 'SPATC1', 'LYN', 'STK17B', 'ZNF790', 'MLLT10', 'DOCK5', 'MAP4K1', 'ZAP70', 'MLKL', 'PRKCQ', 'ASF1A', 'FGR', 'ZNF446']
Refined Community 33: ['BUB1B', 'EYA4', 'CHEK1', 'HOOK3', 'AFG2A', 'SNHG26', 'TMEM130', 'PLK1', 'PBK', 'KRTAP4-4', 'CDK1', 'CDC7', 'EFTUD2', 'PMAIP1', 'PLK4', 'AURKB']
Refined Community 34: ['DOCK7', 'DNM2', 'SETD5', 'ST14', 'FURIN', 'TCEANC', 'TANK', 'FAM216A', 'NUP62', 'MMP16', 'CUX1', 'CAVIN1', 'POSTN', 'CLIC3', 'TINAGL1', 'CAPN15', 'RNPS1', 'ACSF3', 'COG3', 'FLNC', 'THEMIS2', 'TNIP1', 'IFI30', 'SCRIB', 'CASP4', 'TUBB2B', 'SGMS1', 'SBNO2', 'PIK3CG', 'SGCB', 'MAD2L1BP', 'MYC', 'IFITM3P7', 'CDK17', 'ANKLE2', 'ARRDC4', 'CNKSR1', 'CHST15', 'KHDC4', 'PASK', 'CHEK1', 'SEMA3G', 'DAPK2', 'TRRAP', 'CYBA', 'EHBP1L1', 'COL20A1', 'ADGRA2', 'ZNF318', 'LAGE3', 'TCOF1', 'LIMD2', 'SNX31', 'CELSR3', 'WFIKKN2', 'JMJD6', 'ID4', 'RAB35', 'TOX3', 'EWSR1', 'ZNF142', 'ACE', 'TJAP1', 'CLDN6', 'KCNK5', 'CALM2', 'OFD1', 'CDK14', 'PPARD', 'VAMP5', 'CXXC5', 'ZMIZ2', 'GRAMD4', 'MNAT1', 'DDX39B', 'RBM26', 'CDC42SE1', 'RAD51', 'ACTR1A', 'ZNF878', 'DCAKD', 'PRRC2A', 'ENGASE', 'ENSG00000293341', 'BRD2', 'SF1', 'MLXIP', 'BMP5', 'DLX4', 'CGRRF1', 'BAG6', 'BTN2A1', 'SH3BP5-AS1', 'GBA2', 'MCM5', 'CLCN4', 'TYW5', 'POLR1HASP', 'ADAM10', 'MKI67', 'DGKZ', 'NPIPB3', 'KDM1B', 'SMC6', 'MCM3', 'TBRG1', 'TNFRSF25', 'CSNK2B', 'KDM4A', 'ABCA6', 'BPNT2', 'WIPI1', 'MYPOP', 'EXOC3', 'HSPBAP1', 'CALML4', 'RUSC1-AS1', 'NSUN5P1', 'MICALL2', 'AMY2B', 'ITGBL1', 'ENSA', 'NOXO1', 'CCL23', 'CCNJ', 'DHCR24', 'ZNF436-AS1', 'CENPBD1P', 'PPHLN1', 'IFRD1', 'PABPN1', 'TSPAN14', 'FOXK2', 'TBL3', 'DELE1', 'SLC39A10', 'AARD', 'CTDSP1', 'BTAF1', 'CDYL', 'BAMBI', 'PHF11', 'AGO4', 'SHC1', 'MIRLET7D', 'MADD', 'ADGRL4', 'TCIM', 'H6PD', 'LARP1', 'MARCHF6', 'TAPBP', 'CCDC77', 'RPS6KA5', 'RPS6KA1', 'SUPT20H', 'RAPGEF1', 'APBA3', 'PHACTR4', 'CLASRP', 'TCIRG1', 'FGL2', 'CLK3', 'STRN3', 'KCNB2', 'DDX11', 'SCCPDH', 'PSMG4', 'PCGF6', 'FBXL19', 'SPON1', 'SNHG26', 'RBP3', 'CYP26B1', 'C1QTNF8', 'QSOX2', 'SUPT7L', 'TIE1', 'SMC4', 'RPS3', 'CCNL1', 'AKAP13', 'ECHDC1', 'SLC15A4']
Refined Community 35: ['ODC1', 'LAP3', 'ZIK1', 'TCAF1', 'PRRT2', 'PACSIN3', 'STAG2', 'CD300C', 'LPO', 'BAIAP3', 'DNAJB14', 'HIKESHI', 'UQCC1', 'APBA2', 'TSHZ2', 'ZNF217', 'FBP1', 'CCT2', 'PTK6', 'ODR4', 'SPPL2A', 'BNIP3L', 'ERGIC3', 'POSTN', 'MTG2', 'IFT22', 'CACNA1G', 'TMEM87A', 'SPATA24', 'GEMIN6', 'ATP5F1E', 'ZNF148', 'NAAA', 'FBXO43', 'RRP7A', 'EME1', 'MAGT1', 'SGCE', 'BLVRB', 'SRPK3', 'ZNF638', 'SEC23A', 'COPS7B', 'PDP1', 'PPP1R3D', 'KRTAP4-4', 'AKR1A1', 'RHOBTB3', 'PRKAR1A', 'TAF1', 'RAPH1', 'PDE9A', 'LEPROT', 'C11orf52', 'SLC4A7', 'DCAF7', 'ATL3', 'MLLT10', 'U2SURP', 'DPM1', 'CFD', 'SNORA71B', 'AZI2', 'MORF4L2', 'CLMN', 'ZBTB42', 'DDX27', 'STX16', 'F2R', 'ZBTB8A', 'GSPT2', 'FAM216A', 'ADH4', 'BCAS4', 'LINC00491', 'ESR1', 'SEC62', 'PRDM14', 'CACNG4', 'METTL25B', 'UHMK1', 'YWHAB', 'STAU1', 'SHC4', 'DUSP9', 'SYDE2', 'DMAC2', 'FABP9', 'DAAM1', 'TCIRG1', 'GRID1', 'C16orf87', 'MAPK8IP3', 'CDS1', 'AURKA', 'FOXD1', 'ENOX2', 'AAGAB', 'TANK', 'RSPH3', 'GBP5', 'AXL', 'HSPB1', 'NPAS4', 'KRT12', 'POLR2F', 'YIPF5', 'PDP2', 'HSP90AA1', 'TLN1']
Refined Community 36: ['RNF112', 'KIRREL1', 'LINC00115', 'MTNAP1', 'LMNA', 'OR4F6', 'NPM3', 'AMPD2', 'NUP214', 'ESRP1', 'TAF1D', 'MRPL9', 'SBNO2', 'GIGYF1', 'ILF3', 'ARHGAP45', 'IGKV2D-28', 'TNK2', 'ATAD3A', 'SLC39A10', 'HIP1R', 'DDX11', 'NOL8', 'MAPK11', 'DAZAP1', 'SLC26A7', 'SYNJ1', 'MTF2', 'IL4', 'LRRK1', 'CCAR2', 'HNRNPD', 'RANBP3', 'LLGL1', 'LENG8', 'MARK2', 'TRMU', 'SAFB2', 'ARNT2', 'SLC29A4', 'BAZ1B', 'IRX2-DT', 'HSD11B2', 'GABRQ', 'TPM2', 'PIP5K1A', 'SPAG9', 'IVD', 'DHX33', 'SEPTIN6', 'GPER1', 'SHANK2', 'SLC16A7', 'SCUBE1', 'SAMD12', 'PABPN1', 'ADRM1', 'LIMS2', 'TECPR1', 'TNPO2', 'SETD1A', 'MTERF3', 'LIAS', 'CC2D1B', 'THAP2', 'PIK3R1', 'PIM3', 'TTC13', 'PTOV1-AS2', 'PLK1', 'HEMK1', 'MIIP', 'SGSM3', 'NOP16', 'SLC23A2', 'UPB1', 'MAGEA3', 'BICDL2', 'EVA1B', 'AGMAT', 'NDUFB10', 'RFTN1', 'RBBP6', 'LARP1', 'SEMA6B', 'PTBP1', 'CACTIN', 'CDT1', 'APBA1', 'MYO9B', 'HEXD', 'BICRA', 'ENGASE', 'HNRNPU', 'TRAF2', 'FGFR1OP2', 'ZNF358', 'FOLR2', 'FURIN', 'LINC02627', 'AGRN', 'PTP4A3', 'SPPL2B', 'GSN', 'HAUS8', 'PICALM', 'EIF3B', 'MPP7', 'PHPT1', 'APBA3', 'PTDSS1', 'ARHGAP30', 'LDB2', 'ARHGAP17', 'SUV39H1', 'DMWD', 'MRPS25', 'FXR2', 'NFKBIZ', 'MYCBPAP', 'SNAPC4', 'QPCT', 'MBD3', 'PRIM2', 'MAST1', 'DUOX1', 'RNF214', 'INPP1', 'PSPH', 'DGAT1', 'DUOXA2', 'BID', 'PPME1', 'PNMT', 'CD300A', 'SLCO5A1', 'TENM3', 'RAD54L2', 'TRABD', 'GSK3A', 'TTC17', 'FBXL18', 'ANKRD13C-DT', 'POLR2F', 'ELP2', 'CRYGS', 'EMP2', 'LRRC74A', 'IQCG', 'MAGEA4', 'SLC39A3', 'AXL', 'FBXO44', 'OCM2', 'VAV3', 'CPSF7', 'B4GALT1', 'CYBA', 'SMARCA4', 'TNPO3', 'RBM15B', 'MAPK8IP3', 'HRK', 'NPR2', 'ZNF692', 'SCYL2', 'CAMK2B', 'CYP3A7', 'CKAP2L', 'HDAC10', 'COL13A1', 'SMG7', 'RABEP1', 'NOTUM', 'UQCC2', 'MAP2K2', 'PLEKHG1', 'MCTS1', 'LSM14B', 'FKBP1A', 'NSUN5P1', 'CSNK1G2', 'TRAF3IP3', 'FOXO4', 'DRG2', 'CCNB1', 'KMT2D', 'CAPN15', 'TLCD3B', 'MYC', 'ETV1', 'MAPKAP1', 'MICALL2', 'SRCAP', 'IFRD2', 'CRIP1', 'FLNB', 'MAPRE3', 'COPS7B', 'PHKG2', 'LTB4R', 'PTGR2', 'KCNK5', 'NSMAF', 'TEAD4', 'FABP4', 'TMEM201', 'UBAP2L', 'RBM14', 'NEURL4', 'AP5S1', 'MIRLET7D', 'WEE1', 'LINC00482', 'KAT7', 'ARL3', 'CALR', 'SMARCD1', 'SMIM19', 'ATF5', 'GALK1', 'FGF3', 'ATP6V1C1', 'IGHV3-23', 'CR2', 'U2AF2', 'CEP15', 'CUL7', 'PUS1', 'ANKRD54', 'VARS1', 'KHDC4', 'KIF5B', 'SECISBP2', 'SEC24D', 'CD36', 'CD40', 'FGF6', 'ACAP1', 'SGK3', 'RNPS1', 'IGF2BP2', 'INTS15', 'SP140', 'MTUS1', 'L3MBTL2', 'AURKA', 'RAP1GAP2', 'ARPC5L', 'THOC2', 'BRCA2', 'SAP25', 'TLN1', 'CCL2', 'APOL6', 'XPO5', 'SEMA6C', 'EML2', 'CIC', 'BRD1', 'RABGAP1', 'ALPK2', 'FBXO43', 'ZNF213', 'BOC', 'MYL12A', 'ERGIC2', 'H2BC12L', 'NR1H3', 'GTF2H2', 'SFSWAP', 'CD55', 'GPR161', 'LIMK2', 'ZNF611', 'DCLRE1C', 'KMT2E-AS1', 'TCHP', 'HEYL', 'CEACAM6', 'MMADHC', 'ANKHD1', 'R3HDM4', 'CLMP', 'SH3BP5L', 'C11orf54', 'EPN1', 'DES', 'TTF2', 'ZSWIM8', 'CHRNA9', 'MIR3682', 'RFLNB', 'SNN', 'CDC42EP1', 'ABHD13', 'PKP3', 'ZMIZ2', 'NSUN4', 'MAK', 'SPARC', 'TAF1C', 'OPLAH', 'ITGB8', 'USP21', 'CCDC6', 'SLC52A3', 'FAM110C', 'LAD1', 'DZANK1', 'ABCF3', 'SECTM1', 'GCAT', 'CNTROB', 'SEC23IP', 'RASSF7', 'ABCC10', 'NOL9', 'GLDC', 'VMP1', 'GADD45G', 'ZNF436-AS1', 'ZNF43', 'HLA-DRA', 'SYNE1', 'MTSS2', 'ARMC6', 'APOC1', 'MRPS7', 'FAAH', 'FANCA', 'NUTM2E', 'KIF15', 'ATP2B4', 'RHBG', 'CSNK1E']
Refined Community 37: ['AGAP4', 'FGF19', 'TRIM4', 'TRADD', 'PWWP2A', 'TAGAP', 'HIKESHI', 'STAG2', 'ERAP1', 'BTD', 'ST8SIA1', 'ARPP19', 'RSBN1L', 'TM2D3', 'NT5C2', 'TRIP11', 'HSPA4', 'ADD1', 'HEATR3', 'FGR', 'KLF2', 'UNG', 'YIPF5', 'NRARP', 'PCK2', 'GABBR1', 'EMILIN3', 'WSB1', 'TUSC3', 'PIM2', 'SLU7', 'N4BP2', 'MAFF', 'RSPH3', 'LEPROT', 'IGSF1', 'GINM1', 'MAP3K12', 'HMMR', 'FKBP14', 'NCOA3', 'BBC3', 'DNAL1', 'TRAK2', 'BMPR2', 'MRPL41', 'TOMM34', 'ANG', 'C22orf39', 'CYP20A1', 'GPRIN2', 'DENND11', 'MORF4L2', 'UBE2C', 'SIAH1', 'PDP2', 'KCTD20', 'KMT2E-AS1', 'TMEM67', 'ZBTB38', 'F2R', 'AQP11', 'POLR3A', 'APBB2', 'BBX', 'STK32B', 'SLC25A36', 'GNB5', 'APEX1']
Refined Community 38: ['DLEU2', 'S100A9', 'DSG3', 'EVI5', 'TSN', 'COP1', 'AFDN', 'ARHGAP45', 'TRAF5', 'MGRN1', 'NAP1L1', 'ARHGEF7', 'RPS6KB2', 'SRSF11', 'BANP', 'ACOX1', 'DGKE', 'SYNCRIP', 'EHD1', 'METAP2']
Refined Community 39: ['TANK', 'PRRG4', 'RPS13', 'ATP5F1B', 'TOX2', 'CHPF', 'CYP1A1', 'CCDC78', 'COIL', 'POLR3A', 'JUNB', 'MS4A7', 'UBA52', 'RPL19', 'EWSR1', 'ETV6', 'SS18L2', 'TMEM87A', 'ADNP']
Refined Community 40: ['EFR3B', 'BIRC5', 'TPPP', 'CHTF18', 'EGFR', 'LAMP5', 'VIM', 'NADSYN1', 'RABEP2', 'CCND1', 'ACTR1A', 'MFNG', 'STMN3', 'EREG', 'TOP2A', 'MKI67', 'CCNA2']
Refined Community 41: ['UBE2Z', 'IL7', 'RGS1', 'TAL1', 'COMMD5', 'GRHL2', 'EML1', 'MINPP1', 'NEFL', 'SYT17', 'PDE6A', 'AFF1', 'ZNF211', 'SLC29A2', 'H2AC8', 'YTHDC2', 'TRMT6', 'TUBGCP6', 'NAV1', 'ARHGEF6', 'NUP85', 'VRK1', 'LMOD1', 'SBNO1', 'SRSF11', 'GOLGA1', 'RYBP', 'GPC2', 'CSNK1E', 'PAAF1', 'PDXK', 'BRINP2', 'MAST1', 'ADGRG5', 'CD36', 'RASSF7', 'APEX1', 'GFM2', 'RCL1', 'BMP6', 'PMEPA1', 'LAPTM4B']
Refined Community 42: ['NOC2L', 'MICALL2', 'ORMDL2', 'KDM5A', 'PAK2', 'MNAT1', 'PRRG4', 'PRIM1', 'ANGEL1', 'RALY', 'CALU', 'AGL', 'NAF1', 'PRPF39', 'PPP2R5A', 'CCR7', 'TUFM', 'MAPRE1', 'CXCL5', 'AKT1', 'PEF1', 'FDFT1', 'RBBP4', 'MGMT', 'SF3B4', 'IL17RA', 'ZBTB42', 'PPY2P', 'PRR35', 'MAPKAPK2', 'SH3KBP1', 'RUNX1', 'CDC42EP4', 'SLC9A1', 'NT5DC2', 'ILF2', 'NAGA', 'LGR6', 'MCRIP2', 'MAP2K3', 'LOX', 'MLKL', 'TMEM265', 'PNPLA2', 'BAMBI', 'CXCL2', 'MPPED2', 'COPA', 'BBOF1']
Refined Community 43: ['TP53TG5', 'CIDEC', 'LAMB1', 'AMY2B', 'ADIG', 'AP1S2', 'VWF', 'CXCL12', 'ZNF133', 'NR1H3', 'PKM', 'NANS', 'LMO3', 'CHMP6', 'TRIB1', 'CLEC3B', 'SRPX', 'MAP4', 'RDH16', 'SORBS1', 'MAGED2', 'PPP4R3A', 'ZDHHC2', 'ALAS1', 'LUM', 'PGM1', 'H6PD', 'CCDC47', 'USH1G', 'NRP1', 'COL1A1', 'CAVIN3', 'OPRPN', 'MFNG', 'TF', 'NMU', 'HMGB3', 'FEZ2', 'SIGLEC1', 'HHEX', 'ETFB', 'BCKDHB', 'ITFG2', 'REXO1', 'XRRA1', 'ACAA1', 'UCK1', 'BNIP2', 'S100A6', 'HSPB8', 'PIGV', 'GHR', 'HTRA1', 'PRR16', 'MAP2K6', 'COL6A3', 'BNIP3', 'TGFBI', 'KRTCAP2', 'DPT', 'MBNL3', 'GPR152', 'KLF4', 'SULT1A1', 'WLS', 'DLL3', 'PPP2R5A', 'HOTAIRM1', 'MRC1', 'PDE8A', 'POGZ', 'PDXK']
Refined Community 44: ['LIAS', 'RELN', 'RNF149', 'ID2', 'TPD52', 'DDC', 'SERP1', 'MET', 'ACAA1', 'HCN3', 'TERF2', 'CLDN7', 'ATP6V1A', 'MAD1L1', 'CDK1', 'PLET1', 'LSM7', 'RETREG1', 'CACNA1H', 'ARHGEF5', 'LRRFIP1', 'ALDOC', 'LRRN1', 'VARS1', 'KCNN4', 'MAP2K2', 'CYB561', 'NPM3', 'FCRL2', 'ADGRE5', 'IRF6', 'ELL2', 'CCNJ', 'SS18L2', 'SHROOM3', 'PRR14', 'SLC66A2', 'CLDN3', 'ZNF652', 'ANKRD46', 'OSMR', 'FXYD3', 'SRP19', 'CARD6', 'TFAP2C', 'EBP']
Refined Community 45: ['DAB2', 'ERCC2', 'KIF15', 'EIF1', 'MRPL9', 'IRAG1-AS1', 'NRN1', 'SERPINE2', 'NCOR1', 'TMEM63A', 'PNLIPRP3', 'HNRNPA0', 'PIEZO2', 'NYX', 'ZNF470', 'CCT2', 'PBXIP1', 'OGFRL1', 'TMEM117', 'PCYT1A', 'HENMT1', 'GIT2', 'ANP32E', 'RPA2', 'BHLHE40', 'CTSB', 'CHD9', 'MT-ND5', 'FTH1P5', 'BAALC', 'HMGCS2', 'CALM2']
Refined Community 46: ['S100A2', 'KRTAP5-11', 'EGLN1', 'GPSM2', 'PLSCR3', 'DDR1', 'HLA-DPB1', 'UCHL5', 'CST3', 'ABHD16B', 'COL4A6', 'MAPRE2', 'SERPINE2', 'P4HTM', 'GAB2', 'DMAC2L', 'SERPINF1', 'LENEP', 'HCFC2', 'FGD3', 'LOXL2', 'IGFBP4', 'MPZL1', 'RNPS1', 'LTBP3', 'NHERF1', 'ADAM9', 'RBM5', 'ANGPTL4', 'P3H3', 'SERPINC1', 'VWF', 'DPY30', 'F10', 'SRPX', 'CTSB', 'ADGRL3', 'CTSF', 'TRAF5', 'AAGAB', 'POLA1']
Refined Community 47: ['SRPX', 'HBM', 'S100A2', 'CXCL9', 'ZNF567', 'CCN2', 'ITGA7', 'BCL11B', 'TIMP1', 'TINAGL1', 'HTRA1', 'P4HTM', 'COL22A1', 'TMSB15A', 'CAPN15', 'PAPPA', 'RBBP8', 'RALY', 'MARVELD2', 'IGFBP4', 'SH3KBP1', 'HCFC2', 'LOXL2', 'METRN', 'SERPINE2', 'C4A', 'S100A10', 'MFGE8', 'FAM117A']
Refined Community 48: ['SYNE4', 'MMP1', 'LAMA2', 'COL6A6', 'MMP19', 'MCFD2', 'COL4A1', 'ZNF623', 'COL26A1', 'RAB1B', 'CARD6', 'PRPS2', 'ASPN', 'HMCN1', 'COL17A1', 'CEP250', 'NGLY1', 'PICK1', 'RESF1', 'EHBP1', 'PLAT', 'TIMP3', 'COL19A1', 'MAK', 'ETNK2']
Refined Community 49: ['FYB1', 'COL17A1', 'TMEM130', 'FBLN5', 'FLG2', 'POFUT2', 'SLC27A2', 'S100A16', 'COL19A1', 'RNF157', 'VPS26B', 'OGN']
Refined Community 50: ['SMOC2', 'EME2', 'REXO2', 'KIF14', 'CEBPD', 'ZNF436-AS1', 'SHOX2', 'NAF1', 'FBLN1', 'CRYBG1', 'CNKSR1', 'C1S', 'ATP2A3', 'NRK', 'SSR2', 'ANXA5', 'SLIT3', 'DCN']
Refined Community 51: ['WNT11', 'AK5', 'FA2H', 'SHMT2', 'JADE1', 'OCEL1', 'CEP55', 'EIF6', 'RNFT2', 'CENPN', 'UBE2T', 'TNFRSF14', 'TSPYL5', 'PAPSS1', 'FADD', 'FZD6', 'BEX1', 'TOP3A', 'SKA2', 'MYBL2', 'MLLT1', 'NABP1', 'CHPF', 'GPT2', 'TNNC2', 'CDCA8', 'SLC39A6', 'CTPS1', 'EXO1', 'AQP5', 'UBE2I', 'PSMD14', 'PTTG1', 'CAMLG', 'IL20RB', 'MAP6', 'CYP3A43', 'SMARCA4', 'PTN', 'HSPA14', 'CARTPT', 'HOOK2', 'CXCL11', 'SMARCD1', 'RGP1', 'PRAME', 'ATP2C2', 'H3C4', 'EBP', 'CDCA5', 'RNASE9']
Refined Community 52: ['ZCCHC24', 'POP1', 'EIF5AL1', 'HRCT1', 'KCNMA1', 'CD93', 'ZMIZ1', 'RANBP9']
Refined Community 53: ['RBM14', 'ANK2', 'KLHL35', 'KDM4B', 'AEBP2', 'MYBL2', 'LENEP', 'GXYLT2', 'SELENBP1', 'DPP3', 'MYEOV', 'CABP4', 'PAK1', 'ANKRD13D', 'EMSY', 'ZFP30', 'P3H2', 'CD5', 'PITPNM1', 'CHAF1B', 'NFKB2', 'F10', 'TMEM138', 'ZNF318', 'KRTAP5-11', 'RAD54B', 'ZEB1', 'TKFC', 'LRRC32', 'TBC1D19', 'CD248', 'USP35', 'WNT11', 'LY6K', 'YIF1A', 'PHF12', 'PPFIA1', 'PDSS1', 'CDK2AP2', 'ASB8', 'TENM4', 'RPS6KB2', 'CCDC9B', 'TMEM132A', 'RBM4B', 'KLC2', 'TESMIN', 'NAV1', 'PELI3', 'ALDH3B2', 'PPP2R1B', 'CTTN', 'MYO7A', 'FAM216A', 'CFB', 'MUC1', 'CCDC81', 'FBP1', 'CAMK2G', 'POLD3', 'MRPL11', 'ACER3', 'NDUFC2', 'MRGPRF', 'SPTBN2', 'FBLIM1', 'MOGAT2', 'RBBP5', 'ANO1', 'DCST2', 'CCDC89', 'LMO2', 'UNCX', 'SLC39A4', 'RIN1', 'NDUFV1', 'CHRDL2', 'HCFC2', 'THRSP', 'SLCO2B1', 'PICALM', 'GLIS3', 'FGF3', 'LTO1', 'MCOLN3', 'CCS', 'FGF4', 'TMEM134', 'S100A4', 'NDUFV1-DT', 'NME1-NME2', 'ZNF16', 'SDHAF2', 'PDK2', 'CCL11', 'ATAD3B', 'EXT1', 'PER3', 'CSNK1A1L', 'FAM169BP', 'FOSL1', 'CYP3A4', 'KRTAP9-4', 'AIP', 'NEU3', 'HSP90AA1', 'PPP1CA', 'XKR9', 'CKS2', 'GOLGA7', 'CPT1A', 'RCE1', 'TFPI', 'NPY1R', 'PTX3', 'BRMS1', 'FGF18', 'CCDC87', 'AAMDC', 'TBX10', 'ACTN3', 'CDC73', 'JPH3', 'TMSB15B', 'NUDT8', 'MBNL1', 'TMEM126B', 'MTF2', 'KPNA1', 'DLAT', 'VDAC1', 'CORO1B', 'CLNS1A', 'ECI1', 'GSTP1', 'FANCG', 'BLTP2', 'THAP12', 'OMP', 'DBNDD1', 'SLIT3', 'PPME1', 'OR2AT4', 'PCK2', 'RAB22A', 'CCDC83', 'PGA4', 'RSF1', 'B3GNT6', 'CSTF2', 'IGKV1D-37', 'TMEM151A', 'CSF2RA', 'CALCR', 'EED', 'LRFN4', 'NMT2', 'GLI3']
Refined Community 54: ['CPSF6', 'TBK1', 'INTS15', 'IFNG', 'ZNF778', 'DYRK2', 'IL6R', 'PLAGL1', 'FRS2', 'LONP2', 'ABL1', 'CDK4', 'RRS1', 'OS9', 'YEATS4', 'CFL2', 'ASPN', 'CCT2', 'MAP1LC3A', 'TUSC2', 'PIK3R1', 'IL26', 'KRT28', 'SLC35E3', 'ANKRD26', 'FOS', 'CTDSP2', 'LYZ', 'METTL1', 'IL22', 'TNFRSF17', 'NUP107', 'TMEM19', 'GABBR2', 'CNOT2', 'PRKCB', 'XPNPEP1', 'XKR9', 'KCNK5', 'CYP27B1', 'AVIL', 'TSPAN8', 'BEST3', 'TSFM', 'MAPK3', 'NAP1L2']
Refined Community 55: ['EXO1', 'TBC1D3F', 'P2RX2', 'GOLGA3', 'NOC4L', 'POT1', 'PUS1', 'RNF6', 'PGAM5', 'USP42', 'GHR', 'SRSF3', 'NMU', 'SFSWAP', 'DDX51']
Refined Community 56: ['TCFL5', 'ERO1A', 'GNG2', 'PSMC6', 'STYX', 'SNRPA', 'FERMT2', 'COG7', 'NID2', 'EGFR', 'KIF1A', 'SLC25A10', 'CACFD1', 'GNPNAT1']
Refined Community 57: ['BTD', 'ARRDC4', 'CHCHD4', 'LYSMD4', 'LINS1', 'CCDC93', 'WRAP53', 'MEF2A', 'MSX2', 'ALDH1A3', 'SPHK2', 'LRRC28', 'PECAM1', 'FRY', 'KMO', 'IGF1R', 'GLRX', 'BRAT1', 'TARS3', 'SNRPA1', 'CERS3', 'ASB7']
Refined Community 58: ['NOXO1', 'GPC2', 'MYL6B', 'SORL1', 'PDE9A', 'CTSV', 'PHOSPHO1', 'LY6S-AS1', 'ODR4', 'MSRB1', 'PIGQ', 'C1QTNF8', 'TBL3', 'RBBP8', 'NME2', 'NTHL1', 'KIF13B', 'CLDN9', 'WDR3', 'PRR35', 'CHTF18', 'ARHGDIG', 'LPCAT2', 'KREMEN2', 'NOTCH2NLA', 'DUSP8', 'B9D1', 'ANXA4', 'MAPK8IP3', 'DPP3', 'UBBP1', 'PIK3R2', 'TCEA2', 'TNFRSF12A', 'ITIH3', 'ZNF875', 'RDH13', 'TMEM204', 'TMEM130', 'STYK1', 'SNRNP25', 'NHERF2', 'TELO2', 'LMF1', 'MRPS34', 'TUBB3', 'SYNGR3', 'FAHD1', 'HAGHL', 'CXCL13', 'UBE2I', 'CLCN7', 'NME4', 'METTL26', 'POLR3K', 'STAU1', 'CDH15', 'CYB5R3', 'KCTD5', 'PRSS27', 'FBXL16', 'HBM', 'GJA4', 'TPSAB1', 'ZNF598', 'MS4A1', 'BAIAP3', 'FBXO47', 'HBA2', 'ABLIM3', 'THEM4', 'THOC6', 'PRSS21', 'ZNF611', 'NFKBIE', 'TPSG1', 'MAP3K12', 'ZNF213', 'LUC7L', 'MTCL1', 'MPG', 'LIME1', 'GOLGA1', 'MIR3682', 'PTX4', 'MSLNL', 'DNPEP', 'MRPL28', 'UNKL', 'CCNDBP1', 'AXIN1', 'QTRT2', 'BAG4', 'TSC2', 'ZSCAN10', 'MGAT3', 'PDIA2', 'RHOB', 'TAF15', 'PIK3CD', 'SMIM14', 'NDUFA4L2', 'JPT2', 'KDM2B', 'EME2', 'TLCD1', 'LY6D', 'MMP25', 'CLIC2', 'PKMYT1', 'HBQ1', 'GNPTG', 'SRRM2', 'NHLRC4', 'MEIOB', 'CAPN15', 'PDCD4', 'NAP1L1', 'ID3']
Refined Community 59: ['ADGRE5', 'CXCR5', 'LPXN', 'SPIRE2', 'KRT8', 'CDC42BPA', 'PRKCD', 'ZNF276', 'SLC7A8', 'HHAT', 'SPATA33', 'ACSF3', 'RNF166', 'SOX10', 'APRT', 'TCF25', 'TUBB3', 'VPS9D1', 'BANP', 'SYNGR1', 'STAT5B', 'FKBP2', 'CHMP1A', 'C1orf174', 'SNAI3', 'CTBS', 'KLHDC4', 'THRSP', 'MAP1LC3B', 'CPNE7', 'THEMIS2', 'IGF2R', 'SIRPA', 'MC1R', 'CCNL1', 'CCL13', 'PFDN4', 'GAS8', 'PIEZO1', 'COX4I1', 'CDH15', 'PRDM7', 'CD79B', 'EMC8', 'ZFPM1', 'FRZB', 'SPATA2L', 'ICAM2', 'GINS2', 'MAPKAP1', 'CBFA2T3', 'PDZD8']
Refined Community 60: ['DAP', 'EIF5AL1', 'NID1', 'SLC47A1', 'RNF112', 'EPM2AIP1', 'MFAP4', 'FEM1C', 'MAPK7', 'ALDH3A1']
Refined Community 61: ['TGFB2', 'SEZ6', 'FBXO47', 'CBX3', 'BLTP2', 'TP53I3', 'CPD', 'TAOK1', 'SARM1', 'DHRS13', 'ASIC2', 'CCR7', 'TMEM265', 'TDO2', 'AFG3L1P', 'RARS1', 'SPAG5', 'MYO1D', 'COL4A2', 'LAMC2', 'PSMD3', 'RARA', 'SLC27A2', 'PPP2R2C', 'SLC6A3', 'KRT27', 'TCAP', 'SYS1', 'PMEPA1', 'ERBB2', 'CDC6', 'CD24P2', 'RHEB', 'PFKM', 'ZNF572', 'ANTXR1', 'MACC1', 'KRTAP9-4', 'RRAS', 'RPL23AP42', 'FBXL20', 'RPL19', 'TNFSF4', 'GSDMB', 'STARD3', 'MRPL45', 'GPR179', 'KRTAP9-8', 'HIF1A', 'DLX4', 'BLMH', 'FCSK', 'SLC2A1', 'IKZF3', 'PKM', 'PLXDC1', 'FAM222B', 'ELOC', 'IGFBP4', 'PARVA', 'KRTAP4-2', 'MED24', 'RAB5IF', 'TRAF4', 'ARHGEF28', 'SUPT16H', 'EFCAB5', 'PSMD2', 'UBE2G1', 'PIGS', 'RAPGEFL1', 'S100A16', 'EPB41L1', 'ZDHHC2', 'IL6ST', 'TFPI2', 'CORO6', 'ANKRD13B', 'STAC2', 'GJD3', 'NUFIP2', 'HEY1', 'CCL2', 'CEP192', 'GOSR1', 'ZPBP2', 'ACKR3', 'PTPN6', 'DUSP14', 'CCL21', 'NUP85', 'MOCOS', 'S100A9', 'IL18', 'LASP1', 'PPP1R1B', 'WDR90', 'FGG', 'RPL23A', 'ORMDL3', 'PSMB2', 'SDF2', 'CDK12', 'AFG2A', 'FCGR3A', 'NSRP1', 'RPL23', 'ABHD15', 'KRT222', 'HDAC10', 'MIR205', 'HGF', 'BICD1', 'HNF1B', 'BIN1', 'S100A10', 'CMTM3', 'CACNB1', 'TM2D3', 'SYNRG', 'CWC25', 'TLCD1', 'ZNF470', 'FXYD1', 'TBC1D29P', 'HECTD4', 'KDM4B', 'CRYBA1', 'SLC6A4', 'THRA', 'DDX52', 'SLC25A13']
Refined Community 62: ['KIF4A', 'CENPX', 'RNF125', 'COL1A1', 'FAM217B', 'CABP2', 'DGKE', 'KLHL8', 'ZNF652', 'ROMO1', 'SOCS3', 'MORF4L2', 'ANKRD40', 'FGR', 'SMARCE1', 'ACVRL1', 'NUP85', 'ERN1', 'SELENOO', 'NACA2', 'CCDC137', 'ZNF330', 'ABCC8', 'CACNA2D2', 'GET4', 'MAP2K6', 'LLGL2', 'ALPK2', 'ABCA8', 'GNPTG', 'ICE2', 'NHERF1', 'GRAMD2A', 'PAG1', 'PPM1E', 'LUC7L3', 'RRAS2', 'COG1', 'MYCBPAP', 'KCNJ16', 'SNF8', 'TAC4', 'RANBP3', 'CEP95', 'TUBG1', 'ABCC2', 'NKX1-2', 'OCA2', 'REXO1', 'SENP6', 'COL5A2', 'NRIP2', 'CSH2', 'CD19', 'RPL39L', 'EXOC7', 'OIP5', 'VGLL1', 'WFIKKN2', 'NME1-NME2', 'UNK', 'NFE2L1', 'RGS20', 'RAB37', 'HEG1', 'EVPL', 'MKS1', 'THBS1', 'TBC1D31', 'KCNK15', 'TSEN54', 'TNPO3', 'GOLM1', 'TRIM25', 'SOCS7', 'PEX5', 'NPBWR2', 'AEBP2', 'TMEM276', 'NPLOC4', 'ASCL1', 'HNRNPA1', 'HSF5', 'BIRC5', 'ITGAX', 'CHST15', 'CDC42EP4', 'CCDC69', 'CLEC11A', 'ARNT2', 'HOXB7', 'MRPS7', 'KAT7', 'TRAF3', 'RALGPS2', 'NARF', 'INHBA', 'F13A1', 'MIF4GD', 'TRBV5-4', 'UCHL3', 'KCTD2', 'MED13', 'GPS1', 'SORL1', 'ZNF638', 'GATA5', 'CD300LF', 'CCNL1', 'BAIAP2', 'CD300LB', 'GDPD5', 'AHCY', 'INTS7', 'DDX42', 'ACOT2', 'OGFOD3', 'RNFT1', 'SAP18', 'MBTD1', 'CXCR5', 'PRRT3', 'PHOSPHO1', 'ARHGEF4', 'PSMD12', 'PDXK', 'LINC02875', 'TEX19', 'USH1G', 'PRR15L', 'TBX2', 'DLX6', 'GPR142', 'ATR', 'RABEP1', 'PFDN6', 'CD300A', 'HIGD1B', 'ACSM1', 'SLC6A2', 'SYNGR2', 'LYZ', 'STARD3', 'LRRC40', 'SEMA6C', 'TAF2', 'TRIM37', 'SDK2', 'RPS13', 'PNPLA4', 'TSPOAP1', 'NXPH3', 'UTS2R', 'NT5C', 'ELK3', 'SAP30BP', 'ANXA2', 'SLC16A6', 'ADAMTS20', 'BRI3', 'STRADA', 'PRR11', 'ZACN', 'SPAG9', 'NDUFAF8', 'GRIN2C', 'PEX12', 'APOH', 'SKA2', 'CCL11', 'DHX37', 'SMG8', 'IGLL3P', 'SLC38A10', 'CDK5RAP3', 'TRIM65', 'IGF2BP2', 'MMD', 'PCTP', 'MRPL58', 'ORMDL1', 'TMEM94', 'NORAD', 'SERF1A', 'NAT9', 'PDK2', 'HOXB3', 'B4GALNT2', 'NCKAP5L', 'ENTR1', 'UTP18', 'DLX3', 'PNPO', 'CACNG1', 'EIF4A1', 'NRXN2', 'AKR1C1', 'BPTF', 'LAMC3', 'NME1', 'SUPT4H1', 'YPEL2', 'CBR3-AS1', 'NAV3', 'KRTAP4-4', 'CD7', 'VPS26B', 'C1R', 'NME2', 'HOXB5', 'ITGA4', 'CDR2L', 'AATK', 'BRMS1', 'CYBC1', 'TMC6', 'DCST2', 'SGCA', 'TMEM104', 'HEATR6', 'SPOP', 'LMO4', 'PBX1', 'ADGRE5', 'MBNL1', 'TIAL1', 'FOXJ1', 'MARCHF10', 'MPO', 'MRPL27', 'NTRK2', 'TACO1', 'GNA13', 'KATNAL1', 'PBX2', 'ARSG', 'SLC39A11', 'INTS2', 'PRRC2A', 'CCDC74B', 'YWHAQ', 'GDPD1', 'COPZ2', 'MMP1', 'OTOP3', 'MRPL12', 'WDR45B', 'CUL4B', 'RPL38', 'WFS1', 'PDE7B', 'FAAP100', 'GPRC5C', 'CD24', 'TMEM105', 'BTBD17', 'SMURF2', 'CD247', 'ABI3', 'VPS13B', 'DNAJA2', 'SH2B3', 'TNRC6C', 'TMPRSS13', 'RNF43', 'UNC13D', 'BAHCC1', 'SECTM1', 'CACNA1G', 'NDUFV1-DT', 'PPP1R27', 'RNF39', 'KRT25', 'UBE2T', 'ETV5', 'USP36', 'ABCA6', 'ITGA3', 'ABCA10', 'ASPSCR1', 'FTSJ3', 'PRKCA', 'CCDC47', 'OR4F6', 'EPN3', 'CCDC106', 'TMEM130', 'CACNG5', 'LAMA5', 'CERS4', 'LRRC37A3', 'MRPL38', 'ATP5MC1', 'ZNF524', 'EPPK1', 'DUSP16', 'NDUFAF4', 'OAS3', 'LMBRD2', 'EPN2', 'IFIH1', 'GNGT2', 'NFIL3', 'TEX2', 'GPR162', 'TOB1', 'CSNK1D', 'SOX9', 'LRRC59', 'ZNF793', 'PSMC5', 'ASAH1', 'MYLIP', 'IFT88', 'TK1', 'AGTR1', 'AMZ2', 'SUMO2', 'BRD9', 'USHBP1', 'FBXL7', 'HPS5', 'RPS6KB1', 'CEP131', 'FAM209B', 'CABP4', 'TMC8', 'MAP3K3', 'TEX14', 'ACOX1', 'HOXB8', 'PTPRM', 'ZNF20', 'ANXA6']
Refined Community 63: ['RBM47', 'USHBP1', 'OCEL1', 'IGKV2D-28', 'IQCG']
Refined Community 64: ['TLCD1', 'MAN1A1', 'CRACDL', 'ZNF568', 'ZNF781', 'ZNF875', 'ZNF571', 'ANXA1', 'TAPT1', 'ZNF570', 'ZNF420', 'ZNF382', 'SHTN1', 'ZNF793', 'CXCL1', 'HID1', 'GSK3A', 'ZNF383', 'ZNF585B', 'ATAD3A', 'ZFP30', 'CCDC77']
Refined Community 65: ['KLF6', 'EIF3B', 'PRPF31', 'NDUFA3', 'SLC23A2', 'CACNG6', 'CASKIN2']
Refined Community 66: ['LENEP', 'EFNA4', 'ZNF829', 'CENPB', 'EFNA3', 'DCST2', 'BRD1', 'HMGCR', 'RORC', 'KRTCAP2', 'IGHV3-23', 'PPP3CC', 'RUSC1', 'MRPL9', 'FBXL18', 'DPM3', 'HMGB1P1', 'PBXIP1', 'EPM2AIP1', 'EFNA1', 'SLC27A6', 'PKLR', 'DDHD1', 'WSB2', 'VAT1', 'THBS3', 'CLK2', 'RSPH3', 'GPR152', 'FAM210B', 'CFAP69', 'ISG20L2', 'ZBTB7B', 'ENSG00000293341', 'THEM4', 'PMVK', 'GBA1', 'SLC50A1']
Refined Community 67: ['LETM2', 'RASSF5', 'IL10', 'CRISPLD1', 'ALDH4A1', 'DYRK3', 'LMNB1', 'FOXJ1', 'SRGAP2', 'FCMR', 'NUP153', 'IKBKE', 'NBL1']
Refined Community 68: ['TLN2', 'GFRA4', 'IGLV4-60', 'HSPA12B', 'SIGLEC1', 'EXO1', 'SPEF1', 'PRPF31']
Refined Community 69: ['CDK12', 'RBM12', 'DYNLRB1', 'SPAG4', 'MYH7B', 'SYT7', 'ACSF3', 'UQCC1', 'TNFRSF19', 'CHAC1', 'GSS', 'SYNE4', 'ACSS2', 'TOP1', 'APPL2', 'TRPC4AP', 'NCOA6', 'PHF20', 'TSHZ3', 'HOXA5', 'LRRFIP1', 'NDUFAF4', 'NFS1', 'ERGIC3', 'CNBD2', 'CLTC', 'ARHGEF40', 'EPB41L1', 'CCDC89', 'TRAC', 'CEP250']
Refined Community 70: ['TMEM109', 'PEDS1', 'GPR183', 'PRSS16', 'NECTIN3', 'GPX7', 'PIK3C2A', 'TBC1D29P', 'RNF114', 'RUBCN', 'GGA1', 'CRMA', 'LAMA5', 'ZNF778', 'YWHAB', 'FNDC5', 'CASS4', 'ANKFN1', 'STX16', 'SNCG', 'CENPN', 'GDF15', 'HRH3', 'TRIB3', 'PARD6B', 'PARP10', 'TPD52L2', 'CHRNA4', 'NTSR1', 'UBE2V1', 'ZBP1', 'PSMD2', 'FAM209A', 'ZGPAT', 'SAXO4', 'EDN3', 'SYCP2', 'TEDC2', 'PIP4K2C', 'ADRM1', 'ABHD16B', 'ZBTB46', 'GLRX', 'GALR2', 'MATN4', 'PRKAA1', 'BIRC7', 'HMGB1P1', 'SALL4', 'SOX18', 'MMP17', 'BCAS1', 'TERF1', 'GARNL3', 'TOMM34', 'PEDS1-UBE2V1', 'CYP24A1', 'ATP5F1E', 'MPG', 'MSRB3', 'TCFL5', 'ZMYND8', 'UCKL1', 'CXCL14', 'LACTB', 'DPM1', 'TSHZ2', 'ANAPC15', 'SLPI', 'IGBP1', 'PPDPF', 'SCGB1D2', 'DIDO1', 'SLCO2B1', 'DCAKD', 'ARFGEF2', 'WFIKKN1', 'DNAJC5', 'C8orf76', 'CTCFL', 'NPBWR2', 'TCEA1', 'DCBLD2', 'LSM14B', 'PHACTR3', 'HOOK1', 'BMP7', 'MAOA', 'FAM217B', 'TMEM208', 'TMEM132A', 'PDE3A', 'MMP12', 'RAD9A', 'DMWD', 'RTCA', 'ME3', 'PYCR1', 'EP300', 'MID1', 'JTB', 'ITM2A', 'IGKV4-1', 'GULP1', 'GMEB2', 'YEATS2', 'FNDC11', 'RAE1', 'CBLN4', 'LYN', 'PI3', 'TP53TG5', 'PCK1', 'DOK5', 'TSPOAP1', 'ZNF655', 'SS18L1', 'AMPD2', 'PMEPA1', 'RAPGEF5', 'SRMS', 'NPEPL1', 'ALOX5', 'FGF3', 'NFATC2', 'SLC17A9', 'TMEM134', 'PTGFR', 'TNFRSF6B', 'ADNP', 'SEMG1', 'OGN', 'SPTBN2', 'MIR3682', 'TAF4', 'MAGED2', 'FANCF', 'RTF2', 'PTPN1', 'THEMIS2', 'LGALS3BP', 'MLLT1', 'TCEA2', 'ZFP64', 'RPS21', 'ARFRP1', 'WFDC12', 'GID8', 'GNAS']
Refined Community 71: ['YBEY', 'ADARB1', 'AKAP12', 'COL6A2', 'PCNT', 'PIM3', 'KRT4', 'OSGIN2', 'COL6A1', 'COL18A1', 'THOC6', 'NEXN', 'CD7', 'C21orf58', 'WDR6', 'SPATC1L']
Refined Community 72: ['CRELD2', 'SPIDR', 'MAPK11', 'LMNA', 'FBXL16', 'HOXA11', 'HOXB1', 'IL17REL', 'SIDT2', 'ALG12', 'OSBPL3', 'HDAC10', 'PFDN6', 'SELENOO', 'ZBED4', 'RASGRP3', 'BRD1']
Refined Community 73: ['SLC6A18', 'CACNA2D2', 'IRX2', 'LILRB3', 'LPCAT1', 'POGLUT3', 'AHRR', 'SDHA', 'NET1', 'PDCD6', 'SLC9A3', 'IRX4', 'ANKRD40', 'RESF1', 'NGLY1', 'NDUFS6', 'SLC12A7', 'ZNF362', 'CCDC127', 'IRX2-DT', 'ZNF706', 'UBR5', 'NKD2', 'DPY19L4', 'CD19', 'LGALS9']
Refined Community 74: ['CAP2', 'KLHL29', 'ALDH6A1', 'B2M', 'GRIA2', 'GFOD1', 'CENPX', 'ZNF18', 'FAM8A1', 'NUP153', 'TPMT', 'DNAAF11', 'KDM1B', 'MCUR1', 'DEK', 'RNF182', 'MED24', 'IL17RB', 'ID4', 'MSLN', 'PIP4K2B']
Refined Community 75: ['C4orf19', 'ZFP41', 'HOXA6', 'HAS2', 'HOXA10', 'TAC1', 'HOXA4', 'HOXA2', 'HOXA3', 'FBXL16', 'KCNMA1']
Refined Community 76: ['TNRC18', 'MUCL1', 'GET4', 'SNX16', 'WIPI2', 'IQCE', 'INTS8', 'NUDT1', 'TMEM216', 'TTYH3', 'H2AJ', 'CLGN', 'FOXK1', 'ACSS3', 'GNA12', 'ZNF239', 'FAM20C', 'GREB1L', 'ITIH4', 'SAMD4A', 'MAFK', 'SZT2', 'ADAP1', 'FOSL1', 'MAD1L1', 'SS18L2', 'EIF3B', 'ENSG00000280119', 'FBXL18', 'FLT4', 'BRAT1', 'TMEM184A', 'TENM3', 'COX19', 'UNCX', 'CDK1', 'SLC7A5', 'RADIL']
Refined Community 77: ['BRI3', 'CDC42', 'RABGAP1', 'ANKIB1', 'DPT', 'SMAD1', 'WWTR1', 'SDHAF3', 'FZD1', 'PON2', 'CYP3A43', 'TINAGL1', 'TFPI2', 'USP15', 'KCTD7', 'CASD1', 'CCR7', 'GBP3', 'AHNAK', 'KHSRP', 'TASOR2', 'ZNF142', 'AASS', 'SAMD10', 'CHURC1', 'RBM48', 'ATP5MF', 'CYP51A1', 'PON1', 'JHY', 'GSK3B', 'KRI1', 'FAM133B', 'CALCR', 'ASNS', 'RAPGEF5', 'GIMAP6', 'PPP1R9A', 'SGCE', 'TECPR1', 'ETFA', 'ULK1', 'ABCA3', 'ANXA3', 'MEAK7', 'ZSCAN25', 'MTERF1', 'FYN', 'BHLHA15', 'DBI', 'SAMD9', 'VTCN1', 'NPTX2', 'GNG11', 'AZGP1', 'GJC3', 'GALNT1', 'PEX1', 'ASB4', 'CCDC69', 'ATXN2L', 'GTPBP10', 'SLC25A13', 'GHR', 'ADH1B', 'LINC00469', 'TCF7L2', 'MYEF2', 'BUD31', 'COL1A2', 'CYP3A7', 'HEPACAM2', 'ZKSCAN5', 'PDAP1', 'DLX6']
Refined Community 78: ['IDO2', 'PBLD', 'IDO1', 'SFRP1', 'GPAT4', 'HGSNAT', 'TPT1P8', 'PDGFRL', 'AP3M2', 'SLC15A4', 'SLC16A5', 'LRRC37A3', 'UNC5D', 'ITGA1', 'S100A1', 'HMCN1', 'NSD3', 'PLEKHA2', 'PLEC', 'KCNS1', 'ASPHD2', 'MORF4L2', 'PYGO2', 'STAR', 'PPP1R26', 'NKX6-3', 'TCIM', 'LSM1', 'ZNF703', 'FAM171A1', 'VBP1', 'SMIM19', 'RNF170', 'RAB11FIP1', 'CYP1A1', 'ZFAND4', 'GOT1L1', 'PPP1R9A', 'TAGAP', 'LINC03042', 'GINS4', 'CHRNA6', 'PEX19', 'C2orf72', 'KAT6A', 'POTEA', 'DDHD2', 'POLB', 'TMSB15A', 'ADAM18', 'TUBA1C', 'NME7', 'DGLUCY', 'CD79A', 'ADAM2', 'CD47', 'CLCA2']
Refined Community 79: ['SLC26A3', 'RALYL', 'MRPS28', 'CDC27', 'PDP1', 'CIBAR1', 'MARS1', 'TMEM64', 'LINC00482', 'CNN2', 'GRIFIN', 'CHCT1', 'TSPYL5', 'CLVS1', 'TP53INP1', 'KLK8', 'TRPA1', 'RIPK2', 'ROPN1', 'NCAPG', 'SNX16', 'ZFAND1', 'VIRMA', 'GDAP1', 'CCNE2', 'DUS1L', 'PLEKHF2', 'PDS5B', 'ARHGAP25', 'CPQ', 'RFNG', 'KRTAP4-3', 'PEX2', 'RAD54B', 'TNFSF10', 'FABP5', 'BAX', 'SNX31', 'TTYH2', 'POLR2K', 'CRX', 'TMEM132A', 'CSE1L', 'PTPRZ1', 'COL25A1', 'CNBD1', 'FOXC1', 'PABPC4L', 'DCAF4L2', 'CRACDL', 'SLC10A5', 'CHMP4C', 'TUFT1', 'PXMP2', 'CRISPLD1', 'GSTP1', 'MAP3K5', 'ATP6V1C1', 'VPS13B', 'SLCO5A1', 'GSTZ1', 'HSH2D', 'MAN2B2', 'SLC26A7', 'TAP2', 'PGS1', 'ELMO3', 'FBXO43', 'DIPK1A', 'E2F4', 'KCNS2', 'METTL25B', 'ICOS', 'PABPC1', 'NECAB1', 'RUNX1T1', 'PGRMC2', 'SLK', 'FURIN', 'SDCBP', 'ITK', 'ENOX2', 'BCCIP', 'UBE2W', 'HOXB3', 'COMMD1', 'HNF4G', 'FZD6', 'MAGT1', 'RPL30', 'PI15', 'MMP16', 'RAB2A', 'MSRB3', 'CDK2AP1', 'RPL7', 'KLF10', 'STX1A', 'CISH', 'IDH3A', 'TACC2', 'SULF1', 'AZIN1', 'TPD52', 'ELOC', 'ZNF706', 'MSH6', 'CYP7A1', 'HACE1', 'SRRT', 'STMN2', 'DCAF13', 'NKX3-1', 'DSG2', 'COX5B', 'LY96', 'FABP9', 'GDF6', 'GGH', 'TBC1D30', 'ASPH', 'OSR2', 'ADORA2B', 'IL20', 'CCN2', 'NBN', 'INTS8', 'ZNF704', 'ASXL1', 'IMPA1', 'ERICH5', 'EYA1']
Refined Community 80: ['SNCAIP', 'RECQL4', 'SLC4A1AP', 'RNF182', 'DEPTOR', 'SUCLA2', 'NEO1', 'AURKA', 'RPL8', 'SLC52A2', 'FOXH1', 'RNF139', 'ZNF7', 'ZHX1', 'ATP7A', 'SAMD12', 'KCNV1', 'PHF20L1', 'RSU1', 'COMMD5', 'COLEC12', 'RAD21', 'NUDCD1', 'RAPGEF1', 'PTP4A3', 'ST3GAL1', 'JRK', 'MAFA', 'MRPL13', 'SFMBT2', 'PART1', 'TTC12', 'PGRMC2', 'SYBU', 'ZFHX3', 'ARC', 'NCAPG2', 'BASP1', 'NRBP2', 'DMRT1', 'ARHGAP39', 'NME1', 'CEP250', 'SMOC2', 'PIK3CA', 'DTNB-AS1', 'AGO2', 'SLA', 'PCYOX1', 'ZNF572', 'RHOF', 'ZNF16', 'LY6S-AS1', 'HEBP2', 'PACRG', 'GOLGA1', 'MROH5', 'EPPK1', 'CCN3', 'TRPS1', 'ZNF696', 'C8orf33', 'RHPN1', 'WDR75', 'GPIHBP1', 'BCAP31', 'MAX', 'SCRIB', 'HLA-DQA1', 'ZNF517', 'MTSS1', 'GRB10', 'KLHL38', 'EXOSC4', 'MYC', 'EBAG9', 'TMEM126A', 'ADISSP', 'PGR', 'CTSZ', 'FRMD6', 'PAM', 'PTPN22', 'MSRB1', 'FER1L6', 'CALM2', 'AZIN1', 'CIITA', 'MBOAT1', 'ADRM1', 'TMEFF1', 'ZNF623', 'HIKESHI', 'RSF1', 'GPR20', 'RUNX2', 'UHRF1', 'WNT11', 'SP100', 'C8orf76', 'FAM91A1', 'C2CD2', 'CTSV', 'P3H2', 'TRIB1', 'CYP11B1', 'CHRAC1', 'SLC30A8', 'MAPK15', 'VIPR1', 'SCX', 'TAF2', 'LYPD2', 'TOP1MT', 'STAR', 'MB21D2', 'NTAQ1', 'RASGRP3', 'P4HA3', 'UTP23', 'TMEM67', 'ZBTB16', 'TRMT12', 'LPXN', 'LRRC24', 'C8orf82', 'ADCK5', 'HOXB8', 'PRPF31', 'TCEAL9', 'CYP51A1', 'COL14A1', 'SLC39A4', 'PIP4K2B', 'DPT', 'MROH1', 'GRINA', 'PKHD1L1', 'BANF1', 'ITGB8', 'LPP', 'SMC1A', 'SQLE', 'THEM6', 'ZNF34', 'LCN2', 'ANXA13', 'AARD', 'TIGD5', 'MTBP', 'MED13', 'SNTB1', 'CYC1', 'PDK4', 'SOS1', 'CCN4', 'B2M']
Refined Community 81: ['TRBC2', 'SYTL2', 'ANAPC1', 'MYO9B', 'ABHD16B', 'DCAF13', 'CRNKL1', 'NUFIP2', 'LGALS9', 'NAT1', 'CCN4', 'OSGIN2', 'RNASEH2A', 'CACNA1D', 'NID2', 'MT1H', 'PRKDC', 'CHCT1', 'GDF6', 'MYO7A', 'PRPF39', 'NOL9', 'PPM1E', 'NCOA6', 'CLPTM1L', 'RTF2', 'STRBP', 'RET', 'SNRPA', 'SLCO2B1', 'SIGLEC1', 'HOOK3', 'ENTREP3', 'RALY', 'AKAP9', 'ADGRB1', 'QDPR', 'RCE1', 'ZCCHC14', 'RSKR', 'HLA-G', 'JMJD8', 'MTF2', 'PLS3', 'TESMIN', 'MYH7B', 'ATP5IF1', 'FAM106A', 'ATRN', 'GAB2', 'HCN3', 'RADIL', 'STARD10', 'TMEM229B', 'PIGS', 'GRIN2C', 'UQCC1', 'CSNK1A1L', 'ASCL1', 'DNMBP', 'PCNT', 'CACNA1G', 'HOXA3', 'GADD45A', 'BBIP1', 'FANCA', 'OAS2', 'NPR2', 'ENPP1', 'CBX3', 'BAZ2A', 'SPCS2', 'THBS3', 'POP1', 'ZFHX4', 'OCM2', 'HNRNPDL', 'MSLNL', 'ARFRP1', 'KDR', 'ZNF529', 'GGA3', 'FABP4', 'ADRB2', 'ATAD1', 'SMURF2', 'C1D', 'KRT12', 'FGFR1', 'OR4D2', 'ANKRD13D', 'RIMS2', 'ARC', 'TAC4']
Refined Community 82: ['DUOXA2', 'HOXA4', 'CUX1', 'TBC1D7', 'XBP1', 'BCAS4', 'PSMG3', 'CRYGS', 'PCYOX1', 'DCAF4L2', 'TXNDC16', 'POU4F2', 'ETV5', 'NFKB1', 'TWIST2', 'CRX', 'TCF7L1', 'STAT1', 'MEF2C', 'DYRK1A', 'CCL21']
Refined Community 83: ['HOXB3', 'SDC4', 'CDK10', 'PIK3R1', 'WFDC2', 'NEDD4L', 'H2AC6', 'PLAT', 'ZFAND2A', 'KPNA4', 'MIIP', 'MAST3', 'MYLK', 'ITPR1', 'GOLGA2P5', 'ACSM1', 'HACD2', 'PGA3', 'PEX11B', 'SETDB1', 'OS9', 'ARPC5L', 'CASKIN2', 'FAM110B', 'PELI1', 'ARFRP1', 'IL1B', 'XRCC4', 'SAA1', 'AKAP13', 'RPS3', 'CA5A', 'TF', 'HBA1', 'PHF11', 'LTF', 'AK5', 'VTCN1', 'OVOL2', 'S100P', 'OXTR', 'PLEKHB1', 'SERPINA5', 'CELSR1', 'COL4A5', 'TRAPPC10', 'IGFBP2', 'KCTD21', 'TMEM241', 'SERPINA3', 'LTBP4', 'DACH1', 'TMEM132A', 'PPP1CA', 'TAPT1', 'ADGRL4', 'IFT140', 'IGSF1', 'FANCA', 'CACNA1D', 'TDRD12', 'MMP12', 'HMGN4', 'DHX40', 'SMG7', 'ACSF3', 'FLNB', 'LZTS3', 'STC1', 'RPL29P17', 'TMEM87A', 'RCAN2', 'MFAP5', 'ALDH4A1', 'GFRA1', 'DCT', 'BPNT2', 'RBM26', 'FLRT3', 'NPY2R', 'MEIS3P1', 'SCML1', 'MATN2', 'SCUBE2', 'TPMT', 'SYMPK', 'MPPED2', 'UPF3B', 'CITED1', 'ERBB4', 'CAVIN2', 'SMPDL3A', 'ANPEP', 'SMARCD2', 'ID1', 'PKP1', 'IGKV2D-28', 'ATP2B4', 'FOXD1', 'CD151', 'IGF1R', 'NPY1R', 'NSUN5P1', 'HBB', 'INPP4B', 'CLMN', 'TMEM101', 'ALDH3A1', 'CORO1B', 'CAPN3', 'NET1', 'LAMB3', 'DNAJC12', 'ABI3', 'PLPP3', 'FRZB', 'SLC25A37', 'TPM2', 'DLC1', 'SEZ6L2', 'KRT17', 'ABRACL', 'EDN3', 'C1orf21', 'GSTT1', 'ADAP1', 'SLAMF8', 'ECE1', 'SYT17', 'ANKH', 'EREG', 'KREMEN2', 'CEBPG']
Refined Community 84: ['PAFAH1B1', 'TIGD5', 'LRRN3', 'LOXL1', 'EHD1', 'CUL9', 'COL11A1', 'YEATS2', 'PTTG1', 'SOAT1', 'PPDPF', 'PSTPIP1', 'RB1CC1', 'LDLR', 'CXCL13', 'AHI1', 'LAPTM5', 'ADCY7', 'PRKCD', 'SLC16A3', 'EVI2B', 'AFG3L1P', 'GLS2', 'FKBP11', 'FUT9', 'CCL11', 'FYB1', 'IGHV3-33', 'PLAUR', 'SELL', 'CCN4', 'TDO2', 'ENPEP', 'SIRT3', 'SOD2', 'TRAF3IP3', 'ARHGAP25', 'CHIT1', 'CCL3', 'QPCT', 'CORO6', 'CCL4', 'ASPH', 'SPC25', 'IRGQ', 'DIO1', 'IGHM', 'TRIP13', 'CENPX', 'LPXN', 'GLIPR1', 'NPY1R', 'PREP', 'CPM', 'E2F3', 'SEC23B', 'CD79B', 'TRAF1', 'IRX5', 'YAP1', 'CCDC74B', 'PHF20L1', 'RSAD2', 'HOPX', 'CDV3', 'VCAN', 'DNAJC1', 'PPAT', 'CHMP6', 'HHEX', 'P4HA1', 'FANCI', 'RPUSD1', 'C3orf52', 'PCK2', 'PRR14L', 'CXCR3', 'UBN1', 'GUSBP14', 'BANK1', 'PWWP2B', 'IGKV1OR2-108', 'DNA2', 'NUSAP1', 'PLEK', 'IGHV4-61', 'GRB2', 'COL10A1', 'CD44', 'CCR7', 'LRP8', 'MCM4', 'NPL', 'CD86', 'H2BC12L', 'ADAM9', 'TOP2A', 'OLFML1', 'RPA2', 'GPR65', 'POLR3K', 'CD69', 'USP3', 'GPR171', 'TNFAIP3', 'KIF2C', 'GINS1', 'NUP62', 'CTNND2', 'NCF2', 'BLM', 'SMARCA4', 'MYL9', 'CDK1', 'GADD45A', 'CCR1', 'HLA-DRB6', 'PAPOLB', 'SULF1', 'MNDA', 'MMP11', 'COL5A2', 'CXCR4', 'BTK', 'FCMR', 'GP2', 'CKS2', 'CEP55', 'SAMSN1', 'TRBC1', 'NRP1', 'CLASRP', 'HSD17B1', 'CYP24A1', 'ORMDL1', 'IGKV1D-13', 'SPATA2', 'MELK', 'CSH1', 'CHST12', 'HS3ST3A1', 'ISG20', 'AGFG1', 'CCL8', 'DLGAP5', 'BPNT2', 'KMO', 'BRD9', 'CDC42EP3', 'PIM3', 'GPR152', 'EGFL6', 'HRH3', 'IRF8', 'UQCC4', 'MMP12', 'DOK5', 'MAD2L1', 'GATAD1', 'FAM53B', 'PAX6', 'TNFRSF10B', 'APOC1', 'CLEC7A', 'MATK', 'TSFM', 'HHIPL2', 'CEACAM6', 'SAXO4', 'IGLV3-19', 'CST5', 'REEP5', 'PDAP1', 'HCLS1', 'GPSM2', 'CCNB2', 'CMTM3', 'EPN3', 'INTS2', 'PCLAF', 'ADAMDEC1', 'SRR', 'AURKA', 'CCDC69', 'ICAM2', 'CALCRL', 'EPPK1', 'RPA1', 'CREB3L1', 'CEACAM5', 'PLIN2', 'CYP39A1', 'PRKCB', 'HLA-DRB1', 'CD52', 'MMP1', 'TNFAIP6', 'LY96', 'CCND2', 'FZD9', 'CCKBR', 'S100P', 'DOCK2', 'PRLR', 'SEMG2', 'THEMIS2', 'LCK', 'PRKAG1', 'STK17B', 'CAND1', 'MLLT1', 'SELPLG', 'FAM83D', 'SGO2', 'SLC18A2', 'SP140', 'ANKRD54', 'SGCB', 'NCAPG', 'APOBEC3A', 'GPR37', 'TCAF1', 'METTL1', 'KRTAP4-2', 'KATNA1', 'LYZ', 'PBX1', 'CENPF', 'SBK1', 'STOML2', 'RAD1', 'WNT5A', 'PUS1', 'LARP4B', 'PRRX1', 'PRUNE2', 'SOX11', 'XBP1', 'FCGR3B', 'PRSS33', 'ARHGAP15', 'SLAMF8', 'COLEC12', 'LDLRAD4', 'CCNA2', 'SDC2', 'CENPU', 'NARS2', 'TYMS', 'FCGR2A', 'CD163', 'ABHD13', 'MMP19', 'TMEM123', 'ARHGEF4', 'ETV2', 'PHYH', 'ATF1', 'DNAAF11', 'P2RX5', 'POLM', 'CD27', 'LGALS2', 'APOBEC3B', 'SFRP4', 'LRRC15', 'MYT1', 'COL24A1', 'SSR1', 'CYLD', 'CXCL9', 'MPV17L2', 'LIMD2', 'CEMIP', 'PTP4A2', 'C16orf95', 'FCGR3A', 'PTPRC', 'ZNF467', 'SRGN', 'LAMP3', 'PGM2L1', 'TICAM1', 'BIRC5', 'CPNE7', 'LAT2', 'LST1', 'CSF2RB', 'TTC12', 'FPR3', 'GINS3', 'DEK']
Refined Community 85: ['HBEGF', 'ARHGAP25', 'TP53', 'RSBN1L', 'FXR1', 'KLHL29', 'XRCC6', 'RNF41', 'ARMC10', 'RELA', 'CCND1', 'HMG20B', 'CYP26B1', 'TERF1', 'CDC25A', 'MDM2', 'CIITA', 'CDK4', 'DHX9', 'GSDMB', 'BARD1', 'HCFC2', 'BRCA2', 'RAD51', 'SMAD3', 'CREBBP', 'RPL13', 'TRAF1', 'CRTAP', 'CCL5', 'FLNA', 'FANCG', 'ATM', 'SCAMP1', 'BCCIP', 'ALDH1A3', 'RPS2', 'CCNB1', 'ATF1', 'PRKDC', 'MDC1', 'ABCG1', 'CTBP1', 'RPA2', 'SULF1', 'EEF1D', 'RBBP7', 'HEBP1', 'TYMS', 'BAP1', 'CAND1', 'HDAC2', 'CHEK1', 'MDK', 'BLM', 'RBBP4', 'MLH1', 'NBN', 'GBA1', 'YTHDF3', 'NFKBIA', 'MFAP2', 'BACH1', 'WASHC5', 'LUZP1', 'EMSY', 'PABPC4L', 'CAST', 'SLC39A3', 'MYNN', 'SLU7', 'IQCG', 'BOLA1', 'TERF2', 'LMO4', 'THRA', 'STAT5A', 'CCNL1', 'NMI', 'HOXB8', 'JAK2', 'TP53BP1', 'E2F4', 'RB1', 'ZNF652', 'UGT8', 'FADS6', 'RBAK', 'NELFB', 'ZNF706', 'HSF1', 'MSH2', 'FCSK', 'TUBA4A', 'NUFIP1', 'C1R', 'SEM1', 'FHL2', 'MLLT10', 'TMEM126B', 'ACACA']
Refined Community 86: ['BUB3', 'TMED2', 'SSBP2', 'NEMP1', 'NACA2', 'LGALS1', 'SUMO2', 'DDX39B', 'GTF2H2']
Refined Community 87: ['SNRPA', 'MYH10', 'HSPA6', 'MSTO1', 'TULP3', 'TOPBP1', 'ILF3', 'STMN1', 'PCGF2', 'NCAPD2', 'RGS22', 'SS18L2', 'SKP2', 'CDK1', 'PVALB', 'FOXM1', 'UBE2S', 'DCLK1', 'CHST8', 'GNB2', 'NDC80', 'NSUN5', 'NANS', 'ASF1A', 'MRE11', 'ABCC2', 'HMMR', 'ULK4', 'CENPA', 'MAP3K7CL', 'RAD54L', 'TNFSF13', 'SMTN', 'TTF2', 'FXYD3', 'MCM5', 'CEP57', 'CYP3A43', 'RHEB', 'MCM6', 'RFC3', 'RFC4', 'USP1', 'RGPD5', 'AURKB', 'CEP43', 'TFCP2L1', 'MTF2', 'SF1', 'TTC13', 'CCNA1', 'CAD', 'SRSF10', 'PARVA', 'DEK']
Refined Community 88: ['KLF5', 'RPL13', 'SHARPIN', 'CAPNS2', 'C16orf74', 'CDH8', 'ITFG1', 'SHCBP1', 'SGCB', 'MMP2', 'IRX6', 'SQLE', 'GPT2', 'PHKB', 'CLBA1', 'DHRS4-AS1', 'LPCAT2', 'NETO2', 'DYNLRB2', 'SLC6A2', 'USE1', 'TPSD1', 'MAF', 'PWWP2A', 'CEP41', 'SESN3']
Refined Community 89: ['RNF39', 'ADGRE2', 'IL4', 'KATNB1', 'CETP', 'H1-1', 'FCGR2A', 'PRDM7', 'SLC15A4', 'UBE2Z', 'CCL17', 'COMMD5', 'GRK5', 'CNOT1', 'LIN7A', 'OAZ1', 'ORC6', 'CFAP20', 'GOT2', 'OLFML1', 'EXOC7', 'COQ9', 'MED30', 'CYP2E1', 'KRT26', 'C16orf78', 'SPAG5', 'SPIRE2', 'FTO', 'SLC38A7', 'FANCA', 'AKTIP', 'ZFHX4', 'EDNRA', 'MAP4K3', 'HECTD4', 'CBLN1', 'CPNE2', 'GSDME', 'ADGRG1', 'GAS8', 'CCL22', 'VPS35', 'AFG3L1P', 'TTC9', 'VEGFA', 'SLC29A4', 'TGFBR2', 'GPR142', 'SLC9A1', 'POLR2C', 'MC1R', 'KRT10', 'LYN', 'AGR2', 'EFNA4', 'GET4', 'RBL2', 'PLCG2', 'SMAD5', 'DOK4', 'GINS3', 'CDH8', 'TUBB4A']
Refined Community 90: ['ZNF398', 'CDKN3', 'PEX7', 'GADD45G', 'PDCD4', 'DEFB1', 'PSCA', 'KIFC2', 'DNAJC12', 'CD27', 'FBXL5', 'RPA2', 'EPN3', 'ABAT', 'OPN3', 'PROM1', 'ERBB3', 'LETM1', 'KIAA0040', 'FUS', 'TBX3', 'RTN4', 'TFPI', 'CD24P4', 'CHST3', 'GDF15', 'CEBPD', 'OVOL2', 'S100A2', 'AMY1A', 'LEMD2', 'TBC1D8', 'HSPA4', 'PFN2', 'OXR1', 'CELSR1', 'HOXA5', 'PCBP2', 'C1orf54', 'LRBA', 'GMEB2', 'IER5', 'CFD', 'PLAT', 'CLASRP', 'AUTS2', 'ORM2', 'PRSS21', 'MAST4', 'CELSR2', 'GGH', 'CDC73', 'OCA2', 'TTC39A', 'PLK2', 'SEC16A', 'TAB1', 'ABHD11', 'DST', 'PBX1', 'HOOK1', 'ACADSB', 'GALNT7', 'TP63', 'MDK', 'COL28A1', 'FGF5', 'ZDHHC11', 'DAPK1', 'MDM2', 'TPK1', 'PADI2', 'GSE1', 'CSGALNACT1', 'ABCD3', 'SOAT1', 'IFI44', 'SLC27A2', 'CNIH2', 'DSC3', 'PDZK1IP1', 'KAT2B', 'MAGED2', 'RAPGEF5', 'WFS1']
Refined Community 91: ['CD74', 'S100A4', 'TMEM47', 'IGF2BP2', 'SDHAF2', 'ZNRD2', 'LDLRAD4', 'HSD17B11', 'DCBLD2', 'PRR35', 'TRDC', 'MRC2', 'GPIHBP1', 'TGFBR2', 'HCLS1', 'POSTN', 'MYL9', 'TBC1D19', 'CFAP410', 'MSN', 'RCAN2', 'PECAM1', 'IGLC2', 'CERS6', 'FBLN2', 'TRBV5-4', 'IL13RA1', 'CRISPLD2', 'SGCB', 'ABLIM3', 'ANXA1', 'GFPT2', 'LY6K', 'ENTPD3', 'SCG5', 'EMP3', 'LGALS1', 'LY96', 'ABCC4', 'LAMB1', 'RECK', 'TSPAN13', 'TLR7', 'ECM2', 'SIRT1', 'AKR1B1', 'CCN2', 'MAPT', 'MICAL3', 'GLRX', 'ARHGEF40', 'FCER1G', 'TACC3', 'TGFB1I1', 'MYBL1', 'RCBTB2', 'PCDH18', 'BARX1', 'IFT88', 'CHI3L2', 'CREBZF', 'FOXN3', 'TMEM151A', 'TIMP3', 'TNFSF4', 'SPOCK1', 'CD53', 'TRPC1', 'TLE4', 'CES5A', 'HLA-DQA1', 'IRAK3', 'SERPINB8', 'NLK', 'PDGFRA', 'SULF1', 'VPS13B', 'RGS4', 'NOX4', 'TMEM201', 'DPYSL3', 'SLC7A7', 'RARRES2', 'ATL1', 'PXDN', 'CENPBD1P', 'DOK5', 'MERTK', 'DACT1', 'IGF2', 'SLC9A6', 'ATP6AP2', 'GATA3', 'LPXN', 'PALLD', 'GRHL2-DT', 'SERPINA6', 'LYN', 'CTSS', 'GLIPR1', 'SPATA24', 'SPATA2L', 'CLIC4', 'ADAMTS12', 'STRADA', 'TDO2', 'BLNK', 'SERPING1', 'SEPTIN6', 'SAMD4A', 'CLEC2B', 'REPS2', 'TNFRSF1B', 'CASP1', 'FTL', 'HNF4G', 'LEPROT', 'OCM2', 'CYP39A1', 'SHOX2', 'GAK', 'NBL1', 'CERS3', 'STAT3', 'LAMC1', 'VCAN', 'ZNF532', 'PSAP', 'CD300A', 'KRTAP4-3', 'PRNP', 'RGP1', 'HLA-DRA', 'SEPTIN4', 'SIDT1', 'E2F8', 'DACT3', 'SPON1', 'FLI1', 'DENND1B', 'SOBP', 'ATP5F1E', 'NPRL3', 'FYN', 'CDC73', 'ASPN', 'APOE', 'CLDN5', 'HOXB13', 'CILP', 'CAVIN3', 'SERPINH1', 'PTPRC', 'POFUT2', 'CHRNB3', 'NNT', 'MAP1B', 'FZD2', 'C1orf74', 'HCCS', 'ZNF444', 'CPVL', 'ZFPM2', 'CABP4', 'LCP1', 'SLC16A3', 'PNN', 'FN1', 'GPNMB', 'SEM1', 'UBLCP1', 'EPS8L1', 'CSF2RB', 'CHRM3', 'DMD', 'ERG', 'XYLT1', 'SRGAP2', 'SKAP2', 'PLCG2', 'ELF1', 'LOX', 'MAP4K4', 'SLC6A19', 'FITM2', 'HS3ST6', 'MME', 'SARM1', 'ACVR1', 'REEP1', 'COL8A2', 'MFAP2', 'VEZF1', 'TBC1D9', 'TLCD3B', 'ETV5', 'S100A13', 'TATDN2', 'OR12D2', 'ZNF512B', 'RPA1', 'MEIS2', 'NDN', 'NEU3', 'CLEC11A', 'BGN', 'MNDA', 'PDLIM2', 'ALDH4A1', 'PIGS', 'PLAUR', 'CSMD3', 'PIMREG', 'ETS2', 'TMEM71', 'BNC2', 'MMP11', 'FAP', 'PCK2', 'GNAO1', 'MCF2L', 'HLA-DRB6', 'NUAK1', 'MXRA5', 'ACY3', 'C16orf74', 'CDC42', 'ANGPTL2', 'TMEFF1', 'PFDN1', 'PLAU', 'CMC2', 'DYNLRB1', 'RUNX2', 'NKG7', 'THBS4', 'ST3GAL6', 'PDE10A', 'TRIM22', 'KCND2', 'LCP2', 'ADAM19', 'ALOX5', 'MXRA7', 'GPX7', 'DAPK1', 'BRI3', 'C17orf67', 'DOP1B', 'PSMB9', 'ACKR3', 'TPD52', 'COL5A1', 'ITGBL1', 'KMT2D', 'GPR152', 'LONP2', 'DLX5', 'CLEC10A', 'TCL1A', 'NCSTN', 'CLEC7A', 'PKD2', 'TGM2', 'ITIH5', 'PSMG3', 'PROS1', 'SGCD', 'MT1M', 'NAT1', 'CLBA1', 'EMP1', 'SLC2A10', 'TOMM70', 'KRT5', 'DPYD', 'CPB1', 'COL1A2', 'CD34', 'MS4A6A', 'PLCH1', 'CCN1', 'NFIX', 'GIGYF1', 'LOXL1', 'CST5', 'H2BC12L', 'RGS1', 'ADGRF5', 'COL10A1', 'CELF2', 'CFAP418', 'CALHM6', 'ZNF436-AS1', 'LY6E', 'GPR65', 'ZFPM1', 'PFKP', 'CLEC4A', 'CDH11', 'LPCAT1', 'HLA-DRB4', 'KCTD14', 'INHBA', 'PARVA', 'SERPINF1', 'COLEC12', 'LETM2', 'OMP', 'SDC2', 'IL11', 'SPARCL1', 'EXOSC8', 'SLAMF8', 'LUM', 'RHOH', 'HLA-DRB1', 'CCR1', 'ESM1', 'DNM3', 'ZMYND19', 'CTSV', 'FAS', 'IGF1R', 'C1orf54', 'SPHK1', 'ROBO1', 'MXD4', 'FKBP11', 'TRHR', 'FRMD4A', 'ANTXR2', 'KCTD12', 'COL16A1', 'FEZ2', 'TMEM130', 'NOL8', 'BICC1', 'HEY1', 'ATAT1', 'TP53I3', 'ISLR', 'APOC1', 'COL15A1', 'ZBTB5', 'CCND2', 'ETV1', 'KDR', 'WNT5B', 'ANXA5', 'LRRC17', 'TBC1D30']
Refined Community 92: ['CFB', 'PAN2', 'TSNARE1', 'SSTR2', 'TMEM143', 'MYH11', 'TMEM101', 'C1QTNF3', 'GATB', 'PLCB1', 'SLC26A3', 'POGZ', 'EEF1A2', 'KDELR2', 'FAXDC2', 'PTEN', 'TAF1', 'YIPF6', 'FAH', 'DZIP3', 'MFAP5', 'EVL', 'POLR1A', 'CYP4B1', 'HS3ST1', 'TMT1A', 'NRXN1', 'ENTPD3', 'RAB26', 'IGFBP4', 'EPHA10', 'KMO', 'ARHGAP15', 'HES5', 'FABP4', 'CA12', 'SEMA3C', 'PRC1', 'AGAP4', 'MYL10', 'RDH16', 'ITGB8', 'RTN1', 'GDF7', 'SCNN1A', 'KITLG', 'MIR99AHG', 'DCAF10', 'LRRC17', 'SPATA7', 'GPD1L', 'CD44', 'CD36', 'GSE1', 'SPINK5', 'LEP', 'FLNC', 'GPRC5C', 'LPCAT4', 'GSS', 'ATP1A2', 'NFYC', 'HLA-DRA', 'UCP2', 'RHOBTB3', 'HYCC1', 'DCXR', 'PLAAT4', 'S100A7', 'RNF39', 'UBA7', 'HNF1B', 'NT5C', 'GPC1-AS1', 'TBC1D19', 'NEFH', 'RGS16', 'CAMP', 'DHX34', 'GRHL1', 'LDLR', 'HNRNPA1', 'TMC4', 'SACS', 'ISYNA1', 'DPY30', 'RBM4B', 'GGT1', 'GNMT', 'PSD3', 'SEC61A2', 'DIAPH3', 'ZBTB18', 'DDHD2', 'ARHGAP35', 'DNM1', 'CLSTN2', 'EPHX2', 'RRAS', 'ASS1', 'ACACA', 'FERRY3', 'FOXH1', 'MDN1', 'KIAA0040', 'PRKACB', 'GINS3', 'KCNJ3', 'OAS2', 'AREG', 'CTDSP1', 'IRX5', 'HIGD1B', 'C7', 'SNED1', 'MROH5', 'GAGE1', 'MRPL41', 'PTPRN2', 'CLK4', 'SERPINA5', 'JCHAIN', 'RAP1BL', 'CERS6', 'UBL3', 'XAF1', 'CBFA2T3', 'ZNF43', 'CLDN5', 'KCNMA1', 'BATF', 'SCGN', 'KRT19', 'DNAJC1', 'JPH1', 'RHOB', 'ATP6V1A', 'TNFRSF6B', 'ZNF568', 'DGKD', 'AQR', 'HBA1', 'STAU2', 'MALT1', 'PON2', 'PGBD5', 'MAPK3', 'RUNX2', 'TSPAN13', 'TCN1', 'RETREG1', 'MAGED4B', 'ZNF839', 'ZFC3H1', 'CASP8AP2', 'ACSM5', 'MSL1', 'CEP15', 'RNF125', 'FRY', 'ZNF34', 'DDR2', 'DNAJC17', 'C4A', 'LAPTM5', 'CNTNAP2', 'UBE2G1', 'POFUT2', 'DBI', 'LGALS8', 'MCTP2', 'UCN', 'RARRES2', 'CACNG4', 'OMD', 'UGT2B11', 'STAG3', 'PPIP5K1', 'TATDN1', 'CLEC3B', 'LPXN', 'GRIP2', 'PPP3CC', 'MUC6', 'DHRS2', 'COL5A2', 'ERBB3', 'LRRC31', 'MAGED2', 'ANG', 'LARP6', 'VAV3', 'GOT1L1', 'GASK1B', 'GNA14', 'CREBZF', 'RARA', 'KDM4B', 'CREB3L1', 'RCL1', 'AGTR1', 'PGR', 'COQ9', 'DACH1', 'LCORL', 'DGLUCY', 'KRT8', 'SERPINA6', 'LY6H', 'MAGEA2', 'SLC39A3', 'ERCC2', 'ZMYND8', 'PKIA', 'GUSBP3', 'ACADSB', 'C1RL-AS1', 'MCTS1', 'HEMK1', 'MACIR', 'TOM1L1', 'NCOA6', 'ZC3H11A', 'ADH1B', 'NDN', 'NPDC1', 'MLPH', 'WSB1', 'APP', 'RALGPS2', 'EPHA3', 'EGR3', 'TRAC', 'ANKFN1', 'C1D', 'ZNF790', 'HPS1', 'CACNG1', 'PEX11A', 'SNRNP25', 'NEXN', 'CYP19A1', 'SEC16A', 'KIRREL1', 'SPTLC2', 'BIK', 'TREM2', 'CPA3', 'ALDH4A1', 'GOLPH3', 'ELOVL2', 'AARD', 'SEMA3A', 'ATP5F1E', 'PRSS23', 'CRIP2', 'LIAS', 'EVA1B', 'TSKU', 'CDK2AP1', 'CALCRL', 'ZFAND1', 'ANO1', 'MZF1', 'ESRRG', 'DUSP5', 'SULF2', 'CST5', 'MAN1C1', 'KRTAP5-9', 'HOXB2', 'HMCES', 'ATL3', 'PEG3', 'IL10RA', 'DIO1', 'PIK3CA', 'SLC49A3', 'TKT', 'ALDH3B2', 'HOXB3', 'NEDD4L', 'RPL38', 'NAT2', 'CYP21A2', 'FAM217B', 'HGD', 'LAMP5', 'TNNT1', 'APPBP2', 'JAG1', 'SLC9A1', 'TBK1', 'MYO6', 'CST3', 'YIPF5', 'ARL3', 'MGMT', 'MRPS30', 'EVI2A', 'ECM1', 'POU4F2', 'CGN', 'WIPF1', 'PATJ', 'POLM', 'MNAT1', 'SLC1A4', 'PITRM1', 'CACNA2D3', 'WWOX', 'STK39', 'MUC1', 'SOX9', 'NQO1', 'CYB5A', 'SEC14L2', 'CCND1', 'CDC25C', 'SOCS2', 'GPD1', 'NAIP', 'PLEKHF2', 'SLC27A2', 'GPC2', 'GADD45A', 'GIMAP6', 'CRYBG1', 'CPNE2', 'IGFBP7', 'CSAD', 'SLC38A7', 'REXO1', 'RASSF1', 'CRISPLD2', 'HPN', 'PLCL1', 'GZMA', 'DNAAF1', 'DCBLD2', 'SCYL2', 'LONP2', 'PLA1A', 'FURIN', 'PCLO', 'MED30', 'CCL11', 'BICDL1', 'MIPEP', 'SYNE1', 'CCL7', 'DNAJC12', 'AKR7A3', 'LIPE', 'DCLK1', 'SLC25A37', 'HNRNPA2B1', 'THEM5', 'ITPKB', 'TPPP3', 'ECE1', 'SPHK2', 'RHOH', 'P2RX5', 'AIF1', 'PAX2', 'PTPRG', 'SPRED2', 'HPGD', 'TTC17', 'FCMR', 'HHEX', 'C2orf68', 'SGCA', 'MCF2L', 'AGL', 'EFEMP1', 'CHMP2B', 'FCER1A', 'CHST1', 'CASD1', 'BNC1', 'PAPSS1', 'NFIL3', 'GABBR1', 'LTO1', 'SPON1', 'ATP8A1', 'ABCA8', 'CHN2', 'CDK6', 'NARF', 'CLEC2B', 'GNG13', 'CELSR1', 'TARS3', 'SNCG', 'ANKRD33B', 'ATP7B', 'MLXIP', 'SLC7A1', 'ERBB2', 'REPS2', 'MON2', 'CSNK1E', 'GRP', 'OAS3', 'TTC12', 'CCL14', 'PPP1R3C', 'TBL3', 'MEOX2', 'CATSPERB', 'TCF4', 'UTP15', 'ARRB1', 'WWP1', 'DCN', 'SAXO4', 'CYP2B7P', 'B3GNT3', 'PAX6', 'CD69', 'FBP1', 'LINC01949', 'UHRF1', 'CDC27', 'GREB1', 'RHOD', 'FAT1', 'GATA2', 'TRIM68', 'CRIM1', 'TOB1', 'ANAPC5', 'IGF1R', 'COL4A6', 'PTK6', 'CEACAM5', 'MAPKAPK2', 'RPS24', 'GREM1', 'IFT74', 'SELENOP', 'SCGB1D2', 'IKBKB', 'PLIN1', 'KCNK5', 'CHST15', 'EMILIN3', 'KRT26', 'TOX3', 'CFD', 'PCGF6', 'TES', 'NCOA3', 'HES1', 'ITPR1', 'CXCL11', 'POLD4', 'NR2F1', 'PSME4', 'TMEM63A', 'ANXA9', 'KLHL38', 'RAB38', 'RTL10', 'PCBP2', 'SLC44A4', 'TMEM74B', 'ZFP41', 'BICC1', 'ST6GALNAC2', 'SFRP4', 'DDX50', 'RAB17', 'PDXK', 'KATNAL1', 'CAMK2N1', 'ECHDC1', 'FGG', 'BCOR', 'DHX33', 'LETM2', 'MREG', 'C2CD2L', 'CHL1', 'ALOX15B', 'MATN3', 'NBEA', 'PNPLA4', 'UNC119', 'ZNF394', 'PRRT3', 'CCNB2', 'FAM174B', 'HBB', 'ALYREF', 'NME3', 'PCDH18', 'KAT6A', 'TCL1A', 'HMGCS2', 'SMARCA4', 'ITGA7', 'CLK2', 'ABCC8', 'ACKR3', 'SRRT', 'LAGE3', 'CRISP3', 'ADAMTS20', 'PTHLH', 'RNF170', 'ASPA', 'ABCC6', 'PDS5B', 'ZBTB20', 'TRAK1', 'EMCN', 'NFIX', 'ICA1', 'CCT5', 'SLC43A3', 'LIPG', 'SPAG16', 'SRRM2', 'SVEP1', 'SORT1', 'CNR1', 'SALL4', 'KCNMB4', 'AZGP1', 'ARNT', 'DNAJA4', 'PHYH', 'FITM2', 'XRCC4', 'WNT2B', 'RPS20P22', 'MRPS33', 'FAAH', 'TNIK', 'ELAPOR2', 'RABEP1', 'MYT1', 'BLVRA', 'SLC4A7', 'GOT1', 'KIF20A', 'RBM47', 'TM4SF1', 'ZYG11B', 'MPZL2', 'PWWP2A', 'GJA4', 'SSPN', 'INPP5F', 'TLCD3B', 'SERHL2', 'UBA52', 'NPY1R', 'DNAAF11', 'PRLR', 'SEMA3G', 'HELZ', 'POLR2B', 'PLCG2', 'ACAP1', 'QDPR', 'NME5', 'EVI5', 'HDAC11', 'CCDC77', 'SFMBT2', 'DLAT', 'FGFR3', 'ERBB4', 'BMI1', 'CCL4', 'CEP131', 'DPY19L4', 'MYB', 'KLK8', 'MINK1', 'JAK1', 'ALB', 'CYP2E1', 'WIPF2', 'DBF4', 'QTRT2', 'LPP', 'TIMP4', 'PSPC1', 'SPRR3', 'POMT1', 'TUBA1C', 'TSPAN1', 'SLC2A8', 'ZHX1', 'COL4A5', 'C1QBP', 'TBC1D16', 'DYRK3', 'SCUBE2', 'BCAS1', 'SLC11A2', 'SLC16A6', 'GATA3', 'INPP4B', 'FZD5', 'DLG3', 'TAX1BP1', 'FAM171A1', 'TAT', 'KCNAB1', 'ROGDI', 'RAPGEF3', 'HIPK2', 'CPNE1', 'TRIM36', 'IL11', 'PDZK1', 'TRADD', 'RAPGEFL1', 'ATP2A3', 'IFT140', 'IQCG', 'HOXB6', 'CDKAL1', 'STK32B', 'NMT2', 'MYBL2', 'ABCB1', 'KATNIP', 'ZBTB40', 'ABLIM3', 'MAP4K4', 'SPARCL1', 'TAX1BP3', 'MAOA', 'COMP', 'TBX10', 'MLLT3', 'DUOX1', 'RGS5', 'ZNF382', 'DENND1B', 'PNMT', 'CYP1A1', 'HNMT', 'F13A1', 'GHR', 'SIX1', 'ASPH', 'VPS35', 'PLK3', 'TPSAB1', 'ADIPOQ', 'PPP4R1', 'METRN', 'CACNA2D2', 'MAST4', 'CHAD', 'SEPTIN8', 'NELL2', 'PHKG2', 'TNF', 'SENP6', 'DDX27', 'ZKSCAN1', 'CXCL14', 'GPNMB', 'FBLN1', 'MEGF9', 'GPC3', 'FAM234A', 'ZCCHC8', 'IL1R1', 'APLNR', 'SPDEF', 'CCNB1', 'CHRM3', 'ICE2', 'SIGIRR', 'AAGAB', 'INPP5J', 'VMP1', 'THBS4', 'MTSS2', 'TFAP2B', 'NDUFA4', 'PPP2R1A', 'NUSAP1', 'RNF41', 'SMG8', 'PTPN21', 'FAM242E', 'IL6ST', 'E2F4', 'PDK1', 'OVOL2']
Refined Community 93: ['RALGPS2', 'TTYH1', 'ITGBL1', 'FZD9', 'FANCA', 'P4HB', 'CTH', 'MCL1', 'MARVELD1', 'DESI2', 'KIF3C', 'UTP15', 'MX1', 'SERPINH1', 'GJB3', 'ERAP2', 'KLF6', 'TP53INP2', 'RAB26', 'ATP13A3', 'NDUFA4L2', 'SLC2A3', 'PRKY', 'ALKBH5', 'HPSE', 'STIP1', 'ERI2', 'GLS', 'SPAG1', 'CAND2', 'CASP4', 'EPHX3', 'GBP1', 'NUDT1', 'SOX11', 'DEFB1', 'CRYAB', 'RCL1', 'CCNG2', 'ZNF750', 'TFEC', 'DSG2', 'PRR16', 'IGHV4-34', 'SERPINB2', 'DLX5', 'SOD2', 'AMZ1', 'WNT6', 'EIF4H', 'RAP2A', 'ACADSB', 'PSMA5', 'ASPM', 'NFIB', 'CXCL8', 'PAPOLA', 'ZNF330', 'LDHB', 'SLC6A3', 'CLEC7A', 'MFGE8', 'COLEC12', 'HACD1', 'GNPTG', 'MFSD3', 'MMP7', 'YWHAZ', 'OIP5', 'DBN1', 'C16orf95', 'VPS37D', 'RASL10A', 'RHBDF1', 'RBMS1', 'MEF2A', 'ARHGAP25', 'RNFT2', 'LBR', 'KCTD5', 'NAT9', 'ADORA3', 'DSG1', 'NEK2', 'ACTN3', 'FGFR1', 'LTBP1', 'PCDHB3', 'RAD54L', 'ST3GAL6', 'TFAP2B', 'ACTG2', 'TAT', 'PON2', 'CEACAM1', 'E2F3', 'IGF2BP3', 'HPCAL1', 'TRAPPC6A', 'CGN', 'DPYD', 'TIMELESS', 'DNM3', 'BIK', 'YES1', 'LAMTOR5', 'ETV6', 'PRKD3', 'PVR', 'KLK5', 'BICD2', 'CXADR', 'TOX', 'PAK1', 'SRPX2', 'BMAL2', 'CD59', 'CDH19', 'CD14', 'CCL5', 'GAMT', 'PTPN6', 'ARHGEF9', 'PLXNB1', 'SNTB1', 'CHAF1A', 'PI3', 'MSH6', 'PEG10', 'LGALSL', 'SNTG2', 'LYPD1', 'LZTS3', 'AFG2A', 'NMU', 'MREG', 'ATG5', 'HEPH', 'MAGEA9', 'EDN1', 'PLOD2', 'GJC1', 'STXBP6', 'NCALD', 'PCSK1N', 'SDK2', 'GPRC5B', 'CADPS2', 'RUNX1T1', 'TMEFF1', 'FOXQ1', 'MYBL1', 'IFNAR2', 'MDC1', 'NOL3', 'ZNF133', 'SLC16A14', 'ITPKB', 'RSRC1', 'DDIT4', 'ENPP1', 'TMEM158', 'ITGA1', 'FERMT1', 'RBMS3', 'LOX', 'SLC15A1', 'RDH11', 'IQSEC1', 'BRAF', 'CEP170', 'KIAA0319L', 'TNFRSF11B', 'PFN2', 'CD38', 'DCN', 'SLC6A14', 'GDPD5', 'TNNI2', 'CYP2B7P', 'ZNF362', 'SSRP1', 'NR6A1', 'ADAM17', 'DPYSL3', 'PDE6A', 'RTCA', 'YBX1', 'SEMA4A', 'ITM2C', 'GZMH', 'RAB34', 'PART1', 'RGS20', 'PTPN22', 'PSPC1', 'MIRLET7D', 'TNFRSF19', 'FAT1', 'CELSR3', 'UBE2M', 'RSF1', 'STRN3', 'TEP1', 'CFI', 'TAPBPL', 'LHX2', 'ARID5A', 'MAGOH', 'RBL1', 'PDE8A', 'CSAD', 'SMAD9', 'ODC1', 'SLC16A1', 'RETREG1', 'ITPA', 'KBTBD8', 'UBE2Z', 'PLAAT1', 'EN1', 'LRP8', 'GNA14', 'CXXC5', 'UCK1', 'TCF7L1', 'TBC1D7', 'KCNK5', 'FBXO46', 'USP38', 'ANGPTL4', 'MAP1LC3A', 'P2RX2', 'TYMS', 'RRP7A', 'APBA2', 'SCX', 'IQCG', 'PLP2', 'MPHOSPH10P1', 'KLHL24', 'TP53', 'SATB1', 'CXCL10', 'UCK2', 'CHRNA6', 'TBX19', 'PROM1', 'KRT81', 'SLC25A37', 'TNFRSF21', 'MAGEA12', 'MANSC1', 'ST3GAL2', 'LAD1', 'TRIP13', 'CCN2', 'GNGT1', 'S100A10', 'HBEGF', 'DCAKD', 'SLPI', 'MMP11', 'ATF1', 'AMY1A', 'TMEM98', 'LCP1', 'S100A2', 'ANGPT1', 'CDH3', 'DSC3', 'CMTM3', 'RMND5B', 'CREBBP', 'CALHM6', 'HOXA9', 'CCDC106', 'SOX15', 'PKMYT1', 'CYP2D6', 'TNFSF4', 'SMIM10L2A', 'MARCO', 'FABP5', 'SRD5A1', 'SLAMF8', 'BBX', 'WNT5B', 'NUDT11', 'MCUR1', 'BTG3', 'TCN1', 'CHEK1', 'CRABP1', 'IL32', 'SOSTDC1', 'PFKP', 'TRMT12', 'H4C9', 'MPHOSPH9', 'TTC6', 'ATP11A', 'SERPINB3', 'FBXL6', 'PROCR', 'SLC7A5', 'POU5F1', 'TCF7', 'TMEM126A', 'CMC2', 'GLDC', 'CALML5', 'TFF2', 'TNFAIP2', 'CYP11B2', 'SMAD6', 'IL27RA', 'ADAMTS7', 'ELN', 'IGLV4-60', 'SULT2B1', 'AP5Z1', 'MXD4', 'SRGN', 'RUBCNL', 'CRMA', 'EDIL3', 'ELOVL4', 'KIFC1', 'BACH1', 'CDK11B', 'CHST11', 'GUCY1A1', 'SOX10', 'LIPE', 'HJURP', 'GATA5', 'LRFN4', 'ASPN', 'GPM6B', 'PABPC1', 'PRKCA', 'CCL20', 'MAST4', 'CHAC1', 'LRRC32', 'DCPS', 'PPM1E', 'CD44', 'PLCB1', 'ZNF296', 'PLCH1', 'IRX4', 'GGA3', 'PBK', 'TUBB2A', 'UBE2S', 'NRARP', 'ZNF350', 'BCL11A', 'PAM', 'PIK3R2', 'FZD7', 'MLC1', 'GMDS', 'ADORA2B', 'CCNB1', 'TIMP3', 'KCNAB1', 'TK2', 'SULF1', 'NT5DC2', 'NEDD4L', 'AQP9', 'KRT15', 'KLK8', 'MSN', 'PLK2', 'NSMAF', 'ME2', 'RNU6-1016P', 'VLDLR', 'MRGPRF', 'NR1D1', 'DLK1', 'RMND5A', 'CA6', 'TBCD', 'ZFAND2A', 'SPG7', 'TTLL4', 'FFAR2', 'MUC16', 'KRT4', 'EXTL1', 'BCLAF3', 'SMCO4', 'PCP4', 'TPM2', 'CNOT9', 'PYCR1', 'METTL25B', 'LGR5', 'DDX52', 'IL18R1', 'FAM209A', 'NME5', 'TRIM25', 'ORC1', 'CBS', 'TBCK', 'ATP6V0E2', 'RNF145', 'ZFPM2', 'RYR1', 'AK4', 'TUSC3', 'MCM4', 'SERPINF1', 'MIA', 'ATP10D', 'RASA2', 'GPRIN2', 'MALL', 'FAT2', 'SLC26A2', 'PHF19', 'S100A6', 'MT-ND5', 'MTMR2', 'FSCN1', 'DSC2', 'PAPSS2', 'NEK1', 'YTHDF1', 'NCAPH', 'CYP2J2', 'TLN1', 'UCKL1', 'PPP1R14B', 'FLT3LG', 'GAGE1', 'FADD', 'HTR2B', 'FXR1', 'BIRC5', 'PTK2', 'RTP4', 'CLMP', 'TNFRSF1B', 'HTN1', 'ADGRL3', 'UQCC2', 'PVALB', 'ENO1', 'BSPRY', 'JRKL', 'IRX2', 'AHR', 'QPCT', 'PAMR1', 'GABBR2', 'TRA2B', 'CACNA2D2', 'ATP13A4', 'TMSB15B', 'PALS2', 'NES', 'SEC31B', 'SH3BP2', 'MET', 'NCAN', 'CD81', 'ZNF532', 'PKN2', 'KIF1A', 'CNTNAP2', 'SYNCRIP', 'ADGRG6', 'REG1A', 'YBX3', 'FNDC3B', 'GALNT12', 'CA9', 'GATA6', 'SHANK2', 'XKR9', 'HOXA11', 'EPN1', 'ELF4', 'ANXA8', 'GPR19', 'DZIP1', 'PTX3', 'STAMBPL1', 'MRPL28', 'MPZL2', 'IGHV3-7', 'VIPR1', 'ADD2', 'PCGF2', 'RDX', 'PAK1IP1', 'CERS4', 'PHGDH', 'KLK7', 'PERP', 'PLA2G4A', 'AK2', 'CDH2', 'P3H2', 'HRK', 'H2AZ2', 'FGB', 'SELE', 'RHO', 'HSPA4L', 'DMRT1', 'ZNF251', 'TTLL6', 'GOLPH3', 'TBL3', 'ZACN', 'CLCN4', 'CHI3L2', 'TIMM44', 'CDKAL1', 'SSR1', 'SCHIP1', 'SCAF11', 'CDC45', 'PPP1R2', 'PLCB4', 'ZYX', 'SOCS3', 'CDCA3', 'TMEM117', 'PDE9A', 'MID1', 'AGGF1', 'DNAH17', 'SFN', 'IMPA2', 'LPCAT3', 'PICK1', 'CD247', 'H2BC12L', 'CARD19', 'BLOC1S6', 'COL9A3', 'ZNF793', 'MPHOSPH6', 'HELLS', 'ARRB1', 'TMEM94', 'TUBB', 'DDX17', 'IGLL3P', 'KRT6B', 'LRBA', 'SLC43A3', 'MAGEA2', 'NFIX', 'MSLN', 'MOB3B', 'COX11', 'KLHL7', 'EDN2', 'GDF6', 'CCNB2', 'MACO1', 'CPQ', 'FGFBP1', 'ATF5', 'PLIN2', 'EIF4EBP3', 'ERICH5', 'SGK3', 'CALU', 'APH1A', 'DVL2', 'LAMP3', 'FOLR1', 'ZC3H18', 'MEIS2', 'MAGI2', 'MAGED2', 'NFE2L3', 'FUT9', 'SPIB', 'WWTR1', 'S100B', 'CDC25A', 'DENND2A', 'CCDC81', 'ARL4C', 'LAMA4', 'CYTIP', 'NAP1L1', 'CSPG4', 'BIRC3', 'MIR29B1', 'CDK1', 'PEX2', 'DDX5', 'NOTCH2NLA', 'CDYL', 'CHMP3', 'NEIL1', 'ILRUN', 'PLPP3', 'UPP1', 'FOSL1', 'CSRP2', 'S100A9', 'MICALL1', 'ING1', 'IFIH1', 'BCL2A1', 'BAIAP2L1', 'PCK2', 'ZNF124', 'KRT23', 'ZKSCAN5', 'STAT5B', 'ACTL8', 'CHML', 'EPRS1', 'LALBA', 'MPHOSPH8', 'NSUN5', 'ANP32E', 'OBP2B', 'PIPOX', 'ATP1A1', 'PTPRZ1', 'WLS', 'NFIL3', 'CDKN2A', 'MRAS', 'AOAH', 'NPL', 'TGFB3', 'RAPGEF3', 'BRAT1', 'HNRNPU', 'SMPDL3B', 'MMP1', 'CALCRL', 'BARX1', 'TNN', 'S100A1', 'DLAT', 'SIX3', 'CDK17', 'PLAAT2', 'CAPN6', 'TRIM29', 'C1QTNF3', 'CAVIN3', 'NUP155', 'ANKRD6']
Refined Community 94: ['RGL2', 'CEP250', 'SOX10', 'GABRP', 'NMI']
Refined Community 95: ['PKP1', 'GSDMB', 'CYP27B1', 'BLTP2', 'EML2', 'IGKV1D-39', 'ALDH3B2', 'SERHL2', 'CEACAM6', 'PNMT', 'ACSL1', 'CTNND1', 'RPS6KB1', 'HS3ST1', 'TMC5', 'CORO1B', 'IGF2R', 'DHRS7', 'C2orf72', 'WIPF2', 'IL19', 'PPM1D', 'SUPT6H', 'RBAK', 'IGKV1D-13', 'PAFAH1B1', 'RRP7A', 'FA2H', 'AHNAK2', 'MX1', 'CRISP2', 'RNF43', 'HID1', 'FAM210B', 'AP5Z1', 'IGLC2', 'PGAP3', 'KHSRP', 'CYB561', 'SCGB1D2', 'TBC1D1', 'IRAG2', 'CRYBG2', 'LRRC74A', 'MAGEA3', 'IGHV3-23', 'SLC12A1', 'PAPSS2', 'PRMT7', 'PRODH', 'E2F4', 'ISG20', 'NEK5', 'SMIM19', 'DDC', 'TRPM4', 'NACA2', 'GZMM', 'LRRN1', 'SCGB2A2', 'FASN', 'MANSC1', 'PCNT', 'TTYH3', 'FKBP14', 'GALNT6', 'ARL6IP6', 'SRRT', 'SERHL', 'ABCC6', 'KRT24', 'DKK3', 'P4HB', 'REEP5', 'GGT1', 'TM7SF2', 'CYTIP', 'ANKH', 'STAR', 'CABLES2', 'CLIP4', 'ATOSA', 'TFAP2B', 'UBL3', 'PLLP', 'SLC24A3', 'CPD', 'TMPRSS2', 'NUDT15', 'UGT2B11', 'SRPK3', 'FFAR2', 'ATP2C2', 'FGB', 'TAPT1', 'BAG4', 'PROM1', 'NOL3', 'ABCC3', 'LBP', 'QPRT', 'MAB21L4', 'IGHV3-7', 'LPCAT3', 'ZNF211', 'PLK3', 'BAALC', 'FXYD1', 'TSKU', 'PYGL', 'SOX8', 'BHLHE40', 'SPTLC2', 'CRYBG1', 'SQSTM1', 'FUT3', 'IGKV1D-17', 'TNFRSF21', 'CPSF4', 'TRAF4', 'IRF1', 'LDLR', 'MRPS7', 'CAMP', 'CD46', 'SCD', 'CLIC3', 'KIT', 'KMO', 'IL24', 'MAFA', 'KYNU', 'IGFBP5', 'SLC39A4', 'GOT1', 'TXN', 'CCNE2', 'TSPAN8', 'SLC44A4', 'ALDH4A1', 'ALCAM', 'TRIM9', 'CKMT1B', 'CD24P4', 'DRAM2', 'CHRD', 'MSANTD3', 'SLC35B1', 'PABPN1L', 'MED1', 'ORM2', 'ELL2']
Refined Community 96: ['CENPN', 'ANP32E', 'ZNF570', 'DKC1', 'BIRC7', 'CREBBP', 'TPX2', 'KRT16', 'MPP7', 'MLLT11', 'EIF1AX', 'TGFBR3', 'NDC80', 'TOP2A', 'CYRIA', 'CDCA8', 'BYSL', 'SLC7A5']
Refined Community 97: ['MAPK9', 'SULF2', 'G0S2', 'UBA2', 'RIPK2', 'LAPTM5', 'LMOD1', 'CCDC77', 'PBXIP1', 'PSMB9', 'CCN5', 'ZBTB16', 'ITPKB', 'HMGCS2', 'ELOVL2', 'ADIPOQ', 'MYH11', 'CHRDL1', 'TACC2', 'PDCD4', 'RIDA', 'IKBKB', 'NKG7', 'BTG2', 'FAXDC2', 'TAT', 'C1orf21', 'SIAH2', 'OTOP2', 'MS4A7', 'EMC10', 'TMEM204', 'MMP2', 'PLK3', 'FOSB', 'ADIRF', 'HPX', 'CILP', 'NKX3-1', 'PRR15', 'MAOB', 'NAT1', 'CAP2', 'BLOC1S6', 'TMEM138', 'SVEP1', 'NCF2', 'FFAR2', 'PTN', 'RBP4', 'FOXP2', 'SPTAN1', 'RAB5IF', 'TNXA', 'DDX11', 'FGF18', 'CCT2', 'BHLHE23', 'HLF', 'ATP1A2', 'LIPE', 'MTMR11', 'FBLN1', 'RUNX1', 'FHL1', 'ZNF470', 'GFRA1', 'AREG', 'RTN1', 'IGF1', 'OGN', 'PLA2G2A', 'SLC26A3', 'ABCC3', 'ERN1', 'TIGD5', 'TBC1D29P', 'MYB', 'ETFB', 'DUSP1', 'MAP3K12', 'PGR', 'LAMA2', 'LEP', 'JRKL']
Refined Community 98: ['HLA-DRB4', 'PDGFRA', 'GRB14', 'ECHDC1', 'TRA2B', 'DPT', 'ANP32E', 'CYP39A1', 'CLDN8', 'NF2', 'IMPA2', 'CALD1', 'CYTL1', 'CHST7', 'LGALS2', 'LYZ', 'TSPAN15', 'LRRC10', 'KMT2D', 'SH2D5', 'CDKN1B', 'MPHOSPH6', 'DTX3L', 'SERPINH1', 'KRT6B', 'LAD1', 'CRIP2', 'NFIL3', 'CPA3', 'PCDHA9', 'CFAP410', 'BICDL2', 'CD2', 'CCN1', 'WFS1', 'C3', 'MCTS1', 'SLC1A1', 'ACYP1', 'ALDH1A1', 'CRNDE', 'ARPP19', 'SETDB1', 'GNB4', 'RAD54B', 'STK10', 'CYBA', 'SPINT1-AS1', 'PTGFR', 'GPR179', 'GJC3', 'CNOT6', 'DAB2', 'IGF1', 'RAC2', 'ADH1B', 'CLEC4A', 'PDE9A', 'LILRA4', 'TMEM150C', 'PTPRC', 'AQP1', 'LDHB', 'TRAF3', 'SNTG2', 'ALDH1A3', 'SIT1', 'FGL2', 'H6PD', 'CD3G', 'ACTN1', 'GH1', 'HUS1', 'RASSF8', 'TMEM45A', 'CIMIP1', 'GADD45A', 'FAM83B', 'PRKX', 'DEPP1', 'IGHV4-61', 'SBSPON', 'SERPINB3', 'PHGDH', 'ISG20', 'PELI1', 'GABBR1', 'CRLF1', 'HMMR', 'EXTL1', 'PFKP', 'CLCN3', 'KLHL38', 'MME', 'IL7', 'TMEM100', 'KRT15', 'CDCA8', 'HELZ2', 'ROPN1', 'USP32', 'PRKCQ', 'SLC40A1', 'HS3ST3A1', 'GSDMB', 'AKR1B10', 'NLRP1', 'APCDD1L', 'JAM3', 'COX11', 'ACKR1', 'IGF2BP2', 'PI15', 'CXCR4', 'BANK1', 'SAA1', 'SERPING1', 'CSAG2', 'TIE1', 'SLC25A37', 'MAGEA5P', 'KDM7A', 'C7', 'RRAGD', 'BCL11B', 'MYCN', 'KLF5', 'MMP7', 'SEL1L3', 'UBE2E3', 'PTGER4', 'PSCA', 'SLC44A2', 'OGFRL1', 'BRAF', 'TMEM67', 'CDC42', 'GPR19', 'GFRA1', 'C16orf87', 'CCL13', 'ST8SIA1', 'CSGALNACT1', 'METTL25B', 'APC2', 'DSC3', 'MRAS', 'FKBP9', 'GHR', 'GRHL1', 'LINC01973', 'DNAJB4', 'PECAM1', 'MBNL2', 'TNNI2', 'HSD17B11', 'STXBP6', 'MUC16', 'TNFAIP3', 'GATA6', 'SH3GLB2', 'MYH7B', 'COL9A3', 'OCIAD1', 'ZDHHC17', 'SRD5A1', 'XIST', 'SNAPIN', 'ANK2', 'SLC13A2', 'GABRE', 'COL2A1', 'IGHM', 'SFRP1', 'S100A7', 'CYTIP', 'HK3', 'LILRB2', 'TNFRSF21', 'AKR1C1', 'CD7', 'SYNGR1', 'MNDA', 'ADNP', 'SLC4A3', 'MARVELD2', 'CPM', 'ACAP1', 'HLX', 'CD96', 'NSUN5P1', 'MT1F', 'TSPAN8', 'GXYLT2', 'CD48', 'PROM1', 'TRAC', 'RPL23AP42', 'RAB11FIP3', 'NFIA', 'HLA-DQB1', 'KAT8', 'RAI14', 'MFAP2', 'CDH17', 'CD1E', 'HEPH', 'NDRG2', 'IGHA1', 'MMP14', 'RHOV', 'LEPR', 'PADI2', 'SYNM', 'PDP2', 'CHI3L2', 'PHB2', 'DTNB-AS1', 'COL4A2', 'FMNL1', 'ITFG1', 'IGLV3-10', 'CD38', 'NINJ2', 'MRTO4', 'ENPP2', 'FADS6', 'PTN', 'SCRG1', 'CACNG6', 'C9orf152', 'ART3', 'CACNG7', 'PRKCB', 'RGS2', 'RAD9A', 'GP1BB', 'TRIM2', 'ARHGDIB', 'RAP1A', 'FLYWCH1', 'LAMB3', 'VAMP1', 'CXCL2', 'GLIPR1', 'KLRC1', 'WWTR1', 'GABRP', 'CLEC7A', 'TNFRSF4', 'IGHV3-47', 'IGLC2', 'ABCA8', 'TMSB15A', 'PIK3CG', 'KCNB2', 'FYN', 'ROBO1', 'GALNT3', 'HTR2B', 'CERS3', 'WNT3A', 'MYB', 'PDZD2', 'IL1R2', 'ZNF532', 'SOX10', 'PIEZO1', 'SPTAN1', 'SLC6A14', 'CLIC4', 'PPP1R16B', 'NMU', 'MMP3', 'C1orf54', 'FEM1C', 'SQOR', 'CD19', 'FMO2', 'AMD1', 'FNDC3B', 'CSRP2', 'TOX4', 'CEP43', 'FBXO17', 'DKK1', 'ALDH18A1', 'TCL1A', 'CD24P4', 'HOXA5', 'SLC20A2', 'TRAF3IP3', 'RFTN1', 'SLC23A2', 'WFIKKN2', 'SLC43A3', 'BAG2', 'NRN1', 'SLC6A15', 'SNRPN', 'DMD', 'TM6SF1', 'SRPX', 'ANXA8', 'APOC1', 'CD300LB', 'RAB5IF', 'NKG7', 'RBM12B', 'GAS1', 'JRKL', 'SELL', 'HOXB3', 'TAP1', 'CTSW', 'ARHGEF6', 'PROS1', 'CCR7', 'HSPB2', 'LGALS9', 'HSD17B2', 'CXCL11', 'MAP2K6', 'ZBTB46', 'MMP1', 'ANKRD13B', 'NCF1', 'CTSC', 'ADGRG6', 'RASGRP2', 'PLCH1', 'STARD3', 'PTPRCAP', 'HSD17B4', 'INSM1', 'BIN2', 'HSPBAP1', 'MT1X', 'IGFBP5', 'IGKV4-1', 'RGS4', 'TNIK', 'ACOT7', 'C1S', 'MCM5', 'RND3', 'ASH1L', 'GTF3C1', 'MYC', 'RCAN1', 'HOXB9', 'PPM1G', 'GSTA1', 'MCM3AP', 'FAM8A1', 'SP100', 'ATAD3A', 'PAX6', 'FTH1', 'IL6', 'GUCY1A1', 'DSN1', 'NRP1', 'SLC4A11', 'KCNN4', 'LAMP3', 'MPHOSPH8', 'DNMT1', 'CXCL10', 'FUS', 'ZC3HAV1L', 'NOXO1', 'COL4A5', 'GPR183', 'WNT2', 'CAV1', 'ZSCAN18', 'PTK7', 'TAC1', 'SRMS', 'LBP', 'KCNG1', 'THAP1', 'FHOD3', 'ADAMTS20', 'ANKHD1', 'HMGN4', 'RNF138', 'ALCAM', 'DZIP1', 'CEACAM6', 'RARRES1', 'ITGB7', 'BBOX1', 'MIA', 'PSMD12', 'PSD3', 'WNK4', 'VCAN', 'HPCAL1', 'ZC3H12C', 'ATP9A', 'XRCC4', 'PSAT1', 'RNF11', 'PLCB4', 'SLC50A1', 'PWP1', 'CDC25A', 'FAM200B', 'GABBR2', 'CYC1', 'PRKD3', 'POLG2', 'EN1', 'ARHGEF26', 'PDZK1IP1', 'NAB1', 'BRAT1', 'IL33', 'AIFM2', 'ITM2C', 'CD247', 'TM2D3', 'KRTAP9-2', 'SLURP1', 'LGALS7', 'MRPS33', 'PDE6A', 'CRISPLD1', 'GJB3', 'EMCN', 'TPSG1', 'WTAP', 'P2RY14', 'KLK5', 'ADD2', 'ANGPTL4', 'FDPS', 'NECTIN4', 'CALB2', 'SLC18A2', 'RIPK4', 'TRIP6', 'MS4A7', 'THEMIS2', 'SENP6', 'DPP4', 'PTCH1', 'GATA5', 'YBX1', 'ABCB1', 'FERMT2', 'MMP2', 'RNF166', 'TXN', 'KLK10', 'FAM107A', 'UGT8', 'CTDSPL2', 'PRSS12', 'ELOVL2', 'BNIP2', 'PKD2', 'ZFAND6', 'RANGAP1', 'GALNS', 'S100B', 'SLC35A1', 'L1CAM', 'ADAM19', 'IFI35', 'IL21R', 'WIF1', 'GEM', 'CHODL', 'CD27', 'WDR75', 'HIVEP2', 'KLRB1', 'MRPS17', 'FOLR1', 'ZAP70', 'OGN', 'MSN', 'LY9', 'ZACN', 'PML', 'MEOX2', 'CCL14', 'IGHV3-21', 'KLRK1', 'FLT4', 'UNKL', 'COL11A2', 'RUNX3', 'SNX1', 'IL7R', 'NEK11', 'CCN2', 'SMC2', 'TBX19', 'IGKV1D-17', 'LPL', 'PRDM13', 'NUP93', 'DCN', 'GAD2', 'SERPINE2', 'ERAP2', 'CXCL12', 'PPP1R27', 'QTRT2', 'SEMA5B', 'IFI16', 'RAB40C', 'CTSB', 'LYN', 'HOXA9', 'PPP1R2', 'ABCC2', 'NPTX2', 'ANTXR2', 'TNXA', 'FRZB', 'PRELP', 'GRB2', 'PLAUR', 'AFG2B', 'FOLH1', 'ACTG1', 'JCHAIN', 'GNLY', 'VWF', 'WNT5A', 'MCRIP2', 'ID4', 'GPM6B', 'TNKS1BP1', 'MPPED2', 'CXCL9', 'IGHV1-69', 'STAR', 'CNOT2', 'GRAMD2B', 'CCDC6', 'AHNAK2', 'EME2', 'HBB', 'TTC7A', 'KRT23', 'SLC16A1', 'ATRNL1', 'TRIM29', 'MFAP4', 'MARCO', 'KLHL2', 'GPR37', 'SNX9', 'SIRT5', 'EML1', 'DST', 'MAP4K1', 'PLEKHB1', 'TNFRSF17', 'MICAL3', 'ID1', 'CABP2', 'FTL', 'CHRDL1', 'ISG15', 'WNT3', 'C1QC', 'IGF2', 'RERG', 'CASK', 'TNC', 'TLCD1', 'ARID2', 'KRT5', 'CEP41', 'SRGN', 'ELN', 'PTPN22', 'ARHGAP25', 'PRRX2']
Refined Community 99: ['PLEKHF2', 'ZBTB5', 'SELENOO', 'CCBE1', 'B9D1', 'ERBB4', 'KCNJ3', 'MICB', 'MRPS17', 'SCCPDH', 'MEIOB', 'E2F7', 'WWP1', 'SLC6A4', 'GTF2H2C', 'LDLRAD4', 'SIAH2', 'MAFB', 'ENO2', 'ALG8', 'AGTR1', 'FBP1', 'SPOCK1', 'SORD', 'NAT1', 'RDH16', 'CXCL2', 'ZNF142', 'GATA3', 'EPOR', 'TPD52', 'MISP', 'BRD7', 'EEF1A2', 'DNAI7', 'FGF18', 'NDUFV1', 'TNNI3', 'TMEM101', 'LGALS2', 'TESMIN', 'ITIH4', 'RHBDL1', 'CACNA1D', 'GNG13', 'EDN2', 'RAB26', 'SNX11', 'CA12', 'SLC19A2', 'COL8A1', 'GSTM3', 'SPINK4', 'TRIM9', 'S100A8', 'ABCA3', 'TLN1', 'XIST', 'RAB11FIP1', 'TJP3', 'ACADSB', 'SLC1A1', 'STRAP', 'OR4D2', 'CYP3A7', 'REPS2', 'INSM1', 'CACNA2D2', 'CFAP418', 'KLHL22', 'GALNT9', 'TRPA1', 'CCT2', 'PTGER2', 'PPP1R3C', 'GREB1', 'METTL1', 'TNNT1', 'NELL2', 'BMP4', 'CAMK2B', 'RGS10', 'ARMT1', 'NT5E', 'SOX10', 'CABP2', 'KIF1B', 'CRISP2', 'RAB6B', 'DIO1', 'CDK1', 'CHML', 'SERPINA6', 'NAIP', 'ZNF652', 'WNT10A', 'IL13RA1', 'LINC01503', 'CYP2B6', 'ANXA9', 'TPPP3', 'MCCC1', 'CNRIP1', 'ANKIB1', 'MT1F', 'SYMPK', 'PLAC8', 'MRPS30', 'NKX2-5', 'CRABP2', 'SPAG16', 'DNAJC1', 'NUDT6', 'ABCB1', 'CCL7', 'SFPQ', 'CXCL14', 'SERPINA5', 'SERPINA3', 'MSLN', 'CYP2A6', 'HGD', 'RABEP1', 'HPX', 'SYT1', 'PTGES', 'SVEP1', 'KDM4B', 'C4A', 'GUSBP3', 'IFT122', 'PRKDC', 'WNT2B', 'CENPE', 'TBC1D9', 'RRAS2', 'MAPT', 'CNKSR1', 'QDPR', 'EVPL', 'THEMIS2', 'EPHB3', 'NKAIN1', 'CLK3', 'CCND1', 'GLRB', 'TUBB6', 'GLIS3', 'PCYOX1', 'EVA1B', 'PIERCE1', 'ERBB2', 'DACH1', 'CACNG4', 'WFDC2', 'TOX3', 'CCDC106', 'PYCR3', 'RSF1', 'MAFF', 'TBK1', 'ARNT2', 'GBP1', 'CD8A', 'PGM2L1', 'PAN2', 'FTO', 'LMTK2', 'SLC29A4', 'KCTD3']
Refined Community 100: ['PHACTR2', 'EPN3', 'AP1M2', 'CYP2B6', 'MXD4', 'CDS1']
Refined Community 101: ['STK10', 'ANK3', 'DES', 'DOCK2', 'IL33', 'TPD52', 'POLA1', 'RUNX1T1', 'SPIB', 'CCL2', 'CMA1', 'ID3', 'KLK7', 'CNN1', 'TCL1A', 'CLCF1', 'SLU7', 'SLC25A19', 'TIMP2', 'RASA4', 'AQP1', 'NOL11', 'ACTA2', 'PTPRCAP', 'TNFAIP8', 'SLIT3', 'ZNF846', 'SLC1A3', 'C3AR1', 'ICOS', 'CLIC2', 'SETX', 'PTPRB', 'GPR183', 'GMNN', 'SLC4A1AP', 'TNFRSF17', 'CRK', 'CD247', 'PMAIP1', 'MCTS1', 'CACNG7', 'HLX', 'CAVIN2', 'WIPF1', 'SCGB1D2', 'KCTD12', 'NR3C2', 'PEG3', 'CTSK', 'HS3ST6', 'ANXA4', 'SYNE1', 'UHRF1', 'MT1M', 'BIRC3', 'PRDM14', 'RBM47', 'DTNB-AS1', 'ARTN', 'TRIM33', 'RPSAP47', 'MAP4K1', 'ZSCAN10', 'MAPRE2', 'FABP4', 'EMID1', 'IL21R', 'KIT', 'MLLT11', 'PPP1R12C', 'ITIH5', 'RRN3', 'CYTL1', 'CHL1', 'MEF2C', 'CRY2', 'ACACB', 'CLSPN', 'ASAP1', 'MAML2', 'CD79B', 'ZKSCAN5', 'GAPDH', 'DKC1', 'KASH5', 'CD2', 'LAMC3', 'ARHGEF40', 'IGLC2', 'ESCO2', 'CRTAP', 'NR2F1', 'PPP1R16B', 'MOGS', 'RDH10', 'EPB41L2', 'MAGEA2', 'ITM2C', 'TCEA1', 'CYB5A', 'CASP1', 'NCF1', 'VILL', 'CR2', 'TRGC1', 'GJA4', 'C14orf132', 'ARHGAP45', 'KIF13A', 'GOLGA2P5', 'CCL19', 'ADRB2', 'SLC12A7', 'MATK', 'ZACN', 'DST', 'NRXN2', 'LAMA2', 'PPP2R1B', 'KRI1', 'RNF103', 'IRAK2', 'XRCC6', 'CLEC4A', 'VAMP1', 'TRAT1', 'SSBP2', 'NAB1', 'RAD51', 'COL14A1', 'SMNDC1', 'CLC', 'LINC01503', 'ROGDI', 'HLA-DOB', 'SRPX2', 'FLRT2', 'NR1D1', 'FEM1B', 'CCR6', 'LGALS2', 'APH1A', 'TNFRSF1B', 'NPR1', 'RASGRP3', 'MFAP5', 'POU6F1', 'CD244', 'STOML2', 'UBN1', 'WIPI1', 'MS4A1', 'JCHAIN', 'GZMH', 'PIGF', 'UHMK1', 'AR', 'EDN3', 'AOAH', 'RASSF3', 'GLIPR1', 'GATA5', 'PDE2A', 'TOX4', 'PLPP3', 'EIF4H', 'XCL1', 'LGALS4', 'ZIC2', 'HHEX', 'ANGPT2', 'PLCL1', 'LEPR', 'CCDC69', 'SATB1', 'MGAT3', 'FOLR2', 'CHMP2B', 'BANK1', 'UNC13D', 'C1orf116', 'CCDC137', 'GMFG', 'MRFAP1L1', 'SIT1', 'ARHGEF11', 'LAMP3', 'EOLA1', 'CES1', 'CRIM1', 'FHL1', 'CRNDE', 'TNFSF13', 'PAG1', 'MNDA', 'LRRC32', 'NDC80', 'SYMPK', 'MCF2L-AS1', 'ZAP70', 'SCCPDH', 'ODC1', 'CD24P4', 'CD300E', 'CD180', 'C3', 'LTO1', 'DOCK10', 'GPT', 'IGFBP6', 'BUB1B', 'HSPB1', 'XYLT1', 'CSF2RA', 'ABCA8', 'CHRDL1', 'LRCH4', 'BCL6', 'DYNC2H1', 'MTDH', 'CCDC88A', 'RYR1', 'LPCAT4', 'UTS2', 'CENPX', 'ENSG00000293341', 'LST1', 'OFD1', 'RHOF', 'CSGALNACT1', 'SH3BP1', 'GFER', 'IL7R', 'POLD4', 'SERPINB4', 'CDH1', 'CXCR5', 'CASP10', 'GDF7', 'NEK2', 'FEZ2', 'SELL', 'LIME1', 'LRP8', 'KLRK1', 'SLCO3A1', 'VEGFD', 'P2RX5', 'GLMN', 'PPP3CC', 'ORC6', 'COX16', 'ODAM', 'SLC29A4', 'FMO2', 'CDCP1', 'PRKCB', 'BBS4', 'PIK3CD', 'DIO1', 'PBLD', 'HLA-DPA1', 'CTSG', 'ITGAL', 'GIMAP6', 'GPR18', 'SSR2', 'TRDC', 'DSC3', 'SOSTDC1', 'SPATC1L', 'SPRY2', 'SPATA24', 'LINC01140', 'TMEM130', 'ITM2A', 'SELE', 'POU5F1', 'ZBP1', 'SLA', 'ST3GAL2', 'MSN', 'NDUFAF4', 'CXCL16', 'IGKV1D-17', 'APEX1', 'PAICS', 'PALMD', 'ULK2', 'ZPR1', 'CSN3', 'ARHGAP25', 'NAP1L2', 'PTGER4', 'TAB1', 'VPS45', 'FCRL2', 'CD300A', 'SCX', 'ADGRF5', 'DPP4', 'AIF1', 'CD1C', 'CCDC134', 'TCF4', 'DENND1B', 'CD96', 'KDM5A', 'SAP18', 'ESR2', 'MBP', 'DAG1', 'KLK6', 'PABPN1', 'LYN', 'CHIC2', 'PTPRC', 'STK32B', 'SMC4', 'RAB8B', 'LAT', 'CAMTA1', 'GSPT2', 'MATN2', 'CHCHD1', 'MZB1', 'KRT5', 'FZD4', 'ESYT2', 'DNAJC17', 'GIMAP5', 'ITPRID2', 'IFIT3', 'PECAM1', 'CTTN', 'PLA2G2A', 'MEOX2', 'TMOD1', 'CD302', 'TLE4', 'RUBCNL', 'PDE4DIP', 'ADAM28', 'CD3G', 'G0S2', 'SMTN', 'SNCAIP', 'COTL1', 'KIF11', 'SERHL', 'IGLV3-10', 'VNN2', 'PIAS1', 'GZMM', 'EXOSC4', 'CD72', 'IFI16', 'MIR646HG', 'DLL4', 'ACAP1', 'PSMG4', 'PITRM1', 'PTN', 'EPHA4', 'GNPTG', 'NLRP1', 'CXCL12', 'MCM2', 'RBM45', 'TNFAIP2', 'TTLL8', 'ISCU', 'UBE2B', 'ABLIM1', 'ZEB1', 'UBA2', 'ITGA4', 'EMCN', 'CFH', 'HOXA9', 'ICAM2', 'RHOV', 'DEGS2', 'MAST3', 'SPCS3', 'ARID5A', 'CAV1', 'TAC4', 'GZMK', 'GPR171', 'FGFR1', 'ID1', 'PNOC', 'MOGAT2', 'FGG', 'CSF3', 'CHGA', 'GPRC5C', 'LGALSL', 'NAT2', 'MMP10', 'NEURL1', 'KLRB1', 'PNPLA4', 'CLEC2B', 'HOXA7', 'PGM2L1', 'CCDC28A', 'BIN1', 'PTGR2', 'EMILIN3', 'HEPACAM2', 'TNXA', 'HDAC2', 'RAC2', 'PLTP', 'MOB4', 'IGKV3-20', 'CD52', 'NMNAT2', 'RTP4', 'PPIF', 'AARD', 'CRTC3', 'TNFRSF25', 'PIGQ', 'SEL1L3', 'CRYBG2', 'SYNE3', 'SLC16A7', 'GZMA', 'KIF16B', 'ACVRL1', 'ARC', 'PDGFD', 'LPL', 'LSP1P5', 'CD1E', 'CLU', 'CD40LG', 'RBMS1', 'ABCA6', 'IGHV4-34', 'FLI1', 'APOBEC3A', 'FMNL1', 'SKAP2', 'PCOLCE', 'DPT', 'HLA-DRB1', 'CD38', 'RNF44', 'CTCF', 'POU2AF1', 'DHX9', 'CCL5', 'DVL1', 'GVINP1', 'IGF1', 'FYN', 'TMEM265', 'NFYC', 'DPY19L4', 'PELI2', 'IGHM', 'AOC3', 'TRAC', 'BCL11B', 'ARHGAP4', 'KRT14', 'SEPHS2', 'TSC2', 'TRIM65', 'IRAG2', 'RNASE6', 'MTRF1', 'ATP13A4', 'GHR', 'GATA3', 'APOBEC3G', 'CD3D', 'CHEK1', 'GIMAP4']
Refined Community 102: ['HEPACAM2', 'LEFTY2', 'ADK', 'CD244', 'LRP12', 'CAPN6', 'ZNRD2', 'TPK1', 'GPSM2', 'CD24P4', 'LST1', 'MILR1', 'MAP4K1', 'IGLC2', 'BTN3A3', 'YBX3', 'TCEAL2', 'SOX15', 'AZIN1', 'SF3B1', 'ATP1B3', 'PWWP2A', 'ZNF22', 'NANOG', 'ZNF568', 'FZD9', 'CYLD', 'UCHL5', 'IGHV4-61', 'TCIRG1', 'SOCS5', 'KLHL24', 'ZNF587', 'RAB23', 'TUBB2A', 'ID4', 'MC3R', 'FANCA', 'RARRES1', 'TMEM79', 'DAP', 'IGKV1D-37', 'HS3ST6', 'SPAG4', 'PICALM', 'RNMT', 'PYGB', 'MMP14', 'DNTTIP2', 'ITGA6', 'MZF1', 'DSC3', 'VWCE', 'KCNG1', 'TAC1', 'SMNDC1', 'IGKV1D-13', 'CXCL5', 'C17orf58', 'CDKN2A', 'ZIC1', 'CALU', 'KRTAP9-4', 'NDUFS6', 'IFI16', 'SRGN', 'KHSRP', 'CD96', 'GRN', 'SLC7A5', 'CRAMP1', 'CYP39A1', 'HSPD1', 'IRAG1', 'FRY', 'CLDN1', 'RYR1', 'UBE2V1', 'ANKRD27', 'EPB41L2', 'EXOC5', 'CLIP4', 'TTC3', 'RBP1', 'CA5A', 'EIF5A', 'PSAT1', 'ELF5', 'LDHB', 'LTBP3', 'IGHV1-69', 'LYSMD4', 'MIA', 'ANP32E', 'QKI', 'CHRM3', 'PAPOLA', 'TNRC6C', 'PAX6', 'KRT23', 'KRT81', 'KHDRBS3', 'NUCB2', 'COL11A2', 'EPHB3', 'CYTL1', 'CDKN1B', 'ACE2', 'SUPT20H', 'IQCG', 'LMO4', 'CCL20', 'VBP1', 'ZDHHC24', 'FNDC4', 'DHRS4-AS1', 'MED1', 'LYNX1', 'JCAD', 'KANK1', 'GAGE1', 'COL9A3', 'IGKV3-20', 'ADGRB1', 'PADI2', 'SLC16A1', 'RTRAF', 'METTL25B', 'IDI1', 'LRP5', 'CXCR5', 'FZD7', 'ACTL8', 'FGL2', 'S100B', 'WIF1', 'SIX3', 'NUDT15', 'GBP1', 'DNAJB4', 'KCNJ16', 'NKX2-5', 'CHRNA6', 'CRIM1', 'HMGCS2', 'GRID1', 'CASK', 'TSPAN14', 'PFDN6', 'COL4A2', 'SPATA20', 'PLA2G4A', 'DDO', 'SMPDL3B', 'FGF4', 'TNFRSF11B', 'XBP1', 'APBA2', 'TRIM37', 'EDN2', 'CEACAM1', 'NDUFA4L2', 'SLC43A3', 'HSPB8', 'GSTM4', 'KIF15', 'MMP7', 'RUNX3', 'WDR24', 'SYNE1', 'SFRP1', 'EGFR', 'CPPED1', 'SUPT16H', 'NCALD', 'SRD5A1', 'KRT5', 'YWHAE', 'DERL1', 'DLAT', 'FBLIM1', 'GPM6B', 'NCAM1', 'COL2A1', 'PLEKHB1', 'ST8SIA1', 'KCNN4', 'SIGMAR1', 'SLC6A18', 'COL1A1', 'TIMELESS', 'HK1', 'SLC6A4', 'BGN', 'CTSW', 'CCND1', 'TMSB15A', 'TRAF5', 'CDH19', 'THBD', 'SLC12A9', 'EDN1', 'KRT14', 'WEE1', 'KRTAP9-8', 'AFG2A', 'PTPN2', 'ALAS1', 'AMD1', 'GTPBP1', 'TRIM29', 'MCL1', 'RAI2', 'PI3', 'CXCL8', 'GNAI1', 'SLC40A1', 'PEG10', 'PDPN', 'S100A2', 'PDE9A', 'CRYAB', 'WRAP53', 'GAL', 'TMEM45A', 'PPM1E', 'PGBD5', 'ADAMTS17', 'EVI2A', 'IGLV2-14', 'CSN3', 'SLPI', 'PGRMC1', 'HTRA4', 'CEBPB', 'IMPA2', 'WWOX', 'LGR5', 'ITGB4', 'CEP170', 'ACTG2', 'KIF5C', 'PHGDH', 'AURKA', 'RAB37', 'STIP1', 'L1CAM', 'PTGS2', 'DNM2', 'STARD3', 'PFDN4', 'LIMK2', 'SNN', 'MYC', 'MRAS', 'CLDN10', 'BAG2', 'BLTP2', 'GUCY1A2', 'JHY', 'NABP1', 'POFUT2', 'UCHL1', 'IGHV4-34', 'NLRP1', 'TEP1', 'FOLH1', 'MS4A6A', 'TLE1', 'STXBP6', 'MAD1L1', 'IGHV3-23', 'TUBB2B', 'LAMP3', 'CTAG1B', 'GDI2', 'WNT5B', 'CCDC124', 'CRISPLD2', 'FOXO3', 'HERC5', 'ROPN1', 'MTUS1', 'MAGEA12', 'CHI3L2', 'GNPTG', 'PHB2', 'CCDC74B', 'HLA-DQB2', 'BCL11A', 'YBX1', 'KLF5', 'NCSTN', 'FBXO17', 'HLA-DOB', 'POLR2F', 'TMEM158', 'MICALL1', 'ATXN7', 'EMC3', 'IGHG1', 'NFIB', 'TCL1A', 'CHMP4C', 'RFC3', 'DKK1', 'SOX9', 'BMP1', 'IGLV3-25', 'BMPR2', 'WWTR1', 'NUP155', 'IGHV3-7', 'KCNK5', 'ANKRD26', 'KRT17', 'ABI3', 'ERN1', 'PTPRZ1', 'SELPLG', 'KRT6A', 'FERMT1', 'PDK1', 'FIGNL1']
Refined Community 103: ['NPM3', 'CABLES2', 'SLC23A2', 'SCGB1D2', 'SEMA3B', 'SKP2', 'LINC02875', 'DDHD1', 'LIN7A', 'ENPP3', 'GP2', 'DCXR', 'SS18L2', 'TMEM208', 'GALNT7', 'CSPG4', 'PSMC5', 'HMGCS2', 'ATRNL1', 'MAGT1', 'CEACAM5', 'IL24', 'ANXA9', 'GFRA1', 'LINC01949', 'RMND1', 'RTEL1', 'RASSF5', 'SCUBE2', 'EMILIN1', 'CYP2B6', 'TIMP4', 'CAPN9', 'SLC4A8', 'TMEM101', 'OPN3', 'DRAM2', 'CLGN', 'TRAC', 'CDKN1B', 'RAD54L2', 'PDE4DIP', 'MAST4', 'SERPINA5', 'CITED1', 'PPM1B', 'SYT1', 'HIF1A', 'PIERCE1', 'NME4', 'ZNF18', 'PRR14L', 'MUC1', 'LYPD1', 'KAZALD1', 'PRELP', 'PDGFD', 'C1orf131', 'LARS1', 'MRAS', 'PGGHG', 'MXRA7', 'CENPN', 'RTN1', 'NAT1', 'ADH1B', 'MAML2', 'TSPAN1', 'PAH', 'CRIP1', 'MB', 'ATN1', 'FGFR3', 'MSRB1', 'SUMO2', 'HCK', 'PITRM1', 'RECQL4', 'INPP4B', 'MSMB', 'GATA6', 'GSTT2', 'MLPH', 'CELSR1', 'NELL2', 'CHAD', 'BEX5', 'CNNM4', 'SVIL', 'MATN3', 'SIL1', 'P3H1', 'AGR2', 'BCAS1', 'CA12', 'GPC1-AS1', 'DHX34', 'PSD3', 'ACADSB']
Refined Community 104: ['MAST4', 'KLRK1', 'SPATA2', 'THEMIS2', 'MPHOSPH9', 'N4BP2L2', 'CELSR1', 'HOXA9', 'GPR171', 'PLCH1', 'NUDT4', 'MUC16', 'PPARD', 'INHBA', 'DEPTOR', 'COQ7', 'ELAPOR1', 'RAB27B', 'DNAH17', 'DNALI1', 'TAF1D', 'SPDEF', 'TCAP', 'TMEM74B', 'IKBKB', 'ALDH18A1', 'RET', 'PPFIA1', 'DNAJC12', 'CGAS', 'NECTIN3', 'GUSBP3', 'C4A', 'FBF1', 'ZNF790', 'APOL6', 'IVD', 'MET', 'MT1M', 'MAPKBP1', 'RAMP1', 'ITPR1', 'KRTAP9-4', 'RBM47', 'RNF182', 'CLEC7A', 'MCCC2', 'MLPH', 'TFF3', 'CDS1', 'MED28', 'LPXN', 'GATB', 'TBC1D9', 'KRT16', 'GATA3', 'ASTN2', 'SERPINA5', 'ST6GALNAC2', 'IL6ST', 'SLC7A8', 'CXXC4', 'GOLGA2P5', 'SLC44A4', 'NRIP1', 'EVPL', 'ERFE', 'GREB1', 'ARMT1', 'GPC1', 'ERBB4', 'SLC27A2', 'IL6R', 'MAGEA12', 'NAAA', 'CCNO', 'VPS50', 'HEXIM1', 'ZNF91', 'EVL', 'PSD3', 'TBRG1', 'NCOA3']
Refined Community 105: ['CLIP2', 'PPY2P', 'ART3', 'COX19', 'CP', 'VSNL1', 'MMP9', 'CXCL9', 'DLX5', 'CHST3', 'OAS3', 'TMEM87A', 'DPT', 'HOXA9', 'SGO1', 'MCM10', 'CCKAR', 'EGFR', 'KCNK5', 'KASH5', 'SPIDR', 'WWTR1', 'IGFBP4', 'SMG9', 'LMO3', 'POMT1', 'EGFL6', 'LINC02627', 'TTLL4', 'S100A10', 'KRT16', 'RPTOR', 'MMP7', 'KIFC1', 'GM2A', 'P2RX5', 'CENPN', 'SFRP1', 'USP34', 'GDF5', 'SELENBP1']
Refined Community 106: ['CFB', 'CPB1', 'TNXA', 'MOCOS', 'SYDE2', 'CHRD', 'GLRB', 'ARHGAP45', 'DNALI1', 'STC2']
Refined Community 107: ['KYNU', 'CRLF1', 'SPOP', 'PATZ1', 'RAP1B', 'FGB']
Refined Community 108: ['CH25H', 'NAT1', 'POGLUT3', 'TIMP4', 'DNALI1', 'TRRAP', 'NMI', 'SKP1P1', 'NUP62', 'SCGB1D2', 'RAC3', 'FABP4', 'TTC12', 'COL14A1', 'TLE4', 'KLHL28', 'PRRC2C', 'LRRC31', 'COMP', 'ASPN', 'FABP5', 'STAP2', 'PPP1R1A', 'ADIPOQ', 'SCGB2A1', 'TFF1', 'ASXL1', 'SIM1', 'TBC1D19', 'HMGCS2', 'ITGA7', 'IRS1', 'KCNK9', 'GMFG', 'DPT', 'BAG6', 'RNF214', 'LRRN3']
Refined Community 109: ['FOXQ1', 'TMIGD1', 'CSPG4', 'MMP14', 'SUPT4H1', 'GLS', 'FOXD1', 'CTSV', 'ANP32E', 'FHL1', 'PRKDC', 'TNNI2', 'PRPF31', 'PRAME', 'SYNGR2', 'ESRRG', 'KDM6A', 'CDC5L', 'RYR1', 'KPNA4', 'KLK7', 'MUC16']
Refined Community 110: ['HCP5', 'SESN1', 'S100P', 'SLPI', 'IGKV1D-39', 'CDH3', 'DDX39A', 'GABBR1', 'PEG10', 'GSS', 'RASA4', 'MAGEA3', 'SCPEP1', 'PLEKHO1', 'ENSG00000278932', 'FABP7', 'FUS', 'CP', 'ZPBP2', 'CEACAM1', 'PTPRT', 'CCL18', 'MAPK11', 'MIA', 'BAMBI', 'IGKV1D-17', 'SIM1', 'CST1', 'TBX19', 'PRRG4', 'DZIP1', 'PAX2']
Refined Community 111: ['SLC3A2', 'TENM3', 'ISG20', 'SYNGR3', 'CHN2', 'ZCCHC8']
Refined Community 112: ['WDR19', 'CRTC3', 'CFAP69', 'ARHGEF12', 'CRNDE', 'DZANK1', 'NADSYN1', 'LILRB4', 'DRG2', 'LTBP3', 'CASP9', 'STAT5B', 'SECISBP2', 'DYNC2H1', 'STARD13', 'PGRMC2', 'ZNF862', 'SIRT3', 'MAN2B2', 'MPHOSPH8', 'ITGA3', 'ECI2', 'DEAF1', 'GFOD3P', 'CYP2J2', 'SEMA5A', 'FRY', 'CHEK1', 'PKP4', 'NME5', 'NKAIN4', 'RASEF', 'TOP6BL', 'SNX1', 'PLEKHH1', 'FOS', 'SESN1', 'LYNX1', 'RGPD5', 'RANGAP1', 'NBR1', 'CREBZF', 'C6orf62', 'ARHGEF40', 'CIRBP', 'C3', 'F13B', 'IFT88', 'RHOF', 'CLC', 'ZFP36L2', 'LINC00115', 'ECHDC2']
Refined Community 113: ['POLQ', 'RTL10', 'MCM6', 'DLGAP5', 'ITM2C', 'KPNA2', 'KIF11', 'MDM2', 'TMPO', 'EPB41L1', 'TIMELESS', 'KCTD1', 'LAGE3', 'CD300LF', 'TMED2', 'H2AZ1', 'TRIM2', 'CCNE2', 'HSF5', 'PEX6', 'KIF14', 'MFSD3', 'CORO6', 'ANPEP', 'KIF18B', 'TOMM70', 'TROAP', 'H2AZ2', 'TXNRD1', 'PLTP', 'ANAPC1', 'GREB1', 'KIFC1', 'UBE2C', 'UBBP1', 'CASP9', 'CCNB1', 'RAB5IF', 'HCCS', 'MKI67', 'ALDH3A1', 'NECTIN2', 'STMN1', 'CDCA3', 'TRIM65', 'PPP1R16A', 'MCM2', 'NUP93', 'CDC20', 'MRPS17', 'PSCA', 'NEK2', 'HMMR', 'IMPA2', 'H2BC12L', 'CHEK1', 'HGD', 'KIF15', 'HJURP', 'RBBP5', 'MFAP5', 'MAD2L1', 'ACSM5', 'CHST3', 'CDC25A', 'SLC15A3', 'OLFML3', 'GTSE1', 'CMC2', 'STK11', 'RAD1', 'TXN', 'NUDT6', 'CENPN', 'BIRC5', 'NOL8', 'PRKAR2B', 'MRPL15', 'CTSZ', 'SPHK1', 'ZEB2', 'CSNK2A2', 'TMSB15A', 'TFPT', 'CDCA8', 'GNPNAT1', 'FOXM1', 'FEN1', 'E2F1', 'ADCY1', 'MIS18A', 'SNRPF', 'ETFA', 'GASK1B', 'TIGD5', 'CRACD', 'PSMA7', 'OR12D2', 'NGLY1', 'SLC2A3', 'RNASEH2A', 'TAX1BP3', 'TUBA1A', 'JPT1', 'SLC52A2', 'TNNT1', 'GLCCI1', 'NCOA1', 'BUB1', 'CDC45', 'TGM2', 'SPAG5', 'DESI1', 'KIF4A', 'NOC2L', 'DSN1', 'TPX2', 'RACGAP1', 'TUBGCP6', 'PARPBP', 'NTSR1', 'BAG6', 'LDHA', 'LMNB1', 'EPHB3', 'CDKL5', 'CAPN3', 'GATAD2A', 'TEDC2', 'STIP1', 'NUTF2', 'PSMD12', 'ESPL1', 'AURKB', 'MAGEA2', 'ST3GAL6', 'C7orf50', 'LAT2', 'RAD51', 'TTK', 'ITGB1', 'CEP55', 'MIR22HG', 'NCAPH', 'GALR2', 'XPOT', 'NME1', 'CAPNS2', 'KIF20A', 'DENND1B', 'MYBL2', 'POLR2K', 'PDSS1', 'QPRT']
Refined Community 114: ['CA2', 'VEZF1', 'GFRA1', 'PRKD1', 'PIGT', 'DCPS', 'SH3BP4', 'BMPR2', 'CCDC74B', 'ZFHX3', 'TTC28', 'ABHD12', 'ADGRG1', 'ZNF165', 'NEMP1', 'KIF12', 'ZNF750', 'HDGFL3', 'TPPP3', 'PCOLCE', 'CHURC1', 'FXYD1', 'CSNK1A1', 'HSD11B2', 'CSTF2', 'PLEKHB1', 'BLTP2', 'ITGBL1', 'RPL23A', 'C11orf52', 'EMC3', 'NAT2', 'PRKDC', 'RGP1', 'PARD6B', 'WNT3A', 'CTSK', 'UBE2Q2', 'KLHL2', 'SEC14L2', 'MID1IP1']
Refined Community 115: ['RRP1B', 'OGG1', 'CCL2', 'GLMN', 'APOE', 'CPSF7', 'ART3', 'KCNN4', 'UCHL5', 'CPPED1', 'MXRA5', 'CHODL', 'PPP1CB', 'THEMIS2', 'FRAT2', 'CYP26A1', 'DACT1', 'WDR43', 'ZPR1', 'CACTIN', 'PIM2', 'ATAD3A', 'GPR176', 'KRT86', 'KDM2B', 'VPREB3', 'JAG2', 'TPSAB1', 'NANS', 'FCGR1A', 'KCNMA1', 'RRP7A', 'PACSIN3', 'DNTTIP2']
Refined Community 116: ['BTG3', 'S100A10', 'RRS1', 'TNF', 'TMSB10', 'GPRC5B', 'PRKACB', 'CA10', 'YES1', 'SMIM19', 'TRPA1', 'SYNCRIP', 'ESPL1', 'LSM7', 'TTK', 'GDF5', 'LARS1', 'TCN2', 'GPSM2', 'CCL18', 'LPIN1', 'LGALS8', 'ZFC3H1', 'RDX', 'PCLAF', 'CTSS', 'PDXK', 'ATP1B3', 'HRAS', 'PSAT1', 'FBXO47', 'BIRC2', 'KATNA1', 'FIRRM', 'ANKRD27', 'SLC49A3', 'MRAS', 'LDHB', 'FZD3', 'STIL', 'IDO1', 'WDR90', 'IDH2', 'KRT7', 'SERBP1', 'TRIM29', 'NDN', 'FAXC', 'CREB3L2', 'TTC12', 'TTL', 'MYBL2', 'DYRK1A', 'DDR1', 'CYP3A43', 'BAHCC1', 'LMO4', 'GLCCI1', 'MPZL1', 'CYBA', 'RGL2', 'IL17RB', 'LYN', 'TM2D2', 'OAS2', 'PADI2', 'NASP', 'ILF2', 'WARS1', 'GPRIN2', 'GPR20', 'CUL7', 'BOP1', 'CA2', 'PPP1CB', 'CEP57', 'HYAL1', 'LALBA', 'SLC2A6', 'ATP6V1G1', 'TRIM33', 'FZD9', 'IGFBP2', 'GLS2', 'MT-CO2', 'SGCB', 'ACOX2', 'MALL', 'CCL5', 'SUSD6', 'ENO1', 'EDARADD', 'GATA6', 'TNFRSF10B', 'MEAK7', 'THEM6', 'PFKP', 'EGFL6', 'MID1', 'TRAPPC9', 'TATDN2', 'ODC1', 'PSMG1', 'NCK1', 'BUB1', 'CDK2AP1', 'SRGN', 'HDGF', 'HARS1', 'MARCO', 'FAP', 'CSRP2', 'CTNND1', 'NMB', 'CETP', 'MRPS27', 'MCM6', 'NMI', 'MELK', 'PPM1H', 'HJURP', 'GAPDH', 'CHI3L1', 'APOBEC3B', 'IKBKE', 'SYNRG', 'LGALSL', 'PICK1', 'CIAO3', 'RRP1', 'ADK', 'TRIO', 'PLSCR1', 'CASP10', 'VGLL1', 'CLDN10', 'NKAIN3', 'TMEM245', 'ACAA1', 'NUTM2E', 'FAM83E', 'EXOC5', 'UGP2', 'PHGDH', 'MED24', 'NDUFB10', 'CCNA1', 'SAMD4A', 'PDZK1IP1', 'ZNF345', 'RBM7', 'KRT16', 'GMPS', 'HEBP1', 'PIM1', 'PROM1', 'P2RX5', 'THEMIS2', 'KRT23', 'FDX1', 'RSU1', 'CELF2', 'PRIM2', 'ARFIP2', 'POT1', 'DEF8', 'HIF1A', 'RAD54L', 'H2AJ', 'DLGAP4', 'LAMP3', 'GSTT1', 'PDCD5', 'GOLT1A', 'NDRG1', 'GFM2', 'H2BC12L', 'SFT2D2', 'CTSC', 'ATAD3A', 'ADAMTS7', 'TNFRSF21', 'GLIPR1', 'SMCO4', 'MCM5', 'UGT8', 'NR2F1', 'FGF17', 'SRSF7', 'LILRB3', 'GTPBP4', 'NF2', 'FNDC3B', 'NUDT15', 'IQSEC1', 'FAM171A1', 'MAPRE2', 'TNFAIP3', 'ZNF469', 'TM4SF1', 'RILPL2', 'WDR1', 'IFRD1', 'AGFG1', 'TFAP2C', 'CYB5R2', 'RALGPS1', 'YBX2', 'TMEM134', 'CNPY4', 'MTMR2', 'RBMS1', 'CCL4', 'DPM3', 'UTRN', 'DUSP9', 'TUBA4A', 'DESI2', 'BRCA1', 'GTPBP10', 'SLC38A10', 'CLCN4', 'PHPT1', 'PXDN']
Refined Community 117: ['SOX12', 'MYO5C', 'BGN', 'MMP9', 'SYBU', 'BICDL1', 'AQR', 'WDTC1', 'TP53INP1', 'KRT7', 'HDAC7', 'APOH', 'NOTCH3', 'ZNF587', 'KRT18', 'SCAMP1', 'ANGPT2', 'QDPR', 'ESR1', 'ZNF260', 'SYNE4', 'CIRBP', 'HNRNPDL', 'KIF13B', 'RIPK2', 'HHAT', 'CENPN', 'LIN52', 'ZNF16', 'SNX1', 'PTPRT', 'NR1H3', 'WDR6', 'RASSF3', 'IGBP1', 'OVOL2', 'PCGF6', 'CAMTA1', 'KIAA0232', 'SEMA3G', 'STARD10', 'IRS1', 'MED13L', 'NUBP2', 'ZNF7', 'RGP1', 'MEIS3P1', 'ITPK1', 'DDAH2', 'NOTCH4', 'CNOT1', 'BTBD9', 'ENO2', 'C3orf18', 'CELSR1', 'CHRD', 'MAST4', 'PCP4', 'FBXO2', 'POLR3A', 'HEXIM1', 'SLC35E2B', 'KRTAP4-1', 'INTS13', 'ANK3', 'WIPI2', 'CHAD', 'TAGLN', 'FAM234A', 'TUBA3E', 'TMBIM4', 'ODR4', 'LMNB1', 'SKP1P1', 'RNF151', 'HGS', 'BAG1', 'PPP1R26', 'DBNDD2', 'PTP4A2', 'UGCG', 'GGCT', 'ABCA3', 'HTT', 'BTN3A1', 'STYX', 'EVL', 'TCEAL1', 'LZTFL1', 'CXXC5', 'SPATC1L', 'PSCA', 'NEK9', 'PRICKLE1', 'CCDC74B', 'HDAC11', 'ERBB4', 'NFYC', 'RAB11FIP3', 'IFNAR2', 'FBP1', 'TUBB1', 'MYB', 'SLC16A6', 'EEIG1', 'GSTP1', 'PALM2AKAP2', 'SMARCA2', 'FAAH', 'SMARCD1', 'KPNA2', 'NAT1', 'CERS6', 'SCUBE2', 'MINDY1', 'TEPSIN', 'LARP6', 'ABCC2', 'SOCS7', 'LALBA', 'TFF3', 'DELE1', 'KCTD14', 'SPEF1', 'RPL7', 'FAN1', 'ARHGEF4', 'COX6C', 'KCTD3', 'BHLHE40', 'FARP2', 'ATP5F1E', 'ARAP2', 'LMNTD2-AS1', 'CAPN5', 'CFLAR', 'PIGT', 'MBP', 'ZBTB40', 'TPBG', 'TSPAN13', 'EPPK1', 'NPDC1', 'MAD2L1BP', 'TBC1D9', 'THUMPD1', 'CAMK2B', 'IL6ST', 'MRFAP1L1']
Refined Community 118: ['AURKA', 'ESM1', 'PIMREG', 'CENPA', 'ADAMDEC1', 'ELOVL4', 'TES', 'GTF2H2', 'FBXO5', 'ARHGAP30', 'NMB', 'GSTM3', 'S100A13', 'DMAC2L', 'MGAT4A', 'LGALS8', 'SRPX2', 'PLAAT1', 'GGH', 'THBS3', 'TRAF4', 'CXXC5', 'CENPN', 'CTPS1', 'NMU', 'IVNS1ABP', 'EXT1', 'LRRC37A3', 'KDM7A', 'FAU', 'SNX16', 'SHCBP1', 'CADM1', 'HACD2', 'DTL', 'ASNS', 'XRCC3', 'INTS7', 'KRT23', 'GPC1-AS1', 'ZNF18', 'FGFR4', 'TMSB10', 'MPHOSPH6', 'DSCR10', 'MAP2', 'MMP23B', 'DLL3', 'RILPL2', 'TMBIM4', 'SERF1A', 'JCHAIN', 'PAQR3', 'RAD21', 'IGFBP5', 'ACTG1', 'MPRIP', 'HIF1A', 'NR2F6', 'RFC4', 'SLC7A1', 'INAVA', 'ALOX15B', 'SLC2A3', 'DLGAP5', 'DEPDC1', 'LRP6', 'HGS', 'SPC25', 'COL4A2', 'ITGA2', 'PSMD7', 'TMEM65', 'KIF21A', 'ARMC1', 'MAPK7', 'RGS4', 'KIF14', 'MELK', 'SMC4', 'RAB6B', 'AQP3', 'MATN4', 'UCHL5', 'BUB1', 'PSMD2', 'ZNF878', 'TRIP13', 'ARHGEF5', 'BNIP3', 'OXCT1', 'TBCA', 'SULT1A1', 'CSE1L', 'PFDN4', 'TSPYL5', 'TMEM45A', 'LEFTY2', 'PRAME', 'STK3', 'DCK', 'PTDSS1', 'SLC20A2', 'MTDH', 'SACS', 'RRAGD', 'EFNA1', 'HJURP', 'LYPD3']
Refined Community 119: ['MAP6', 'MC1R', 'TRAFD1', 'ALDH6A1', 'CTCF', 'SEC14L2', 'METRN', 'ALDH4A1', 'PDIA4', 'MMADHC', 'RPS4X', 'TGFB1', 'TFAP2B', 'KIF3B', 'ACADS', 'QDPR', 'FBP1', 'DNM1', 'CIRBP', 'CCNB2', 'SUV39H1', 'MYRIP', 'MRPL36', 'RBM47', 'KIAA1217', 'PLEKHA1', 'RAB27B', 'BTG2', 'STK32B', 'IGFBP5', 'LIX1L', 'PEX12', 'IL13RA1', 'KRT18', 'KATNA1', 'UBE2I', 'COL4A6', 'PPFIBP2', 'BBC3', 'TBX3', 'SPEF1', 'IP6K2', 'COL8A1', 'MYLIP', 'DHX16']
Refined Community 120: ['ELF1', 'TSPYL5', 'NMU', 'ALDH4A1', 'NDC80', 'CMC2', 'FTSJ3', 'STK32B', 'WNT5B', 'TGFB3', 'PRC1', 'LMTK3', 'RFC4', 'ARHGEF2', 'SCUBE2', 'C1RL-AS1', 'FOXK2', 'MS4A7', 'TMEM74B', 'LPCAT1', 'FANCA', 'CD247', 'EXT1', 'MIR100HG', 'OXCT1', 'LARS1', 'TMEM65', 'CD19', 'ECI2', 'SLC2A3', 'HEG1', 'NOP16', 'FGF18', 'ORC6', 'HMGA1', 'WARS1', 'PDGFRL', 'EPB41L1', 'MELK', 'CBR1', 'SERPINE1', 'CSPG4', 'IGFBP5', 'CCNE2', 'GDPD1', 'SLC6A4', 'DAAM1', 'ARHGAP5', 'AP2B1']
Refined Community 121: ['MXD4', 'DTNB-AS1', 'PSTPIP1', 'TBL3', 'FUT3', 'CEP57', 'ACOT11', 'GABRQ', 'TNFSF10', 'ZNF362', 'NDUFAF8', 'PKP2', 'ODAM', 'PCK2', 'PARP4']
Refined Community 122: ['ABLIM1', 'METTL25B', 'OR12D2', 'TIMP2', 'ZFP36L2', 'VIPR1', 'CD44', 'GABRP', 'ASH1L', 'VNN1', 'AP2A2', 'SLC35A1', 'MCCC2', 'CNKSR1', 'PDLIM1', 'ST6GAL1', 'CFAP69', 'LINC01482', 'RNF103', 'TATDN1', 'PHF6', 'NKAIN1', 'TESPA1', 'MMP23B', 'TACC2', 'CLDN9', 'ARHGDIB', 'FKBP14', 'PHF11', 'DUSP4', 'BICD1', 'MIA']
Refined Community 123: ['NR2F1-AS1', 'BBS1', 'SUPT16H', 'PPP1CC', 'H4C8', 'CCNE2', 'POLR1C', 'PGAP6', 'BID', 'NCAPG2', 'C4orf54', 'ATAD2', 'KITLG', 'GTSE1', 'EPN1', 'HLA-DPA1', 'SLC7A5', 'POLQ', 'MYRF', 'PRKX', 'SCGB1D2', 'ZCCHC8']
Refined Community 124: ['CDC25A', 'CAVIN2', 'INTS2', 'GADD45A', 'GJB3', 'CASP1', 'JAKMIP1', 'DKK1', 'MAFF', 'MAD2L1BP', 'OLFML1', 'VCL', 'CTNNA1', 'FOSL1', 'IRX5', 'HBEGF', 'SMURF2', 'SNHG26', 'PIK3R2', 'WDR6', 'DHX34', 'MIEN1', 'NELL2', 'CAV1', 'GDF15', 'THBS1', 'SLC7A5', 'KRT17', 'RAB22A', 'CCL14', 'MSANTD3', 'SPHK1', 'EPHA2', 'UPP1', 'GNPTG']
Refined Community 125: ['SLC7A11', 'PSPH', 'PRICKLE1', 'NFAT5', 'STAR', 'RHBG', 'PSAT1']
Refined Community 126: ['HSD17B1', 'FZD2', 'BARX1', 'BAK1', 'NFKB2', 'CTNNB1', 'TNFRSF14', 'FGF4', 'HS3ST1', 'GBA1', 'USP1', 'CSNK2A2', 'FGF9', 'PHYH', 'PIK3R1', 'AMY2B', 'FGF23', 'KRAS', 'MAP2K1', 'PPP1R15B', 'ESR2', 'NBN', 'ARHGAP35', 'NCOA1', 'ZNF467', 'FOS', 'DIXDC1', 'ARAF', 'PIM2', 'NYX', 'FGF20', 'CRABP2', 'FGF6', 'GPR176', 'CCN1', 'CSNK1A1', 'RAB2A', 'PTEN', 'ASPH', 'MED1', 'NPW', 'WNT7A', 'AXIN1', 'PIK3R3', 'SLC47A1', 'PIDD1', 'TNFSF11', 'LRP5', 'TTPA', 'MIA', 'ANG', 'SLC16A6', 'PCMTD2', 'NOTCH1', 'ATOSA', 'ATM', 'FGF22', 'TMEM109', 'WNT10A', 'MIF4GD', 'EPB41L4A', 'RPS6KB2', 'HOXB3', 'PARP1', 'PRIM1', 'KCNMA1', 'CCDC124', 'CSTB', 'PITRM1', 'GBE1', 'PIK3CA', 'SHANK2', 'SLK', 'PIM3', 'RAD50', 'CDK6', 'ORMDL1', 'GREM1', 'CSNK2B', 'LY6S-AS1', 'FABP9', 'CSNK2A1', 'NPTX2', 'EGFR', 'HES1', 'WNT16', 'CELSR3', 'CD1C', 'WNT5B', 'ESR1', 'SPON2', 'HES5', 'CLU', 'EGF', 'CRYBA1', 'MAPK1', 'PIGS', 'DLL1', 'HEY1', 'AKT2', 'GREB1L', 'BANF1', 'JUN', 'NME1', 'FZD1', 'MRE11', 'FGF17', 'GID4', 'SRSF7', 'SHC2', 'P2RY6', 'WNT3', 'PPY2P', 'APC2', 'KLK6', 'EVI2B', 'HRAS', 'PIK3R2', 'KIT', 'NDUFB10', 'PTBP1', 'FGF19', 'DLL4', 'GSK3B', 'SHC3', 'GADD45G', 'WNT1', 'FGF16', 'WNT7B', 'E2F1', 'KIDINS220', 'IL7', 'CMA1', 'RAP1GAP2', 'SFTPC', 'WNT3A', 'FGF2', 'FTH1', 'TOMM34', 'LKAAEAR1', 'ATG12', 'EDN2', 'NOTCH3', 'PPP1R3D', 'GADD45B', 'HEYL', 'NRAS', 'FZD8', 'ABCA6', 'MOCOS', 'PGR', 'RAD51', 'BRCA2', 'PIK3CD']
Refined Community 127: ['CCNE1', 'SNHG3', 'PIK3C2B', 'HRAS', 'MTOR', 'ZBTB10', 'PIK3CD', 'SSTR5', 'RAF1', 'CDK7', 'SELENOS', 'IRS1', 'DEPTOR', 'INS', 'RCAN2', 'PIK3C2G', 'CYP19A1', 'IL17REL', 'PIK3R6', 'PNPLA4', 'EIF1', 'GPAM', 'CDK4', 'NDUFS8', 'PIK3R3', 'ZFP2', 'CCND1', 'OSBPL2', 'PIK3C2A', 'PIK3R1', 'FAM171A1', 'HID1', 'CDK2', 'PIK3CB', 'MCF2', 'COL16A1', 'STYX', 'NRAS']
Refined Community 128: ['HK1', 'CLIC2', 'HMG20B', 'PDP1', 'PDK1', 'ARHGAP8', 'TRA2A', 'LDHA']
Refined Community 129: ['ASH1L', 'EGFR', 'ATM', 'DHCR24', 'RHEB', 'TRAPPC2L', 'ADISSP', 'CHEK2', 'RPP38', 'SYDE2', 'BRCA1', 'SAR1B', 'AURKA', 'NFKB1', 'BAK1', 'RAP1A', 'SERHL', 'SMARCA4', 'MYL9', 'RNF220', 'ADAMTS12', 'CDK12', 'RHO', 'SMAD6', 'NKD2', 'BID', 'PARD6B', 'NAB1', 'ZNF655', 'SPINK5', 'MSH2', 'SERF2', 'RSPH1', 'HIPK2', 'FILIP1', 'NOXA1', 'CXCL8', 'NFYC', 'PIGR', 'LEPROT', 'ITK', 'DYNLRB2', 'PALMD', 'PLEKHG1', 'CDC42', 'ZNF358', 'NUP85', 'PATZ1', 'RASGEF1A', 'TGFBR2', 'BAD', 'PI4KA', 'GSK3A', 'ATR', 'RAPGEF3', 'RALGAPA1', 'MDM2', 'BLM', 'SIRT1', 'NTAQ1', 'RNF166', 'SDHA', 'ODC1', 'EML1', 'C11orf54', 'SRPK1', 'ATF1', 'HMGCR', 'SOX10', 'ZNF43', 'CDC25B', 'FOXO1', 'RB1', 'RBFOX2', 'SGPL1', 'SP1', 'FHL2', 'HLA-DQB2', 'HHEX', 'FAM83A', 'PRPF31', 'USP15', 'CDK7', 'MAP3K13', 'RAD50', 'CASP8', 'RAC1', 'DDX27', 'PIR', 'DAGLA', 'BARD1', 'TP53', 'PROCA1', 'FRY', 'MIR29B1', 'NFU1', 'FADS6', 'ANXA2', 'TMEM192', 'VEGFA', 'IMPA1', 'PLK1', 'BAG6', 'SLC6A4', 'UBE2F', 'AR', 'TSC1', 'GGH', 'DCAKD', 'GAL', 'C17orf78', 'GDI1', 'BRCA2', 'NF1', 'BACH1', 'CTNNB1', 'PHB1', 'MYT1', 'CHUK', 'MBNL3', 'HDAC1', 'SYCP2', 'BAIAP3', 'CASP9', 'FOSL1', 'FOLR1', 'POLR2A', 'IQCG', 'CPNE7', 'MTOR', 'JUN', 'DSC3', 'ATP2C2', 'EDAR', 'NPW', 'TSC2', 'ATP11B', 'PDP1', 'FOXO3', 'DAB2', 'CREB1', 'APOBEC3G', 'KRAS', 'POU2AF1', 'BAX', 'ZMIZ1', 'OGN', 'SERPINE2', 'CD1E', 'PLA2G7', 'LAMTOR5', 'BOP1', 'CDH1', 'WDR19', 'RAD51', 'SMAD4']
Refined Community 130: ['HBM', 'ZNF611', 'MLST8', 'MAP2K2', 'GASK1B', 'CDK1', 'ACTG1', 'MAST1', 'SIX1', 'PIK3CA', 'USH1G', 'GRB2', 'MAP2K1', 'P2RY14', 'PTEN', 'TAGAP', 'SMAD6', 'TRIM46', 'RAF1', 'IRS1', 'SLC11A2', 'AKT2', 'RPS6KB1', 'PARP1', 'MTOR', 'POMK', 'INPP4B', 'VBP1', 'MT1X', 'AKT1', 'ADGRA2', 'FGF21', 'EIF4EBP1', 'POTEA', 'HRAS', 'PDK1', 'DEPTOR', 'MAPK1']
Refined Community 131: ['WASHC5', 'HRAS', 'H2BC12L', 'RAB31', 'PIK3R4', 'ERBB3', 'PIK3R5', 'SPOPL', 'F2RL1', 'HSH2D', 'ABHD11', 'PIK3C2B', 'PIK3CG', 'EGFR', 'CTCFL', 'PHLDB2', 'PIR', 'PIK3C2G', 'AKT2', 'DDX50', 'IGF2BP3', 'MAGEA5P', 'PIK3R2', 'RAF1', 'AKT1', 'NRAS']
Refined Community 132: ['IGF1', 'COL2A1', 'TTK', 'PLAUR', 'DSE', 'CNIH2', 'MAPK8', 'TMEM123', 'KIF1A', 'BAG2', 'PLIN2', 'LAD1', 'NRK', 'SIRPA', 'RIN3', 'KIF4A', 'NCK1', 'GTPBP4', 'BCL2', 'PAPOLA', 'KCNK5', 'TCN2']
Refined Community 133: ['PAG1', 'SLC7A8', 'LMNA', 'PLAAT2', 'CRIP1', 'GDPD1', 'RNF170', 'CELSR1', 'INPP5J', 'FH', 'DDB1', 'MYO5C', 'SLC39A6', 'SKP1', 'RETREG1', 'GNB1', 'AGGF1', 'LONP2', 'RNF149', 'CASP8AP2', 'ARMT1', 'ELOVL5', 'IFI44', 'REEP5', 'GPC3', 'NSUN4']
Refined Community 134: ['C2CD2', 'SOX11', 'SFRP1', 'CORO1B', 'RAB11FIP3', 'FBXL5', 'SLC9A6', 'LYNX1', 'PRR15L', 'TLE4', 'NPIPB3', 'CFAP45', 'FABP5', 'TBX19', 'TPSB2', 'PROM1', 'PPP3CC', 'PLCH1', 'TUBB6', 'LBR', 'TLN1', 'ARHGEF9', 'STK32B', 'MFAP2']
Refined Community 135: ['ZNF532', 'MAD1L1', 'SEC63', 'ACAT2', 'CACNA2D2', 'NCR3', 'PLD1', 'SMCO4', 'SAA1', 'FREM2', 'ALDH3B2', 'GSTP1', 'SOS1', 'CRIM1', 'BMAL2', 'VEZF1', 'SGCE', 'ZBTB42', 'NVL', 'SOX8', 'UBE2J1', 'FSCN1', 'RRP1B', 'LAMB3', 'ACOT9', 'PPP1CB', 'ZYG11B', 'PLS3', 'PALS2', 'CHI3L1', 'PRKD3', 'HSPA14', 'MTBP', 'C1QC', 'UGT8', 'PKP1', 'CCN4', 'ATP11B', 'ITSN2', 'ANXA1', 'PRNP', 'COMT', 'CAVIN2', 'NEIL1', 'SNX3', 'ODC1', 'RSRC1']
Refined Community 136: ['ZNF12', 'MAP9', 'LETM1', 'DNMT1', 'CPT1A', 'SIRT3', 'PACSIN3', 'TP53TG1', 'MCCC1', 'ALAD', 'HMGCL', 'PPP1R12C', 'CDH11', 'KDELR2', 'AGR2', 'DACH1', 'TMEM123', 'IVD', 'NEK9', 'PSMC6', 'BBOF1', 'PEX3', 'UGCG', 'PPP1CA', 'MSX2', 'COQ7', 'FRAT1', 'IFT46', 'RAB26', 'SIRPAP1', 'MOAP1', 'GPR65']
Refined Community 137: ['BMP1', 'BUB3', 'CA12', 'TFF3', 'TRIP10', 'PSPH', 'SLC7A8', 'RAB26', 'SNX8', 'KRT23', 'AHNAK', 'NUP155', 'SLK', 'KDM4B', 'SCNN1A', 'SULT2B1', 'SERPINE2', 'CAMK2G', 'SCUBE2', 'GAS1', 'XBP1', 'PAK1', 'WWP1', 'KCNK15', 'WFS1', 'ABAT', 'MAGED2', 'SMARCA5', 'APBB2', 'NAT1', 'AKTIP', 'GSTM3', 'ANGEL1', 'IL17REL', 'KRT18', 'NUMA1']
Refined Community 138: ['VCAM1', 'MFSD5', 'RAP2A', 'SEC61A2', 'NDUFA3', 'COMMD5', 'RIPOR3', 'SELPLG', 'SLC38A7', 'PTEN', 'RBM5', 'IRF6', 'SOX4', 'C17orf100', 'RBBP4', 'F13A1', 'RANBP1', 'HPS1', 'A2M', 'DNAJC13', 'DDR2', 'OSBP', 'GYG1', 'AAGAB', 'ORMDL1', 'ARHGDIA', 'CACHD1', 'NPRL3', 'ARRDC1', 'RAMAC', 'MSC', 'MAP1LC3B', 'POFUT2', 'CSNK1A1', 'ANTKMT', 'SLC38A10', 'DDX5', 'CD3D', 'KLHL25', 'SCNN1A', 'GALNT1', 'WLS', 'NTRK2', 'PRSS23', 'DCAF6', 'ZNF461', 'PRKCD', 'NCBP1', 'CCDC50', 'ERGIC2', 'MPHOSPH8', 'TRMT13', 'SETMAR', 'CX3CL1', 'C4orf54', 'JAK1', 'COL1A1', 'IGLV4-60', 'RNF41', 'TCEANC', 'ELF5', 'CAST', 'KRTCAP2', 'ARHGAP32', 'EYA3', 'SAP18', 'ARMC8', 'CEACAM1', 'TACSTD2', 'TNFAIP2', 'KIF14', 'FMNL2', 'MON2', 'CGRRF1', 'TMEM30A', 'PM20D2', 'NRIP2', 'TIRAP', 'SPX', 'MNDA', 'SEMA7A', 'TMSB10', 'IFI27', 'SOX10', 'PFKFB4', 'TLE4', 'APCDD1L', 'GZMA', 'PRKDC', 'ZNF385A', 'MAPRE1', 'ELK3', 'ZNF18', 'ZMIZ2', 'CSTF1', 'KCNN4', 'ATP5IF1', 'SPECC1L', 'NDRG4', 'TYROBP', 'LTBP4', 'MDFIC', 'ITPA', 'NKX6-3', 'BPHL', 'RFC2', 'XBP1', 'CYB5R3', 'CNDP2', 'UGCG', 'MMP3', 'SRR', 'TAPBP', 'TSHZ3', 'POLR3K', 'LMNB1', 'IL1R2', 'MTMR11', 'NPR3', 'TMEM229B', 'BSCL2', 'DAG1', 'SAT1', 'NXPH3', 'NEK4', 'RNF24', 'PMP22', 'CCKAR', 'HELLS', 'CSF1R', 'ANXA9', 'IL18', 'RNF166', 'SGO2', 'MARCHF10', 'CRYZL1', 'NRXN2', 'MXD4']
Refined Community 139: ['BMPR1B', 'ERGIC2', 'CPQ', 'PRPS2', 'ATF2', 'WNT7B', 'MPHOSPH8', 'PAICS', 'ADAMTS20', 'SMIM15', 'PCMTD2', 'DACT1', 'NOL4L', 'PHF6', 'CACNG6', 'SNRPB2', 'ODF1', 'ZNF22', 'XRCC5', 'BRD7', 'N4BP2', 'TTYH2', 'CCT5', 'FH', 'PIP4P2', 'CEP95', 'SFPQ', 'CDCA7', 'EIF4B', 'SERPINH1', 'SPRR3', 'C1orf131', 'GTPBP10', 'TADA2A', 'DLAT', 'TP63', 'ZNF875', 'IRX6', 'CLDN4', 'NSMCE4A', 'ITGA2', 'TMEM126B', 'ACYP1', 'LAMB2', 'INTS2', 'MSH6', 'HNRNPLL', 'CDCA8', 'RAC3', 'DYSF', 'NUP107', 'RPS3', 'AGFG1', 'CDC42SE1', 'LUZP1', 'WDR75', 'SMAD2', 'POLR3K', 'ILF2', 'USP1', 'MYH14', 'ACADS', 'THAP12', 'PLSCR4', 'APPBP2', 'S100A9', 'H6PD', 'CTTNBP2', 'STC1', 'MMP3', 'ARMC10', 'CCN4', 'CTNNAL1', 'DSG3', 'ACTR6', 'PBX2', 'PAFAH1B1', 'CMA1', 'CD24P2', 'PTGS2', 'PPP1R16A', 'BCOR', 'FGF2', 'LLGL1', 'SET', 'NFIA', 'TCL1A', 'NBL1', 'RPA3', 'PIGF', 'ATAD1', 'TEP1', 'ACTG1', 'PWP1', 'STK26', 'ZNF367', 'WIF1', 'LRBA', 'RFX3', 'SETX', 'YIPF6', 'PSAT1', 'ZBTB44', 'ACTL6A', 'POU2F1', 'NUP155', 'CYB561A3', 'DENND2D', 'MCL1', 'EID1', 'SSB', 'RDH10', 'BET1', 'GASK1B', 'GNG11', 'RGPD5', 'CGA', 'FAR2', 'EFCAB7', 'CRYZ', 'MOB4', 'PGA5', 'IGFBP3', 'NDUFA4', 'GALNS', 'PTEN', 'FAM76B', 'EYA3', 'NET1', 'FKBP2', 'ZEB2', 'CATSPERB', 'RNF7', 'MAN1A1', 'PLK3', 'GAK', 'TSPOAP1', 'PIGU', 'POLR1A', 'NUDCD1', 'PRELID3B', 'TMEM117', 'ZNF529', 'CD38', 'C9orf152', 'PRR16', 'EVL', 'RNF6', 'HMMR', 'PSMA7', 'LZTFL1', 'GIT1', 'ADK', 'NBN', 'SHMT1', 'ZNF24', 'CCR2', 'SPTLC2', 'RGS10', 'TULP3', 'RPE', 'C1orf74', 'AZIN1', 'EPB41', 'PAXBP1', 'MTHFD1', 'GFOD2', 'MCEE', 'AKAP9', 'ZNF136', 'WDR3', 'ZNF160', 'ZNF532', 'SYNCRIP', 'RIDA', 'COX11', 'RAP1BL', 'ZNF267', 'ECT2', 'SMARCA5', 'FKBP11', 'CD6', 'COX5B', 'COPG1', 'SNCAIP', 'TMEM64', 'SMCHD1', 'MME', 'SLC35E2B', 'TRPS1', 'TRAPPC10', 'SCUBE2', 'CNOT6', 'AP1S3', 'EDAR', 'PIAS1', 'ITPRID2', 'WEE1', 'NUP85', 'F8', 'NFYB', 'PLEKHF2', 'LIN7A', 'RAPGEFL1', 'UGDH', 'CSTF2', 'PCDHB3', 'EME2', 'TNIK', 'CTSG', 'MYNN', 'SLF1', 'FMO5', 'RNF44', 'TRDMT1', 'KITLG', 'DEPDC1', 'SBK1', 'GMFG', 'CDKL5', 'AURKB', 'BAK1', 'KCTD15', 'COMP', 'ARHGAP17', 'METAP2', 'IFNGR1', 'CD2AP', 'SERTAD4', 'CTH', 'H2AC6', 'SGO2', 'BBIP1', 'YIF1A', 'TSPAN2', 'BCL2', 'SHC2', 'CCR1', 'CENPF', 'SLC7A1', 'POLR2K', 'TOP2A', 'GRHL2', 'CLK4', 'AUTS2', 'ZNF770', 'TTF2', 'CHST1', 'TP53BP2', 'LMBRD2', 'NLK', 'ITGA6', 'ARL6IP6', 'CEP43', 'ZNF362', 'COL11A2', 'SMC4', 'MTCL2', 'BCLAF3', 'PVR', 'ADGRE2', 'ORC5', 'ZNF239', 'GSDMD', 'LGALSL', 'RELL1', 'VNN2', 'HSPA6', 'ARHGEF4', 'CXCL10', 'EDARADD', 'ATAD2', 'UBE2D2', 'DNAJA1', 'UBLCP1', 'TRIP13', 'CDK1', 'TAF2', 'IGF1R', 'ILF3', 'GFOD1', 'MIS18BP1', 'NPIPB3', 'ARSG', 'STAT3', 'AEBP2', 'PLEKHB1', 'SMURF2', 'CLDN12', 'RIOK3', 'PSPC1', 'RAD21', 'SUCLA2', 'UBA2', 'CA12', 'ARAP2', 'PTPN14', 'MEMO1', 'RASSF5', 'BCAP31', 'APOC1', 'CLDN3', 'CEP192', 'AFG2A', 'KBTBD8', 'CXXC4', 'PDP1', 'RPS6KB2', 'RPL15', 'LCORL', 'SSBP1', 'CYP3A5', 'TSTD1', 'IPO8', 'ZNF317', 'ECI2', 'KLF9', 'IGLL3P', 'ARMC1', 'PPP4R3B', 'OSBPL2', 'PPP1R18', 'PROM1', 'CLNS1A', 'TCEANC', 'PLIN2', 'UTP15', 'TLE4', 'CAVIN2', 'TRMT6', 'KIAA0232', 'PTK2', 'EPB41L2', 'MARCHF6', 'USP34', 'PPP2R1A', 'NUMA1', 'AOC3', 'ASXL1', 'TRPM3', 'NEO1', 'SLC52A2', 'IDH3A', 'POLR1C', 'PDS5B', 'CD24', 'RPA1', 'TIMM21', 'SKP1P1', 'TRRAP', 'CAPZA2', 'KLK8', 'PUS7L', 'SKA2', 'L3MBTL2', 'PIR', 'BUB1B', 'CTSC', 'FARSB', 'PIN1', 'APCDD1L', 'RPRD2', 'CCNDBP1', 'HGH1', 'KRT18', 'CAD', 'TRAF3', 'RPUSD1', 'STRBP', 'NANP', 'CEP55', 'SAP30', 'PATZ1', 'MPHOSPH10P1', 'SZT2', 'NFU1', 'TMEM216', 'DNAJC9', 'RNMT', 'POLR2B', 'MAGED2', 'DLK1', 'HMGA1', 'TIMM17A', 'KIF16B', 'HMGB2', 'TASP1', 'SLC39A11', 'TTC9C', 'HNRNPA1', 'WDR11', 'POLA1', 'CPSF6', 'PCDH18', 'EMID1', 'ZRANB2', 'RFC2', 'PDK1', 'UNG', 'PRODH', 'SGO1', 'CENPX', 'ANKRD11', 'F3', 'CALCRL', 'RPS13', 'CDC25A', 'SEMA3C', 'CD4', 'PRR14L', 'HCLS1', 'ZNF703', 'CAV2', 'CPSF7', 'GPM6B', 'LRRC40', 'IGF2BP3', 'RBBP6', 'NSUN5P1', 'PTPN2', 'STARD7', 'DDR1', 'MAGEA4', 'RGP1', 'FAM199X', 'CASP3', 'EPS8', 'ANP32E', 'H2AZ1', 'TFRC', 'TROAP', 'UBA3', 'ZCCHC8']
""" 
    
    detailed_df = run_optimized_analysis(
        input_text=input_text,
        api_key=os.environ.get("DEEPSEEK_API_KEY"), 
        detailed_csv="Breast Cancer Annotation partial.csv"
    )
    
    pd.set_option('display.max_colwidth', None)
    print("\nPathway Analysis Results (Clean Text):")

    main_columns = ['Community', 'Process_With_Enrichment', 'Confidence_With_Enrichment', 
                   'Process_Without_Enrichment', 'Confidence_Without_Enrichment', 
                   'Final_Process', 'Final_Confidence']
    available_main_columns = [col for col in main_columns if col in detailed_df.columns]
    print(detailed_df[available_main_columns])
    
    print(f"\nTotal communities analyzed: {len(detailed_df)}")
    print(f"Available columns: {list(detailed_df.columns)}")
    
    if 'Contributing_Genes_With_Enrichment' in detailed_df.columns:
        print(f"\nSample contributing genes (cleaned):")
        for idx, row in detailed_df.head(3).iterrows():
            print(f"Community {row['Community']}: {row.get('Contributing_Genes_With_Enrichment', 'N/A')}")
```

### Clean Process Name Columns — AML Annotation

Removes trailing embedded confidence-score artifacts from `Process_With_Enrichment` and
`Process_Without_Enrichment`, e.g.:

- `Name ([0.42])`
- `Name (0.68)`
- `Name (confidence score0.65)`
- `Name (confidence score: 0.65)`

Genuine parenthetical content in the middle of a name (e.g. `... (SREBP) Signaling`) is left alone,
since the pattern only matches a trailing parenthetical block that contains a number.

```python
import re
import pandas as pd

COLUMNS_TO_CLEAN = [
    "Process_With_Enrichment",
    "Process_Without_Enrichment",
]

SCORE_SUFFIX_PATTERN = re.compile(
    r"""
    \s*
    \(+
    \s*
    (?:confidence\s*score\s*:?\s*)?
    \[?\s*
    (-?\d*\.\d+|-?\d+)
    \s*\]?
    \s*
    \)+
    \s*$
    """,
    re.IGNORECASE | re.VERBOSE,
)


def clean_process_name(value):
    if not isinstance(value, str):
        return value

    return SCORE_SUFFIX_PATTERN.sub("", value).strip()


def clean_annotation_file(input_path: str, output_path: str):
    df = pd.read_csv(input_path)
    df.columns = [c.strip() for c in df.columns]

    for col in COLUMNS_TO_CLEAN:
        if col not in df.columns:
            print(f"Skipping '{col}' - column not found")
            continue

        df[col] = df[col].apply(clean_process_name)
        print(f"{col}: cleaned")

    df.to_csv(output_path, index=False)
    print(f"Saved cleaned file to {output_path}")

    return df



INPUT_CSV = "Breast Cancer Annotation partial.csv"
OUTPUT_CSV = "Breast Cancer Annotation partial.csv"

df = clean_annotation_file(INPUT_CSV, OUTPUT_CSV)
df.head()
```

### Validation Pipeline

```python
import pandas as pd
import json
import os
import re
import time
import requests
from typing import List, Dict, Any, Optional, Tuple
from tqdm import tqdm

UNKNOWN_PROCESS_LABEL = "unknown process"

def _is_unknown(process_name: str) -> bool:
    """Return True if the process name is a variant of 'unknown process'."""
    return process_name.strip().lower() == UNKNOWN_PROCESS_LABEL


class GeneSetValidator:
    def __init__(
        self,
        gene_sets_path: str,
        papers_csv: str = "AML_Paper_DB.csv",
        api_key: Optional[str] = None,
        output_path: str = "validated_gene_sets.csv",
    ):
        self.gene_sets_path = gene_sets_path
        self.papers_csv     = papers_csv
        self.output_path    = output_path
        self.api_key        = api_key or os.environ.get("DEEPSEEK_API_KEY")

        if not self.api_key:
            print("Warning: No DeepSeek API key provided.")

        self.gene_sets_df      = None
        self.papers_df         = None
        self.validated_results = []

        # ── Load Clarivate JIF journal list ───────────────────────────────────
        try:
            jif_df = pd.read_csv("journals_filtered_JIF_ge_4.csv", sep=",")
            jif_df.columns = [c.strip() for c in jif_df.columns]

            self._hq_issn_set  = set()
            self._hq_eissn_set = set()
            self.TOP_JOURNALS  = []

            for _, jr in jif_df.iterrows():
                issn  = str(jr.get("ISSN",  "")).strip().replace("-", "").upper()
                eissn = str(jr.get("eISSN", "")).strip().replace("-", "").upper()
                name  = str(jr.get("Journal name", "")).strip()

                if issn  and issn  != "NAN": self._hq_issn_set.add(issn)
                if eissn and eissn != "NAN": self._hq_eissn_set.add(eissn)
                if name:                     self.TOP_JOURNALS.append(name)

            self._top_journal_set_normalized = {
                self.normalize_journal(j) for j in self.TOP_JOURNALS
            }

            print(f"Loaded {len(jif_df)} high-quality journals from Clarivate JIF CSV "
                  f"({len(self._hq_issn_set)} ISSNs, {len(self._hq_eissn_set)} eISSNs)")

        except FileNotFoundError:
            print("Warning: journals_filtered_JIF_ge_4.csv not found. No quality filter applied.")
            self._hq_issn_set                = set()
            self._hq_eissn_set               = set()
            self.TOP_JOURNALS                = []
            self._top_journal_set_normalized = set()


    def normalize_journal(self, name) -> str:
        if pd.isna(name):
            return ""
        name = str(name).lower()
        name = name.split(" : ")[0].split(" - ")[0]
        name = re.sub(r"[^\w\s]", "", name)
        name = re.sub(r"\s+", " ", name)
        return name.strip()

    def _normalize_id(self, val) -> str:
        return str(val).strip().replace("-", "").upper() if pd.notna(val) else ""

    def _is_high_quality(self, row) -> bool:
        """Match in priority order: ISSN -> eISSN -> exact normalized journal name only."""
        issn = self._normalize_id(row.get("issn"))
        if issn and issn in self._hq_issn_set:
            return True
        eissn = self._normalize_id(row.get("eissn"))
        if eissn and eissn in self._hq_eissn_set:
            return True
        norm = self.normalize_journal(row.get("journal", ""))
        if norm and norm in self._top_journal_set_normalized:
            return True
        return False

    def _load_papers_for_genes(self, genes: List[str]) -> pd.DataFrame:
        if self.papers_df is None or self.papers_df.empty:
            return pd.DataFrame()

        genes_upper = {g.upper() for g in genes if g}
        mask   = self.papers_df["gene"].str.upper().isin(genes_upper)
        subset = self.papers_df[mask].copy()

        if subset.empty:
            return subset

        if "full_text" not in subset.columns:
            subset["full_text"] = "Full text not available via PMC"

        subset["matched_journal"] = subset.apply(
            lambda row: "HQ" if self._is_high_quality(row) else None, axis=1
        )

        hq  = subset["matched_journal"].notna().sum()
        tot = len(subset)
        print(f"  Paper filter: {hq}/{tot} high-quality papers for gene set")
        return subset


    def load_data(self):
        print("Loading gene sets data...")
        self.gene_sets_df = pd.read_csv(self.gene_sets_path)
        self.gene_sets_df.columns = [col.strip() for col in self.gene_sets_df.columns]
        print(f"Loaded {len(self.gene_sets_df)} gene sets from {self.gene_sets_path}")

        print(f"Loading AML paper database from {self.papers_csv}...")
        if not os.path.exists(self.papers_csv):
            print(f"Paper CSV '{self.papers_csv}' not found.")
            self.papers_df = pd.DataFrame()
        else:
            self.papers_df = pd.read_csv(self.papers_csv, low_memory=False)
            self.papers_df.columns = [col.strip() for col in self.papers_df.columns]
            if "gene" not in self.papers_df.columns:
                print("'gene' column not found in paper CSV - gene filtering disabled.")
                self.papers_df = pd.DataFrame()
            else:
                self.papers_df["gene"] = self.papers_df["gene"].astype(str).str.strip()
                print(f"Loaded {len(self.papers_df)} papers ({self.papers_df['gene'].nunique()} unique genes)")


    def extract_genes_from_set(self, gene_set_row: pd.Series) -> List[str]:
        genes = []
        possible_gene_columns = [
            "Genes_String", "Contributing_Genes", "Genes", "Final_Contributing_Genes",
        ]
        for col in possible_gene_columns:
            if col in gene_set_row and pd.notna(gene_set_row[col]):
                gene_data = gene_set_row[col]
                if isinstance(gene_data, list):
                    genes = gene_data
                elif isinstance(gene_data, str):
                    if "," in gene_data:
                        genes = [g.strip() for g in gene_data.split(",")]
                    elif ";" in gene_data:
                        genes = [g.strip() for g in gene_data.split(";")]
                    else:
                        genes = [gene_data.strip()]
                break
        return [g for g in genes if g and g.strip()]


    def extract_gene_related_abstracts(
        self,
        genes: List[str],
        papers_df: pd.DataFrame,
        max_papers: int = 50,
    ) -> Tuple[List[Dict], List[str]]:
        """
        Filter to HQ papers only, then prioritise by gene_count descending.

        gene_count = number of query genes found in the COMBINED
        abstract + full_text corpus (never short-circuited by the gene column).
        The gene column is used only as a safety net to ensure the primary
        gene is never missed if the text search fails.

        """
        if isinstance(genes, str):
            genes = [g.strip() for g in genes.split(",")]

        genes_set = {g.upper() for g in genes if g}

        hq_papers = []

        for _, paper in papers_df.iterrows():

            # ── HQ gate ───────────────────────────────────────────────────────
            if pd.isna(paper.get("matched_journal")):
                continue

            abstract  = str(paper.get("abstract",  ""))
            full_text = str(paper.get("full_text", ""))

            if abstract  in ("nan", "Abstract not available", ""):
                abstract = ""
            if full_text in ("nan", "Full text not available via PMC", ""):
                full_text = ""

            if not abstract and not full_text:
                continue

            # ── Always scan abstract + full_text combined corpus ──────────────
            search_corpus = (abstract + " " + full_text).strip()
            mentioned_genes = [
                g for g in genes
                if g and re.search(rf"\b{re.escape(g)}\b", search_corpus, re.IGNORECASE)
            ]
            gene_from_col = str(paper.get("gene", "")).strip()
            if gene_from_col and gene_from_col.upper() in genes_set:
                if gene_from_col not in mentioned_genes:
                    mentioned_genes.append(gene_from_col)

            if not mentioned_genes:
                continue

            truncated_abstract  = (abstract[:500]  + "...") if len(abstract)  > 500  else abstract
            truncated_full_text = (full_text[:3000] + "...") if len(full_text) > 3000 else full_text

            hq_papers.append({
                "title":           paper.get("title",   "No title"),
                "authors":         paper.get("authors", "No authors"),
                "journal":         paper.get("journal", "No journal"),
                "year":            paper.get("year",    "Unknown"),
                "abstract":        truncated_abstract,
                "full_text":       truncated_full_text,
                "doi":             paper.get("doi",     "No DOI"),
                "pmid":            paper.get("pmid",    "No PMID"),
                "genes_mentioned": mentioned_genes,
                "gene_count":      len(mentioned_genes),
            })

        hq_papers.sort(key=lambda x: x["gene_count"], reverse=True)
        relevant_papers = hq_papers[:max_papers]

        top_journal_papers_used = [
            f"Paper {idx}: '{p['title']}', {p['authors']}"
            for idx, p in enumerate(relevant_papers, 1)
        ]

        return relevant_papers, top_journal_papers_used

    def create_validation_prompt(
        self,
        gene_set_id: str,
        genes: List[str],
        process_with_enrichment: str,
        confidence_with_enrichment: float,
        process_without_enrichment: str,
        confidence_without_enrichment: float,
        analysis_with_enrichment: str,
        analysis_without_enrichment: str,
        relevant_papers: List[Dict],
    ) -> str:

        genes_str      = ", ".join(genes) if isinstance(genes, list) else str(genes)
        papers_section = ""
        for i, paper in enumerate(relevant_papers, 1):
            papers_section += (
                f"PAPER {i}:\n"
                f"Title: {paper['title']}\n"
                f"Authors: {paper['authors']}\n"
                f"Journal: {paper['journal']} ({paper['year']})\n"
                f"Genes Mentioned: {', '.join(paper['genes_mentioned'])}\n"
                f"Abstract: {paper['abstract']}\n"
                f"\n"
            )

        return f"""You are a scientific expert in genomics and bioinformatics tasked with validating gene set analysis results using ONLY the provided literature.

GENE SET ID: {gene_set_id}
GENES: {genes_str}

ORIGINAL ANALYSIS WITH ENRICHMENT:
Process Name: {process_with_enrichment}
Original Confidence Score: {confidence_with_enrichment}
Analysis: {analysis_with_enrichment[:1000]}...

ORIGINAL ANALYSIS WITHOUT ENRICHMENT:
Process Name: {process_without_enrichment}
Original Confidence Score: {confidence_without_enrichment}
Analysis: {analysis_without_enrichment[:1000]}...

PROVIDED LITERATURE (USE ONLY THESE STUDIES):
{papers_section}

VALIDATION TASK:
Based STRICTLY on the provided literature above, evaluate both analyses and provide updated confidence scores.

**ABSOLUTE REQUIREMENT FOR PAPER CITATIONS:**
Whenever you reference a paper anywhere in your response, always use this exact format:
"Paper X (FirstAuthorLastName et al.)"
For example: "Paper 3 (Nakamura et al.) shows..." or "supported by Paper 1 (Chen et al.) and Paper 4 (Okafor et al.)"
Do NOT use bare numbers like "Paper 3" alone, and do NOT spell out full titles or complete author lists in-line — the first-author-et-al form is sufficient everywhere in the response.

CRITICAL REQUIREMENTS:
1. Use ONLY the provided papers - do not add external knowledge
2. For each process, check if the genes are supported by the literature
3. Provide updated confidence scores based on evidence strength
4. Select the better-supported process as the final choice
5. **MANDATORY:  When referencing papers by number anywhere in your analysis text, always use the "Paper X (FirstAuthor et al.)" format defined above — never bare numbers, never full citations..**
6. The final selected process MUST be either:
   (a) exactly one of the two original process names, OR
   (b) "Neither process" ONLY if both updated confidence scores (Updated Confidence With Enrichment, Updated Confidence Without Enrichment) are <= 0.05.

7. The final confidence MUST follow:
   - If a process is selected: Final Confidence = the updated confidence of that selected process.
   - If "Neither process" is selected: Final Confidence = max(Updated Confidence With Enrichment, Updated Confidence Without Enrichment).

FORMAT YOUR RESPONSE EXACTLY AS FOLLOWS:

VALIDATION OF ENRICHMENT ANALYSIS:
Evidence Assessment: [Detailed assessment based strictly on provided papers]
Original Confidence: {confidence_with_enrichment}
Updated Confidence: [Your revised score 0.00-1.00 based on literature evidence]
Supporting Papers: [List paper numbers that support this process]

VALIDATION OF DIRECT ANALYSIS:
Evidence Assessment: [Detailed assessment based strictly on provided papers]
Original Confidence: {confidence_without_enrichment}
Updated Confidence: [Your revised score 0.00-1.00 based on literature evidence]
Supporting Papers: [List paper numbers that support this process]

FINAL PROCESS SELECTION:
Selected Process: [Choose the better-supported process name]
Final Confidence: [The updated confidence score for your selected process]
Selection Reasoning: [Explain why this process has stronger literature support]

CONFLICT ANALYSIS:
Review all papers for contradictory evidence about gene functions, pathway assignments, or experimental results.
CONFLICTING_EVIDENCE_FOUND: [TRUE/FALSE]
CONFLICT_DESCRIPTION: [Brief description of any conflicts found, or "No conflicts detected"]

SUPPORTING CITATIONS:
[List citations in format: "Title, Authors" for papers that support the final selected process]

VALIDATION ANALYSIS TEXT:
[Comprehensive summary of all changes made, reasoning for confidence adjustments, and evidence from the provided studies that led to the final process selection.
Use "Paper X (FirstAuthor et al.)" format for any paper references.]"""

    def call_deepseek_api(self, prompt: str) -> str:
        if not self.api_key:
            raise ValueError("DeepSeek API key is required.")

        api_url = "https://api.deepseek.com/v1/chat/completions"
        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type":  "application/json",
        }
        data = {
            "model": "deepseek-chat",
            "messages": [
                {
                    "role":    "system",
                    "content": (
                        "You are a scientific expert in genomics and bioinformatics "
                        "specializing in gene set analysis validation. "
                        "Base your analysis strictly on the provided literature."
                    ),
                },
                {"role": "user", "content": prompt},
            ],
            "temperature": 0,
            "max_tokens":  8000,
        }

        max_retries = 3
        retry_delay = 5

        for attempt in range(max_retries):
            try:
                response = requests.post(api_url, headers=headers, json=data, timeout=180)

                if response.status_code == 200:
                    return response.json()["choices"][0]["message"]["content"]

                elif response.status_code == 429:
                    wait_time = int(response.headers.get("Retry-After", retry_delay * 2))
                    print(f"Rate limited. Waiting {wait_time}s...")
                    time.sleep(wait_time)

                elif 500 <= response.status_code < 600:
                    print(f"Server error {response.status_code}. Retrying in {retry_delay}s...")
                    time.sleep(retry_delay)
                    retry_delay *= 2

                else:
                    raise Exception(f"API Error {response.status_code}: {response.text}")

            except requests.exceptions.Timeout:
                print(f"Timeout on attempt {attempt + 1}/{max_retries}")
                if attempt < max_retries - 1:
                    time.sleep(retry_delay)
                    retry_delay *= 2
                else:
                    raise Exception(f"API timeout after {max_retries} attempts")

            except requests.exceptions.RequestException as e:
                if attempt < max_retries - 1:
                    time.sleep(retry_delay)
                    retry_delay *= 2
                else:
                    raise Exception(f"Request failed after {max_retries} attempts: {e}")

        raise Exception(f"Failed to get a valid response after {max_retries} attempts")

    def parse_llm_response(self, response: str) -> Dict[str, Any]:
        results = {}

        m = re.search(
            r"VALIDATION OF ENRICHMENT ANALYSIS:.*?Updated Confidence:\s*([\d.]+)",
            response, re.DOTALL,
        )
        if m:
            results["confidence_with_enrichment_after"] = float(m.group(1))

        m = re.search(
            r"VALIDATION OF DIRECT ANALYSIS:.*?Updated Confidence:\s*([\d.]+)",
            response, re.DOTALL,
        )
        if m:
            results["confidence_without_enrichment_after"] = float(m.group(1))

        m = re.search(r"Selected Process:\s*(.+?)(?=\n|$)", response)
        if m:
            results["final_process"] = m.group(1).strip()

        m = re.search(r"Final Confidence:\s*([\d.]+)", response)
        if m:
            results["final_confidence"] = float(m.group(1))

        m = re.search(r"CONFLICTING_EVIDENCE_FOUND:\s*(TRUE|FALSE)", response, re.IGNORECASE)
        results["conflicting_evidence_found"] = (
            m.group(1).upper() == "TRUE" if m else False
        )

        m = re.search(
            r"CONFLICT_DESCRIPTION:\s*(.+?)(?=\n\n|SUPPORTING CITATIONS:|$)",
            response, re.DOTALL,
        )
        results["conflict_description"] = m.group(1).strip() if m else "No conflicts detected"

        m = re.search(
            r"SUPPORTING CITATIONS:\s*(.+?)(?=\n\n|VALIDATION ANALYSIS TEXT:|$)",
            response, re.DOTALL,
        )
        if m:
            results["supporting_citations"] = [
                line.strip()
                for line in m.group(1).strip().split("\n")
                if line.strip() and not line.strip().startswith(("-", "*"))
            ]
        else:
            results["supporting_citations"] = []

        m = re.search(r"VALIDATION ANALYSIS TEXT:\s*(.+?)$", response, re.DOTALL)
        if m:
            results["validation_analysis_text"] = m.group(1).strip()

        return results

    def validate_gene_set(self, gene_set_row: pd.Series) -> Dict[str, Any]:
        set_id = str(
            gene_set_row.get("Community", "")
            or gene_set_row.get("Set_ID", "")
            or gene_set_row.get("community", "")
            or gene_set_row.get("index", "")
        )

        genes = self.extract_genes_from_set(gene_set_row)
        if not genes:
            print(f"  No genes found for set {set_id}")
            return self._error_result(set_id, "No genes found for this set")

        process_with    = str(gene_set_row.get("Process_With_Enrichment", "")
                              or gene_set_row.get("Final_Process", ""))
        conf_with       = float(gene_set_row.get("Confidence_With_Enrichment", 0)
                                or gene_set_row.get("Final_Confidence", 0))
        process_without = str(gene_set_row.get("Process_Without_Enrichment", ""))
        conf_without    = float(gene_set_row.get("Confidence_Without_Enrichment", 0))

        analysis_with    = str(gene_set_row.get("Analysis_Text_With_Enrichment", "")
                               or gene_set_row.get("Pathway_Reasoning_With_Enrichment", "")
                               or gene_set_row.get("Full_Analysis", ""))
        analysis_without = str(gene_set_row.get("Analysis_Text_Without_Enrichment", "")
                               or gene_set_row.get("Pathway_Reasoning_Without_Enrichment", ""))

        # ── Unknown process guard ─────────────────────────────────────────────
        with_is_unknown    = _is_unknown(process_with)
        without_is_unknown = _is_unknown(process_without)

        if with_is_unknown and without_is_unknown:
            print(f"  Both processes are 'unknown process' for set {set_id} - skipping.")
            return self._both_unknown_result(set_id, genes, gene_set_row)

        if with_is_unknown:
            print(f"  'With enrichment' is unknown process for set {set_id} - zeroing out, validating direct only.")
            process_with  = UNKNOWN_PROCESS_LABEL
            conf_with     = 0.0
            analysis_with = "Unknown process - not validated."

        if without_is_unknown:
            print(f"  'Without enrichment' is unknown process for set {set_id} - zeroing out, validating enrichment only.")
            process_without  = UNKNOWN_PROCESS_LABEL
            conf_without     = 0.0
            analysis_without = "Unknown process - not validated."

        papers_df = self._load_papers_for_genes(genes)

        relevant_papers, top_journal_info = self.extract_gene_related_abstracts(
            genes, papers_df, max_papers=50
        )

        if not relevant_papers:
            print(f"  No matching high-quality papers for set {set_id}")
            return self._no_papers_result(set_id, genes, gene_set_row)

        prompt = self.create_validation_prompt(
            gene_set_id=set_id,
            genes=genes,
            process_with_enrichment=process_with,
            confidence_with_enrichment=conf_with,
            process_without_enrichment=process_without,
            confidence_without_enrichment=conf_without,
            analysis_with_enrichment=analysis_with,
            analysis_without_enrichment=analysis_without,
            relevant_papers=relevant_papers,
        )

        os.makedirs("prompts", exist_ok=True)
        with open(f"prompts/prompt_{set_id}.txt", "w", encoding="utf-8") as f:
            f.write(prompt)

        try:
            response = self.call_deepseek_api(prompt)

            os.makedirs("responses", exist_ok=True)
            with open(f"responses/response_{set_id}.txt", "w", encoding="utf-8") as f:
                f.write(response)

            parsed = self.parse_llm_response(response)

            conf_with_after    = parsed.get("confidence_with_enrichment_after",   conf_with)
            conf_without_after = parsed.get("confidence_without_enrichment_after", conf_without)
            if with_is_unknown:
                conf_with_after = 0.0
            if without_is_unknown:
                conf_without_after = 0.0

            return {
                "Set_ID":                                set_id,
                "Genes":                                 genes,
                "Process_With_Enrichment_Original":      process_with,
                "Process_Without_Enrichment_Original":   process_without,
                "Confidence_With_Enrichment_Before":     conf_with,
                "Confidence_Without_Enrichment_Before":  conf_without,
                "Confidence_With_Enrichment_After":      conf_with_after,
                "Confidence_Without_Enrichment_After":   conf_without_after,
                "Final_Process":                         parsed.get("final_process",   process_with),
                "Final_Confidence":                      parsed.get("final_confidence", conf_with),
                "Validation_Analysis_Text":              parsed.get("validation_analysis_text", "No analysis provided"),
                "Supporting_Citations":                  parsed.get("supporting_citations", []),
                "Conflicting_Evidence_Found":            parsed.get("conflicting_evidence_found", False),
                "Conflict_Description":                  parsed.get("conflict_description", "No conflicts detected"),
                "Total_Papers_Found":                    len(relevant_papers),
            }

        except Exception as e:
            print(f"  API error for set {set_id}: {e}")
            return self._error_result(set_id, f"API error: {e}", genes, gene_set_row)

    def validate_all_gene_sets(self):
        if self.gene_sets_df is None:
            self.load_data()

        results    = []
        total_sets = len(self.gene_sets_df)
        print(f"\nStarting validation of {total_sets} gene sets...")

        pbar = tqdm(
            total=total_sets,
            desc="Validating gene sets",
            bar_format="{desc}: {percentage:3.0f}%|{bar}| {n_fmt}/{total_fmt} [{elapsed}<{remaining}]",
            ncols=100,
        )

        for i, (_, row) in enumerate(self.gene_sets_df.iterrows()):
            set_id = str(
                row.get("Community", "")
                or row.get("Set_ID", "")
                or row.get("community", "")
                or i
            )
            try:
                result = self.validate_gene_set(row)
                results.append(result)
            except Exception as e:
                print(f"\nUnexpected error on set {set_id}: {e}")

            pbar.update(1)
            pbar.set_postfix_str(f"Set {set_id}")

            if (i + 1) % 5 == 0:
                self.validated_results = results
                self.save_results(f"{self.output_path}.partial")

        pbar.close()
        self.validated_results = results
        self.save_results(self.output_path)
        print(f"\nValidation complete! Processed {len(results)} gene sets")
        return results

    def save_results(self, output_path: str = None):
        if not self.validated_results:
            print("No results to save")
            return

        path = output_path or self.output_path
        df   = pd.DataFrame(self.validated_results)

        required_columns = [
            "Set_ID", "Genes",
            "Process_With_Enrichment_Original", "Process_Without_Enrichment_Original",
            "Confidence_With_Enrichment_Before", "Confidence_Without_Enrichment_Before",
            "Confidence_With_Enrichment_After",  "Confidence_Without_Enrichment_After",
            "Final_Process", "Final_Confidence",
            "Validation_Analysis_Text", "Supporting_Citations",
            "Conflicting_Evidence_Found", "Conflict_Description",
            "Total_Papers_Found",
        ]
        for col in required_columns:
            if col not in df.columns:
                df[col] = None
        df = df[required_columns]
        df.to_csv(path, index=False)
        print(f"Saved {len(df)} validated results to {path}")


    def generate_summary_report(self) -> str:
        if not self.validated_results:
            return "No validation results available"

        total = len(self.validated_results)

        conflict_count     = sum(1 for r in self.validated_results if r.get("Conflicting_Evidence_Found"))
        total_paper_counts = [r.get("Total_Papers_Found", 0)           for r in self.validated_results]
        final_confidences  = [r.get("Final_Confidence", 0)             for r in self.validated_results]

        conf_delta_enrich = []
        conf_delta_direct = []
        for r in self.validated_results:
            be, ae = r.get("Confidence_With_Enrichment_Before", 0),   r.get("Confidence_With_Enrichment_After", 0)
            bd, ad = r.get("Confidence_Without_Enrichment_Before", 0), r.get("Confidence_Without_Enrichment_After", 0)
            if be > 0 and ae > 0: conf_delta_enrich.append(ae - be)
            if bd > 0 and ad > 0: conf_delta_direct.append(ad - bd)

        lines = [
            "Gene Set Validation Summary Report",
            "=" * 50,
            f"Total gene sets processed: {total}",
            "",
            "CONFLICT ANALYSIS:",
            f"  With conflicts:    {conflict_count} ({conflict_count/total*100:.1f}%)",
            f"  Without conflicts: {total - conflict_count} ({(total-conflict_count)/total*100:.1f}%)",
            "",
            "PAPER DISCOVERY:",
            f"  Avg papers/set:  {sum(total_paper_counts)/len(total_paper_counts):.2f}",
            f"  Sets with papers: {sum(1 for c in total_paper_counts if c > 0)} ({sum(1 for c in total_paper_counts if c > 0)/total*100:.1f}%)",
            "",
            "CONFIDENCE:",
            f"  Avg final confidence: {sum(final_confidences)/len(final_confidences):.2f}",
            f"  Range: {min(final_confidences):.2f} - {max(final_confidences):.2f}",
        ]
        if conf_delta_enrich:
            lines.append(f"  Avg enrichment confidence delta: {sum(conf_delta_enrich)/len(conf_delta_enrich):+.3f}")
        if conf_delta_direct:
            lines.append(f"  Avg direct confidence delta:     {sum(conf_delta_direct)/len(conf_delta_direct):+.3f}")

        return "\n".join(lines)


    def _error_result(
        self,
        set_id: str,
        error_msg: str,
        genes: List[str] = None,
        row: pd.Series = None,
    ) -> Dict[str, Any]:
        genes = genes or []
        return {
            "Set_ID":                                set_id,
            "Genes":                                 genes,
            "Process_With_Enrichment_Original":      row.get("Process_With_Enrichment", "") if row is not None else "",
            "Process_Without_Enrichment_Original":   row.get("Process_Without_Enrichment", "") if row is not None else "",
            "Confidence_With_Enrichment_Before":     row.get("Confidence_With_Enrichment", 0) if row is not None else 0,
            "Confidence_Without_Enrichment_Before":  row.get("Confidence_Without_Enrichment", 0) if row is not None else 0,
            "Confidence_With_Enrichment_After":      0,
            "Confidence_Without_Enrichment_After":   0,
            "Final_Process":                         "ERROR",
            "Final_Confidence":                      0,
            "Validation_Analysis_Text":              error_msg,
            "Supporting_Citations":                  [],
            "Conflicting_Evidence_Found":            False,
            "Conflict_Description":                  "Error occurred during validation",
            "Total_Papers_Found":                    0,
        }

    def _no_papers_result(
        self, set_id: str, genes: List[str], row: pd.Series
    ) -> Dict[str, Any]:
        conf_with = float(row.get("Confidence_With_Enrichment", 0) or row.get("Final_Confidence", 0))
        return {
            "Set_ID":                                set_id,
            "Genes":                                 genes,
            "Process_With_Enrichment_Original":      str(row.get("Process_With_Enrichment", "") or row.get("Final_Process", "")),
            "Process_Without_Enrichment_Original":   str(row.get("Process_Without_Enrichment", "")),
            "Confidence_With_Enrichment_Before":     conf_with,
            "Confidence_Without_Enrichment_Before":  float(row.get("Confidence_Without_Enrichment", 0)),
            "Confidence_With_Enrichment_After":      conf_with,
            "Confidence_Without_Enrichment_After":   float(row.get("Confidence_Without_Enrichment", 0)),
            "Final_Process":                         str(row.get("Process_With_Enrichment", "") or row.get("Final_Process", "")),
            "Final_Confidence":                      conf_with,
            "Validation_Analysis_Text":              "No high-quality papers found for this gene set",
            "Supporting_Citations":                  [],
            "Conflicting_Evidence_Found":            False,
            "Conflict_Description":                  "No high-quality papers available",
            "Total_Papers_Found":                    0,
        }

    def _both_unknown_result(
        self, set_id: str, genes: List[str], row: pd.Series
    ) -> Dict[str, Any]:
        return {
            "Set_ID":                                set_id,
            "Genes":                                 genes,
            "Process_With_Enrichment_Original":      UNKNOWN_PROCESS_LABEL,
            "Process_Without_Enrichment_Original":   UNKNOWN_PROCESS_LABEL,
            "Confidence_With_Enrichment_Before":     0,
            "Confidence_Without_Enrichment_Before":  0,
            "Confidence_With_Enrichment_After":      0,
            "Confidence_Without_Enrichment_After":   0,
            "Final_Process":                         UNKNOWN_PROCESS_LABEL,
            "Final_Confidence":                      0,
            "Validation_Analysis_Text":              "Both processes are unknown - validation skipped.",
            "Supporting_Citations":                  [],
            "Conflicting_Evidence_Found":            False,
            "Conflict_Description":                  "Both processes unknown - no validation performed",
            "Total_Papers_Found":                    0,
        }



def main():
    gene_sets_path = "Breast Cancer Annotation partial.csv"
    papers_csv     = "BC_Paper_DB.csv"
    output_path    = "Breast Cancer Annotation partial.csv"

    print("BC Gene Set Validation Pipeline")
    print("=" * 50)

    api_key = os.environ.get("DEEPSEEK_API_KEY")
    if not api_key:
        print("No DEEPSEEK_API_KEY env var found.")
        print("Set it with:  export DEEPSEEK_API_KEY=your_key")
        return

    print("API key found")

    validator = GeneSetValidator(
        gene_sets_path=gene_sets_path,
        papers_csv=papers_csv,
        api_key=api_key,
        output_path=output_path,
    )

    validator.load_data()

    test_mode = False

    if test_mode:
        print("\nTest mode - validating first gene set only")
        result = validator.validate_gene_set(validator.gene_sets_df.iloc[0])
        validator.validated_results = [result]
        validator.save_results("test_validation.csv")
        for k, v in result.items():
            print(f"  {k}: {len(v) if isinstance(v, list) else v}")
    else:
        validator.validate_all_gene_sets()

    report = validator.generate_summary_report()
    print("\n" + report)
    with open("validation_summary.txt", "w") as f:
        f.write(report)

    print("\nDone!")


if __name__ == "__main__":
    main()
```
