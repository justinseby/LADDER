```python
import pandas as pd
```

```python
# Load file
df = pd.read_csv("ALL_cell_lines_diff_proteins_upregulated.txt", sep="\t")
gene_sets = {}
summary = []

for drug in df["drug"].unique():

    sub = df[df["drug"] == drug]
    filtered = sub[sub["pval"] < 0.01]

    up = filtered[filtered["m_diff"] > 0]["protein"].tolist()
    if len(up) > 200:
        filtered = sub[sub["adjpval"] < 0.01]
        up = filtered[filtered["m_diff"] > 0]["protein"].tolist()
    if len(up) >= 10:
        gene_sets[f"{drug}_UP"] = up

        summary.append({
            "drug": drug,
            "n_up": len(up)
        })
summary_df = pd.DataFrame(summary)
```

```python
gene_sets_df = pd.DataFrame([
    {
        "gene_set": name,
        "genes": ", ".join([f"'{g}'" for g in genes])
    }
    for name, genes in gene_sets.items()
])

gene_sets_df.to_csv("drug_gene_sets.csv", index=False)
```
